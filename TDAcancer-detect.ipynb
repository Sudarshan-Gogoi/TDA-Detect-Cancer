```python
"""Optimized Biomarker Selection
=============================

Biomarker Selection Module

TDA-based biomarker-panel selection using a four-stage constrained random search.
Methodological overview
-----------------------
1. Creates an updated workbook with four mutually exclusive data roles:
   - Reference: the first configured number of normal patients.
   - Regression Fit Calibration 1: a configured fraction of the remaining
     normal patients; this partition is used to fit the regression model and
     calculate the 1D reference coordinates.
   - Biomarker-Panel Calibration and Selection: configured fractions of the
     remaining normal and cancer cohorts; this partition is used for panel and
     threshold selection.
   - Test: the remaining patients, held out until final evaluation.
2. Evaluates exactly TOTAL_PANEL_BUDGET unique biomarker panels, each containing
   PANEL_SIZE biomarkers, distributed across four search stages.
3. Stage 1 performs balanced, low-overlap exploration across the biomarker pool.
4. Stage 2 samples panels using biomarker recurrence weights derived from the
   strongest specificity-feasible panels evaluated so far.
5. Stage 3 performs balanced exploration within a pool of recurring biomarkers.
6. Stage 4 explores one-biomarker substitutions around high-performing panels.
7. Selects the final panel from all evaluated stages, subject to the configured
   pooled Calibration 1 + Calibration 2 normal-specificity constraint.
8. For each panel, computes TDA features from a Vietoris-Rips filtration built
   from a correlation-derived dissimilarity matrix, D = 1 - |C|. The regression
   model and 1D reference coordinates are fitted/calculated using Calibration 1
   normal patients only.
9. Searches candidate thresholds for Regression Difference Percentage and
   Distance %(1) using an OR decision rule on Calibration 2. Among feasible
   threshold pairs, the search maximizes macro-sensitivity across the cancer
   cohorts represented in the workbook.
10. Evaluates the selected panel and frozen thresholds on the held-out Test
    partition after panel selection is complete.

TDA features calculated
-----------------------
- 0-Dim Hole Min Life Range
- 0-Dim Hole Max Life Range
- Median_Death_0D
- Std_Death_0D
- Avg_Death_0D
- Average Birth (1D)
- Average Death (1D)

The script does not calculate or export 0D/1D hole counts or the 1D maximum
lifetime-range feature.

Dependencies
------------
Install the third-party packages from a terminal or notebook cell before
running this script (skip packages that are already installed):

    python -m pip install --upgrade numpy pandas openpyxl scikit-learn gudhi

Python standard-library modules (pathlib, math, and random) do not require
installation. The script reads an Excel workbook and writes Excel/CSV outputs.

Configuration notes
-------------------
- Set the input/output paths, worksheet name, biomarker names, cohort labels,
  and column names in the USER-DEFINED SETTINGS section.
- Cancer cohorts are detected from the Cohort column; the normal cohort is
  identified by NORMAL_COHORT.
- The log transformation requires selected biomarker values to be finite and
  strictly positive for the reference and evaluated patients.
- GUDHI's RipsComplex constructor names its input keyword `distance_matrix`.
  The values supplied to that API argument in this script are a symmetric,
  nonnegative dissimilarity matrix; no triangle-inequality guarantee is assumed.
  The API keyword is retained for compatibility.

Reproducibility
---------------
The random seed, panel-search budget, phase weights, split fractions, and
threshold-search settings are defined in the USER-DEFINED SETTINGS section.
"""

from pathlib import Path
import math
import random

import gudhi
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression


# =============================================================================
# USER-DEFINED SETTINGS
# Change these values when applying the function to another dataset.
# Algorithmic tuning parameters below this section retain their defaults.
# =============================================================================

# Input/output files
clinical_data_path = Path(
    r"D:\Cancer Detection folder\Biomarker Selection\clinical cancer validation data.xlsx"
)
sheet_name = "Normal and Cancer"
output_folder_path = Path(r"D:\Cancer Detection folder\Biomarker Selection")

# Dataset-specific biomarker universe (names must match workbook column names).
BIOMARKERS = [
    "Tenascin C", "sAXL", "sE-selectin", "sHGFR/c-Met", "sHer2", "sHer3",
    "sIL-6Ra", "sNeuropilin-1", "sPECAM-1", "sVEGFR1", "sVEGFR3",
    "sc-kit/SCFR", "CA-125", "CA 15-3", "CA 19-9", "CEA", "FGF2", "HE4",
    "HGF", "IL-6", "IL-8", "Leptin", "MIF", "OPN", "Prolactin", "SCF",
    "TNFa", "TRAIL", "Total PSA", "VEGF", "b-HCG", "sFAS", "Cathepsin D",
    "FAP alpha", "Ferritin", "Galectin-3", "IGFBP3", "MIA", "MPO", "SHBG",
    "TIMP1", "TIMP2"
]

# Cohort names and their display/sort order.
# Keep the normal cohort in this mapping; all other keys are treated as cancer cohorts.
NORMAL_COHORT = "normal"
cohort_order = {"normal": 0, "bladder": 1, "ovarian": 2, "pancreatic": 3}
CANCER_TYPES = [name.casefold() for name in cohort_order if name.casefold() != NORMAL_COHORT.casefold()]

# Dataset-specific column names.
COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
STAGE_COL = "AJCC Stage"
DATA_TYPE_COL = "Data Type"

# Data split and model constraints.
# Edit the two fractions below whenever needed. They are fractions, not whole percentages:
#   0.20 = 20%, 0.30 = 30%, 0.50 = 50%.
# For NORMAL patients only, the first N_REFERENCE_NORMAL rows are reserved as Reference;
# the configured Calibration 1 and Calibration 2 fractions are applied to the remaining
# normal patients, with Test receiving the remainder.
# CANCER patients are NEVER assigned to Calibration 1: CALIBRATION2_FRACTION of each
# cancer cohort is assigned to Biomarker-Panel Calibration and Selection, and the rest
# is Test. Thus Calibration 1 contains only healthy/normal patients.
N_REFERENCE_NORMAL = 59
CALIBRATION1_FRACTION = 0.15
CALIBRATION2_FRACTION = 0.50
FPR = 0.01  # Require pooled Calibration 1 + 2 normal specificity strictly > 99.00%
SENSITIVITY_TARGET = 0.99  # Diagnostic target only; specificity is the hard constraint
PANEL_SIZE = 8
TOTAL_PANEL_BUDGET = 3000
RANDOM_SEED = 20261017

# Maximum number of candidate thresholds to test for EACH detection score:
# - One set of candidates is generated for Regression Difference Percentage (k1).
# - One set is generated for Distance %(1) (k2).
# - The threshold search evaluates up to MAX_THRESHOLD_CANDIDATES x
#   MAX_THRESHOLD_CANDIDATES candidate pairs for the OR decision rule.
# - Lower values generally run faster but explore fewer threshold combinations;
#   higher values search more combinations and may take longer.
MAX_THRESHOLD_CANDIDATES = 80

# OPTIONAL USER-DEFINED THRESHOLD SEARCH BOUNDS (inclusive)
# Leave ALL FOUR values as None to use the original unbounded candidate search.
# To restrict the search, provide all four numeric values.
# Example bounded search: K1_MIN=-1.0, K1_MAX=5.0, K2_MIN=0.0, K2_MAX=10.0
K1_MIN = None  # Regression Difference Percentage lower bound, or None
K1_MAX = None  # Regression Difference Percentage upper bound, or None
K2_MIN = None  # Distance %(1) lower bound, or None
K2_MAX = None  # Distance %(1) upper bound, or None

# Output names (saved inside output_folder_path).
UPDATED_WORKBOOK_PATH = clinical_data_path.with_name(
    "clinical cancer validation data_Updated.xlsx"
)
FEATURE_VECTOR_PATH = output_folder_path / "FeatureVectorCancerDet.csv"
SEARCH_RESULTS_PATH = output_folder_path / "Optimized_Biomarker_Selection_Results.csv"
SELECTED_PANEL_SUMMARY_CSV_PATH = (
    output_folder_path / "SelectedPanel_Calibration2_Summary.csv"
)
RECURRENCE_TABLE_PATH = output_folder_path / "Biomarker_Frequency_Top_Feasible_Panels.csv"


# Algorithm defaults: normally leave these unchanged unless tuning the search.
PHASE_WEIGHTS = {
    "balanced_diverse_exploration": 0.40,
    "sensitivity_guided_search": 0.30,
    "recurrence_informed_balanced_search": 0.20,
    "best_panel_neighborhoods": 0.10,
}
CANDIDATE_POOL_SIZE = 300
TOP_PANEL_FRACTION = 0.20
RECURRENCE_POOL_SIZE = 16
PRINT_EACH_PANEL_PERFORMANCE = False


def calculate_phase_budgets(total_budget, phase_weights):
    if not isinstance(total_budget, int) or total_budget < 1:
        raise ValueError("TOTAL_PANEL_BUDGET must be a positive integer.")
    weight_sum = sum(phase_weights.values())
    if not math.isclose(weight_sum, 1.0, rel_tol=0.0, abs_tol=1e-9):
        raise ValueError("PHASE_WEIGHTS must sum to 1.0.")
    exact = {name: total_budget * weight for name, weight in phase_weights.items()}
    budgets = {name: math.floor(value) for name, value in exact.items()}
    remaining = total_budget - sum(budgets.values())
    order = sorted(
        phase_weights,
        key=lambda name: (exact[name] - budgets[name], phase_weights[name]),
        reverse=True,
    )
    for name in order[:remaining]:
        budgets[name] += 1
    return budgets


PHASE_BUDGETS = calculate_phase_budgets(TOTAL_PANEL_BUDGET, PHASE_WEIGHTS)

# Original linear-regression feature set and TDA features are retained.
REGRESSION_PREDICTORS = [
    "0-Dim Hole Min Life Range",
    "0-Dim Hole Max Life Range",
    "Median_Death_0D",
    "Std_Death_0D",
]
FEATURE_COLUMNS = [
    "0-Dim Hole Min Life Range",
    "0-Dim Hole Max Life Range",
    "Median_Death_0D",
    "Std_Death_0D",
    "Avg_Death_0D",
    "Average Birth (1D)",
    "Average Death (1D)",
]


# -------------------- WORKBOOK / SPLIT --------------------


def validate_split_fractions():
    """Validate configurable calibration fractions; Test is the remainder."""
    for name, value in (("CALIBRATION1_FRACTION", CALIBRATION1_FRACTION),
                        ("CALIBRATION2_FRACTION", CALIBRATION2_FRACTION)):
        if not isinstance(value, (int, float)) or not np.isfinite(value):
            raise ValueError(f"{name} must be a finite number between 0 and 1.")
        if value < 0 or value > 1:
            raise ValueError(f"{name} must be between 0 and 1; received {value}.")
    total_calibration = CALIBRATION1_FRACTION + CALIBRATION2_FRACTION
    if total_calibration >= 1.0:
        raise ValueError(
            "CALIBRATION1_FRACTION + CALIBRATION2_FRACTION must be less than 1.0 "
            "so that Test has a positive fraction. "
            f"Received {CALIBRATION1_FRACTION} + {CALIBRATION2_FRACTION}."
        )


def create_updated_workbook(source_path, updated_path, sheet):
    """Create reproducible splits and detect cancer cohorts from the source dataset."""
    global CANCER_TYPES
    validate_split_fractions()
    source_path, updated_path = Path(source_path), Path(updated_path)
    if not source_path.exists():
        raise FileNotFoundError(f"Source workbook not found: {source_path}\nChange clinical_data_path.")
    sheets = pd.read_excel(source_path, sheet_name=None)
    if sheet not in sheets:
        raise ValueError(f"Sheet {sheet!r} not found. Available: {list(sheets)}")
    df = sheets[sheet].copy()
    required = [COHORT_COL, PATIENT_ID_COL, STAGE_COL] + BIOMARKERS
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns in source workbook: {missing}")
    if DATA_TYPE_COL in df.columns:
        df = df.drop(columns=[DATA_TYPE_COL])

    cohort = df[COHORT_COL].astype(str).str.strip().str.casefold()
    if df[COHORT_COL].isna().any():
        raise ValueError(f"Column {COHORT_COL!r} contains missing cohort labels.")
    if cohort.eq("").any():
        raise ValueError(f"Column {COHORT_COL!r} contains blank cohort labels.")
    observed_cohorts = list(dict.fromkeys(cohort.tolist()))
    CANCER_TYPES = [name for name in observed_cohorts if name and name != NORMAL_COHORT.casefold()]
    if not CANCER_TYPES:
        raise ValueError(
            f"No cancer cohorts were detected in {COHORT_COL!r}; expected at least one cohort besides {NORMAL_COHORT!r}."
        )
    print("Automatically detected cancer types:", ", ".join(CANCER_TYPES))
    normal_indices = df.index[cohort.eq(NORMAL_COHORT.casefold())].tolist()
    if len(normal_indices) <= N_REFERENCE_NORMAL:
        raise ValueError(f"Need more than {N_REFERENCE_NORMAL} normal patients; found {len(normal_indices)}.")

    role = {}
    reference_indices = normal_indices[:N_REFERENCE_NORMAL]
    for idx in reference_indices:
        role[idx] = "Reference"

    rng = np.random.RandomState(RANDOM_SEED)
    counts = {"reference_normal": len(reference_indices)}

    # Remaining normal patients: configurable Calibration 1 and Calibration 2
    # percentages; Test receives all remaining patients. Reference normals are
    # excluded from the percentage calculations.
    remaining_normals = list(normal_indices[N_REFERENCE_NORMAL:])
    rng.shuffle(remaining_normals)
    n1 = int(np.floor(len(remaining_normals) * CALIBRATION1_FRACTION))
    n2 = int(np.floor(len(remaining_normals) * CALIBRATION2_FRACTION))
    for idx in remaining_normals[:n1]:
        role[idx] = "Regression Fit Calibration 1"
    for idx in remaining_normals[n1:n1+n2]:
        role[idx] = "Biomarker-Panel Calibration and Selection"
    for idx in remaining_normals[n1+n2:]:
        role[idx] = "Test"
    counts.update({
        "normal_regression_fit_calibration1": n1,
        "normal_calibration2": n2,
        "normal_test": len(remaining_normals) - n1 - n2,
    })

    # Cancer cohorts are split ONLY between Calibration 2 and Test.
    # Cancer patients must never be labelled Regression Fit Calibration 1.
    for cancer in CANCER_TYPES:
        indices = df.index[cohort.eq(cancer.casefold())].tolist()
        if not indices:
            raise ValueError(f"No rows found for cancer cohort {cancer!r}.")
        shuffled = list(indices)
        rng.shuffle(shuffled)
        n2c = int(np.floor(len(shuffled) * CALIBRATION2_FRACTION))
        for idx in shuffled[:n2c]:
            role[idx] = "Biomarker-Panel Calibration and Selection"
        for idx in shuffled[n2c:]:
            role[idx] = "Test"
        counts[f"{cancer}_calibration2"] = n2c
        counts[f"{cancer}_test"] = len(shuffled) - n2c

    df[DATA_TYPE_COL] = pd.Series(role)
    role_order = {
        "Reference": 0,
        "Regression Fit Calibration 1": 1,
        "Biomarker-Panel Calibration and Selection": 2,
        "Test": 3,
    }
    df["_source_order"] = np.arange(len(df))
    df["_role_order"] = df[DATA_TYPE_COL].map(role_order)
    df["_cohort_order"] = (
        df[COHORT_COL].astype(str).str.strip().str.casefold()
        .map({str(k).casefold(): v for k, v in cohort_order.items()}).fillna(99)
    )
    df = df.sort_values(
        ["_role_order", "_cohort_order", "_source_order"], kind="stable"
    ).drop(columns=["_source_order", "_role_order", "_cohort_order"]).reset_index(drop=True)
    sheets[sheet] = df
    updated_path.parent.mkdir(parents=True, exist_ok=True)
    with pd.ExcelWriter(updated_path, engine="openpyxl") as writer:
        for sheet_name_out, sheet_df in sheets.items():
            sheet_df.to_excel(writer, sheet_name=sheet_name_out, index=False)

    test_fraction = 1.0 - CALIBRATION1_FRACTION - CALIBRATION2_FRACTION
    print(f"Updated workbook created: {updated_path}")
    print(
        "Normal split after the separately reserved Reference group: "
        f"{CALIBRATION1_FRACTION:.1%} Calibration 1 (normal patients only) / "
        f"{CALIBRATION2_FRACTION:.1%} Calibration 2 / {test_fraction:.1%} Test."
    )
    print(
        "Cancer split: "
        f"{CALIBRATION2_FRACTION:.1%} Calibration 2 / "
        f"{1.0 - CALIBRATION2_FRACTION:.1%} Test; cancer patients are never assigned to Calibration 1."
    )
    print(
        f"The first {N_REFERENCE_NORMAL} normal patients are separately labelled Reference; "
        "the percentages for normal patients apply only after those reference rows are removed."
    )
    print("\nData Type counts:")
    print(df[DATA_TYPE_COL].value_counts().to_string())
    print("\nCohort x Data Type counts:")
    print(pd.crosstab(df[COHORT_COL], df[DATA_TYPE_COL]).to_string())
    return updated_path, counts


def calculate_tda_features(reference_df, patient_row, panel):
    """Calculate the retained 0D and 1D persistence features for one patient."""
    panel = list(panel)
    reference_values = reference_df[panel].to_numpy(dtype=float, copy=False)
    patient_values = patient_row[panel].to_numpy(dtype=float, copy=False).reshape(1, -1)
    values = np.vstack((reference_values, patient_values))

    patient_id = patient_row.get(PATIENT_ID_COL, "unknown")
    if not np.isfinite(values).all():
        raise ValueError(
            f"Non-finite selected biomarker value for patient {patient_id}."
        )
    if (values <= 0).any():
        raise ValueError(
            "Log transformation requires strictly positive values. "
            f"Check selected biomarkers for patient {patient_id}."
        )

    logged_values = np.log(values)
    correlation_matrix = np.corrcoef(logged_values, rowvar=False)
    dissimilarity_matrix = 1.0 - np.abs(correlation_matrix)

    if not np.isfinite(dissimilarity_matrix).all():
        raise ValueError(
            "Invalid correlation-derived dissimilarity matrix for patient "
            f"{patient_id}."
        )

    # Enforce numerical symmetry, a zero diagonal, and nonnegative values.
    dissimilarity_matrix = (
        dissimilarity_matrix + dissimilarity_matrix.T
    ) / 2.0
    np.fill_diagonal(dissimilarity_matrix, 0.0)
    dissimilarity_matrix = np.clip(dissimilarity_matrix, 0.0, 1.0)

    # GUDHI's public API calls this argument `distance_matrix`; the supplied
    # array is the correlation-derived dissimilarity matrix defined above.
    rips_complex = gudhi.RipsComplex(distance_matrix=dissimilarity_matrix)
    simplex_tree = rips_complex.create_simplex_tree(max_dimension=2)
    persistence = simplex_tree.persistence()

    finite_pairs = [
        (dimension, (birth, death))
        for dimension, (birth, death) in persistence
        if np.isfinite(death)
    ]
    one_dimensional_pairs = [
        pair for dimension, pair in finite_pairs if dimension == 1
    ]
    zero_dimensional_pairs = [
        pair for dimension, pair in finite_pairs if dimension == 0
    ]

    # 1D birth/death means are retained for the reference-coordinate score.
    # No 1D hole count or 1D maximum lifetime-range feature is calculated.
    if one_dimensional_pairs:
        average_birth_1d = float(
            np.mean([birth for birth, _ in one_dimensional_pairs])
        )
        average_death_1d = float(
            np.mean([death for _, death in one_dimensional_pairs])
        )
    else:
        average_birth_1d = 0.0
        average_death_1d = 0.0

    # Retain the 0D lifetime-range and death-distribution features used by the
    # existing regression model and feature analysis. No 0D hole count is computed.
    if zero_dimensional_pairs:
        deaths_0d = [death for _, death in zero_dimensional_pairs]
        lifetimes_0d = [death - birth for birth, death in zero_dimensional_pairs]
        min_lifetime_0d = float(min(lifetimes_0d))
        max_lifetime_0d = float(max(lifetimes_0d))
        average_death_0d = float(np.mean(deaths_0d))
        median_death_0d = float(np.median(deaths_0d))
        std_death_0d = float(np.std(deaths_0d))
    else:
        min_lifetime_0d = 0.0
        max_lifetime_0d = 0.0
        average_death_0d = 0.0
        median_death_0d = 0.0
        std_death_0d = 0.0

    return {
        "0-Dim Hole Min Life Range": min_lifetime_0d,
        "0-Dim Hole Max Life Range": max_lifetime_0d,
        "Median_Death_0D": median_death_0d,
        "Std_Death_0D": std_death_0d,
        "Avg_Death_0D": average_death_0d,
        "Average Birth (1D)": average_birth_1d,
        "Average Death (1D)": average_death_1d,
    }


def get_role_mask(df, role_name):
    return df[DATA_TYPE_COL].astype(str).str.strip().str.casefold().eq(
        role_name.casefold()
    )



def calculate_panel_features(df, reference_df, panel, include_test=False):
    """Extract features only for the requested development roles."""
    allowed = (
        get_role_mask(df, "Regression Fit Calibration 1")
        | get_role_mask(df, "Biomarker-Panel Calibration and Selection")
    )
    if include_test:
        allowed = allowed | get_role_mask(df, "Test")
    rows = []
    for idx, patient in df.loc[allowed].iterrows():
        record = {
            "Source_Row_Index": idx,
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
            COHORT_COL: patient[COHORT_COL],
            STAGE_COL: patient.get(STAGE_COL, np.nan),
            DATA_TYPE_COL: patient[DATA_TYPE_COL],
        }
        record.update(calculate_tda_features(reference_df, patient, panel))
        rows.append(record)
    return pd.DataFrame(rows)



def evaluate_panel(df, reference_df, panel, save_features=False):
    """
    Fit regression and thresholds on Calibration 1 NORMALS only.
    Evaluate the frozen rule on the independent Calibration 2 partition.
    """
    features = calculate_panel_features(df, reference_df, panel, include_test=False)
    if features.empty:
        raise ValueError("No Calibration 1 / Calibration 2 patients were available.")

    role = features[DATA_TYPE_COL].astype(str).str.strip().str.casefold()
    cohort = features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal1_normals = features[
        role.eq("regression fit calibration 1")
        & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()
    cal2 = features[
        role.eq("biomarker-panel calibration and selection")
    ].copy()
    if len(cal1_normals) < 2:
        raise ValueError("At least two Calibration 1 normal patients are required.")
    if cal2.empty:
        raise ValueError("Calibration 2 is empty.")

    model = LinearRegression().fit(
        cal1_normals[REGRESSION_PREDICTORS],
        cal1_normals["Avg_Death_0D"],
    )
    features["Predicted Avg_Death_0D"] = model.predict(features[REGRESSION_PREDICTORS])
    features["Difference"] = features["Avg_Death_0D"] - features["Predicted Avg_Death_0D"]
    denom = features["Avg_Death_0D"].abs().replace(0, np.nan)
    features["Regression Difference Percentage"] = (
        features["Difference"].abs() / denom
    ) * 100.0

    ref_birth = float(cal1_normals["Average Birth (1D)"].mean())
    ref_death = float(cal1_normals["Average Death (1D)"].mean())
    features["Distance %(1)"] = np.sqrt(
        (features["Average Birth (1D)"] - ref_birth) ** 2
        + (features["Average Death (1D)"] - ref_death) ** 2
    ) * 100.0

    # IMPORTANT: cal1_normals was copied before the derived columns were added.
    # Refresh it now so it contains Regression Difference Percentage and Distance %(1).
    cal1_normals = features.loc[
        role.eq("regression fit calibration 1")
        & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()

    # ------------------------------------------------------------------
    # REDUCED, VECTORIZED exhaustive threshold-pair search for the OR rule.
    #
    # Thresholds are still optimized over candidate values, rather than set
    # directly to the pooled-normal maxima. To reduce runtime, the candidate
    # grid is deterministically downsampled to at most MAX_THRESHOLD_CANDIDATES
    # values per score. Specificity is constrained on the pooled Calibration 1
    # + Calibration 2 normal patients, strictly above 99.00%.
    # ------------------------------------------------------------------
    cal2_mask = role.eq("biomarker-panel calibration and selection")
    cal2_features = features.loc[cal2_mask].copy()
    cal2_cohort = cal2_features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal2_normals = cal2_features[cal2_cohort.eq(NORMAL_COHORT.casefold())]
    cal2_cancers = cal2_features[cal2_cohort.isin(CANCER_TYPES)]

    if cal2_normals.empty:
        raise ValueError("Calibration 2 contains no normal patients for specificity calibration.")
    if cal2_cancers.empty:
        raise ValueError("Calibration 2 contains no cancer patients for threshold selection.")

    cal1_reg_norm = cal1_normals["Regression Difference Percentage"].to_numpy(dtype=float)
    cal1_dist_norm = cal1_normals["Distance %(1)"].to_numpy(dtype=float)
    cal2_reg_norm = cal2_normals["Regression Difference Percentage"].to_numpy(dtype=float)
    cal2_dist_norm = cal2_normals["Distance %(1)"].to_numpy(dtype=float)
    pooled_reg_norm = np.concatenate([cal1_reg_norm, cal2_reg_norm])
    pooled_dist_norm = np.concatenate([cal1_dist_norm, cal2_dist_norm])

    cancer_reg = cal2_cancers["Regression Difference Percentage"].to_numpy(dtype=float)
    cancer_dist = cal2_cancers["Distance %(1)"].to_numpy(dtype=float)
    cancer_cohort_values = (
        cal2_cancers[COHORT_COL].astype(str).str.strip().str.casefold().to_numpy()
    )
    specificity_target = 1.0 - FPR

    def reduced_boundary_candidates(
        normal_scores, cancer_scores, max_candidates, min_value=None, max_value=None, name="threshold"
    ):
        """Build original score-boundary candidates, optionally restricting their range."""
        combined = np.concatenate([
            np.asarray(normal_scores, dtype=float),
            np.asarray(cancer_scores, dtype=float),
        ])
        combined = combined[np.isfinite(combined)]
        if combined.size == 0:
            raise ValueError("No finite scores available for threshold search.")
        if not isinstance(max_candidates, int) or max_candidates < 2:
            raise ValueError("MAX_THRESHOLD_CANDIDATES must be an integer >= 2.")

        # This is the original candidate-generation procedure.
        unique_values = np.unique(combined)
        boundaries = np.unique(np.concatenate([
            unique_values, np.nextafter(unique_values, np.inf)
        ]))

        # Apply bounds only when a complete min/max pair is provided.
        bounded = min_value is not None and max_value is not None
        if bounded:
            if not np.isfinite(min_value) or not np.isfinite(max_value):
                raise ValueError(f"{name} bounds must be finite numbers.")
            if min_value > max_value:
                raise ValueError(
                    f"{name}_MIN ({min_value}) must be <= {name}_MAX ({max_value})."
                )
            boundaries = boundaries[
                (boundaries >= float(min_value)) & (boundaries <= float(max_value))
            ]
            # Include exact bounds, even when no observed score lies there.
            boundaries = np.unique(np.concatenate([
                boundaries, np.array([float(min_value), float(max_value)])
            ]))

        if boundaries.size > max_candidates:
            # Keep the original deterministic downsampling behavior.
            indices = np.linspace(0, boundaries.size - 1, max_candidates, dtype=int)
            boundaries = boundaries[np.unique(indices)]
        return boundaries

    # Bounds are active only when all four are supplied. If all are None, this
    # exactly follows the original code's unbounded threshold candidate search.
    bound_values = (K1_MIN, K1_MAX, K2_MIN, K2_MAX)
    bounds_provided = [value is not None for value in bound_values]
    if any(bounds_provided) and not all(bounds_provided):
        raise ValueError(
            "Provide all four bounds (K1_MIN, K1_MAX, K2_MIN, K2_MAX), or leave all four as None."
        )
    USE_THRESHOLD_BOUNDS = all(bounds_provided)

    k1_candidates = reduced_boundary_candidates(
        pooled_reg_norm, cancer_reg, MAX_THRESHOLD_CANDIDATES,
        K1_MIN if USE_THRESHOLD_BOUNDS else None,
        K1_MAX if USE_THRESHOLD_BOUNDS else None, "K1"
    )
    k2_candidates = reduced_boundary_candidates(
        pooled_dist_norm, cancer_dist, MAX_THRESHOLD_CANDIDATES,
        K2_MIN if USE_THRESHOLD_BOUNDS else None,
        K2_MAX if USE_THRESHOLD_BOUNDS else None, "K2"
    )

    # Evaluate each k1 against all k2 values in vectorized NumPy operations.
    n_normals = len(pooled_reg_norm)
    best_pair = None
    best_key = None
    best_macro_sensitivity = -1.0
    best_overall_sensitivity = -1.0
    best_specificity = -1.0
    best_fp = None
    best_tn = None

    cancer_reg_pos = cancer_reg[:, None] > k1_candidates[None, :]
    cancer_dist_pos = cancer_dist[:, None] > k2_candidates[None, :]
    active_cal2_cancers = [
        cancer for cancer in CANCER_TYPES
        if np.any(cancer_cohort_values == cancer.casefold())
    ]
    per_type_masks = {
        cancer: cancer_cohort_values == cancer.casefold()
        for cancer in active_cal2_cancers
    }

    for i, k1 in enumerate(k1_candidates):
        normal_positive = (
            (pooled_reg_norm > k1)[:, None]
            | (pooled_dist_norm[:, None] > k2_candidates[None, :])
        )
        fp_counts = normal_positive.sum(axis=0)
        specificities = 1.0 - fp_counts / n_normals
        feasible = np.flatnonzero(specificities > specificity_target + 1e-12)
        if feasible.size == 0:
            continue

        cancer_positive_by_k2 = cancer_reg_pos[:, i, None] | cancer_dist_pos[:, feasible]
        overall_sens = cancer_positive_by_k2.mean(axis=0)
        type_sens = []
        for cancer in active_cal2_cancers:
            mask = per_type_masks[cancer]
            type_sens.append(cancer_positive_by_k2[mask].mean(axis=0))
        if not type_sens:
            continue
        macro_sens = np.mean(np.vstack(type_sens), axis=0)

        # Lexicographic objective matches original intent: macro sensitivity,
        # overall sensitivity, specificity, then larger thresholds.
        for j_local, j in enumerate(feasible):
            candidate_key = (
                float(macro_sens[j_local]),
                float(overall_sens[j_local]),
                float(specificities[j]),
                float(k1),
                float(k2_candidates[j]),
            )
            if best_key is None or candidate_key > best_key:
                best_key = candidate_key
                best_pair = (float(k1), float(k2_candidates[j]))
                best_macro_sensitivity = candidate_key[0]
                best_overall_sensitivity = candidate_key[1]
                best_specificity = candidate_key[2]
                best_fp = int(fp_counts[j])
                best_tn = int(n_normals - best_fp)

    if best_pair is None:
        if USE_THRESHOLD_BOUNDS:
            raise RuntimeError(
                "No k1/k2 pair within the user-defined bounds satisfied the pooled "
                f"Calibration 1 + Calibration 2 specificity requirement (> {specificity_target:.2%}). "
                f"Current bounds: k1=[{K1_MIN}, {K1_MAX}], k2=[{K2_MIN}, {K2_MAX}]. "
                "Adjust the bounds or review the specificity target."
            )
        # Preserve the original unbounded fallback: thresholds just above the
        # maximum pooled normal scores guarantee zero false positives.
        finite_reg = pooled_reg_norm[np.isfinite(pooled_reg_norm)]
        finite_dist = pooled_dist_norm[np.isfinite(pooled_dist_norm)]
        if finite_reg.size == 0 or finite_dist.size == 0:
            raise ValueError("Pooled calibration normal scores contain no finite values.")
        k1_fallback = float(np.nextafter(np.max(finite_reg), np.inf))
        k2_fallback = float(np.nextafter(np.max(finite_dist), np.inf))
        best_pair = (k1_fallback, k2_fallback)
        fallback_positive = (cancer_reg > k1_fallback) | (cancer_dist > k2_fallback)
        sens_by_type = [
            float(fallback_positive[per_type_masks[cancer]].mean())
            for cancer in CANCER_TYPES if per_type_masks[cancer].any()
        ]
        best_macro_sensitivity = (
            float(np.mean(sens_by_type)) if len(sens_by_type) == len(CANCER_TYPES) else float("nan")
        )
        best_overall_sensitivity = float(fallback_positive.mean()) if fallback_positive.size else float("nan")
        best_fp, best_tn = 0, n_normals
        best_specificity = 1.0
    best_threshold, distance_threshold = best_pair

    # Report each partition's specificity descriptively, but only the pooled
    # specificity is a threshold-selection constraint.
    cal1_positive_final = (
        (cal1_normals["Regression Difference Percentage"] > best_threshold)
        | (cal1_normals["Distance %(1)"] > distance_threshold)
    )
    best_cal1_specificity = float((~cal1_positive_final).mean())

    features["Result(Cancer)"] = np.where(
        (features["Regression Difference Percentage"] > best_threshold)
        | (features["Distance %(1)"] > distance_threshold),
        "Positive", "Negative"
    )
    cal1_normal_positive_final = (
        (cal1_normals["Regression Difference Percentage"] > best_threshold)
        | (cal1_normals["Distance %(1)"] > distance_threshold)
    )
    cal2_normal_positive = (
        (cal2_normals["Regression Difference Percentage"] > best_threshold)
        | (cal2_normals["Distance %(1)"] > distance_threshold)
    )
    combined_normal_positive_final = np.concatenate([
        cal1_normal_positive_final.to_numpy(dtype=bool),
        cal2_normal_positive.to_numpy(dtype=bool),
    ])
    tn = int((~combined_normal_positive_final).sum())
    fp = int(combined_normal_positive_final.sum())
    specificity_combined = (
        tn / (tn + fp) if (tn + fp) else float("nan")
    )
    specificity_cal2 = (
        float((~cal2_normal_positive).mean()) if len(cal2_normal_positive) else float("nan")
    )
    if not specificity_combined > specificity_target + 1e-12:
        raise RuntimeError(
            "Selected thresholds failed the pooled Calibration 1 + Calibration 2 "
            "specificity constraint."
        )

    cancer_results = {}
    sensitivities = []
    all_predictions = (
        (cal2_cancers["Regression Difference Percentage"] > best_threshold)
        | (cal2_cancers["Distance %(1)"] > distance_threshold)
    )
    for cancer in CANCER_TYPES:
        subset = cal2_cancers[
            cal2_cancers[COHORT_COL].astype(str).str.strip().str.casefold().eq(cancer)
        ]
        pred = (
            (subset["Regression Difference Percentage"] > best_threshold)
            | (subset["Distance %(1)"] > distance_threshold)
        )
        sens = float(pred.mean()) if len(subset) else float("nan")
        if len(subset):
            sensitivities.append(sens)
        cancer_results[cancer] = {
            "n": int(len(subset)),
            "sensitivity": sens,
            "true_positives": int(pred.sum()),
        }
    macro_sensitivity = float(np.mean(sensitivities)) if sensitivities else float("nan")
    overall_sensitivity = float(all_predictions.mean()) if len(cal2_cancers) else float("nan")

    if save_features:
        export_features = features.copy()
        export_features.insert(
            0, "Unnamed: 0",
            export_features["Source_Row_Index"].map(lambda value: f"P_{int(value)}")
        )
        requested_columns = [
            "Unnamed: 0", PATIENT_ID_COL, COHORT_COL, STAGE_COL, DATA_TYPE_COL,
            *FEATURE_COLUMNS, "Predicted Avg_Death_0D", "Difference",
            "Regression Difference Percentage", "Distance %(1)", "Result(Cancer)",
        ]
        export_features[[c for c in requested_columns if c in export_features.columns]].to_csv(
            FEATURE_VECTOR_PATH, index=False
        )

    return {
        "n_biomarkers": len(panel),
        "panel": list(panel),
        "regression_threshold": float(best_threshold),
        "distance_threshold": float(distance_threshold),
        "threshold_bounds_enabled": bool(USE_THRESHOLD_BOUNDS),
        "k1_min_search_bound": float(K1_MIN) if USE_THRESHOLD_BOUNDS else None,
        "k1_max_search_bound": float(K1_MAX) if USE_THRESHOLD_BOUNDS else None,
        "k2_min_search_bound": float(K2_MIN) if USE_THRESHOLD_BOUNDS else None,
        "k2_max_search_bound": float(K2_MAX) if USE_THRESHOLD_BOUNDS else None,
        "k1_candidate_count": int(len(k1_candidates)),
        "k2_candidate_count": int(len(k2_candidates)),
        "threshold_pairs_considered_max": int(len(k1_candidates) * len(k2_candidates)),
        "calibration1_specificity": float(best_cal1_specificity),
        "calibration2_specificity": float(specificity_cal2),
        "specificity": float(specificity_combined),
        "combined_calibration_specificity": float(specificity_combined),
        "false_positives": fp,
        "true_negatives": tn,
        "macro_sensitivity": macro_sensitivity,
        "overall_cancer_sensitivity": overall_sensitivity,
        "cancer_results": cancer_results,
        "regression_fit_normal_n": int(len(cal1_normals)),
        "calibration2_normal_n": int(len(cal2_normals)),
        "calibration_cancer_n": int(len(cal2_cancers)),
        "meets_sensitivity_target": bool(
            np.isfinite(macro_sensitivity) and macro_sensitivity > SENSITIVITY_TARGET
        ),
        "threshold_rule": (
            "Positive iff Regression Difference Percentage > regression_threshold "
            "OR Distance %(1) > distance_threshold"
        ),
    }, features


def canonical_panel(panel):
    return tuple(sorted(panel, key=BIOMARKERS.index))


def panel_key(panel):
    return tuple(panel)


def random_panel(rng):
    return canonical_panel(rng.sample(BIOMARKERS, PANEL_SIZE))


def balanced_diverse_panels(rng, count, used, marker_pool=None, prior_counts=None):
    """Generate unique panels while balancing marker appearances and reducing overlap.

    In Stage 1, marker_pool is all 42 biomarkers and prior_counts starts at zero,
    so all markers have approximately equal expected representation. In Stage 3,
    the pool is restricted to recurring markers found among the strongest feasible
    panels, and representation is balanced within that pool.
    """
    pool = list(marker_pool or BIOMARKERS)
    if len(pool) < PANEL_SIZE:
        pool = list(BIOMARKERS)
    counts = {b: int((prior_counts or {}).get(b, 0)) for b in pool}
    selected = []
    selected_set = set(used)
    max_attempts = max(100000, count * 1000)

    while len(selected) < count:
        candidates = set()
        attempts = 0
        while len(candidates) < CANDIDATE_POOL_SIZE and attempts < max_attempts:
            attempts += 1
            candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            if candidate not in selected_set:
                candidates.add(candidate)
        if not candidates:
            # This is only expected if the candidate space is nearly exhausted.
            candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            if candidate in selected_set:
                # Fall back to the full marker universe to maintain the exact budget.
                pool = list(BIOMARKERS)
                candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
                while candidate in selected_set:
                    candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            candidates.add(candidate)

        def balance_diversity_key(candidate):
            projected = dict(counts)
            for b in candidate:
                if b in projected:
                    projected[b] += 1
            values = list(projected.values())
            # First balance total appearances; then minimize overlap with earlier
            # panels from this stage; randomized final tie-break avoids ordering bias.
            imbalance = float(np.std(values)) if values else 0.0
            max_count = max(values) if values else 0
            if selected:
                overlaps = [len(set(candidate) & set(old)) for old in selected]
                mean_overlap = float(np.mean(overlaps))
                max_overlap = max(overlaps)
            else:
                mean_overlap = max_overlap = 0.0
            return (imbalance, max_count, mean_overlap, max_overlap, rng.random())

        # IMPORTANT FOR REPRODUCIBILITY:
        # `candidates` is a set, whose iteration order can vary between Python
        # processes. The key function consumes seeded RNG values for tie-breaking;
        # therefore, iterate in a stable sorted order before calling it.
        chosen = min(sorted(candidates), key=balance_diversity_key)
        selected.append(chosen)
        selected_set.add(chosen)
        # Keep the shared cross-stage registry synchronized. Without this line,
        # panels generated in this stage are not counted in `used`, even though
        # they are appended to all_rows later.
        used.add(chosen)
        for b in chosen:
            if b in counts:
                counts[b] += 1

    return selected


def top_feasible_rows(panel_rows):
    feasible = [item for item in panel_rows if item[1]["specificity"] > (1.0 - FPR) + 1e-12]
    feasible.sort(key=lambda item: item[1]["macro_sensitivity"], reverse=True)
    if not feasible:
        return []
    n_top = max(1, int(math.ceil(len(feasible) * TOP_PANEL_FRACTION)))
    return feasible[:n_top]


def biomarker_weights(panel_rows):
    """Estimate recurring-marker weights from the strongest feasible panels."""
    top = top_feasible_rows(panel_rows)
    counts = {b: 1.0 for b in BIOMARKERS}  # small baseline keeps exploration possible
    for panel, _summary in top:
        for biomarker in panel:
            counts[biomarker] += 1.0
    return counts


def weighted_panel(rng, weights):
    """Sample a unique-size panel without replacement, weighted by marker frequency."""
    available = list(BIOMARKERS)
    chosen = []
    for _ in range(PANEL_SIZE):
        total = sum(max(0.001, weights[b]) for b in available)
        pick = rng.random() * total
        cumulative = 0.0
        selected = available[-1]
        for biomarker in available:
            cumulative += max(0.001, weights[biomarker])
            if cumulative >= pick:
                selected = biomarker
                break
        chosen.append(selected)
        available.remove(selected)
    return canonical_panel(chosen)


def make_unique_weighted_panels(rng, count, used, weights):
    panels = []
    attempts = 0
    max_attempts = max(10000, count * 200)
    while len(panels) < count and attempts < max_attempts:
        attempts += 1
        candidate = weighted_panel(rng, weights)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    while len(panels) < count:
        candidate = random_panel(rng)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    return panels


def make_neighborhood_panels(rng, count, used, panel_rows):
    """Generate swaps around the best feasible panels, with random unique fill."""
    feasible = [item for item in panel_rows
                if item[1]["specificity"] > (1.0 - FPR) + 1e-12
                and np.isfinite(item[1]["macro_sensitivity"])]
    ranked_source = feasible if feasible else panel_rows
    ranked = sorted(ranked_source,
                    key=lambda item: item[1]["macro_sensitivity"]
                    if np.isfinite(item[1]["macro_sensitivity"]) else -1.0,
                    reverse=True)
    panels = []
    for panel, _summary in ranked:
        candidates = local_neighbors(panel)
        rng.shuffle(candidates)
        for candidate in candidates:
            if candidate not in used:
                used.add(candidate)
                panels.append(candidate)
                if len(panels) >= count:
                    return panels
    while len(panels) < count:
        candidate = random_panel(rng)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    return panels

def score_panel(df, reference_df, panel, cache):
    key = panel_key(panel)
    if key not in cache:
        summary, _ = evaluate_panel(df, reference_df, panel, save_features=False)
        cache[key] = summary
    return cache[key]


def rank_key(summary):
    # Specificity the configured specificity target is enforced as a hard feasibility constraint in
    # evaluate_panel. Among feasible panels, rank ONLY by macro-sensitivity.
    return summary["macro_sensitivity"]


def local_neighbors(panel):
    panel_set = set(panel)
    not_in_panel = [b for b in BIOMARKERS if b not in panel_set]
    neighbors = set()
    for removed in panel:
        for added in not_in_panel:
            candidate = canonical_panel((panel_set - {removed}) | {added})
            if len(candidate) == PANEL_SIZE:
                neighbors.add(candidate)
    return sorted(
        neighbors,
        key=lambda p: tuple(BIOMARKERS.index(x) for x in p),
    )



def evaluate_selected_panel_on_test(df, reference_df, panel, thresholds):
    """Evaluate the selected panel once on the untouched Test partition."""
    test_df = df[get_role_mask(df, "Test")].copy()
    if test_df.empty:
        print("No Test patients were available.")
        return None

    # Fit the regression model and reference coordinates from Calibration 1 normals,
    # then apply the already frozen thresholds to Test patients.
    cal1_features = calculate_panel_features(df, reference_df, panel, include_test=False)
    role = cal1_features[DATA_TYPE_COL].astype(str).str.strip().str.casefold()
    cohort = cal1_features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal1_normals = cal1_features[
        role.eq("regression fit calibration 1") & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()
    model = LinearRegression().fit(
        cal1_normals[REGRESSION_PREDICTORS], cal1_normals["Avg_Death_0D"]
    )
    test_features = []
    for idx, patient in test_df.iterrows():
        rec = {
            "Source_Row_Index": idx,
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
            COHORT_COL: patient[COHORT_COL],
            STAGE_COL: patient.get(STAGE_COL, np.nan),
            DATA_TYPE_COL: patient[DATA_TYPE_COL],
        }
        rec.update(calculate_tda_features(reference_df, patient, panel))
        test_features.append(rec)
    test_features = pd.DataFrame(test_features)
    test_features["Predicted Avg_Death_0D"] = model.predict(test_features[REGRESSION_PREDICTORS])
    test_features["Difference"] = (
        test_features["Avg_Death_0D"] - test_features["Predicted Avg_Death_0D"]
    )
    denom = test_features["Avg_Death_0D"].abs().replace(0, np.nan)
    test_features["Regression Difference Percentage"] = (
        test_features["Difference"].abs() / denom
    ) * 100.0
    ref_birth = float(cal1_normals["Average Birth (1D)"].mean())
    ref_death = float(cal1_normals["Average Death (1D)"].mean())
    test_features["Distance %(1)"] = np.sqrt(
        (test_features["Average Birth (1D)"] - ref_birth) ** 2
        + (test_features["Average Death (1D)"] - ref_death) ** 2
    ) * 100.0
    reg_t = thresholds["regression_threshold"]
    dist_t = thresholds["distance_threshold"]
    positive = (
        (test_features["Regression Difference Percentage"] > reg_t)
        | (test_features["Distance %(1)"] > dist_t)
    )
    test_features["Result(Cancer)"] = np.where(positive, "Positive", "Negative")
    cohort_values = test_features[COHORT_COL].astype(str).str.strip().str.casefold()
    # Macro-sensitivity is the unweighted mean of sensitivities across
    # cancer cohorts represented in the Test partition.
    sens_values = []
    for cancer in CANCER_TYPES:
        mask = cohort_values.eq(cancer)
        if mask.any():
            sens_values.append(float(positive[mask].mean()))

    macro_sens = float(np.mean(sens_values)) if sens_values else float("nan")

    # Test specificity is calculated using only normal patients in the Test partition.
    normal_mask = cohort_values.eq(NORMAL_COHORT.casefold())
    test_specificity = (
        float((~positive[normal_mask]).mean()) if normal_mask.any() else float("nan")
    )

    print(f"Test macro-sensitivity: {macro_sens:.4%}")
    print(f"Test specificity:       {test_specificity:.4%}")

    # Retain the patient-level prediction export without printing per-cancer results.
    test_features.to_csv(
        output_folder_path / "SelectedPanel_FinalTest_Predictions.csv",
        index=False
    )
    return {
        "macro_sensitivity": macro_sens,
        "specificity": test_specificity,
        "n_test": int(len(test_features)),
        "n_test_normal": int(normal_mask.sum()),
    }


def Optimized_Biomarker_Selection():
    """Run the staged, constrained TDA biomarker-selection procedure.

    Dataset-specific values are configured in the USER-DEFINED SETTINGS section.
    Returns the selected biomarker panel and its calibration summary.
    """
    output_folder_path.mkdir(parents=True, exist_ok=True)
    validate_split_fractions()
    bound_values = (K1_MIN, K1_MAX, K2_MIN, K2_MAX)
    bounds_provided = [value is not None for value in bound_values]
    if any(bounds_provided) and not all(bounds_provided):
        raise ValueError(
            "Provide all four bounds (K1_MIN, K1_MAX, K2_MIN, K2_MAX), or leave all four as None."
        )
    if all(bounds_provided):
        for name, value in zip(("K1_MIN", "K1_MAX", "K2_MIN", "K2_MAX"), bound_values):
            if not isinstance(value, (int, float)) or not np.isfinite(value):
                raise ValueError(f"{name} must be a finite number or None when bounds are disabled.")
        if K1_MIN > K1_MAX:
            raise ValueError(f"K1_MIN ({K1_MIN}) must be <= K1_MAX ({K1_MAX}).")
        if K2_MIN > K2_MAX:
            raise ValueError(f"K2_MIN ({K2_MIN}) must be <= K2_MAX ({K2_MAX}).")
    if not isinstance(MAX_THRESHOLD_CANDIDATES, int) or MAX_THRESHOLD_CANDIDATES < 2:
        raise ValueError("MAX_THRESHOLD_CANDIDATES must be an integer >= 2.")
    # Do not validate CANCER_TYPES here: it is populated automatically from the
    # workbook Cohort column inside create_updated_workbook(), which runs below.
    # cohort_order is only a preferred sorting map and must not control which
    # cancer cohorts are included in analysis.
    if len(BIOMARKERS) != len(set(BIOMARKERS)):
        raise ValueError("BIOMARKERS must contain unique names.")
    if PANEL_SIZE < 1 or PANEL_SIZE > len(BIOMARKERS):
        raise ValueError("PANEL_SIZE must be between 1 and the number of BIOMARKERS.")
    if sum(PHASE_BUDGETS.values()) != TOTAL_PANEL_BUDGET:
        raise ValueError("The phase budgets must sum exactly to TOTAL_PANEL_BUDGET.")

    updated_path, _ = create_updated_workbook(
        clinical_data_path, UPDATED_WORKBOOK_PATH, sheet_name
    )
    df = pd.read_excel(updated_path, sheet_name=sheet_name)
    reference_df = df[
        get_role_mask(df, "Reference")
        & df[COHORT_COL].astype(str).str.strip().str.casefold().eq(NORMAL_COHORT.casefold())
    ].copy()
    if len(reference_df) != N_REFERENCE_NORMAL:
        raise ValueError(
            f"Expected {N_REFERENCE_NORMAL} reference normals; found {len(reference_df)}."
        )

    rng = random.Random(RANDOM_SEED)
    cache = {}
    used = set()
    all_rows = []
    stage_by_panel = {}

    def evaluate_phase(label, panels, budget):
        if len(panels) != budget:
            raise RuntimeError(f"{label}: expected {budget} panels, got {len(panels)}")
        phase_rows = []
        for i, panel in enumerate(panels, start=1):
            summary = score_panel(df, reference_df, panel, cache)
            phase_rows.append((panel, summary))
            all_rows.append((panel, summary))
            stage_by_panel[panel] = label

            # Print one compact global progress line every 100 evaluated panels,
            # rather than printing every panel's detailed performance.
            completed = len(all_rows)
            if completed % 100 == 0 or completed == TOTAL_PANEL_BUDGET:
                feasible = [
                    row_summary["macro_sensitivity"]
                    for _, row_summary in all_rows
                    if row_summary["specificity"] > (1.0 - FPR) + 1e-12
                    and np.isfinite(row_summary["macro_sensitivity"])
                ]
                best_so_far = max(feasible) if feasible else float("nan")
                print(
                    f"{completed}/{TOTAL_PANEL_BUDGET} | "
                    f"Stage: {label} | "
                    f"best pooled-specificity-feasible macro-sensitivity so far: {best_so_far:.4f}"
                )
        return phase_rows

    print("\n" + "=" * 78)
    print(f"OPTIMIZED BIOMARKER SELECTION: exactly {TOTAL_PANEL_BUDGET} unique {PANEL_SIZE}-marker panels")
    print(f"Fixed seed: {RANDOM_SEED} | Pooled Calibration 1 + 2 normal specificity target: strictly > {1.0 - FPR:.0%} | macro-sensitivity target: > {SENSITIVITY_TARGET:.0%}")
    for stage, budget in PHASE_BUDGETS.items():
        print(f"  {stage}: {budget}")
    print("=" * 78)

    # STAGE 1: all 42 biomarkers are treated equally at the outset. A greedy
    # balancing criterion keeps their appearance counts as even as possible,
    # while minimizing overlap between panels. No outcome-guided marker weights.
    stage1_panels = balanced_diverse_panels(
        rng, PHASE_BUDGETS["balanced_diverse_exploration"], used,
        marker_pool=BIOMARKERS, prior_counts={b: 0 for b in BIOMARKERS}
    )
    used.update(stage1_panels)
    evaluate_phase("1/4 balanced diverse exploration", stage1_panels,
                   PHASE_BUDGETS["balanced_diverse_exploration"])

    # STAGE 2: rank only feasible panels by macro-sensitivity, then estimate marker
    # recurrence from the top feasible fraction and use it to guide new combinations.
    weights = biomarker_weights(all_rows)
    stage2_panels = make_unique_weighted_panels(
        rng, PHASE_BUDGETS["sensitivity_guided_search"], used, weights
    )
    evaluate_phase("2/4 sensitivity-guided search", stage2_panels,
                   PHASE_BUDGETS["sensitivity_guided_search"])

    # STAGE 3: identify recurring biomarkers in the top feasible panels from Stages
    # 1–2, then repeat balanced/low-overlap exploration within the recurrence-informed
    # pool. This keeps the marker frequencies balanced within that selected pool.
    top_before_stage3 = top_feasible_rows(all_rows)
    recurrence_counts = {b: 0 for b in BIOMARKERS}
    for panel, _summary in top_before_stage3:
        for b in panel:
            recurrence_counts[b] += 1
    recurring_pool = [b for b, _count in sorted(
        recurrence_counts.items(), key=lambda item: (-item[1], BIOMARKERS.index(item[0]))
    )[:RECURRENCE_POOL_SIZE]]
    if len(recurring_pool) < PANEL_SIZE:
        recurring_pool = list(BIOMARKERS)
    stage3_panels = balanced_diverse_panels(
        rng, PHASE_BUDGETS["recurrence_informed_balanced_search"], used,
        marker_pool=recurring_pool, prior_counts={b: 0 for b in recurring_pool}
    )
    evaluate_phase("3/4 recurrence-informed balanced search", stage3_panels,
                   PHASE_BUDGETS["recurrence_informed_balanced_search"])

    # STAGE 4: one-biomarker swaps around the strongest panels from all previous
    # stages, with random unique panels as a fallback. The final selection below
    # still compares every feasible panel from all four stages.
    stage4_panels = make_neighborhood_panels(
        rng, PHASE_BUDGETS["best_panel_neighborhoods"], used, all_rows
    )
    evaluate_phase("4/4 best-panel neighborhood search", stage4_panels,
                   PHASE_BUDGETS["best_panel_neighborhoods"])

    if len(used) != TOTAL_PANEL_BUDGET or len(all_rows) != TOTAL_PANEL_BUDGET:
        raise RuntimeError(
            f"Search budget error: evaluated {len(all_rows)} panels, expected {TOTAL_PANEL_BUDGET}."
        )
    if len(set(panel for panel, _ in all_rows)) != TOTAL_PANEL_BUDGET:
        raise RuntimeError("Duplicate panels were evaluated; every panel must be unique.")

    # IMPORTANT: the winning panel is selected across all four stages, not only Stage 4.
    specificity_feasible_rows = [
        item for item in all_rows
        if item[1]["specificity"] > (1.0 - FPR) + 1e-12
        and np.isfinite(item[1]["macro_sensitivity"])
    ]
    finite_rows = [
        item for item in all_rows
        if np.isfinite(item[1]["macro_sensitivity"])
    ]
    if not finite_rows:
        raise RuntimeError("All evaluated panels had undefined macro-sensitivity.")
    # Detection-based panel selection:
    # 1. Keep only panels whose pooled Calibration 1 + Calibration 2 normal specificity is strictly above 1 - FPR.
    # 2. Among those feasible panels, select the panel with the highest
    #    macro-sensitivity across the cancer cohorts.
    # 3. Do not require macro-sensitivity itself to exceed 99%; it is the
    #    optimization objective, while specificity is the hard constraint.
    if specificity_feasible_rows:
        ranked_rows = sorted(
            specificity_feasible_rows,
            key=lambda item: (
                item[1]["macro_sensitivity"],
                item[1]["overall_cancer_sensitivity"],
                item[1]["specificity"],
            ),
            reverse=True,
        )
    else:
        # This is a diagnostic fallback only. Under the exhaustive threshold
        # search, thresholds above all scores should normally yield 100% specificity.
        ranked_rows = sorted(
            finite_rows,
            key=lambda item: (
                item[1]["macro_sensitivity"],
                item[1]["overall_cancer_sensitivity"],
                item[1]["specificity"],
            ),
            reverse=True,
        )
        print(
            "\\nWARNING: No panel met the strict pooled Calibration 1 + Calibration 2 "
            "specificity constraint. The selected panel is diagnostic only."
        )
    best_panel, _ = ranked_rows[0]

    output_rows = []
    for panel, summary in sorted(
        all_rows, key=lambda item: item[1]["macro_sensitivity"], reverse=True
    ):
        row = {
            "stage": stage_by_panel.get(panel, "unknown"),
            "panel": " | ".join(panel),
            "n_biomarkers": summary["n_biomarkers"],
            "specificity": summary["specificity"],
            "calibration1_specificity_descriptive": summary["calibration1_specificity"],
            "calibration2_specificity_descriptive": summary["calibration2_specificity"],
            "combined_calibration_specificity": summary["specificity"],
            "meets_combined_specificity_constraint": summary["specificity"] > (1.0 - FPR) + 1e-12,
            "false_positives": summary["false_positives"],
            "true_negatives": summary["true_negatives"],
            "macro_sensitivity": summary["macro_sensitivity"],
            "overall_cancer_sensitivity": summary["overall_cancer_sensitivity"],
            "regression_threshold": summary["regression_threshold"],
            "distance_threshold": summary["distance_threshold"],
        }
        # Automatically add sample size, sensitivity, and TP columns for every
        # cancer cohort detected in the workbook; no cohort names are hard-coded.
        for cancer in CANCER_TYPES:
            cancer_result = summary["cancer_results"].get(
                cancer, {"n": 0, "sensitivity": float("nan"), "true_positives": 0}
            )
            safe_name = "".join(ch if ch.isalnum() else "_" for ch in cancer).strip("_")
            row[f"{safe_name}_n"] = cancer_result["n"]
            row[f"{safe_name}_sensitivity"] = cancer_result["sensitivity"]
            row[f"{safe_name}_true_positives"] = cancer_result["true_positives"]
        output_rows.append(row)
    results_frame = pd.DataFrame(output_rows)
    results_frame["meets_sensitivity_target"] = results_frame["macro_sensitivity"] > SENSITIVITY_TARGET
    results_frame.to_csv(SEARCH_RESULTS_PATH, index=False)
    print(f"\nAll {len(all_rows)} evaluated panels saved: {SEARCH_RESULTS_PATH}")

    print("\nTOP 5 PANELS RANKED BY CALIBRATION 2 MACRO-SENSITIVITY (pooled calibration specificity feasible first)")
    top5_candidates = specificity_feasible_rows if specificity_feasible_rows else finite_rows
    top5 = sorted(top5_candidates, key=lambda item: item[1]["macro_sensitivity"], reverse=True)[:5]
    for rank, (panel, summary) in enumerate(top5, start=1):
        print(f"\nRank {rank}: {list(panel)!r}")
        print(f"  Regression Difference Percentage threshold: {summary['regression_threshold']:.6g}")
        print(f"  Distance %(1) threshold: {summary['distance_threshold']:.6g}")
        print(f"  Calibration 1 specificity (descriptive): {summary['calibration1_specificity']:.4%}")
        print(f"  Calibration 2 macro-sensitivity: {summary['macro_sensitivity']:.4%}")
        print(f"  Calibration 2 specificity (descriptive): {summary['calibration2_specificity']:.4%}")
        print(f"  Combined Calibration 1 + 2 normal specificity: {summary['specificity']:.4%}")
        print(f"  Meets pooled specificity target: {summary['specificity'] > (1.0 - FPR) + 1e-12}")
    if not specificity_feasible_rows:
        print("\\nWARNING: No panel met the pooled calibration specificity constraint; displayed panels are diagnostic only.")

    best_summary, _ = evaluate_panel(df, reference_df, best_panel, save_features=True)
    best_summary["panel"] = list(best_panel)
    best_summary["winning_stage"] = stage_by_panel.get(best_panel, "unknown")
    best_summary["search_budget"] = TOTAL_PANEL_BUDGET
    best_summary["phase_budgets"] = PHASE_BUDGETS
    best_summary["random_seed"] = RANDOM_SEED
    best_summary["calibration1_fraction"] = CALIBRATION1_FRACTION
    best_summary["calibration2_fraction"] = CALIBRATION2_FRACTION
    best_summary["test_fraction"] = 1.0 - CALIBRATION1_FRACTION - CALIBRATION2_FRACTION
    best_summary["n_reference_normal"] = N_REFERENCE_NORMAL
    best_summary["selection_rule"] = (
        "For each panel, select the Regression Difference Percentage and Distance %(1) "
        "thresholds using the OR detection rule, requiring pooled specificity strictly above the configured target "
        f"({(1.0 - FPR):.4%}) across the combined Calibration 1 and Calibration 2 normal patients, and maximizing macro-sensitivity across "
        "cancer cohorts. Select the final panel from "
        "specificity-feasible panels by highest macro-sensitivity; overall sensitivity and specificity "
        "are tie-breakers. The Test partition is not used for panel or threshold selection."
    )
    best_summary["recurrence_informed_stage3_pool"] = recurring_pool
    best_summary["top_feasible_panel_count_for_recurrence"] = len(top_before_stage3)

    # Evaluate the selected panel on the untouched Test partition and include
    # both Test metrics in the selected-panel summary output.
    test_summary = evaluate_selected_panel_on_test(
        df, reference_df, best_panel, best_summary
    )
    if test_summary is not None:
        best_summary["test_macro_sensitivity"] = test_summary["macro_sensitivity"]
        best_summary["test_specificity"] = test_summary["specificity"]
        best_summary["n_test"] = test_summary["n_test"]
        best_summary["n_test_normal"] = test_summary["n_test_normal"]
    else:
        best_summary["test_macro_sensitivity"] = float("nan")
        best_summary["test_specificity"] = float("nan")
        best_summary["n_test"] = 0
        best_summary["n_test_normal"] = 0

    # Marker frequencies in top feasible panels across ALL four stages.
    top_panels = top_feasible_rows(all_rows)
    recurrence_all = {b: sum(b in panel for panel, _ in top_panels) for b in BIOMARKERS}
    best_summary["biomarker_frequency_in_top_feasible_panels"] = dict(
        sorted(recurrence_all.items(), key=lambda kv: (-kv[1], BIOMARKERS.index(kv[0])))
    )
    flat_summary = {}
    for key, value in best_summary.items():
        if isinstance(value, dict):
            for subkey, subvalue in value.items():
                if isinstance(subvalue, dict):
                    for nested_key, nested_value in subvalue.items():
                        flat_summary[f"{key}_{subkey}_{nested_key}"] = nested_value
                elif isinstance(subvalue, (list, tuple)):
                    flat_summary[f"{key}_{subkey}"] = "; ".join(map(str, subvalue))
                else:
                    flat_summary[f"{key}_{subkey}"] = subvalue
        elif isinstance(value, (list, tuple)):
            flat_summary[key] = "; ".join(map(str, value))
        else:
            flat_summary[key] = value
    pd.DataFrame([flat_summary]).to_csv(SELECTED_PANEL_SUMMARY_CSV_PATH, index=False)

    # Save a separate recurrence table to make the frequency analysis easy to inspect.
    recurrence_path = RECURRENCE_TABLE_PATH
    pd.DataFrame([
        {"biomarker": b, "count_in_top_panels": recurrence_all[b],
         "number_of_top_feasible_panels": len(top_panels),
         "frequency": (recurrence_all[b] / len(top_panels)) if top_panels else 0.0}
        for b in sorted(BIOMARKERS, key=lambda x: (-recurrence_all[x], BIOMARKERS.index(x)))
    ]).to_csv(recurrence_path, index=False)

    best_macro_sensitivity = best_summary["macro_sensitivity"]
    # Report every feasible panel tied for the best macro-sensitivity.
    # Macro-sensitivity is calculated from finite sample proportions, so exact
    # equality is appropriate for panels evaluated on the same calibration data.
    tied_best_rows = [
        (panel, summary) for panel, summary in ranked_rows
        if np.isclose(
            summary["macro_sensitivity"], best_macro_sensitivity,
            rtol=0.0, atol=1e-12
        )
    ]

    print("\n" + "=" * 78)
    print("BEST BIOMARKER PANEL — SELECTED ACROSS ALL FOUR STAGES")
    print("=" * 78)
    print(f"Winning stage: {stage_by_panel.get(best_panel, 'unknown')}")
    print(f"Final biomarker set: {list(best_panel)!r}")
    print(f"Calibration 2 macro-sensitivity: {best_macro_sensitivity:.4f}")
    print(f"Meets macro-sensitivity >99% (diagnostic only): {best_macro_sensitivity > SENSITIVITY_TARGET}")
    print(f"Calibration 1 specificity: {best_summary['calibration1_specificity']:.4f}")
    print(f"Calibration 2 specificity (descriptive): {best_summary['calibration2_specificity']:.4f}")
    print(f"Combined Calibration 1 + 2 normal specificity: {best_summary['specificity']:.4f}")
    print(f"Meets pooled specificity > target: {best_summary['specificity'] > (1.0 - FPR) + 1e-12}")
    print(f"Test macro-sensitivity: {best_summary['test_macro_sensitivity']:.4%}")
    print(f"Test specificity:       {best_summary['test_specificity']:.4%}")
    print(f"Total unique panels evaluated: {len(all_rows)}")

    if len(tied_best_rows) > 1:
        print("\nOther biomarker panels with the same best macro-sensitivity:")
        for tied_panel, tied_summary in tied_best_rows:
            if tied_panel == best_panel:
                continue
            print(f"  Biomarker set: {list(tied_panel)!r}")
            print(f"  Macro-sensitivity: {tied_summary['macro_sensitivity']:.4f}")
            print(f"  Calibration specificity: {tied_summary['specificity']:.4f}")
            print(f"  Winning stage: {stage_by_panel.get(tied_panel, 'unknown')}")
    else:
        print("\nNo other feasible panel achieved the same best macro-sensitivity.")

    print("\nOUTPUT FILES AND LOCATIONS")
    print("-" * 78)
    print(f"Updated workbook:                   {UPDATED_WORKBOOK_PATH}")
    print(f"All panel search results:           {SEARCH_RESULTS_PATH}")
    print(f"Selected-panel calibration summary: {SELECTED_PANEL_SUMMARY_CSV_PATH}")
    print(f"Top-panel biomarker frequencies:    {RECURRENCE_TABLE_PATH}")
    print(f"Selected-panel feature vectors:     {FEATURE_VECTOR_PATH}")
    print("\nThe regression model and Distance %(1) reference coordinates were constructed using Calibration 1 normals only.")
    print(f"Final k1/k2 thresholds were selected using the OR rule and pooled specificity strictly above {1.0 - FPR:.4%} across Calibration 1 + Calibration 2 normals combined.")
    print("Cancer types and per-type sensitivity outputs were detected automatically from the dataset Cohort column.")
    print("Calibration 2 cancers were used for panel sensitivity-based selection; Test was untouched until final evaluation.")




if __name__ == "__main__":
    output_folder_path.mkdir(parents=True, exist_ok=True)
    Optimized_Biomarker_Selection()
```

```python
"""
TDAcancer-detect-1
=================

Threshold selection and independent Test-set evaluation for the
TDAcancer-detect cancer detection model.

Workflow
--------
1. Assign the first 59 Normal patients, in workbook order, to the fixed Reference set.
2. Fit the regression model using Normal patients in Regression Fit Calibration 1.
3. Construct the biomarker dissimilarity matrix from D = 1 - |C|, where C is the
   Pearson correlation matrix of log-transformed biomarker measurements.
4. Select the k1 and k2 thresholds using the configured OR decision rule:
       Positive if (Regression Difference Percentage > k1)
                   OR (Distance %(1) > k2)
5. Require the Normal specificity in Calibration 1 and Calibration 2 to be
   strictly greater than SPECIFICITY_TARGET.
6. Maximize macro-sensitivity across the Calibration 2 cancer cohorts, applying
   the configured tie-breaking criteria.
7. Evaluate the selected thresholds on the independent Test partition.
8. Export the results workbook, prediction and metric CSV files, and ROC plot.

TDA features
------------
- 0-Dim Hole Min Life Range
- 0-Dim Hole Max Life Range
- Median_Death_0D
- Std_Death_0D
- Avg_Death_0D
- Average Birth (1D)
- Average Death (1D)

Dependencies
------------
Install the required packages with:

    python -m pip install numpy pandas gudhi scikit-learn openpyxl matplotlib

The matrix D = 1 - |C| is a dissimilarity matrix. It is symmetric and has a
zero diagonal after numerical cleanup; the script does not assume that it
satisfies the triangle inequality. GUDHI's RipsComplex interface accepts this
matrix through its ``distance_matrix`` argument.

Configuration
-------------
Set the workbook path and dataset-specific parameters in the USER SETTINGS
section before running the script.
"""

from pathlib import Path

import numpy as np
import pandas as pd
import gudhi
from sklearn.linear_model import LinearRegression
from sklearn.metrics import roc_curve, roc_auc_score



# ---------------------------------------------------------------------
# USER SETTINGS
# ---------------------------------------------------------------------
# The ONLY source/input workbook: the original file without Data Type assignments.
ORIGINAL_INPUT_FILE = Path(
    r"D:\Cancer Detection folder\Biomarker Selection"
    r"\clinical cancer validation data.xlsx"
)

# This file is generated automatically from ORIGINAL_INPUT_FILE before analysis.
UPDATED_INPUT_FILE = ORIGINAL_INPUT_FILE.with_name(
    "clinical cancer validation data_Updated.xlsx"
)
SHEET_NAME = "Normal and Cancer"

# Combined results workbook. Neither the original nor the generated working
# workbook is overwritten by the analysis results.
OUTPUT_EXCEL = UPDATED_INPUT_FILE.with_name(
    "clinical cancer validation data_Updated_with_Test_Performance.xlsx"
)

# CSV outputs retained from the original test-evaluation code.
OUTPUT_DIR = UPDATED_INPUT_FILE.parent
PREDICTIONS_CSV = OUTPUT_DIR / "TDA_test_predictions.csv"
METRICS_CSV = OUTPUT_DIR / "TDA_test_metrics.csv"
STAGE_METRICS_CSV = OUTPUT_DIR / "TDA_test_stage_metrics.csv"
ROC_CURVE_CSV = OUTPUT_DIR / "TDA_test_ROC_curve.csv"
ROC_PLOT_FILE = OUTPUT_DIR / "TDA_test_ROC_curve.png"


# ---------------------------------------------------------------------
# WORKBOOK PREPROCESSING
# ---------------------------------------------------------------------
CALIBRATION1_FRACTION = 0.15
CALIBRATION2_FRACTION = 0.50
SPLIT_RANDOM_SEED = 20261017


def create_updated_workbook():
    """
    Load the original clinical validation workbook and assign Data Type roles.
    The first N_REFERENCE_NORMAL normal patients remain the fixed Reference set.
    Calibration fractions are applied to the remaining normal patients.
    Cancer patients are split between Calibration 2 and Test only.
    The resulting workbook is saved to UPDATED_INPUT_FILE, which the analysis then reads.
    """
    if not ORIGINAL_INPUT_FILE.exists():
        raise FileNotFoundError(
            f"Original clinical validation workbook not found: {ORIGINAL_INPUT_FILE}\\n"
            "Update ORIGINAL_INPUT_FILE or place the original workbook beside this script."
        )

    if not (0 <= CALIBRATION1_FRACTION <= 1):
        raise ValueError("CALIBRATION1_FRACTION must be between 0 and 1.")
    if not (0 <= CALIBRATION2_FRACTION <= 1):
        raise ValueError("CALIBRATION2_FRACTION must be between 0 and 1.")
    if CALIBRATION1_FRACTION + CALIBRATION2_FRACTION >= 1:
        raise ValueError(
            "CALIBRATION1_FRACTION + CALIBRATION2_FRACTION must be less than 1 "
            "so that the normal Test group has patients."
        )

    all_sheets = pd.read_excel(ORIGINAL_INPUT_FILE, sheet_name=None)
    if SHEET_NAME not in all_sheets:
        raise ValueError(
            f"Sheet {SHEET_NAME!r} not found in original workbook. "
            f"Available sheets: {list(all_sheets)}"
        )

    df = all_sheets[SHEET_NAME].copy()
    if COHORT_COL not in df.columns:
        raise ValueError(f"Required column {COHORT_COL!r} is missing from the original workbook.")

    if DATA_TYPE_COL in df.columns:
        df = df.drop(columns=[DATA_TYPE_COL])

    cohort = df[COHORT_COL].astype(str).str.strip().str.casefold()
    if df[COHORT_COL].isna().any() or cohort.eq("").any():
        raise ValueError(f"Column {COHORT_COL!r} contains missing or blank cohort labels.")

    normal_mask = cohort.eq(NORMAL_LABEL.casefold())
    normal_indices = df.index[normal_mask].tolist()
    if len(normal_indices) <= N_REFERENCE_NORMAL:
        raise ValueError(
            f"Need more than {N_REFERENCE_NORMAL} normal patients; found {len(normal_indices)}."
        )

    role = {}
    reference_indices = normal_indices[:N_REFERENCE_NORMAL]
    for idx in reference_indices:
        role[idx] = REFERENCE_ROLE

    rng = np.random.RandomState(SPLIT_RANDOM_SEED)

    # Normal Calibration 1 and Calibration 2 percentages are calculated from
    # the normal patients remaining after the fixed Reference group.
    remaining_normals = list(normal_indices[N_REFERENCE_NORMAL:])
    rng.shuffle(remaining_normals)
    n_cal1 = int(np.floor(len(remaining_normals) * CALIBRATION1_FRACTION))
    n_cal2 = int(np.floor(len(remaining_normals) * CALIBRATION2_FRACTION))

    for idx in remaining_normals[:n_cal1]:
        role[idx] = CAL1_ROLE
    for idx in remaining_normals[n_cal1:n_cal1 + n_cal2]:
        role[idx] = CAL2_ROLE
    for idx in remaining_normals[n_cal1 + n_cal2:]:
        role[idx] = TEST_ROLE

    # Cancer cohorts are split between Calibration 2 and Test; never Calibration 1.
    cancer_cohorts = list(dict.fromkeys(
        cohort.loc[~normal_mask].tolist()
    ))
    for cancer in cancer_cohorts:
        indices = df.index[cohort.eq(cancer)].tolist()
        shuffled = list(indices)
        rng.shuffle(shuffled)
        n_cal2_cancer = int(np.floor(len(shuffled) * CALIBRATION2_FRACTION))
        for idx in shuffled[:n_cal2_cancer]:
            role[idx] = CAL2_ROLE
        for idx in shuffled[n_cal2_cancer:]:
            role[idx] = TEST_ROLE

    df[DATA_TYPE_COL] = pd.Series(role)
    role_order = {
        REFERENCE_ROLE: 0,
        CAL1_ROLE: 1,
        CAL2_ROLE: 2,
        TEST_ROLE: 3,
    }
    df["_original_order"] = np.arange(len(df))
    df["_role_order"] = df[DATA_TYPE_COL].map(role_order)
    df = (
        df.sort_values(["_role_order", "_original_order"], kind="stable")
          .drop(columns=["_original_order", "_role_order"])
          .reset_index(drop=True)
    )

    all_sheets[SHEET_NAME] = df
    UPDATED_INPUT_FILE.parent.mkdir(parents=True, exist_ok=True)
    with pd.ExcelWriter(UPDATED_INPUT_FILE, engine="openpyxl") as writer:
        for sheet_out, sheet_df in all_sheets.items():
            sheet_df.to_excel(writer, sheet_name=str(sheet_out)[:31], index=False)

    print("Original clinical validation workbook:", ORIGINAL_INPUT_FILE)
    print("Updated workbook created for the existing analysis:", UPDATED_INPUT_FILE)
    print(f"Fixed Reference normal patients: {len(reference_indices)}")
    print(
        f"Normal split after Reference: {CALIBRATION1_FRACTION:.1%} Calibration 1 / "
        f"{CALIBRATION2_FRACTION:.1%} Calibration 2 / "
        f"{1 - CALIBRATION1_FRACTION - CALIBRATION2_FRACTION:.1%} Test"
    )
    print(
        f"Cancer split: {CALIBRATION2_FRACTION:.1%} Calibration 2 / "
        f"{1 - CALIBRATION2_FRACTION:.1%} Test"
    )
    print("\\nCohort x Data Type counts:")
    print(pd.crosstab(df[COHORT_COL], df[DATA_TYPE_COL]).to_string())


COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
STAGE_COL = "AJCC Stage"
DATA_TYPE_COL = "Data Type"
NORMAL_LABEL = "normal"

REFERENCE_ROLE = "Reference"
CAL1_ROLE = "Regression Fit Calibration 1"
CAL2_ROLE = "Biomarker-Panel Calibration and Selection"
TEST_ROLE = "Test"

N_REFERENCE_NORMAL = 59
SPECIFICITY_TARGET = 0.95  # Both calibration normal specificities must be > 95%.

# USER-DEFINED THRESHOLD SEARCH BOUNDS
# k1 threshold for "Regression Difference Percentage"; inclusive range.
K1_MIN = -1.0
K1_MAX = 3.0

# k2 threshold for "Distance %(1)"; inclusive range.
K2_MIN = 0.0
K2_MAX = 7.0

PANEL = ['sIL-6Ra', 'CA-125', 'CA 19-9', 'IL-6', 'IL-8', 'OPN', 'TRAIL', 'Total PSA']

REGRESSION_PREDICTORS = [
    "0-Dim Hole Min Life Range",
    "0-Dim Hole Max Life Range",
    "Median_Death_0D",
    "Std_Death_0D",
]

FEATURE_COLUMNS = [
    "0-Dim Hole Min Life Range",
    "0-Dim Hole Max Life Range",
    "Median_Death_0D",
    "Std_Death_0D",
    "Avg_Death_0D",
    "Average Birth (1D)",
    "Average Death (1D)",
]


# ---------------------------------------------------------------------
# Shared helpers
# ---------------------------------------------------------------------
def normalized_text(series):
    return series.astype(str).str.strip().str.casefold()


def safe_numeric_frame(frame, columns, description):
    out = frame[columns].apply(pd.to_numeric, errors="coerce")
    if out.isna().any().any():
        bad_rows = out.index[out.isna().any(axis=1)].tolist()
        raise ValueError(
            f"{description} contains missing/non-numeric values in required "
            f"biomarker columns. Example row indices: {bad_rows[:10]}"
        )
    values = out.to_numpy(dtype=float)
    if not np.isfinite(values).all():
        raise ValueError(f"{description} contains NaN or infinite biomarker values.")
    if (values <= 0).any():
        raise ValueError(
            f"{description} contains zero/negative values; log transformation "
            "requires every selected biomarker to be strictly positive."
        )
    return out


def calculate_tda_features(reference_df, patient_row):
    """Calculate TDA features from the reference cohort and one patient."""
    combined = pd.concat(
        [
            reference_df[PANEL].astype(float),
            patient_row[PANEL].astype(float).to_frame().T,
        ],
        ignore_index=True,
    )
    values = combined.to_numpy(dtype=float)

    if not np.isfinite(values).all():
        raise ValueError(
            f"Non-finite biomarker values for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )
    if (values <= 0).any():
        raise ValueError(
            "All selected biomarker values must be strictly positive for "
            "the log transformation. Patient: "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )

    logged_values = np.log(values)
    correlation_matrix = np.corrcoef(logged_values, rowvar=False)
    if not np.isfinite(correlation_matrix).all():
        raise ValueError(
            f"Invalid correlation matrix for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}. Check constant "
            "biomarker columns and input values."
        )

    # Biomarker dissimilarity: D_ij = 1 - |C_ij|.
    dissimilarity_matrix = 1.0 - np.abs(correlation_matrix)
    dissimilarity_matrix = (
        dissimilarity_matrix + dissimilarity_matrix.T
    ) / 2.0
    np.fill_diagonal(dissimilarity_matrix, 0.0)
    dissimilarity_matrix = np.clip(dissimilarity_matrix, 0.0, 1.0)

    # GUDHI's API parameter is named distance_matrix; the supplied matrix
    # is the dissimilarity matrix defined above.
    rips_complex = gudhi.RipsComplex(
        distance_matrix=dissimilarity_matrix
    )
    simplex_tree = rips_complex.create_simplex_tree(max_dimension=2)
    persistence = simplex_tree.persistence()

    finite_intervals = [
        (dimension, (birth, death))
        for dimension, (birth, death) in persistence
        if np.isfinite(death)
    ]
    intervals_0d = [
        interval for dimension, interval in finite_intervals if dimension == 0
    ]
    intervals_1d = [
        interval for dimension, interval in finite_intervals if dimension == 1
    ]

    # Retain only the average birth and average death summaries for 1D intervals.
    if intervals_1d:
        average_birth_1d = float(
            np.mean([birth for birth, _ in intervals_1d])
        )
        average_death_1d = float(
            np.mean([death for _, death in intervals_1d])
        )
    else:
        average_birth_1d = 0.0
        average_death_1d = 0.0

    # Retain the 0D lifetime and death-distribution summaries used by the model.
    if intervals_0d:
        deaths_0d = [death for _, death in intervals_0d]
        lifetimes_0d = [
            death - birth for birth, death in intervals_0d
        ]
        min_life_0d = float(min(lifetimes_0d))
        max_life_0d = float(max(lifetimes_0d))
        average_death_0d = float(np.mean(deaths_0d))
        median_death_0d = float(np.median(deaths_0d))
        std_death_0d = float(np.std(deaths_0d))
    else:
        min_life_0d = 0.0
        max_life_0d = 0.0
        average_death_0d = 0.0
        median_death_0d = 0.0
        std_death_0d = 0.0

    return {
        "0-Dim Hole Min Life Range": min_life_0d,
        "0-Dim Hole Max Life Range": max_life_0d,
        "Median_Death_0D": median_death_0d,
        "Std_Death_0D": std_death_0d,
        "Avg_Death_0D": average_death_0d,
        "Average Birth (1D)": average_birth_1d,
        "Average Death (1D)": average_death_1d,
    }


def build_features(reference_df, rows_to_process):
    records = []
    for idx, patient in rows_to_process.iterrows():
        record = {
            "Source_Row_Index": idx,
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
            COHORT_COL: patient[COHORT_COL],
            STAGE_COL: patient.get(STAGE_COL, np.nan),
            DATA_TYPE_COL: patient[DATA_TYPE_COL],
        }
        record.update(calculate_tda_features(reference_df, patient))
        records.append(record)
    if not records:
        return pd.DataFrame()
    # Preserve original Excel row indices so Test predictions are written back
    # to the correct patient rows in the copied workbook.
    return pd.DataFrame(records).set_index("Source_Row_Index", drop=False)


def add_scores(features, regression_model, ref_birth, ref_death):
    """Add regression residual percentage and distance from Calibration 1 reference."""
    features = features.copy()
    predicted = regression_model.predict(features[REGRESSION_PREDICTORS])
    features["Predicted Avg_Death_0D"] = predicted
    features["Difference"] = features["Avg_Death_0D"] - predicted
    denominator = features["Avg_Death_0D"].abs().replace(0, np.nan)
    features["Regression Difference Percentage"] = (
        features["Difference"].abs() / denominator
    ) * 100.0
    features["Distance %(1)"] = np.sqrt(
        (features["Average Birth (1D)"] - ref_birth) ** 2
        + (features["Average Death (1D)"] - ref_death) ** 2
    ) * 100.0
    return features


def candidate_thresholds(normal_scores, cancer_scores, min_value, max_value, name):
    """
    Generate decision-boundary candidates within inclusive user-defined bounds.

    Observed scores and their next representable float above them are included
    when they fall within [min_value, max_value]. Exact range endpoints are
    also included so the search evaluates the bounds themselves.
    """
    if not np.isfinite(min_value) or not np.isfinite(max_value):
        raise ValueError(f"{name} bounds must be finite numbers.")
    if min_value > max_value:
        raise ValueError(
            f"{name}_MIN ({min_value}) must be <= {name}_MAX ({max_value})."
        )

    combined = np.concatenate(
        [
            np.asarray(normal_scores, dtype=float),
            np.asarray(cancer_scores, dtype=float),
        ]
    )
    combined = combined[np.isfinite(combined)]
    if len(combined) == 0:
        raise ValueError("No finite scores available for threshold search.")

    unique_values = np.unique(combined)
    candidates = np.unique(
        np.concatenate([unique_values, np.nextafter(unique_values, np.inf)])
    )
    candidates = candidates[
        (candidates >= min_value) & (candidates <= max_value)
    ]
    candidates = np.unique(np.concatenate(([min_value, max_value], candidates)))
    return candidates


def wilson_ci(successes, total, z=1.959963984540054):
    """Return Wilson 95% confidence interval as proportions."""
    if total == 0:
        return (np.nan, np.nan)
    p = successes / total
    denom = 1 + z * z / total
    centre = (p + z * z / (2 * total)) / denom
    half = z * np.sqrt(
        (p * (1 - p) / total) + (z * z / (4 * total * total))
    ) / denom
    return max(0.0, centre - half), min(1.0, centre + half)


def metric_row(metric_name, successes, total):
    """Format a metric with numerator, denominator, percentage and Wilson CI."""
    if total == 0:
        return {
            "Metric": metric_name,
            "Numerator": int(successes),
            "Denominator": int(total),
            "Rate (%)": np.nan,
            "95% CI Lower (%)": np.nan,
            "95% CI Upper (%)": np.nan,
        }
    low, high = wilson_ci(int(successes), int(total))
    return {
        "Metric": metric_name,
        "Numerator": int(successes),
        "Denominator": int(total),
        "Rate (%)": 100 * successes / total,
        "95% CI Lower (%)": 100 * low,
        "95% CI Upper (%)": 100 * high,
    }


# ---------------------------------------------------------------------
# Main workflow
# ---------------------------------------------------------------------
def main():
    # Rebuild the updated workbook from the original source before running the existing analysis.
    create_updated_workbook()

    if K1_MIN > K1_MAX:
        raise ValueError(f"K1_MIN ({K1_MIN}) must be <= K1_MAX ({K1_MAX}).")
    if K2_MIN > K2_MAX:
        raise ValueError(f"K2_MIN ({K2_MIN}) must be <= K2_MAX ({K2_MAX}).")

    if not UPDATED_INPUT_FILE.exists():
        raise FileNotFoundError(
            f"Generated updated workbook not found: {UPDATED_INPUT_FILE}\n"
            "The preprocessing step should create it from ORIGINAL_INPUT_FILE first."
        )

    all_sheets = pd.read_excel(UPDATED_INPUT_FILE, sheet_name=None)
    if SHEET_NAME not in all_sheets:
        raise ValueError(
            f"Sheet {SHEET_NAME!r} not found. Available sheets: {list(all_sheets)}"
        )
    df = all_sheets[SHEET_NAME].copy()

    required = [
        COHORT_COL, PATIENT_ID_COL, STAGE_COL, DATA_TYPE_COL, *PANEL
    ]
    missing = [col for col in required if col not in df.columns]
    if missing:
        raise ValueError(f"Missing required workbook columns: {missing}")

    # Normalize role/cohort labels without changing the original workbook data.
    role = normalized_text(df[DATA_TYPE_COL])
    cohort = normalized_text(df[COHORT_COL])

    # Fixed reference: first 59 Normal patients in original workbook order,
    # matching the test-evaluation code. These rows are not used as calibration
    # or test observations, even if their Data Type labels are incorrect.
    normal_positions = np.flatnonzero(cohort.eq(NORMAL_LABEL).to_numpy())
    if len(normal_positions) < N_REFERENCE_NORMAL:
        raise ValueError(
            f"Need at least {N_REFERENCE_NORMAL} Normal rows for the fixed "
            f"reference set; found {len(normal_positions)}."
        )
    reference_indices = df.index[normal_positions[:N_REFERENCE_NORMAL]]
    reference_df = df.loc[reference_indices].copy()

    # Check reference labels diagnostically; selection is still by first 59
    # Normal rows in workbook order, as required by the original test code.
    reference_role_counts = normalized_text(reference_df[DATA_TYPE_COL]).value_counts()
    if not normalized_text(reference_df[DATA_TYPE_COL]).eq(
        REFERENCE_ROLE.casefold()
    ).all():
        print(
            "WARNING: Not all first 59 Normal reference rows are labelled "
            f"{REFERENCE_ROLE!r} in Data Type. They will still be used as the "
            "fixed reference set and excluded from calibration/test rows."
        )
        print("Reference Data Type counts:", reference_role_counts.to_dict())

    reference_index_set = set(reference_indices)
    is_reference = df.index.to_series().isin(reference_index_set).to_numpy()

    cal1_mask = (
        role.eq(CAL1_ROLE.casefold()).to_numpy()
        & cohort.eq(NORMAL_LABEL).to_numpy()
        & ~is_reference
    )
    cal2_mask = (
        role.eq(CAL2_ROLE.casefold()).to_numpy()
        & ~is_reference
    )
    test_mask = (
        role.eq(TEST_ROLE.casefold()).to_numpy()
        & ~is_reference
    )

    cal1_normals = df.loc[cal1_mask].copy()
    cal2 = df.loc[cal2_mask].copy()
    test_df = df.loc[test_mask].copy()

    cal2_cohort = normalized_text(cal2[COHORT_COL])
    cal2_normals = cal2.loc[cal2_cohort.eq(NORMAL_LABEL)].copy()
    cal2_cancers = cal2.loc[~cal2_cohort.eq(NORMAL_LABEL)].copy()

    if len(cal1_normals) < 2:
        raise ValueError(
            f"At least two Normal patients labelled {CAL1_ROLE!r} are required."
        )
    if cal2_normals.empty:
        raise ValueError(
            f"No Normal patients found with Data Type {CAL2_ROLE!r}."
        )
    if cal2_cancers.empty:
        raise ValueError(
            f"No cancer patients found with Data Type {CAL2_ROLE!r}."
        )
    if test_df.empty:
        raise ValueError(f"No rows found with Data Type {TEST_ROLE!r}.")

    print("Loaded generated updated workbook for analysis:", UPDATED_INPUT_FILE)
    print("Fixed reference normal patients:", len(reference_df))
    print("Calibration 1 normal patients:", len(cal1_normals))
    print("Calibration 2 normal patients:", len(cal2_normals))
    print("Calibration 2 cancer patients:", len(cal2_cancers))
    print("Test patients:", len(test_df))
    print("\nCalibration role counts:")
    print(pd.crosstab(df[COHORT_COL], df[DATA_TYPE_COL]).to_string())

    # Validate biomarker values for all rows that will be processed.
    reference_numeric = safe_numeric_frame(
        reference_df, PANEL, "Fixed reference Normal rows"
    )
    reference_df.loc[:, PANEL] = reference_numeric

    # Build TDA features for Cal1 normals, all Cal2 rows, and Test rows.
    cal1_features = build_features(reference_df, cal1_normals)
    cal2_features = build_features(reference_df, cal2)
    test_features = build_features(reference_df, test_df)

    # Fit regression on Calibration 1 NORMAL patients only.
    regression_model = LinearRegression().fit(
        cal1_features[REGRESSION_PREDICTORS].astype(float),
        cal1_features["Avg_Death_0D"].astype(float),
    )

    # The Distance %(1) reference coordinates are the means of Calibration 1
    # normal Average Birth (1D) and Average Death (1D).
    ref_birth = float(cal1_features["Average Birth (1D)"].mean())
    ref_death = float(cal1_features["Average Death (1D)"].mean())

    cal1_features = add_scores(
        cal1_features, regression_model, ref_birth, ref_death
    )
    cal2_features = add_scores(
        cal2_features, regression_model, ref_birth, ref_death
    )
    test_features = add_scores(
        test_features, regression_model, ref_birth, ref_death
    )

    # Separate Calibration 2 normals/cancers while retaining feature alignment.
    cal2_cohort_features = normalized_text(cal2_features[COHORT_COL])
    normal_features = cal2_features[
        cal2_cohort_features.eq(NORMAL_LABEL)
    ].copy()
    cancer_features = cal2_features[
        ~cal2_cohort_features.eq(NORMAL_LABEL)
    ].copy()

    score_columns = ["Regression Difference Percentage", "Distance %(1)"]
    for features, label in [
        (cal1_features, "Calibration 1 normal"),
        (normal_features, "Calibration 2 normal"),
        (cancer_features, "Calibration 2 cancer"),
        (test_features, "Test"),
    ]:
        score_values = features[score_columns].to_numpy(dtype=float)
        if not np.isfinite(score_values).all():
            raise ValueError(
                f"{label} has missing/non-finite threshold scores. Check "
                "Average Death (0D), regression denominator, and biomarker data."
            )

    reg_cal1_norm = cal1_features[
        "Regression Difference Percentage"
    ].to_numpy(float)
    dist_cal1_norm = cal1_features["Distance %(1)"].to_numpy(float)
    reg_cal2_norm = normal_features[
        "Regression Difference Percentage"
    ].to_numpy(float)
    dist_cal2_norm = normal_features["Distance %(1)"].to_numpy(float)
    reg_cancer = cancer_features[
        "Regression Difference Percentage"
    ].to_numpy(float)
    dist_cancer = cancer_features["Distance %(1)"].to_numpy(float)

    # Candidate thresholds cover both calibration normal groups and Cal2 cancers,
    # restricted to the user-defined inclusive k1/k2 bounds.
    reg_all_normals = np.concatenate([reg_cal1_norm, reg_cal2_norm])
    dist_all_normals = np.concatenate([dist_cal1_norm, dist_cal2_norm])
    k1_candidates = candidate_thresholds(
        reg_all_normals, reg_cancer, K1_MIN, K1_MAX, "K1"
    )
    k2_candidates = candidate_thresholds(
        dist_all_normals, dist_cancer, K2_MIN, K2_MAX, "K2"
    )

    print("\nBOUNDED k1/k2 threshold search")
    print(f"k1 search range: [{K1_MIN}, {K1_MAX}]")
    print(f"k2 search range: [{K2_MIN}, {K2_MAX}]")
    print("k1 candidate count:", len(k1_candidates))
    print("k2 candidate count:", len(k2_candidates))
    print(
        "Maximum candidate pairs to examine:",
        f"{len(k1_candidates) * len(k2_candidates):,}",
    )
    print(
        f"Requirement: Calibration 1 and Calibration 2 normal specificity "
        f"must each be strictly greater than {SPECIFICITY_TARGET:.2%}."
    )

    cancer_cohorts = sorted(
        cancer_features[COHORT_COL].astype(str).str.strip().unique(),
        key=lambda value: value.casefold(),
    )
    cancer_cohort_values = (
        cancer_features[COHORT_COL].astype(str).str.strip().str.casefold().to_numpy()
    )

    best = None
    feasible_count = 0
    specificity_target = SPECIFICITY_TARGET

    # Exhaustive search across every candidate pair. Decision rule is OR.
    for k1 in k1_candidates:
        cal1_reg_positive = reg_cal1_norm > k1
        cal2_reg_positive = reg_cal2_norm > k1
        cancer_reg_positive = reg_cancer > k1

        for k2 in k2_candidates:
            cal1_normal_positive = cal1_reg_positive | (dist_cal1_norm > k2)
            cal2_normal_positive = cal2_reg_positive | (dist_cal2_norm > k2)

            cal1_specificity = float((~cal1_normal_positive).mean())
            cal2_specificity = float((~cal2_normal_positive).mean())

            if not (
                cal1_specificity > specificity_target + 1e-12
                and cal2_specificity > specificity_target + 1e-12
            ):
                continue

            feasible_count += 1
            cancer_positive = cancer_reg_positive | (dist_cancer > k2)

            per_cohort = {}
            sensitivities = []
            for cancer in cancer_cohorts:
                mask = cancer_cohort_values == cancer.casefold()
                n = int(mask.sum())
                tp = int(cancer_positive[mask].sum())
                sensitivity = tp / n if n else np.nan
                per_cohort[cancer] = {
                    "n": n,
                    "TP": tp,
                    "FN": n - tp,
                    "sensitivity": sensitivity,
                }
                if n:
                    sensitivities.append(sensitivity)

            if not sensitivities:
                continue

            macro_sensitivity = float(np.mean(sensitivities))
            overall_sensitivity = float(cancer_positive.mean())
            cal2_fp = int(cal2_normal_positive.sum())
            cal2_tn = int(len(cal2_normal_positive) - cal2_fp)

            candidate = {
                "k1": float(k1),
                "k2": float(k2),
                "cal1_specificity": cal1_specificity,
                "cal2_specificity": cal2_specificity,
                "minimum_calibration_specificity": min(
                    cal1_specificity, cal2_specificity
                ),
                "macro_sensitivity": macro_sensitivity,
                "overall_sensitivity": overall_sensitivity,
                "FP": cal2_fp,
                "TN": cal2_tn,
                "per_cohort": per_cohort,
            }

            # Deterministic tie-breaking, consistent with the calibration script.
            current_key = (
                candidate["macro_sensitivity"],
                candidate["overall_sensitivity"],
                candidate["minimum_calibration_specificity"],
                candidate["k1"],
                candidate["k2"],
            )
            best_key = (
                (
                    best["macro_sensitivity"],
                    best["overall_sensitivity"],
                    best["minimum_calibration_specificity"],
                    best["k1"],
                    best["k2"],
                )
                if best is not None
                else None
            )
            if best is None or current_key > best_key:
                best = candidate

    if best is None:
        raise RuntimeError(
            "No k1/k2 pair achieved BOTH Calibration 1 and Calibration 2 "
            f"normal specificity strictly above {specificity_target:.2%}. "
            "No thresholds were relaxed. Check the target and calibration data."
        )

    k1 = best["k1"]
    k2 = best["k2"]

    # Apply automatically selected thresholds to Calibration 1, Calibration 2,
    # and Test using the same OR rule. Test data were not used to choose k1/k2.
    for features in (cal1_features, cal2_features, test_features):
        features["Positive Conditions Count"] = (
            (features["Regression Difference Percentage"] > k1).astype(int)
            + (features["Distance %(1)"] > k2).astype(int)
        )
        features["Cancer_Prediction"] = np.where(
            features["Positive Conditions Count"] >= 1, "Positive", "Negative"
        )
        features["Result(Cancer)"] = features["Cancer_Prediction"]

    cal1_features["Actual_Label"] = "Normal"
    cal2_features["Actual_Label"] = np.where(
        normalized_text(cal2_features[COHORT_COL]).eq(NORMAL_LABEL),
        "Normal",
        "Cancer",
    )

    # Recalculate calibration specificity from the actual selected predictions.
    cal1_fp = int(cal1_features["Cancer_Prediction"].eq("Positive").sum())
    cal1_tn = int(len(cal1_features) - cal1_fp)
    cal1_specificity = cal1_tn / len(cal1_features)

    cal2_normal_mask = normalized_text(cal2_features[COHORT_COL]).eq(NORMAL_LABEL)
    cal2_normal_predictions = cal2_features.loc[
        cal2_normal_mask, "Cancer_Prediction"
    ].eq("Positive")
    cal2_fp = int(cal2_normal_predictions.sum())
    cal2_tn = int(len(cal2_normal_predictions) - cal2_fp)
    cal2_specificity = cal2_tn / (cal2_tn + cal2_fp)

    # Calibration 2 cancer cohort metrics (same result as the threshold script).
    cohort_rows = []
    for cancer, result in best["per_cohort"].items():
        cohort_rows.append(
            {
                "Cohort": cancer,
                "n": result["n"],
                "TP": result["TP"],
                "FN": result["FN"],
                "Sensitivity": result["sensitivity"],
            }
        )
    cal2_cohort_metrics = pd.DataFrame(cohort_rows)

    threshold_summary = pd.DataFrame(
        [
            {
                "biomarker_panel": " | ".join(PANEL),
                "decision_rule": (
                    "(Regression Difference Percentage > k1) OR "
                    "(Distance %(1) > k2)"
                ),
                "k1_Regression_Difference_Percentage": k1,
                "k2_Distance_1": k2,
                "specificity_target": specificity_target,
                "specificity_requirement": (
                    f"Calibration 1 AND Calibration 2 each strictly greater "
                    f"than {specificity_target:.2%}"
                ),
                "Calibration1_specificity": cal1_specificity,
                "Calibration1_TN": cal1_tn,
                "Calibration1_FP": cal1_fp,
                "Calibration1_normal_n": len(cal1_features),
                "Calibration2_specificity": cal2_specificity,
                "Calibration2_TN": cal2_tn,
                "Calibration2_FP": cal2_fp,
                "Calibration2_normal_n": len(cal2_normal_predictions),
                "Calibration1_specificity_requirement_met": (
                    cal1_specificity > specificity_target
                ),
                "Calibration2_specificity_requirement_met": (
                    cal2_specificity > specificity_target
                ),
                "Minimum_calibration_specificity": min(
                    cal1_specificity, cal2_specificity
                ),
                "Calibration2_macro_sensitivity": best["macro_sensitivity"],
                "Calibration2_overall_cancer_sensitivity": best["overall_sensitivity"],
                "Calibration2_cancer_n": len(cancer_features),
                "k1_min_search_bound": K1_MIN,
                "k1_max_search_bound": K1_MAX,
                "k2_min_search_bound": K2_MIN,
                "k2_max_search_bound": K2_MAX,
                "k1_candidate_count": len(k1_candidates),
                "k2_candidate_count": len(k2_candidates),
                "total_candidate_pairs": len(k1_candidates) * len(k2_candidates),
                "feasible_threshold_pairs": feasible_count,
                "Reference_normal_n": len(reference_df),
                "Regression_model_training": (
                    "Regression Fit Calibration 1 normal patients only"
                ),
                "Threshold_selection": (
                    "Both Calibration 1 and Calibration 2 normal specificities "
                    "strictly above target; maximize Calibration 2 cancer "
                    "macro-sensitivity"
                ),
                "Test_used_for_threshold_selection": False,
            }
        ]
    )

    # ----------------------- TEST-only evaluation -----------------------
    test_features["Ground Truth"] = np.where(
        normalized_text(test_features[COHORT_COL]).eq(NORMAL_LABEL),
        "Negative",
        "Positive",
    )

    # Continuous ROC score for the locked OR rule.
    # Scale each threshold excess by its SD among Calibration 1 normals only.
    # The maximum standardized excess is positive exactly when either OR
    # condition is exceeded, while retaining a continuous ranking for ROC.
    reg_cal1_sd = float(
        cal1_features["Regression Difference Percentage"].std(ddof=0)
    )
    dist_cal1_sd = float(cal1_features["Distance %(1)"].std(ddof=0))
    if not np.isfinite(reg_cal1_sd) or reg_cal1_sd <= 0:
        raise ValueError(
            "Calibration 1 normal SD for Regression Difference Percentage "
            "must be finite and greater than zero to construct the continuous ROC score."
        )
    if not np.isfinite(dist_cal1_sd) or dist_cal1_sd <= 0:
        raise ValueError(
            "Calibration 1 normal SD for Distance %(1) must be finite and "
            "greater than zero to construct the continuous ROC score."
        )

    test_features["ROC Continuous Score"] = np.maximum(
        (test_features["Regression Difference Percentage"] - k1) / reg_cal1_sd,
        (test_features["Distance %(1)"] - k2) / dist_cal1_sd,
    )

    y_test = test_features["Ground Truth"].eq("Positive").astype(int).to_numpy()
    roc_scores_test = test_features["ROC Continuous Score"].to_numpy(dtype=float)
    if len(np.unique(y_test)) != 2:
        raise ValueError(
            "A test-only ROC curve requires both Normal and Cancer observations "
            "in the Test set."
        )
    if not np.isfinite(roc_scores_test).all():
        raise ValueError("The Test ROC continuous scores contain non-finite values.")

    test_fpr, test_tpr, test_roc_thresholds = roc_curve(y_test, roc_scores_test)
    test_roc_auc = float(roc_auc_score(y_test, roc_scores_test))
    test_roc_curve_df = pd.DataFrame({
        "False Positive Rate": test_fpr,
        "True Positive Rate": test_tpr,
        "ROC Threshold": test_roc_thresholds,
    })
    test_roc_summary_df = pd.DataFrame([{
        "Evaluation Set": "Test only",
        "AUC": test_roc_auc,
        "Cancer patients": int(y_test.sum()),
        "Normal patients": int((y_test == 0).sum()),
        "Continuous score": (
            "max((Regression Difference Percentage - k1) / Calibration 1 normal SD, "
            "(Distance %(1) - k2) / Calibration 1 normal SD)"
        ),
        "k1": k1,
        "k2": k2,
        "Calibration 1 regression-score SD": reg_cal1_sd,
        "Calibration 1 distance-score SD": dist_cal1_sd,
    }])

    # Save a publication-ready ROC plot using Test data only.
    import matplotlib.pyplot as plt
    plt.figure(figsize=(7, 6))
    plt.plot(
        test_fpr, test_tpr, linewidth=2,
        label=f"Test ROC (AUC = {test_roc_auc:.3f})"
    )
    plt.plot([0, 1], [0, 1], linestyle="--", linewidth=1, label="Chance")
    plt.xlim(0, 1)
    plt.ylim(0, 1.02)
    plt.xlabel("False Positive Rate (1 - Specificity)")
    plt.ylabel("True Positive Rate (Sensitivity)")
    plt.title("TDAcancer-detect: Test-set ROC Curve")
    plt.legend(loc="lower right")
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.savefig(ROC_PLOT_FILE, dpi=300, bbox_inches="tight")
    plt.close()
    test_features["TP"] = (
        test_features["Ground Truth"].eq("Positive")
        & test_features["Result(Cancer)"].eq("Positive")
    )
    test_features["FN"] = (
        test_features["Ground Truth"].eq("Positive")
        & test_features["Result(Cancer)"].eq("Negative")
    )
    test_features["TN"] = (
        test_features["Ground Truth"].eq("Negative")
        & test_features["Result(Cancer)"].eq("Negative")
    )
    test_features["FP"] = (
        test_features["Ground Truth"].eq("Negative")
        & test_features["Result(Cancer)"].eq("Positive")
    )

    metric_rows = []
    test_normal_mask = normalized_text(test_features[COHORT_COL]).eq(NORMAL_LABEL)
    cancer_test = test_features.loc[~test_normal_mask]
    normal_test = test_features.loc[test_normal_mask]

    metric_rows.append(
        metric_row("Overall sensitivity", int(cancer_test["TP"].sum()), len(cancer_test))
    )
    metric_rows.append(
        metric_row("Overall specificity", int(normal_test["TN"].sum()), len(normal_test))
    )
    metric_rows.append(
        metric_row(
            "Overall accuracy",
            int((test_features["TP"] | test_features["TN"]).sum()),
            len(test_features),
        )
    )

    cancer_types = sorted(cancer_test[COHORT_COL].dropna().unique().tolist())
    for cancer_type in cancer_types:
        subset = test_features[test_features[COHORT_COL].eq(cancer_type)]
        metric_rows.append(
            metric_row(
                f"{cancer_type} sensitivity",
                int(subset["TP"].sum()),
                len(subset),
            )
        )
        # For a binary cancer-vs-normal prediction, TP / all cancer patients
        # is numerically equivalent to sensitivity.
        metric_rows.append(
            metric_row(
                f"{cancer_type} detection accuracy",
                int(subset["TP"].sum()),
                len(subset),
            )
        )

    stage_rows = []
    stage_data = cancer_test.copy()
    stage_data[STAGE_COL] = (
        stage_data[STAGE_COL].fillna("Unknown").astype(str).str.strip()
    )
    for stage in sorted(stage_data[STAGE_COL].unique().tolist()):
        subset = stage_data[stage_data[STAGE_COL].eq(stage)]
        stage_rows.append(
            {
                "Cancer Type": "All cancer types",
                STAGE_COL: stage,
                **metric_row(
                    "Stage-wise sensitivity",
                    int(subset["TP"].sum()),
                    len(subset),
                ),
            }
        )
        for cancer_type in cancer_types:
            type_stage = subset[subset[COHORT_COL].eq(cancer_type)]
            if type_stage.empty:
                continue
            stage_rows.append(
                {
                    "Cancer Type": cancer_type,
                    STAGE_COL: stage,
                    **metric_row(
                        "Stage-wise sensitivity",
                        int(type_stage["TP"].sum()),
                        len(type_stage),
                    ),
                }
            )

    test_metrics_df = pd.DataFrame(metric_rows)
    test_stage_metrics_df = pd.DataFrame(stage_rows)

    # Output feature columns preserve all useful diagnostics.
    feature_export_cols = [
        "Source_Row_Index",
        PATIENT_ID_COL,
        COHORT_COL,
        STAGE_COL,
        DATA_TYPE_COL,
        *FEATURE_COLUMNS,
        "Predicted Avg_Death_0D",
        "Difference",
        "Regression Difference Percentage",
        "Distance %(1)",
        "ROC Continuous Score",
        "Positive Conditions Count",
        "Cancer_Prediction",
        "Result(Cancer)",
        "Actual_Label",
        "Ground Truth",
        "TP",
        "FN",
        "TN",
        "FP",
    ]

    cal2_export = cal2_features[
        [col for col in feature_export_cols if col in cal2_features.columns]
    ].copy()
    cal1_export = cal1_features[
        [col for col in feature_export_cols if col in cal1_features.columns]
    ].copy()
    test_export = test_features[
        [col for col in feature_export_cols if col in test_features.columns]
    ].copy()

    # Retain CSV deliverables from the original test-evaluation script.
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    test_export.to_csv(PREDICTIONS_CSV, index=False)
    test_metrics_df.to_csv(METRICS_CSV, index=False)
    test_stage_metrics_df.to_csv(STAGE_METRICS_CSV, index=False)
    test_roc_curve_df.to_csv(ROC_CURVE_CSV, index=False)

    # Add predictions only to Test rows in the copied original worksheet.
    df_out = df.copy()
    output_columns = [
        "Predicted Avg_Death_0D",
        "Difference",
        "Regression Difference Percentage",
        "Distance %(1)",
        "Positive Conditions Count",
        "Cancer_Prediction",
        "Result(Cancer)",
        "Ground Truth",
        "TP",
        "FN",
        "TN",
        "FP",
    ]
    numeric_columns = {
        "Predicted Avg_Death_0D",
        "Difference",
        "Regression Difference Percentage",
        "Distance %(1)",
        "Positive Conditions Count",
    }
    boolean_columns = {"TP", "FN", "TN", "FP"}
    for col in output_columns:
        if col not in df_out.columns:
            if col in numeric_columns:
                df_out[col] = pd.Series(np.nan, index=df_out.index, dtype="float64")
            else:
                df_out[col] = pd.Series(index=df_out.index, dtype="object")

    # Ensure all output columns are blank for non-Test rows before assigning tests.
    for col in output_columns:
        if col in numeric_columns:
            df_out[col] = pd.to_numeric(df_out[col], errors="coerce")
            df_out.loc[:, col] = np.nan
        else:
            df_out[col] = df_out[col].astype("object")
            df_out.loc[:, col] = None

    for col in output_columns:
        if col in test_features.columns:
            df_out.loc[test_features.index, col] = test_features[col].reindex(
                test_features.index
            )

    # Preserve all original sheets and add every output from both source scripts.
    all_sheets[SHEET_NAME] = df_out
    all_sheets["Threshold Summary"] = threshold_summary
    all_sheets["Calibration2 Cancer Metrics"] = cal2_cohort_metrics
    all_sheets["Calibration2 Predictions"] = cal2_export
    all_sheets["Calibration1 Diagnostics"] = cal1_export
    all_sheets["Test Predictions"] = test_export
    all_sheets["Test Metrics"] = test_metrics_df
    all_sheets["Test Stage Metrics"] = test_stage_metrics_df
    all_sheets["Test ROC Curve"] = test_roc_curve_df
    all_sheets["Test ROC Summary"] = test_roc_summary_df

    # Excel sheet names are limited to 31 characters; these names are unique
    # after truncation in this workbook.
    with pd.ExcelWriter(OUTPUT_EXCEL, engine="openpyxl") as writer:
        for sheet_name_out, sheet_df in all_sheets.items():
            sheet_df.to_excel(
                writer,
                sheet_name=str(sheet_name_out)[:31],
                index=False,
            )

    # ----------------------------- Console report -----------------------------
    print("\n" + "=" * 78)
    print("AUTOMATICALLY SELECTED k1 AND k2 — BOUNDED OR DECISION RULE")
    print("=" * 78)
    print(f"k1 (Regression Difference Percentage; allowed [{K1_MIN}, {K1_MAX}]): {k1:.12g}")
    print(f"k2 (Distance %(1); allowed [{K2_MIN}, {K2_MAX}]): {k2:.12g}")
    print(
        "Decision: Positive if Regression Difference Percentage > k1 "
        "OR Distance %(1) > k2"
    )
    print(f"Calibration 1 normal specificity: {cal1_specificity:.4%} "
          f"({cal1_tn}/{cal1_tn + cal1_fp})")
    print(f"Calibration 2 normal specificity: {cal2_specificity:.4%} "
          f"({cal2_tn}/{cal2_tn + cal2_fp})")
    print(f"Required specificity for EACH calibration group: > {specificity_target:.4%}")
    print(f"Calibration 2 macro-sensitivity: {best['macro_sensitivity']:.4%}")
    print(f"Calibration 2 overall cancer sensitivity: {best['overall_sensitivity']:.4%}")
    print(f"Feasible threshold pairs checked: {feasible_count}")
    print("\nCALIBRATION 2 SENSITIVITY BY CANCER COHORT")
    print(cal2_cohort_metrics.to_string(index=False))

    print("\nTEST-SET PERFORMANCE ONLY")
    print(test_metrics_df.to_string(index=False, formatters={
        "Rate (%)": lambda x: f"{x:.2f}" if pd.notna(x) else "NA",
        "95% CI Lower (%)": lambda x: f"{x:.2f}" if pd.notna(x) else "NA",
        "95% CI Upper (%)": lambda x: f"{x:.2f}" if pd.notna(x) else "NA",
    }))

    print(f"\nTEST-SET ROC-AUC (continuous score): {test_roc_auc:.6f}")
    print("ROC plot:", ROC_PLOT_FILE)

    print("\nSTAGE-WISE TEST-SET SENSITIVITY")
    if test_stage_metrics_df.empty:
        print("No stage data available for cancer test rows.")
    else:
        print(test_stage_metrics_df.to_string(index=False))

    print("\nSaved files:")
    print("Combined Excel workbook:", OUTPUT_EXCEL)
    print("Test predictions CSV:", PREDICTIONS_CSV)
    print("Test metrics CSV:", METRICS_CSV)
    print("Test stage metrics CSV:", STAGE_METRICS_CSV)
    print("Test ROC curve CSV:", ROC_CURVE_CSV)
    print("Test ROC plot:", ROC_PLOT_FILE)
    print("Original input workbook was not overwritten.")
    
    return (
        threshold_summary,
        cal2_cohort_metrics,
        cal2_export,
        cal1_export,
        test_export,
        test_metrics_df,
        test_stage_metrics_df,
        test_roc_curve_df,
        test_roc_summary_df,
    )


if __name__ == "__main__":
    main()
```

```python
"""
TDAcancer-detect-2: multi-target one-vs-rest tissue localization.

Workflow
--------
1. Read the clinical validation workbook and retain the fixed 59 Reference normal patients.
2. Across each repeated random split, partition cancer true positives within each cohort into
   10% Localization Calibration 1, 55% independent Localization Calibration 2, and 35%
   held-out localization Test.
3. Select 10% of normal true negatives for Localization Calibration 1 to calculate the
   normal feature centroid used by Distance %(3). Exclude the remaining normal true negatives.
4. Exclude normal false positives and cancer false negatives from localization.
5. For each target cancer type, classify that cohort as positive and the other cancer cohorts
   as negative. Normal patients are used only to construct the Calibration-1 reference centroid.
6. Evaluate all unique three-biomarker panels using the GUDHI-based TDA feature calculation
   and Distance %(3) threshold procedure.
7. Use independent Calibration 2 to tune the threshold and select one panel per target,
   subject to the configured strict specificity constraint. Rank eligible panels by
   Calibration-2 Recall/Sensitivity, with specificity and precision as tie-breakers.
8. Evaluate the selected panel and locked threshold once on the held-out Test set.
9. Repeat the patient-level partition and analysis for the configured number of random splits.
   Summarize held-out performance and biomarker recurrence across successful splits.
10. Save the results workbook and run log in the TDAcancer-detect-2 output folder.

Interpretation
--------------
Localization performance is conditional on the first-stage cancer detection result and the
specified localization groups. Cancer true positives are divided into Calibration 1,
independent Calibration 2, and held-out Test subsets. Normal true negatives are sampled
separately for the Calibration-1 reference centroid; remaining normal true negatives,
normal false positives, and cancer false negatives are excluded from localization evaluation.
Calibration 2 is used for threshold tuning and biomarker-panel selection. Held-out Test
patients are not used for either selection step.
"""

# Install dependencies:
# python -m pip install numpy pandas gudhi openpyxl

from pathlib import Path
from itertools import combinations
import math
import warnings

import gudhi
import numpy as np
import pandas as pd


# =============================================================================
# USER-DEFINED SETTINGS
# =============================================================================

BASE_DIR = Path(r"D:\Cancer Detection folder\Biomarker Selection")
INPUT_WORKBOOK = BASE_DIR / "clinical cancer validation data_Updated_with_Test_Performance.xlsx"
SHEET_NAME = "Normal and Cancer"
# All localization outputs are saved separately in this folder.
OUTPUT_DIR = BASE_DIR / "TDAcancer-detect-2"
# Consolidated workbook containing the main results and audit tables.
RESULTS_WORKBOOK = OUTPUT_DIR / "TDAcancer-detect-2_10_Random_Splits_MeanPerformance_BiomarkerRecurrence.xlsx"
# Updated copy of the input workbook with localization predictions and combined results.
UPDATED_SOURCE_WORKBOOK = OUTPUT_DIR / "TDAcancer-detect-2_Updated.xlsx"

COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
DATA_TYPE_COL = "Data Type"
REFERENCE_ROLE = "Reference"
ORIGINAL_TEST_ROLE = "Test"

# Keep this as 59 for the supplied workbook. The code checks that exactly 59
# normal patients are labelled Reference; it does not select a new reference set.
N_REFERENCE_NORMAL = 59

# User-defined target cancer types. Each target is run independently as one-vs-rest.
# For each target, all other observed cohorts (including Normal) are negatives.
# Use the exact cohort labels found in the workbook.
TARGET_CANCER_TYPES = ["Bladder", "Ovarian", "Pancreatic"]

# Enter the candidate biomarkers exactly as they appear in the Excel headers.
PRESELECTED_BIOMARKERS = ['sIL-6Ra', 'CA-125', 'CA 19-9', 'IL-6', 'IL-8', 'OPN', 'TRAIL', 'Total PSA']
PANEL_SIZE = 3
TOP_N_TO_REPORT = 1

# Cancer true-positive patients are partitioned within each cohort:
# 10% Calibration 1, 55% independent Calibration 2, and the remaining 35% Test.
LOCALIZATION_CAL1_FRACTION = 0.10
LOCALIZATION_CAL2_FRACTION = 0.55

# A separate 10% of normal true-negative patients is used in Calibration 1
# to calculate the normal feature centroid for Distance %(3). The remaining
# normal true-negative patients are excluded from localization evaluation.
NORMAL_CAL1_FRACTION = 0.10

# Calibration 2 is independent of Calibration 1 and is used for threshold tuning.
REUSE_CAL1_FOR_CAL2 = False
RANDOM_SEED = 20261109
N_RANDOM_SPLITS = 10  # Number of repeated patient-level localization partitions.

# Localization threshold selection.
SPECIFICITY_TARGET = 0.90
OFFSETS = list(range(-10, 11))  # -10,...,+10, inclusive.
# Score scale is Distance %(3), the Euclidean distance from the normal
# Calibration-1 feature centroid multiplied by 100.

PRINT_EVERY = 5

# Excel workbook column names for the original 0D features are created internally:
FEATURE_MIN_LIFE = "0-Dim Hole Min Life Range"
FEATURE_RANGE_DEATH = "Range_Death_0D"
DISTANCE_COL = "Distance %(3)"


# =============================================================================
# GENERAL HELPERS
# =============================================================================

def norm_text(value):
    return str(value).strip().casefold()


def validate_settings(df):
    if not INPUT_WORKBOOK.exists():
        raise FileNotFoundError(
            f"Input workbook not found: {INPUT_WORKBOOK}\n"
            "Edit INPUT_WORKBOOK and OUTPUT_DIR in USER-DEFINED SETTINGS."
        )
    if not isinstance(PANEL_SIZE, int) or PANEL_SIZE != 3:
        raise ValueError("This localization script is designed for 3-biomarker panels; set PANEL_SIZE = 3.")
    if len(PRESELECTED_BIOMARKERS) < PANEL_SIZE:
        raise ValueError(
            f"Need at least {PANEL_SIZE} preselected biomarkers; "
            f"received {len(PRESELECTED_BIOMARKERS)}."
        )
    if len(set(PRESELECTED_BIOMARKERS)) != len(PRESELECTED_BIOMARKERS):
        raise ValueError("PRESELECTED_BIOMARKERS contains duplicates.")
    missing = [b for b in PRESELECTED_BIOMARKERS if b not in df.columns]
    if missing:
        raise ValueError(f"Biomarker columns not found in workbook: {missing}")
    required = [COHORT_COL, DATA_TYPE_COL, PATIENT_ID_COL, "Result(Cancer)"]
    missing_required = [c for c in required if c not in df.columns]
    if missing_required:
        raise ValueError(f"Required workbook columns are missing: {missing_required}")
    if not (0 < LOCALIZATION_CAL1_FRACTION < 1):
        raise ValueError("LOCALIZATION_CAL1_FRACTION must be between 0 and 1.")
    if not (0 < LOCALIZATION_CAL2_FRACTION < 1):
        raise ValueError("LOCALIZATION_CAL2_FRACTION must be between 0 and 1.")
    if LOCALIZATION_CAL1_FRACTION + LOCALIZATION_CAL2_FRACTION >= 1:
        raise ValueError(
            "LOCALIZATION_CAL1_FRACTION + LOCALIZATION_CAL2_FRACTION must be less than 1 "
            "so that patients remain for the held-out Test set."
        )
    if not (0 < NORMAL_CAL1_FRACTION <= 1):
        raise ValueError("NORMAL_CAL1_FRACTION must be greater than 0 and at most 1.")
    if not (0 < SPECIFICITY_TARGET < 1):
        raise ValueError("SPECIFICITY_TARGET must be between 0 and 1.")


def make_localization_split(df, random_seed=None):
    """
    Construct independent localization Calibration-1, Calibration-2 and Test groups.

    Cancer true positives are partitioned separately within each cancer cohort:
      - 10% Localization Calibration 1 (starting threshold construction),
      - 55% Localization Calibration 2 (independent threshold tuning and panel selection),
      - remaining 35% Localization Test (locked final evaluation).

    Normal true negatives are sampled separately: 10% are assigned to Calibration 1
    to calculate the normal feature centroid used by Distance %(3); the remainder
    are excluded. Normal false positives and cancer false negatives are excluded.
    The fixed 59 Reference normal patients remain unchanged.
    """
    out = df.copy()
    role = pd.Series("Excluded", index=out.index, dtype="object")
    source_class = pd.Series("Excluded", index=out.index, dtype="object")

    original_role = out[DATA_TYPE_COL].astype(str).str.strip().str.casefold()
    cohort_norm = out[COHORT_COL].astype(str).str.strip().str.casefold()
    result_norm = out["Result(Cancer)"].astype(str).str.strip().str.casefold()

    is_reference = original_role.eq(REFERENCE_ROLE.casefold())
    is_original_test = original_role.eq(ORIGINAL_TEST_ROLE.casefold())
    is_normal = cohort_norm.eq("normal")
    is_cancer = out[COHORT_COL].notna() & ~is_normal
    detection_positive = result_norm.eq("positive")
    detection_negative = result_norm.eq("negative")

    is_cancer_tp = is_original_test & is_cancer & detection_positive
    is_normal_tn = is_original_test & is_normal & detection_negative
    is_normal_fp = is_original_test & is_normal & detection_positive
    is_cancer_fn = is_original_test & is_cancer & detection_negative

    ref_normals = out.loc[is_reference & is_normal]
    if len(ref_normals) != N_REFERENCE_NORMAL:
        raise ValueError(
            f"Expected exactly {N_REFERENCE_NORMAL} Reference normal rows; "
            f"found {len(ref_normals)}. The fixed reference workbook may have changed."
        )

    role.loc[is_reference] = "Reference"
    source_class.loc[is_reference] = "Fixed Reference Normal"
    source_class.loc[is_cancer_tp] = "Cancer True Positive"
    source_class.loc[is_normal_tn] = "Normal True Negative"
    source_class.loc[is_normal_fp] = "Normal False Positive"
    source_class.loc[is_cancer_fn] = "Cancer False Negative"

    seed = RANDOM_SEED if random_seed is None else int(random_seed)
    rng = np.random.RandomState(seed)
    split_counts = []

    def allocate_cancer_three_way(indices, cohort_label):
        """Allocate a cohort into Cal-1, independent Cal-2 and held-out Test."""
        shuffled = np.asarray(list(indices), dtype=object)
        rng.shuffle(shuffled)
        n_total = len(shuffled)
        if n_total < 3:
            raise ValueError(
                f"Cohort '{cohort_label}' has only {n_total} cancer true-positive "
                "patients in the original detection Test set. At least 3 are required "
                "to allocate non-empty Calibration 1, Calibration 2 and Test groups."
            )

        n_cal1 = int(round(n_total * LOCALIZATION_CAL1_FRACTION))
        n_cal1 = min(max(n_cal1, 1), n_total - 2)

        n_cal2 = int(round(n_total * LOCALIZATION_CAL2_FRACTION))
        n_cal2 = min(max(n_cal2, 1), n_total - n_cal1 - 1)

        cal1_idx = list(shuffled[:n_cal1])
        cal2_idx = list(shuffled[n_cal1:n_cal1 + n_cal2])
        test_idx = list(shuffled[n_cal1 + n_cal2:])

        role.loc[cal1_idx] = "Localization Calibration 1"
        role.loc[cal2_idx] = "Localization Calibration 2"
        role.loc[test_idx] = "Test"

        split_counts.append({
            "Cohort": cohort_label,
            "Detection_Test_Class": "Cancer True Positive",
            "Original_Test_Patients": n_total,
            "Localization_Calibration_1": len(cal1_idx),
            "Localization_Calibration_2": len(cal2_idx),
            "Localization_Test": len(test_idx),
            "Excluded_Remainder": 0,
            "Calibration_1_Fraction": len(cal1_idx) / n_total,
            "Calibration_2_Fraction": len(cal2_idx) / n_total,
            "Test_Fraction": len(test_idx) / n_total,
        })

    cancer_tp_idx = out.index[is_cancer_tp].tolist()
    if len(cancer_tp_idx) == 0:
        raise ValueError(
            "No cancer true-positive patients were found in the original detection Test set."
        )

    for cohort in sorted(cohort_norm.loc[cancer_tp_idx].unique()):
        cohort_idx = [i for i in cancer_tp_idx if cohort_norm.loc[i] == cohort]
        allocate_cancer_three_way(cohort_idx, cohort)

    # Normal true negatives: 10% are used only to define the normal centroid.
    normal_tn_idx = out.index[is_normal_tn].tolist()
    if len(normal_tn_idx) == 0:
        raise ValueError(
            "No normal true-negative patients were found in the original detection "
            "Test set; these are required to calculate the Distance %(3) reference centroid."
        )

    shuffled_normal = np.asarray(normal_tn_idx, dtype=object)
    rng.shuffle(shuffled_normal)
    n_normal_cal1 = int(round(len(shuffled_normal) * NORMAL_CAL1_FRACTION))
    n_normal_cal1 = min(max(n_normal_cal1, 1), len(shuffled_normal))
    normal_cal1_idx = list(shuffled_normal[:n_normal_cal1])
    role.loc[normal_cal1_idx] = "Localization Calibration 1"

    split_counts.append({
        "Cohort": "Normal",
        "Detection_Test_Class": "Normal True Negative",
        "Original_Test_Patients": len(normal_tn_idx),
        "Localization_Calibration_1": len(normal_cal1_idx),
        "Localization_Calibration_2": 0,
        "Localization_Test": 0,
        "Excluded_Remainder": len(normal_tn_idx) - len(normal_cal1_idx),
        "Calibration_1_Fraction": len(normal_cal1_idx) / len(normal_tn_idx),
        "Calibration_2_Fraction": 0.0,
        "Test_Fraction": 0.0,
    })

    normal_fp_idx = out.index[is_normal_fp].tolist()
    role.loc[normal_fp_idx] = "Excluded"
    split_counts.append({
        "Cohort": "Normal",
        "Detection_Test_Class": "Normal False Positive",
        "Original_Test_Patients": len(normal_fp_idx),
        "Localization_Calibration_1": 0,
        "Localization_Calibration_2": 0,
        "Localization_Test": 0,
        "Excluded_Remainder": len(normal_fp_idx),
        "Calibration_1_Fraction": 0.0,
        "Calibration_2_Fraction": 0.0,
        "Test_Fraction": 0.0,
    })

    role.loc[is_cancer_fn] = "Excluded"

    out["Localization Role"] = role
    out["Localization Detection Class"] = source_class
    out["Used_For_Localization_Experiment"] = out["Localization Role"].isin(
        ["Reference", "Localization Calibration 1",
         "Localization Calibration 2", "Test"]
    )

    if out.loc[is_cancer_tp, "Localization Role"].eq("Excluded").any():
        raise RuntimeError("Some cancer true-positive Test patients were not assigned.")
    if out.loc[is_normal_fp, "Localization Role"].ne("Excluded").any():
        raise RuntimeError("Normal false-positive patients must be excluded from localization.")
    if out.loc[is_normal_tn, "Localization Role"].eq("Test").any():
        raise RuntimeError("Normal true-negative patients must not be assigned to localization Test.")
    if out.loc[is_cancer_fn, "Localization Role"].ne("Excluded").any():
        raise RuntimeError("Cancer false-negative patients must be excluded from localization.")

    return out, pd.DataFrame(split_counts)



def binary_truth(df, target, negatives):
    cohort = df[COHORT_COL].astype(str).str.strip().str.casefold()
    target_n = norm_text(target)
    neg_n = {norm_text(x) for x in negatives}
    included = cohort.eq(target_n) | cohort.isin(neg_n)
    y = cohort.eq(target_n).astype(int)
    return included, y


def wilson_ci(successes, trials, confidence=0.95):
    """Return the two-sided Wilson score confidence interval for a binomial proportion."""
    if trials <= 0:
        return (np.nan, np.nan)

    # 1.959963984540054 is the two-sided 95% normal quantile.
    from statistics import NormalDist
    z = NormalDist().inv_cdf(0.5 + confidence / 2.0)
    p_hat = successes / trials
    denominator = 1.0 + (z ** 2 / trials)
    centre = (p_hat + z ** 2 / (2.0 * trials)) / denominator
    half_width = (
        z * math.sqrt((p_hat * (1.0 - p_hat) / trials) + (z ** 2 / (4.0 * trials ** 2)))
        / denominator
    )
    return (max(0.0, centre - half_width), min(1.0, centre + half_width))


def bootstrap_f1_ci(y_true, y_pred, confidence=0.95, n_bootstrap=5000, seed=20261019):
    """Percentile-bootstrap CI for F1, resampling test observations with replacement."""
    y_true = np.asarray(y_true, dtype=int)
    y_pred = np.asarray(y_pred, dtype=int)
    if len(y_true) == 0:
        return (np.nan, np.nan)

    rng = np.random.default_rng(seed)
    n = len(y_true)
    values = np.empty(n_bootstrap, dtype=float)

    for b in range(n_bootstrap):
        idx = rng.integers(0, n, size=n)
        yt = y_true[idx]
        yp = y_pred[idx]
        tp_b = int(((yt == 1) & (yp == 1)).sum())
        fp_b = int(((yt == 0) & (yp == 1)).sum())
        fn_b = int(((yt == 1) & (yp == 0)).sum())
        precision_b = tp_b / (tp_b + fp_b) if tp_b + fp_b else 0.0
        recall_b = tp_b / (tp_b + fn_b) if tp_b + fn_b else 0.0
        values[b] = (
            2.0 * precision_b * recall_b / (precision_b + recall_b)
            if precision_b + recall_b else 0.0
        )

    alpha = 1.0 - confidence
    return (
        float(np.quantile(values, alpha / 2.0)),
        float(np.quantile(values, 1.0 - alpha / 2.0)),
    )


def calculate_metrics(y_true, y_pred, include_confidence_intervals=False):
    y_true = np.asarray(y_true, dtype=int)
    y_pred = np.asarray(y_pred, dtype=int)
    tp = int(((y_true == 1) & (y_pred == 1)).sum())
    fp = int(((y_true == 0) & (y_pred == 1)).sum())
    fn = int(((y_true == 1) & (y_pred == 0)).sum())
    tn = int(((y_true == 0) & (y_pred == 0)).sum())
    precision = tp / (tp + fp) if tp + fp else 0.0
    recall = tp / (tp + fn) if tp + fn else 0.0
    f1 = (2 * precision * recall / (precision + recall)
          if precision + recall else 0.0)
    accuracy = (tp + tn) / (tp + fp + fn + tn) if tp + fp + fn + tn else 0.0
    specificity = tn / (tn + fp) if tn + fp else 0.0

    result = {
        "TP": tp, "FP": fp, "FN": fn, "TN": tn,
        "Recall": recall, "Precision": precision, "F1": f1,
        "Accuracy": accuracy, "Specificity": specificity,
        "N": int(len(y_true)),
    }

    # Confidence intervals are calculated for the held-out Test evaluation
    # of the single calibration-selected panel for each target and split.
    if include_confidence_intervals:
        recall_ci = wilson_ci(tp, tp + fn)
        precision_ci = wilson_ci(tp, tp + fp)
        specificity_ci = wilson_ci(tn, tn + fp)
        accuracy_ci = wilson_ci(tp + tn, len(y_true))
        f1_ci = bootstrap_f1_ci(y_true, y_pred)

        result.update({
            "Recall_CI_Lower": recall_ci[0], "Recall_CI_Upper": recall_ci[1],
            "Precision_CI_Lower": precision_ci[0], "Precision_CI_Upper": precision_ci[1],
            "F1_CI_Lower": f1_ci[0], "F1_CI_Upper": f1_ci[1],
            "Accuracy_CI_Lower": accuracy_ci[0], "Accuracy_CI_Upper": accuracy_ci[1],
            "Specificity_CI_Lower": specificity_ci[0], "Specificity_CI_Upper": specificity_ci[1],
        })

    return result


# =============================================================================
# TDA FEATURE CALCULATION
# =============================================================================

def calculate_tda_features(reference_df, patient_row, panel):
    panel = list(panel)
    ref_values = reference_df[panel].apply(pd.to_numeric, errors="coerce")
    patient_values = pd.to_numeric(patient_row[panel], errors="coerce").to_frame().T
    combined = pd.concat([ref_values, patient_values], ignore_index=True)
    values = combined.to_numpy(dtype=float)

    if not np.isfinite(values).all():
        raise ValueError(
            f"Missing/non-finite biomarker value for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}, panel={panel}"
        )
    if (values <= 0).any():
        raise ValueError(
            "The supplied method uses log concentrations; all selected biomarker "
            f"values must be > 0. Patient={patient_row.get(PATIENT_ID_COL, 'unknown')}, "
            f"panel={panel}"
        )

    logged = np.log(values)
    corr = np.corrcoef(logged, rowvar=False)
    if np.ndim(corr) == 0:
        corr = np.array([[1.0]])
    dissimilarity_matrix = 1.0 - np.abs(corr)
    dissimilarity_matrix = (
        dissimilarity_matrix + dissimilarity_matrix.T
    ) / 2.0
    np.fill_diagonal(dissimilarity_matrix, 0.0)
    dissimilarity_matrix = np.clip(dissimilarity_matrix, 0.0, 1.0)
    if not np.isfinite(dissimilarity_matrix).all():
        raise ValueError(
            f"Invalid correlation/dissimilarity matrix for panel {panel}."
        )

    rips = gudhi.RipsComplex(distance_matrix=dissimilarity_matrix)
    simplex_tree = rips.create_simplex_tree(max_dimension=2)
    persistence = simplex_tree.persistence()
    finite = [
        (dim, (birth, death))
        for dim, (birth, death) in persistence
        if np.isfinite(death)
    ]
    holes0 = [pair for dim, pair in finite if dim == 0]

    if holes0:
        deaths0 = [death for _, death in holes0]
        lifetimes0 = [death - birth for birth, death in holes0]
        min_life0 = float(min(lifetimes0))
        range_death0 = float(max(deaths0) - min(deaths0))
    else:
        min_life0 = 0.0
        range_death0 = 0.0

    return {
        FEATURE_MIN_LIFE: min_life0,
        FEATURE_RANGE_DEATH: range_death0,
    }


def extract_features_for_panel(df, reference_df, panel, cache):
    key = tuple(panel)
    if key in cache:
        return cache[key].copy()

    rows = []
    for idx, patient in df.iterrows():
        # The fixed Reference rows define the baseline point cloud. Rows outside the
        # selected calibration/test subset are not evaluated in this experiment.
        role_name = str(patient["Localization Role"]).strip().casefold()
        if role_name in {"reference", "excluded"}:
            continue
        feat = calculate_tda_features(reference_df, patient, panel)
        rows.append({
            "_row_index": idx,
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
            COHORT_COL: patient[COHORT_COL],
            "Localization Role": patient["Localization Role"],
            FEATURE_MIN_LIFE: feat[FEATURE_MIN_LIFE],
            FEATURE_RANGE_DEATH: feat[FEATURE_RANGE_DEATH],
        })
    features = pd.DataFrame(rows).set_index("_row_index", drop=True)

    # Use only normal true-negative patients assigned to Localization Calibration 1
    # to calculate the normal feature centroid for Distance %(3). The fixed 59
    # Reference patients remain the baseline matrix used to compute each patient's
    # TDA features; normal false positives and localization Test patients do not
    # contribute to this centroid.
    cal1_normal_mask = (
        features["Localization Role"].astype(str).str.casefold().eq("localization calibration 1")
        & features[COHORT_COL].astype(str).str.strip().str.casefold().eq("normal")
    )
    cal1_normal_features = features.loc[cal1_normal_mask]
    if cal1_normal_features.empty:
        raise ValueError(
            "Localization Calibration 1 contains no normal true-negative patients; "
            "a normal 0D reference point cannot be constructed."
        )
    ref_point = np.array([
        cal1_normal_features[FEATURE_MIN_LIFE].mean(),
        cal1_normal_features[FEATURE_RANGE_DEATH].mean(),
    ], dtype=float)

    xy = features[[FEATURE_MIN_LIFE, FEATURE_RANGE_DEATH]].to_numpy(dtype=float)
    features[DISTANCE_COL] = np.sqrt(((xy - ref_point) ** 2).sum(axis=1)) * 100.0
    features.attrs["reference_point"] = tuple(ref_point.tolist())
    cache[key] = features
    return features.copy()


# =============================================================================
# THRESHOLD CONSTRUCTION AND CALIBRATION-2 PANEL SCORING
# =============================================================================

def get_target_specific_base_k3(features, df, target):
    """
    Calibration-1 construction:
      k3_base = ceil(mean Distance %(3) among target-tissue Calibration-1 patients).
    Distance %(3) is measured from the centroid of normal true-negative patients
    assigned to Localization Calibration 1. Calibration-1 target labels are used
    only to construct this starting threshold.
    """
    role = features["Localization Role"].astype(str).str.casefold()
    cohort = features[COHORT_COL].astype(str).str.strip().str.casefold()
    target_mask = (
        role.eq("localization calibration 1")
        & cohort.eq(norm_text(target))
    )
    scores = features.loc[target_mask, DISTANCE_COL].dropna().to_numpy(dtype=float)
    if len(scores) == 0:
        raise ValueError(
            f"No {target} patients in Localization Calibration 1; "
            "increase LOCALIZATION_CAL1_FRACTION or check the cohort label."
        )
    return int(np.ceil(float(np.mean(scores)))), len(scores)


def tune_k3_on_calibration2(features, df, target, negatives, k3_base):
    """
    Search k3 = k3_base + offset for each configured value in OFFSETS.
    Among candidates whose Calibration-2 specificity is STRICTLY greater than
    SPECIFICITY_TARGET, select the threshold with the highest Recall/Sensitivity.
    Resolve ties using Specificity, Precision, then proximity to offset zero.
    If none qualify, apply the same ranking to all candidates and mark
    specificity_constraint_met=False.
    """
    role = features["Localization Role"].astype(str).str.casefold()
    # Calibration 2 is an independent patient subset, separate from the
    # Calibration-1 patients used to construct the starting threshold.
    cal2_idx = features.index[role.eq("localization calibration 2")]
    cal2_features = features.loc[cal2_idx]
    cal2_df = df.loc[cal2_idx]

    included, y = binary_truth(cal2_df, target, negatives)
    included = included.to_numpy(dtype=bool)
    y = y.to_numpy(dtype=int)[included]
    scores = cal2_features[DISTANCE_COL].to_numpy(dtype=float)[included]
    if len(y) == 0 or len(np.unique(y)) < 2:
        raise ValueError(
            "Localization Calibration 2 must contain both target positives and "
            "negative-cohort patients for specificity/Recall-based selection."
        )

    rows = []
    for offset in OFFSETS:
        threshold = float(k3_base + offset)
        pred = (scores > threshold).astype(int)
        metrics = calculate_metrics(y, pred)
        rows.append({
            "k3_base": int(k3_base),
            "offset": int(offset),
            "k3_threshold": threshold,
            **metrics,
            "specificity_constraint_met": bool(metrics["Specificity"] > SPECIFICITY_TARGET),
        })

    eligible = [r for r in rows if r["specificity_constraint_met"]]
    if eligible:
        chosen = max(
            eligible,
            key=lambda r: (r["Recall"], r["Specificity"], r["Precision"], -abs(r["offset"]))
        )
        feasible = True
    else:
        chosen = max(
            rows,
            key=lambda r: (r["Recall"], r["Specificity"], r["Precision"], -abs(r["offset"]))
        )
        feasible = False

    chosen = dict(chosen)
    chosen["specificity_constraint_met"] = feasible
    chosen["n_cal1_target"] = None
    return chosen, pd.DataFrame(rows)


def score_panel(panel, df, reference_df, target, negatives, feature_cache):
    features = extract_features_for_panel(df, reference_df, panel, feature_cache)
    k3_base, n_cal1_target = get_target_specific_base_k3(features, df, target)
    chosen, offset_table = tune_k3_on_calibration2(
        features, df, target, negatives, k3_base
    )
    chosen["n_cal1_target"] = n_cal1_target
    chosen["panel"] = " | ".join(panel)
    chosen["biomarkers"] = tuple(panel)
    chosen["reference_point_min_life"] = features.attrs["reference_point"][0]
    chosen["reference_point_range_death"] = features.attrs["reference_point"][1]
    return chosen, features, offset_table


# =============================================================================
# FINAL TEST EVALUATION AND MAIN
# =============================================================================

def evaluate_test(features, df, target, negatives, threshold):
    role = features["Localization Role"].astype(str).str.casefold()
    test_idx = features.index[role.eq("test")]
    test_df = df.loc[test_idx]
    test_features = features.loc[test_idx]
    included, y = binary_truth(test_df, target, negatives)
    included = included.to_numpy(dtype=bool)
    y = y.to_numpy(dtype=int)[included]
    scores = test_features[DISTANCE_COL].to_numpy(dtype=float)[included]
    pred = (scores > threshold).astype(int)
    metrics = calculate_metrics(y, pred, include_confidence_intervals=True)
    metrics["k3_threshold"] = float(threshold)
    metrics["target_tissue"] = target
    metrics["negative_cohorts"] = " | ".join(negatives)
    return metrics



def evaluate_detection_localization_intersection(test_predictions, source_df, target):
    """
    Evaluate combined detection + localization only among original Test patients
    whose detection model output Result(Cancer) == Positive.

    Intersection positive = detection positive AND localization positive.
    The resulting performance matrix excludes all detection-negative Test patients.
    """
    if "Result(Cancer)" not in source_df.columns:
        raise KeyError("Input workbook must contain the detection column 'Result(Cancer)'.")
    if PATIENT_ID_COL not in test_predictions.columns or PATIENT_ID_COL not in source_df.columns:
        raise KeyError(f"Both tables must contain join key {PATIENT_ID_COL!r}.")

    source_test_mask = source_df[DATA_TYPE_COL].astype(str).str.strip().str.casefold().eq(ORIGINAL_TEST_ROLE.casefold())
    detection = source_df.loc[
        source_test_mask, [PATIENT_ID_COL, COHORT_COL, DATA_TYPE_COL, "Result(Cancer)"]
    ].drop_duplicates(subset=[PATIENT_ID_COL], keep="first").copy()
    detection["Detection_Result"] = detection["Result(Cancer)"].astype(str).str.strip()
    detection["Detection_Positive"] = (
        detection["Detection_Result"].str.casefold().eq("positive")
    ).astype(int)

    loc = test_predictions.copy()
    if "Localization Role" in loc.columns:
        loc = loc[loc["Localization Role"].astype(str).str.casefold().eq("test")].copy()

    # Detection_Positive may already be present in the localization prediction
    # table. Drop it before merging so the source detection result remains the
    # single authoritative Detection_Positive column (rather than being renamed
    # to Detection_Positive_Localization / Detection_Positive_Detection).
    for duplicate_col in ("Detection_Positive", "Combined_Intersection_Positive"):
        if duplicate_col in loc.columns:
            loc = loc.drop(columns=duplicate_col)

    joined = loc.merge(
        detection[[PATIENT_ID_COL, COHORT_COL, "Detection_Result", "Detection_Positive"]],
        on=PATIENT_ID_COL,
        how="left",
        suffixes=("_Localization", "_Detection"),
        validate="one_to_one",
    )
    cohort_loc = f"{COHORT_COL}_Localization"
    cohort_det = f"{COHORT_COL}_Detection"
    if cohort_det in joined.columns:
        joined["Actual_Cancer_Type"] = joined[cohort_det].fillna(joined.get(cohort_loc))
    elif cohort_loc in joined.columns:
        joined["Actual_Cancer_Type"] = joined[cohort_loc]
    else:
        joined["Actual_Cancer_Type"] = joined[COHORT_COL]

    joined["Detection_Positive"] = joined["Detection_Positive"].fillna(0).astype(int)
    joined["Localization_Positive"] = pd.to_numeric(
        joined["Predicted_Binary_Label"], errors="coerce"
    ).fillna(0).astype(int)
    joined["Combined_Intersection_Positive"] = (
        joined["Detection_Positive"].eq(1) & joined["Localization_Positive"].eq(1)
    ).astype(int)

    # Restrict all combined metrics/matrices to detection-positive original Test patients.
    detected = joined[joined["Detection_Positive"].eq(1)].copy()
    if detected.empty:
        by_type = pd.DataFrame(columns=[
            "Cancer Type", "Detection-Positive Patients",
            "Localization-Positive Among Detection Positives",
            "Combined Intersection Positives",
            "Localization-Negative Among Detection Positives",
            "Intersection Rate Among Detection Positives (%)"
        ])
        metrics = calculate_metrics([], [])
        metrics.update({
            "target_tissue": target,
            "evaluation_subset": "Detection-positive original Test patients only",
            "note": "No original Test patients had Result(Cancer) == Positive."
        })
    else:
        summary = []
        for cancer_type, group in detected.groupby("Actual_Cancer_Type", dropna=False, sort=True):
            n = int(len(group))
            loc_positive = int(group["Localization_Positive"].sum())
            intersection = int(group["Combined_Intersection_Positive"].sum())
            summary.append({
                "Cancer Type": cancer_type,
                "Detection-Positive Patients": n,
                "Localization-Positive Among Detection Positives": loc_positive,
                "Combined Intersection Positives": intersection,
                "Localization-Negative Among Detection Positives": n - loc_positive,
                "Intersection Rate Among Detection Positives (%)":
                    100.0 * intersection / n if n else 0.0,
            })
        by_type = pd.DataFrame(summary)
        # Add an overall denominator across all detection-positive Test patients.
        overall_n = int(len(detected))
        overall_intersection = int(detected["Combined_Intersection_Positive"].sum())
        overall_loc_positive = int(detected["Localization_Positive"].sum())
        overall_row = {
            "Cancer Type": "OVERALL",
            "Detection-Positive Patients": overall_n,
            "Localization-Positive Among Detection Positives": overall_loc_positive,
            "Combined Intersection Positives": overall_intersection,
            "Localization-Negative Among Detection Positives": overall_n - overall_loc_positive,
            "Intersection Rate Among Detection Positives (%)":
                100.0 * overall_intersection / overall_n if overall_n else 0.0,
        }
        by_type = pd.concat([by_type, pd.DataFrame([overall_row])], ignore_index=True)

        y_true = (
            detected["Actual_Cancer_Type"].astype(str).str.strip().str.casefold()
            == norm_text(target)
        ).astype(int).to_numpy()
        y_pred = detected["Localization_Positive"].to_numpy(dtype=int)
        metrics = calculate_metrics(y_true, y_pred)
        metrics.update({
            "target_tissue": target,
            "evaluation_subset": "Detection-positive original Test patients only",
            "detection_positive_patients_evaluated": int(len(detected)),
            "combined_intersection_positives": int(
                detected["Combined_Intersection_Positive"].sum()
            ),
            "detection_positive_target_tissue_patients": int(y_true.sum()),
            "note": (
                "Conditional localization metrics among patients already called cancer-positive "
                "by Result(Cancer); not whole-Test-set metrics."
            ),
        })

    wanted = [
        PATIENT_ID_COL, "Actual_Cancer_Type", "Detection_Result", "Detection_Positive",
        "Localization_Positive", "Combined_Intersection_Positive", DISTANCE_COL,
        "k3_threshold", "Predicted_Label"
    ]
    return detected[[c for c in wanted if c in detected.columns]].copy(), by_type, pd.DataFrame([metrics])


def _safe_filename(value):
    """Return a filesystem-safe label for per-target output files."""
    cleaned = "".join(ch if ch.isalnum() or ch in "-_" else "_" for ch in str(value).strip())
    return cleaned.strip("_") or "Target"


def _metric_ratio(numerator, denominator, value):
    """Format a metric with its numerator/denominator and percentage."""
    if denominator == 0:
        return f"{numerator} / {denominator} (NA)"
    return f"{numerator} / {denominator} ({100.0 * value:.1f}%)"


def _add_counted_metrics(metrics):
    """Add numerator/denominator fields for transparent reporting."""
    tp, fp, fn, tn = (int(metrics.get(k, 0)) for k in ("TP", "FP", "FN", "TN"))
    n = int(metrics.get("N", tp + fp + fn + tn))
    f1_den = 2 * tp + fp + fn
    metrics.update({
        "Sensitivity_numerator": tp,
        "Sensitivity_denominator": tp + fn,
        "Precision_numerator": tp,
        "Precision_denominator": tp + fp,
        "Specificity_numerator": tn,
        "Specificity_denominator": tn + fp,
        "Accuracy_numerator": tp + tn,
        "Accuracy_denominator": n,
        "F1_numerator": 2 * tp,
        "F1_denominator": f1_den,
        "Sensitivity_numerator_denominator": f"{tp} / {tp + fn}",
        "Precision_numerator_denominator": f"{tp} / {tp + fp}",
        "Specificity_numerator_denominator": f"{tn} / {tn + fp}",
        "Accuracy_numerator_denominator": f"{tp + tn} / {n}",
        "F1_numerator_denominator": f"{2 * tp} / {f1_den}",
        "Sensitivity_display": _metric_ratio(tp, tp + fn, float(metrics.get("Recall", 0.0))),
        "Precision_display": _metric_ratio(tp, tp + fp, float(metrics.get("Precision", 0.0))),
        "Specificity_display": _metric_ratio(tn, tn + fp, float(metrics.get("Specificity", 0.0))),
        "Accuracy_display": _metric_ratio(tp + tn, n, float(metrics.get("Accuracy", 0.0))),
        "F1_display": _metric_ratio(2 * tp, f1_den, float(metrics.get("F1", 0.0))),
    })
    return metrics


def _count_split_samples(split_df, target):
    """Count target/rest samples in calibration and held-out sets for one target."""
    cohort = split_df[COHORT_COL].astype(str).str.strip().str.casefold()
    role = split_df["Localization Role"].astype(str).str.strip().str.casefold()
    target_mask = cohort.eq(norm_text(target))
    cal1_mask = role.eq("localization calibration 1")
    cal2_mask = role.eq("localization calibration 2")
    test_mask = role.eq("test")
    return {
        "target_tissue": target,
        "calibration1_total_n": int(cal1_mask.sum()),
        "calibration1_target_n": int((cal1_mask & target_mask).sum()),
        "calibration1_rest_n": int((cal1_mask & ~target_mask).sum()),
        "calibration2_total_n": int(cal2_mask.sum()),
        "calibration2_target_n": int((cal2_mask & target_mask).sum()),
        "calibration2_rest_n": int((cal2_mask & ~target_mask).sum()),
        "heldout_test_total_n": int(test_mask.sum()),
        "heldout_test_target_n": int((test_mask & target_mask).sum()),
        "heldout_test_rest_n": int((test_mask & ~target_mask).sum()),
    }


def main():
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    log_path = OUTPUT_DIR / "TDAcancer-detect-2_10_Random_Splits_Run_Log.txt"
    log_path.write_text("TDAcancer-detect-2 repeated random-split run log\n", encoding="utf-8")

    def log(message=""):
        print(message)
        with log_path.open("a", encoding="utf-8") as handle:
            handle.write(str(message) + "\n")

    df = pd.read_excel(INPUT_WORKBOOK, sheet_name=SHEET_NAME)
    validate_settings(df)

    if df[PATIENT_ID_COL].isna().any():
        missing_count = int(df[PATIENT_ID_COL].isna().sum())
        raise ValueError(
            f"{PATIENT_ID_COL} contains {missing_count} missing value(s). "
            "Fill missing IDs before running so patient records can be audited."
        )

    id_text = df[PATIENT_ID_COL].astype(str).str.strip()
    duplicate_mask = id_text.duplicated(keep=False)
    if duplicate_mask.any():
        duplicate_values = id_text.loc[duplicate_mask].value_counts()
        examples = duplicate_values.head(10).to_dict()
        log(
            f"WARNING: {int(duplicate_mask.sum())} rows have repeated values in "
            f"{PATIENT_ID_COL}. Examples (ID: row count): {examples}. "
            "These values are retained as supplied; row indices distinguish records."
        )

    observed_cohorts = sorted(df[COHORT_COL].dropna().astype(str).str.strip().unique())
    if not TARGET_CANCER_TYPES:
        raise ValueError("TARGET_CANCER_TYPES must contain at least one cohort label.")

    resolved_targets = []
    for requested in TARGET_CANCER_TYPES:
        matches = [cohort for cohort in observed_cohorts if norm_text(cohort) == norm_text(requested)]
        if not matches:
            raise ValueError(
                f"Target cancer type {requested!r} not found. Available cohorts: {observed_cohorts}"
            )
        actual = matches[0]
        if norm_text(actual) == "normal":
            raise ValueError("TARGET_CANCER_TYPES should contain cancer types, not 'Normal'.")
        if norm_text(actual) not in {norm_text(x) for x in resolved_targets}:
            resolved_targets.append(actual)

    if not resolved_targets:
        raise ValueError("No valid target cancer types were selected.")
    if not isinstance(N_RANDOM_SPLITS, int) or N_RANDOM_SPLITS < 1:
        raise ValueError("N_RANDOM_SPLITS must be a positive integer.")

    all_combos = [
        tuple(sorted(panel, key=lambda x: PRESELECTED_BIOMARKERS.index(x)))
        for panel in combinations(PRESELECTED_BIOMARKERS, PANEL_SIZE)
    ]

    all_test_performance = []
    all_top_panels = []
    all_sample_counts = []
    all_split_counts = []
    all_split_assignments = []

    log("\nTDAcancer-detect-2: repeated random-split one-vs-rest tissue localization")
    log(f"Input workbook: {INPUT_WORKBOOK}")
    log(f"Output folder: {OUTPUT_DIR}")
    log(f"Number of random splits: {N_RANDOM_SPLITS}")
    log(f"Base random seed: {RANDOM_SEED}")
    log(
        f"Cancer true-positive allocation: Calibration 1={LOCALIZATION_CAL1_FRACTION:.0%}, "
        f"independent Calibration 2={LOCALIZATION_CAL2_FRACTION:.0%}, "
        f"Test=remaining {1 - LOCALIZATION_CAL1_FRACTION - LOCALIZATION_CAL2_FRACTION:.0%}"
    )
    log(f"Normal true-negative Calibration-1 centroid fraction: {NORMAL_CAL1_FRACTION:.0%}")
    log(f"Candidate biomarkers: {PRESELECTED_BIOMARKERS}")
    log(f"Three-biomarker panels per search: {len(all_combos)}")
    log(f"Best calibration-selected panel(s) evaluated per cancer type per split: {TOP_N_TO_REPORT}")
    log("Calibration 1 and Calibration 2 contain disjoint patient subsets.")
    log("Calibration 1 constructs the normal centroid and target-specific starting threshold.")
    log("Independent Calibration 2 tunes thresholds and selects the best biomarker panel.")
    log(
        f"Within the strict >{SPECIFICITY_TARGET:.0%} specificity constraint, "
        "threshold and panel ranking prioritize Recall/Sensitivity, then "
        "Specificity, then Precision."
    )
    log("Normal false positives, cancer false negatives, and remaining normal true negatives are excluded.")
    log("The same patient-level partition is used across target-specific one-vs-rest analyses within each split.")
    log(
        f"For each target and split, one panel is selected using Calibration-2 "
        f"specificity > {SPECIFICITY_TARGET:.0%}, then highest Recall/Sensitivity, "
        "Specificity and Precision as tie-breakers."
    )
    log("The selected calibration panel and its threshold are locked before held-out Test evaluation.")
    log("Test performance is averaged across split-specific calibration-selected panels, irrespective of panel identity.")
    log("Biomarker recurrence is calculated only from the single calibration-selected panel in each split.")
    log("No held-out Test metric is used for panel selection or threshold tuning.")

    for split_number in range(1, N_RANDOM_SPLITS + 1):
        split_seed = RANDOM_SEED + split_number - 1
        log("\n" + "#" * 88)
        log(f"RANDOM SPLIT {split_number}/{N_RANDOM_SPLITS} | seed={split_seed}")
        log("#" * 88)

        split_df, split_summary = make_localization_split(df, random_seed=split_seed)
        split_summary.insert(0, "Split", split_number)
        all_split_counts.append(split_summary)

        split_assignment = split_df.copy()
        split_assignment.insert(0, "Split", split_number)
        all_split_assignments.append(split_assignment)

        reference_df = split_df[
            split_df["Localization Role"].astype(str).str.casefold().eq("reference")
            & split_df[COHORT_COL].astype(str).str.strip().str.casefold().eq("normal")
        ].copy()
        if len(reference_df) != N_REFERENCE_NORMAL:
            raise ValueError(
                f"Split {split_number}: expected {N_REFERENCE_NORMAL} fixed Reference patients, "
                f"found {len(reference_df)}."
            )

        role_norm = split_df["Localization Role"].astype(str).str.casefold()
        class_norm = split_df["Localization Detection Class"].astype(str).str.casefold()
        log(
            f"Calibration-1 cancer true positives: "
            f"{int((role_norm.eq('localization calibration 1') & class_norm.eq('cancer true positive')).sum())}"
        )
        log(
            f"Independent Calibration-2 cancer true positives: "
            f"{int((role_norm.eq('localization calibration 2') & class_norm.eq('cancer true positive')).sum())}"
        )
        log(
            f"Calibration-1 normal true negatives: "
            f"{int((role_norm.eq('localization calibration 1') & class_norm.eq('normal true negative')).sum())}"
        )
        log(
            f"Held-out cancer true positives: "
            f"{int((role_norm.eq('test') & class_norm.eq('cancer true positive')).sum())}"
        )
        log(
            f"Excluded normal false positives: "
            f"{int((class_norm.eq('normal false positive') & role_norm.eq('excluded')).sum())}"
        )

        # Feature cache is deliberately reset for every split because the normal
        # Calibration-1 centroid can change when the patient allocation changes.
        feature_cache = {}

        for target in resolved_targets:
            # Normal is used only to construct the Calibration-1 reference centroid;
            # it is excluded from the one-vs-rest negative cohorts for localization.
            negatives = [
                cohort for cohort in observed_cohorts
                if norm_text(cohort) not in {norm_text(target), "normal"}
            ]
            sample_counts = _count_split_samples(split_df, target)
            sample_counts["Split"] = split_number
            all_sample_counts.append(sample_counts)

            log("\n" + "=" * 78)
            log(f"Split {split_number} | Target tissue: {target}")
            log(f"Negative cohorts: {negatives}")
            log(
                f"Calibration 1 samples: {sample_counts['calibration1_total_n']} total; "
                f"target={sample_counts['calibration1_target_n']}; "
                f"rest={sample_counts['calibration1_rest_n']}"
            )
            log(
                f"Independent Calibration 2 samples: {sample_counts['calibration2_total_n']} total; "
                f"target={sample_counts['calibration2_target_n']}; "
                f"rest={sample_counts['calibration2_rest_n']}"
            )
            log(
                f"Held-out test samples: {sample_counts['heldout_test_total_n']} total; "
                f"target={sample_counts['heldout_test_target_n']}; "
                f"rest={sample_counts['heldout_test_rest_n']}"
            )

            if sample_counts["calibration1_target_n"] == 0:
                log(f"WARNING: Skipping {target} in split {split_number}: no target Calibration-1 patients.")
                continue
            if sample_counts["calibration2_target_n"] == 0 or sample_counts["calibration2_rest_n"] == 0:
                log(
                    f"WARNING: Skipping {target} in split {split_number}: independent Calibration 2 "
                    "does not contain both target and non-target cancer cohorts."
                )
                continue
            if sample_counts["heldout_test_target_n"] == 0:
                log(
                    f"WARNING: {target} in split {split_number} has no target patients in held-out Test; "
                    "target sensitivity cannot be estimated."
                )
                continue

            scored_rows = []
            for panel_number, panel in enumerate(all_combos, start=1):
                try:
                    summary, features, offset_table = score_panel(
                        panel, split_df, reference_df, target, negatives, feature_cache
                    )
                    summary["panel_tuple"] = panel
                    summary["specificity_constraint_met"] = bool(summary["specificity_constraint_met"])
                    scored_rows.append(summary)
                except Exception as exc:
                    warnings.warn(
                        f"Split={split_number}; target={target}; panel failed {panel}: {exc}"
                    )
                    scored_rows.append({
                        "panel": " | ".join(panel),
                        "panel_tuple": panel,
                        "k3_base": np.nan,
                        "k3_threshold": np.nan,
                        "Recall": np.nan,
                        "Precision": np.nan,
                        "F1": np.nan,
                        "Accuracy": np.nan,
                        "Specificity": np.nan,
                        "TP": np.nan,
                        "FP": np.nan,
                        "FN": np.nan,
                        "TN": np.nan,
                        "specificity_constraint_met": False,
                        "error": str(exc),
                    })

                if panel_number % PRINT_EVERY == 0 or panel_number == len(all_combos):
                    valid = [r for r in scored_rows if np.isfinite(r.get("Recall", np.nan))]
                    if valid:
                        best = max(
                            valid,
                            key=lambda r: (
                                bool(r.get("specificity_constraint_met", False)),
                                r["Recall"],
                                r.get("Specificity", 0.0),
                                r.get("Precision", 0.0),
                            ),
                        )
                        log(
                            f"Split {split_number} {target}: evaluated "
                            f"{panel_number}/{len(all_combos)} panels; current best "
                            f"calibration Recall={best['Recall']:.3f}; panel={best['panel']}"
                        )

            target_results = pd.DataFrame(scored_rows)
            if (
                target_results.empty
                or not np.isfinite(
                    pd.to_numeric(target_results.get("k3_threshold"), errors="coerce")
                ).any()
            ):
                log(f"ERROR: No valid biomarker panels for {target} in split {split_number}.")
                continue

            target_results["specificity_constraint_met"] = (
                target_results["specificity_constraint_met"].fillna(False).astype(bool)
            )
            target_results["_eligible"] = target_results["specificity_constraint_met"]
            target_results["_recall_sort"] = (
                pd.to_numeric(target_results["Recall"], errors="coerce").fillna(-1)
            )
            target_results["_specificity_sort"] = (
                pd.to_numeric(target_results["Specificity"], errors="coerce").fillna(-1)
            )
            target_results["_precision_sort"] = (
                pd.to_numeric(target_results["Precision"], errors="coerce").fillna(-1)
            )
            target_results = target_results.sort_values(
                ["_eligible", "_recall_sort", "_specificity_sort", "_precision_sort"],
                ascending=[False, False, False, False],
                kind="stable",
            ).reset_index(drop=True)
            target_results["Rank"] = np.arange(1, len(target_results) + 1)
            target_results = target_results.drop(
                columns=[
                    "_eligible", "_recall_sort", "_specificity_sort", "_precision_sort"
                ],
                errors="ignore",
            )

            best_calibration_panel = target_results[
                np.isfinite(pd.to_numeric(target_results["k3_threshold"], errors="coerce"))
            ].head(1).copy()
            if best_calibration_panel.empty:
                log(f"ERROR: No valid calibration-selected panel for {target} in split {split_number}.")
                continue

            for _, panel_row in best_calibration_panel.iterrows():
                panel_text = str(panel_row["panel"])
                panel = tuple(panel_text.split(" | "))
                threshold = float(panel_row["k3_threshold"])
                features = extract_features_for_panel(
                    split_df, reference_df, panel, feature_cache
                )
                test_idx = features.index[
                    features["Localization Role"].astype(str).str.casefold().eq("test")
                ]
                if len(test_idx) == 0:
                    warnings.warn(
                        f"No held-out test patients for split={split_number}, "
                        f"target={target}, panel={panel_text}"
                    )
                    continue

                metrics = evaluate_test(features, split_df, target, negatives, threshold)
                metrics.update({
                    "Split": split_number,
                    "Split_Seed": split_seed,
                    "Target_Tissue": target,
                    "Rank": int(panel_row["Rank"]),
                    "Selected_Panel": panel_text,
                    "k3_base": panel_row.get("k3_base", np.nan),
                    "k3_offset": panel_row.get("offset", np.nan),
                    "k3_threshold": threshold,
                    "Calibration_Recall": panel_row.get("Recall", np.nan),
                    "Calibration_Specificity": panel_row.get("Specificity", np.nan),
                    "Calibration_Precision": panel_row.get("Precision", np.nan),
                    "Calibration_F1": panel_row.get("F1", np.nan),
                    "Calibration_Specificity_Constraint_Met": bool(
                        panel_row["specificity_constraint_met"]
                    ),
                    **sample_counts,
                    "Threshold_Tuning_Reused_Calibration_Subset": REUSE_CAL1_FOR_CAL2,
                })
                metrics = _add_counted_metrics(metrics)
                all_test_performance.append(metrics)

                panel_record = panel_row.drop(labels=["panel_tuple"], errors="ignore").to_dict()
                panel_record.update({
                    "Split": split_number,
                    "Split_Seed": split_seed,
                    "Target_Tissue": target,
                })
                all_top_panels.append(panel_record)

            split_top_metrics = [
                r for r in all_test_performance
                if r.get("Split") == split_number and norm_text(r.get("Target_Tissue")) == norm_text(target)
            ]
            if split_top_metrics:
                brief = pd.DataFrame(split_top_metrics)[
                    ["Rank", "Selected_Panel", "Calibration_Recall", "Calibration_Specificity",
                     "Calibration_Precision", "Calibration_F1", "Recall", "Specificity",
                     "Precision", "F1", "Accuracy"]
                ]
                log(
                    f"\nHELD-OUT TEST PERFORMANCE OF THE CALIBRATION-SELECTED PANEL — "
                    f"Split {split_number}, {target}:\n{brief.to_string(index=False)}"
                )

    # -------------------------------------------------------------------------
    # Each split already contains exactly one panel selected using calibration
    # performance. Carry that panel's held-out Test metrics forward directly;
    # do not rank or select panels using Test performance.
    # -------------------------------------------------------------------------
    performance_df = pd.DataFrame(all_test_performance)
    top_panels_df = pd.DataFrame(all_top_panels)
    sample_counts_df = pd.DataFrame(all_sample_counts)
    split_counts_df = (
        pd.concat(all_split_counts, ignore_index=True) if all_split_counts else pd.DataFrame()
    )
    split_assignments_df = (
        pd.concat(all_split_assignments, ignore_index=True)
        if all_split_assignments else pd.DataFrame()
    )

    best_panel_rows = []
    mean_performance_rows = []
    biomarker_recurrence_rows = []

    if not performance_df.empty:
        for metric in ["F1", "Precision", "Recall", "Specificity", "Accuracy"]:
            performance_df[metric] = pd.to_numeric(performance_df[metric], errors="coerce")

        # There should be one row per target and split: the panel selected on
        # calibration data and evaluated once on the held-out Test patients.
        for _, selected_row in performance_df.iterrows():
            best = selected_row.to_dict()
            best["Best_Panel_Selection_Criterion"] = (
                f"Highest Calibration-2 Recall/Sensitivity among panels satisfying "
                f"the strict calibration specificity constraint (>{SPECIFICITY_TARGET:.0%}); "
                "ties resolved by Calibration-2 Specificity, Precision, then threshold-offset proximity"
            )
            best_panel_rows.append(best)

        best_panel_df = pd.DataFrame(best_panel_rows)

        if not best_panel_df.empty:
            # Average performance across the best panel from each successful split,
            # without requiring the same biomarker combination to recur.
            performance_metrics = ["Recall", "Specificity", "Precision", "F1", "Accuracy"]

            for target in resolved_targets:
                target_best = best_panel_df[
                    best_panel_df["Target_Tissue"].astype(str).map(norm_text).eq(norm_text(target))
                ].copy()
                if target_best.empty:
                    continue

                row = {
                    "Target_Tissue": target,
                    "Successful_Splits": int(target_best["Split"].nunique()),
                    "Requested_Splits": int(N_RANDOM_SPLITS),
                    "Mean_Performance_Based_On": (
                        "Held-out Test performance of the single best panel selected using calibration data in each split"
                    ),
                }
                # Calculate the arithmetic mean of the split-wise performance metrics.
                # Each successful random split receives equal weight:
                #   Mean metric = sum(split-wise metric values) / number of
                #                 successful splits.
                # If every requested split is successful, this denominator equals
                # N_RANDOM_SPLITS. Failed/skipped splits are not treated as zero.
                successful_splits = int(target_best["Split"].nunique())
                row["Mean_Performance_Denominator_Splits"] = successful_splits

                for metric in ["Recall", "Specificity", "Precision", "F1", "Accuracy"]:
                    values = pd.to_numeric(target_best[metric], errors="coerce").dropna()
                    row[f"Mean_{metric}"] = (
                        float(values.sum() / successful_splits)
                        if successful_splits > 0 and len(values) == successful_splits
                        else (float(values.mean()) if len(values) else np.nan)
                    )

                # Additionally calculate the mean Calibration-2 Recall and F1
                # for the calibration-selected panel in each successful split.
                # Use the same arithmetic-mean procedure and split denominator
                # as the held-out Test metrics above.
                for calibration_metric in ["Calibration_Recall", "Calibration_F1"]:
                    values = pd.to_numeric(
                        target_best[calibration_metric], errors="coerce"
                    ).dropna()
                    row[f"Mean_{calibration_metric}"] = (
                        float(values.sum() / successful_splits)
                        if successful_splits > 0 and len(values) == successful_splits
                        else (float(values.mean()) if len(values) else np.nan)
                    )

                # Add integer numerator/denominator values to the final mean
                # performance table. These are the rounded arithmetic means of
                # the split-wise confusion-matrix counts, not pooled patient counts.
                mean_tp = int(np.floor(pd.to_numeric(target_best["TP"], errors="coerce").mean() + 0.5))
                mean_fp = int(np.floor(pd.to_numeric(target_best["FP"], errors="coerce").mean() + 0.5))
                mean_fn = int(np.floor(pd.to_numeric(target_best["FN"], errors="coerce").mean() + 0.5))
                mean_tn = int(np.floor(pd.to_numeric(target_best["TN"], errors="coerce").mean() + 0.5))

                mean_metric_counts = {
                    "Sensitivity": (mean_tp, mean_tp + mean_fn),
                    "Specificity": (mean_tn, mean_tn + mean_fp),
                    "Precision": (mean_tp, mean_tp + mean_fp),
                    "Accuracy": (mean_tp + mean_tn, mean_tp + mean_fp + mean_fn + mean_tn),
                    "F1": (2 * mean_tp, 2 * mean_tp + mean_fp + mean_fn),
                }
                for metric_name, (numerator, denominator) in mean_metric_counts.items():
                    row[f"Mean_{metric_name}_Numerator"] = int(numerator)
                    row[f"Mean_{metric_name}_Denominator"] = int(denominator)
                    row[f"Mean_{metric_name}_Numerator_Denominator"] = (
                        f"{int(numerator)} / {int(denominator)}"
                    )

                # Keep Mean_Sensitivity as an explicit sensitivity column while
                # retaining Mean_Recall for compatibility with existing output.
                row["Mean_Sensitivity"] = row["Mean_Recall"]
                mean_performance_rows.append(row)

                # Count a biomarker once per split if it appears in that split's
                # single best biomarker combination.
                biomarker_split_sets = {
                    biomarker: set() for biomarker in PRESELECTED_BIOMARKERS
                }
                for _, best_row in target_best.iterrows():
                    split_id = int(best_row["Split"])
                    for biomarker in str(best_row["Selected_Panel"]).split(" | "):
                        biomarker_split_sets.setdefault(biomarker, set()).add(split_id)

                ordered_biomarkers = sorted(
                    biomarker_split_sets,
                    key=lambda biomarker: (
                        -len(biomarker_split_sets[biomarker]),
                        PRESELECTED_BIOMARKERS.index(biomarker)
                        if biomarker in PRESELECTED_BIOMARKERS
                        else len(PRESELECTED_BIOMARKERS),
                    ),
                )
                successful_splits = int(target_best["Split"].nunique())
                for biomarker_rank, biomarker in enumerate(ordered_biomarkers, start=1):
                    recurrence_count = len(biomarker_split_sets[biomarker])
                    biomarker_recurrence_rows.append({
                        "Target_Tissue": target,
                        "Biomarker_Rank": biomarker_rank,
                        "Biomarker": biomarker,
                        "Splits_With_Biomarker_In_Best_Panel": recurrence_count,
                        "Successful_Splits": successful_splits,
                        "Recurrence_Percentage": (
                            100.0 * recurrence_count / successful_splits
                            if successful_splits else np.nan
                        ),
                    })

    best_panel_df = pd.DataFrame(best_panel_rows)
    mean_performance_df = pd.DataFrame(mean_performance_rows)
    biomarker_recurrence_df = pd.DataFrame(biomarker_recurrence_rows)

    log("\n" + "=" * 88)
    log("BEST BIOMARKER COMBINATION SELECTED USING CALIBRATION — HIGHEST RECALL, THEN SPECIFICITY AND PRECISION")
    if not best_panel_df.empty:
        log(
            best_panel_df[
                ["Split", "Target_Tissue", "Selected_Panel", "F1",
                 "Recall", "Specificity", "Precision", "Accuracy"]
            ].to_string(index=False)
        )

    log("\nMEAN TISSUE-LOCALIZATION PERFORMANCE ACROSS RANDOM SPLITS")
    log(
        "Mean sensitivity, specificity, precision, F1-score, and accuracy are "
        "arithmetic means of the split-wise held-out Test metrics. Mean Calibration-2 "
        "Recall and F1 are calculated separately from the calibration metrics of the "
        "selected panel in each split. All means are divided by the number of successful "
        "random splits for each cancer type. Numerators and denominators in the mean "
        "performance table are calculated from the rounded mean split-wise confusion counts."
    )
    if not mean_performance_df.empty:
        log(mean_performance_df.to_string(index=False))

    log("\nCANCER-SPECIFIC BIOMARKER RECURRENCE IN THE BEST PANELS")
    if not biomarker_recurrence_df.empty:
        for target in resolved_targets:
            subset = biomarker_recurrence_df[
                biomarker_recurrence_df["Target_Tissue"].astype(str).map(norm_text).eq(norm_text(target))
            ]
            if not subset.empty:
                log(f"\n{target}:")
                log(
                    subset[
                        ["Biomarker", "Splits_With_Biomarker_In_Best_Panel",
                         "Successful_Splits", "Recurrence_Percentage"]
                    ].to_string(index=False)
                )

    workbook_sheets = {
        "Mean Localization Performance": mean_performance_df,
        "Best Panel Each Split": best_panel_df,
        "Biomarker Recurrence": biomarker_recurrence_df,
        "Selected Panel Test Results": performance_df,
        "Calibration Selected Panel": top_panels_df,
        "Sample Counts All Splits": sample_counts_df,
        "Split Counts": split_counts_df,
        "Split Assignment": split_assignments_df,
    }

    with pd.ExcelWriter(RESULTS_WORKBOOK, engine="openpyxl") as writer:
        for sheet_name, table in workbook_sheets.items():
            table.to_excel(writer, sheet_name=sheet_name[:31], index=False)

        text_headers = {
            "F1_numerator_denominator", "Sensitivity_numerator_denominator",
            "Precision_numerator_denominator", "Specificity_numerator_denominator",
            "Accuracy_numerator_denominator", "F1_display", "Sensitivity_display",
            "Precision_display", "Specificity_display", "Accuracy_display",
            "Mean_Sensitivity_Numerator_Denominator", "Mean_Specificity_Numerator_Denominator",
            "Mean_Precision_Numerator_Denominator", "Mean_Accuracy_Numerator_Denominator",
            "Mean_F1_Numerator_Denominator",
        }
        for worksheet in writer.book.worksheets:
            header_map = {cell.value: cell.column for cell in worksheet[1]}
            for header, col_idx in header_map.items():
                if header in text_headers:
                    for row_idx in range(2, worksheet.max_row + 1):
                        cell = worksheet.cell(row=row_idx, column=col_idx)
                        cell.number_format = "@"
                        if cell.value is not None:
                            cell.value = str(cell.value)

    log("\nRun completed.")
    log(f"Random splits completed/requested: {N_RANDOM_SPLITS}")
    log(f"Consolidated Excel workbook: {RESULTS_WORKBOOK}")
    log(f"Run log saved to: {log_path}")
    log(f"All outputs saved in: {OUTPUT_DIR}")
    log(
        "Important workbook sheets: Mean Localization Performance; Best Panel Each Split; "
        "Biomarker Recurrence; Selected Panel Test Results; Calibration Selected Panel."
    )


if __name__ == "__main__":
    main()
```

```python
"""
Mahalanobis-distance baseline for binary cancer detection.

This standalone script evaluates the existing patient split in the updated
clinical workbook. It does not create, modify, or reshuffle patient partitions.

Expected Data Type roles:
    Reference
    Regression Fit Calibration 1
    Biomarker-Panel Calibration and Selection
    Test

Protocol:
    1. Fit the healthy-reference distribution using Reference and Calibration 1
       normal patients only.
    2. Select the Mahalanobis-distance threshold using Calibration 2 normal
       patients only, requiring specificity strictly greater than 95%.
    3. Apply the locked threshold once to the unchanged Test partition.
    4. Use the same eight biomarkers and natural-log transformation throughout.

No Test patients are used to fit the covariance model or select the operating
threshold. Test ROC summaries are descriptive only and do not alter predictions.

Installation:
    python -m pip install numpy pandas matplotlib scikit-learn openpyxl
"""

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.covariance import LedoitWolf
from sklearn.metrics import (
    roc_auc_score,
    roc_curve,
    confusion_matrix,
    accuracy_score,
    balanced_accuracy_score,
)

# =============================================================================
# USER SETTINGS — update paths only if your files are stored elsewhere
# =============================================================================
INPUT_WORKBOOK = Path(
    r"D:\Cancer Detection folder\Biomarker Selection\clinical cancer validation data_Updated.xlsx"
)
SHEET_NAME = "Normal and Cancer"
OUTPUT_WORKBOOK = INPUT_WORKBOOK.with_name(
    "Mahalanobis_8Biomarker_Cancer_Detection_95pctSpecificity.xlsx"
)
OUTPUT_ROC_PLOT = INPUT_WORKBOOK.with_name(
    "Mahalanobis_8Biomarker_Test_ROC.png"
)

COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
DATA_TYPE_COL = "Data Type"
NORMAL_LABEL = "normal"

BIOMARKERS = [
    "sIL-6Ra",
    "CA-125",
    "CA 19-9",
    "IL-6",
    "IL-8",
    "OPN",
    "TRAIL",
    "Total PSA",
]

TARGET_SPECIFICITY = 0.95
EPS = 1e-12

def role_mask(df, role):
    return df[DATA_TYPE_COL].astype(str).str.strip().str.casefold().eq(role.casefold())


def cohort_values(df):
    return df[COHORT_COL].astype(str).str.strip().str.casefold()


def log_biomarkers(df, columns):
    """Return log-transformed biomarker matrix; require finite positive values."""
    x = df[columns].apply(pd.to_numeric, errors="coerce").to_numpy(dtype=float)
    if not np.isfinite(x).all():
        bad_rows, bad_cols = np.where(~np.isfinite(x))
        examples = [
            f"row index {df.index[r]}, biomarker {columns[c]}"
            for r, c in list(zip(bad_rows, bad_cols))[:10]
        ]
        raise ValueError(
            "Missing/non-finite biomarker values found. Do not silently impute "
            "test data in this benchmark. Resolve using a prespecified training-only "
            "imputation strategy. Examples: " + "; ".join(examples)
        )
    if (x <= 0).any():
        bad_rows, bad_cols = np.where(x <= 0)
        examples = [
            f"row index {df.index[r]}, biomarker {columns[c]}"
            for r, c in list(zip(bad_rows, bad_cols))[:10]
        ]
        raise ValueError(
            "Log transformation requires strictly positive concentrations. "
            "Resolve zero/negative values using a scientifically justified, "
            "predefined procedure. Examples: " + "; ".join(examples)
        )
    return np.log(x)


def mahalanobis_scores(x, mean_vector, precision_matrix):
    delta = x - mean_vector
    squared = np.einsum("ij,jk,ik->i", delta, precision_matrix, delta)
    return np.sqrt(np.maximum(squared, 0.0))


def choose_threshold_for_specificity(normal_scores, target_specificity):
    """
    Choose the smallest threshold for rule score > threshold that attains at
    least target specificity on calibration normals. This maximizes sensitivity
    among thresholds meeting the empirical calibration specificity constraint.
    """
    normal_scores = np.asarray(normal_scores, dtype=float)
    normal_scores = normal_scores[np.isfinite(normal_scores)]
    if len(normal_scores) == 0:
        raise ValueError("No finite Calibration 2 normal scores for thresholding.")

    candidates = np.unique(normal_scores)
    candidates = np.concatenate(
        [candidates, np.nextafter(candidates, np.inf), [np.inf]]
    )
    candidates = np.unique(candidates)
    feasible = []
    for threshold in candidates:
        specificity = float(np.mean(normal_scores <= threshold))
        if specificity > target_specificity:
            feasible.append((float(threshold), specificity))
    if not feasible:
        return float(np.inf), 1.0
    return min(feasible, key=lambda pair: pair[0])


def safe_auc(y_true, scores):
    if len(np.unique(y_true)) < 2:
        return float("nan")
    return float(roc_auc_score(y_true, scores))


def main():
    if not INPUT_WORKBOOK.exists():
        raise FileNotFoundError(
            f"Existing split workbook not found:\n{INPUT_WORKBOOK}\n"
            "Set INPUT_WORKBOOK to the already-created *_Updated.xlsx workbook. "
            "This script intentionally does not regenerate or reshuffle the split."
        )

    df = pd.read_excel(INPUT_WORKBOOK, sheet_name=SHEET_NAME)
    required = [COHORT_COL, PATIENT_ID_COL, DATA_TYPE_COL, *BIOMARKERS]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(
            f"Missing required columns in sheet {SHEET_NAME!r}: {missing}"
        )

    # Validate that this is the already-created split; never create a new split here.
    expected_roles = {
        "reference",
        "regression fit calibration 1",
        "biomarker-panel calibration and selection",
        "test",
    }
    observed_roles = set(
        df[DATA_TYPE_COL].astype(str).str.strip().str.casefold().unique()
    )
    missing_roles = expected_roles - observed_roles
    if missing_roles:
        raise ValueError(
            "The workbook does not contain the expected existing split roles: "
            f"{sorted(missing_roles)}. Observed roles: {sorted(observed_roles)}. "
            "Use the updated workbook created by the biomarker-selection script."
        )

    cohorts = cohort_values(df)
    normals = cohorts.eq(NORMAL_LABEL.casefold())

    # Fit the healthy-reference distribution on the 59 reference normals plus
    # Calibration 1 normals. Do not include Calibration 2 or Test in this fit.
    fit_mask = (
        (role_mask(df, "Reference") | role_mask(df, "Regression Fit Calibration 1"))
        & normals
    )
    cal2_normal_mask = (
        role_mask(df, "Biomarker-Panel Calibration and Selection") & normals
    )
    test_mask = role_mask(df, "Test")

    fit_df = df.loc[fit_mask].copy()
    cal2_normals_df = df.loc[cal2_normal_mask].copy()
    test_df = df.loc[test_mask].copy()

    if len(fit_df) < 3:
        raise ValueError(f"Too few healthy fit patients: n={len(fit_df)}")
    if len(cal2_normals_df) < 1:
        raise ValueError("No Calibration 2 normal patients are available.")
    if test_df.empty:
        raise ValueError("The existing Test partition is empty.")
    if not (~cohort_values(test_df).eq(NORMAL_LABEL.casefold())).any():
        raise ValueError("Test partition has no non-normal patients.")
    if not cohort_values(test_df).eq(NORMAL_LABEL.casefold()).any():
        raise ValueError("Test partition has no normal patients.")

    # Use the same eight selected biomarkers throughout. Log transform is applied
    # consistently to fit, calibration and test measurements.
    x_fit = log_biomarkers(fit_df, BIOMARKERS)
    x_cal2_normals = log_biomarkers(cal2_normals_df, BIOMARKERS)
    x_test = log_biomarkers(test_df, BIOMARKERS)

    # Ledoit-Wolf regularized covariance is more stable with correlated biomarkers.
    # Fit only on healthy Reference + Calibration 1 patients.
    covariance_estimator = LedoitWolf().fit(x_fit)
    mean_vector = covariance_estimator.location_
    precision_matrix = covariance_estimator.precision_

    cal2_normal_scores = mahalanobis_scores(
        x_cal2_normals, mean_vector, precision_matrix
    )
    test_scores = mahalanobis_scores(x_test, mean_vector, precision_matrix)

    threshold, cal2_specificity = choose_threshold_for_specificity(
        cal2_normal_scores, TARGET_SPECIFICITY
    )

    # Strict > decision rule. Threshold was selected without looking at Test.
    y_test = (
        ~cohort_values(test_df).eq(NORMAL_LABEL.casefold())
    ).astype(int).to_numpy()
    test_pred = (test_scores > threshold).astype(int)

    test_normal_mask = y_test == 0
    test_cancer_mask = y_test == 1
    tn, fp, fn, tp = confusion_matrix(y_test, test_pred, labels=[0, 1]).ravel()

    test_specificity = (
        float(tn / (tn + fp)) if (tn + fp) else float("nan")
    )
    test_sensitivity = (
        float(tp / (tp + fn)) if (tp + fn) else float("nan")
    )
    test_auc = safe_auc(y_test, test_scores)

    # Descriptive Test-ROC sensitivity at specificity strictly greater than the
    # calibration target. This is not the locked-threshold sensitivity and must
    # not be used to select or revise the operating threshold.
    fpr, tpr, roc_thresholds = roc_curve(y_test, test_scores)
    specificity_curve = 1.0 - fpr
    eligible = np.where(specificity_curve > TARGET_SPECIFICITY)[0]
    if len(eligible):
        # Among descriptive Test ROC points satisfying the strict specificity
        # criterion, report the point with the highest sensitivity.
        best_idx = eligible[np.argmax(tpr[eligible])]
        roc_sens_at_target = float(tpr[best_idx])
        roc_spec_at_target = float(specificity_curve[best_idx])
        roc_threshold_at_target = float(roc_thresholds[best_idx])
    else:
        roc_sens_at_target = float("nan")
        roc_spec_at_target = float("nan")
        roc_threshold_at_target = float("nan")

    # Prediction table includes every patient in the same existing Test partition.
    predictions = test_df.copy()
    predictions["Mahalanobis_Distance"] = test_scores
    predictions[
        "Mahalanobis_Threshold_Calibrated_StrictlyGreaterThanTargetSpecificity"
    ] = threshold
    predictions["Mahalanobis_Prediction"] = np.where(
        test_pred == 1, "Positive", "Negative"
    )
    predictions["True_Label_Binary"] = np.where(y_test == 1, "Cancer", "Normal")
    predictions["Correct_Prediction"] = (test_pred == y_test)

    cohort_test = cohort_values(test_df)
    per_cohort_rows = []
    for cohort in sorted(cohort_test.unique()):
        mask = cohort_test.eq(cohort).to_numpy()
        n = int(mask.sum())
        if cohort == NORMAL_LABEL.casefold():
            positives = int(test_pred[mask].sum())
            per_cohort_rows.append({
                "Cohort": cohort,
                "n": n,
                "True_Positive": np.nan,
                "False_Negative": np.nan,
                "False_Positive": positives,
                "True_Negative": int((test_pred[mask] == 0).sum()),
                "Sensitivity": np.nan,
                "Specificity": float((test_pred[mask] == 0).mean()),
            })
        else:
            true_positives = int(test_pred[mask].sum())
            false_negatives = int(n - true_positives)
            per_cohort_rows.append({
                "Cohort": cohort,
                "n": n,
                "True_Positive": true_positives,
                "False_Negative": false_negatives,
                "False_Positive": np.nan,
                "True_Negative": np.nan,
                "Sensitivity": float(true_positives / n) if n else float("nan"),
                "Specificity": np.nan,
            })

    summary_rows = [
        ("Method", "Mahalanobis distance; Ledoit-Wolf regularized covariance"),
        ("Biomarkers", " | ".join(BIOMARKERS)),
        ("Biomarker count", len(BIOMARKERS)),
        ("Input transformation", "Natural log of positive biomarker concentrations"),
        ("Healthy fit set", "Reference normals + Calibration 1 normals"),
        ("Healthy fit n", len(fit_df)),
        ("Reference n", int((role_mask(df, "Reference") & normals).sum())),
        (
            "Calibration 1 normal n",
            int(
                (
                    role_mask(df, "Regression Fit Calibration 1")
                    & normals
                ).sum()
            ),
        ),
        ("Calibration 2 normal n used for threshold", len(cal2_normals_df)),
        ("Test n", len(test_df)),
        ("Test normal n", int(test_normal_mask.sum())),
        ("Test cancer n", int(test_cancer_mask.sum())),
        ("Calibration specificity requirement (strictly greater than)", TARGET_SPECIFICITY),
        ("Calibrated Mahalanobis threshold", threshold),
        ("Observed Calibration 2 normal specificity", cal2_specificity),
        ("Test AUC", test_auc),
        ("Test sensitivity at locked threshold", test_sensitivity),
        (
            "Test macro-sensitivity across cancer types",
            float(
                np.mean(
                    [
                        test_pred[cohort_test.eq(c).to_numpy()].mean()
                        for c in sorted(cohort_test.unique())
                        if c != NORMAL_LABEL.casefold()
                        and cohort_test.eq(c).any()
                    ]
                )
            ),
        ),
        ("Test specificity at locked threshold", test_specificity),
        ("Test accuracy at locked threshold", float(accuracy_score(y_test, test_pred))),
        ("Test balanced accuracy", float(balanced_accuracy_score(y_test, test_pred))),
        ("Test true positives", int(tp)),
        ("Test false negatives", int(fn)),
        ("Test true negatives", int(tn)),
        ("Test false positives", int(fp)),
        ("Descriptive Test ROC sensitivity at specificity > target", roc_sens_at_target),
        ("Specificity at selected descriptive test ROC point", roc_spec_at_target),
        ("Test ROC threshold for descriptive point (NOT used for predictions)", roc_threshold_at_target),
        ("Covariance estimator", "LedoitWolf"),
        ("Threshold rule", "Cancer positive if Mahalanobis distance > calibrated threshold"),
        ("Data split", "Loaded existing Data Type roles; no resplitting"),
    ]
    summary_df = pd.DataFrame(summary_rows, columns=["Metric", "Value"])
    cohort_df = pd.DataFrame(per_cohort_rows)

    # Continuous ROC coordinates and thresholds are calculated from Test data only.
    roc_df = pd.DataFrame(
        {
            "False Positive Rate": fpr,
            "True Positive Rate": tpr,
            "Specificity": specificity_curve,
            "Mahalanobis Distance Threshold": roc_thresholds,
        }
    )

    # Publication-quality ROC plot. The locked calibration threshold is not changed.
    plt.figure(figsize=(7, 7))
    plt.plot(
        fpr,
        tpr,
        linewidth=2.2,
        label=f"Mahalanobis model (Test AUC = {test_auc:.3f})",
    )
    plt.plot([0, 1], [0, 1], linestyle="--", linewidth=1.5,
             label="No-discrimination reference (AUC = 0.500)")
    plt.xlim(0.0, 1.0)
    plt.ylim(0.0, 1.05)
    plt.xlabel("False Positive Rate (1 - Specificity)", fontsize=12)
    plt.ylabel("True Positive Rate (Sensitivity)", fontsize=12)
    plt.title("Mahalanobis Cancer Detection: Test-Set ROC Curve", fontsize=13)
    plt.legend(loc="lower right", frameon=True)
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.savefig(OUTPUT_ROC_PLOT, dpi=300, bbox_inches="tight")
    plt.close()

    # Single consolidated workbook; no extra CSV files.
    OUTPUT_WORKBOOK.parent.mkdir(parents=True, exist_ok=True)
    with pd.ExcelWriter(OUTPUT_WORKBOOK, engine="openpyxl") as writer:
        summary_df.to_excel(writer, sheet_name="Summary", index=False)
        cohort_df.to_excel(writer, sheet_name="Per_Cohort_Test", index=False)
        predictions.to_excel(writer, sheet_name="Test_Predictions", index=False)
        roc_df.to_excel(writer, sheet_name="Test_ROC_Curve", index=False)

    print("\nMAHALANOBIS CANCER DETECTION — FINAL TEST RESULTS")
    print(f"Biomarkers: {BIOMARKERS}")
    print(f"Healthy fit patients: {len(fit_df)}")
    print(f"Calibration 2 normal patients: {len(cal2_normals_df)}")
    print(f"Test patients: {len(test_df)}")
    print(f"Required calibration specificity: strictly greater than {TARGET_SPECIFICITY:.1%}")
    print(f"Calibrated distance threshold: {threshold:.8g}")
    print(f"Calibration 2 observed specificity: {cal2_specificity:.2%}")
    print(f"Test AUC: {test_auc:.4f}")
    print(f"Test sensitivity at locked threshold: {test_sensitivity:.2%}")
    print(f"Test specificity at locked threshold: {test_specificity:.2%}")
    print(f"TP={tp}, FN={fn}, TN={tn}, FP={fp}")
    print("\nPer-cancer-type test sensitivity:")
    for row in per_cohort_rows:
        if str(row["Cohort"]).casefold() != NORMAL_LABEL.casefold():
            print(
                f"  {row['Cohort']}: n={row['n']}, "
                f"sensitivity={row['Sensitivity']:.2%}, "
                f"TP={row['True_Positive']}, FN={row['False_Negative']}"
            )
    print(f"\nSaved results workbook:\n{OUTPUT_WORKBOOK}")
    print(f"Saved Test ROC curve plot:\n{OUTPUT_ROC_PLOT}")


if __name__ == "__main__":
    main()
```

```python
"""
Logistic-regression baseline for binary cancer detection.

This standalone script evaluates the same existing patient split used by the
other baseline models. It does not create, modify, or reshuffle patient
partitions.

Protocol:
    1. Apply a natural-log transformation to the eight specified biomarkers.
    2. Train logistic regression using labeled normal and cancer patients in
       the existing Biomarker-Panel Calibration and Selection partition.
    3. Generate stratified out-of-fold probabilities within that partition and
       select the decision threshold using calibration-normal probabilities
       only, requiring specificity strictly greater than 95%.
    4. Refit the final pipeline on the complete calibration/training partition.
    5. Apply the locked threshold once to the unchanged Test partition.

No Test patients are used for model fitting or threshold selection. The
out-of-fold specificity is an empirical calibration estimate, not a guarantee
of Test-set specificity. Test ROC summaries are descriptive only and do not
alter the locked predictions.

Installation:
    python -m pip install numpy pandas matplotlib scikit-learn openpyxl
"""

from pathlib import Path
import warnings

import numpy as np
import pandas as pd
from sklearn.exceptions import ConvergenceWarning
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    roc_auc_score,
    roc_curve,
    confusion_matrix,
    accuracy_score,
    balanced_accuracy_score,
)
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt


# =============================================================================
# USER SETTINGS — update paths only if your files are stored elsewhere
# =============================================================================
INPUT_WORKBOOK = Path(
    r"D:\Cancer Detection folder\Biomarker Selection\clinical cancer validation data_Updated.xlsx"
)
SHEET_NAME = "Normal and Cancer"
OUTPUT_WORKBOOK = INPUT_WORKBOOK.with_name(
    "LogisticRegression_8Biomarker_Cancer_Detection_95pctSpecificity.xlsx"
)
OUTPUT_ROC_PLOT = INPUT_WORKBOOK.with_name(
    "LogisticRegression_Test_ROC_Curve.png"
)

COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
STAGE_COL = "AJCC Stage"
DATA_TYPE_COL = "Data Type"
NORMAL_LABEL = "normal"

BIOMARKERS = [
    "sIL-6Ra",
    "CA-125",
    "CA 19-9",
    "IL-6",
    "IL-8",
    "OPN",
    "TRAIL",
    "Total PSA",
]

TARGET_SPECIFICITY = 0.95
RANDOM_SEED = 20261017
MAX_CV_FOLDS = 5


def role_mask(df, role):
    return (
        df[DATA_TYPE_COL]
        .astype(str)
        .str.strip()
        .str.casefold()
        .eq(role.casefold())
    )


def cohort_values(df):
    return df[COHORT_COL].astype(str).str.strip().str.casefold()


def log_biomarkers(df, columns):
    """Return log-transformed biomarker matrix; require finite positive values."""
    x = df[columns].apply(pd.to_numeric, errors="coerce").to_numpy(dtype=float)
    if not np.isfinite(x).all():
        bad_rows, bad_cols = np.where(~np.isfinite(x))
        examples = [
            f"row index {df.index[r]}, biomarker {columns[c]}"
            for r, c in list(zip(bad_rows, bad_cols))[:10]
        ]
        raise ValueError(
            "Missing/non-finite biomarker values found. Do not silently impute "
            "test data in this benchmark. Resolve using a prespecified "
            "training-only imputation strategy. Examples: " + "; ".join(examples)
        )
    if (x <= 0).any():
        bad_rows, bad_cols = np.where(x <= 0)
        examples = [
            f"row index {df.index[r]}, biomarker {columns[c]}"
            for r, c in list(zip(bad_rows, bad_cols))[:10]
        ]
        raise ValueError(
            "Log transformation requires strictly positive concentrations. "
            "Resolve zero/negative values using a scientifically justified, "
            "predefined procedure. Examples: " + "; ".join(examples)
        )
    return np.log(x)


def safe_auc(y_true, scores):
    if len(np.unique(y_true)) < 2:
        return float("nan")
    return float(roc_auc_score(y_true, scores))


def choose_threshold_for_specificity(normal_scores, target_specificity):
    """
    Choose the smallest threshold for the rule score > threshold that attains
    strictly greater than target_specificity on the supplied calibration-normal
    out-of-fold scores. This maximises sensitivity subject to the strict
    empirical calibration-specificity constraint.
    """
    normal_scores = np.asarray(normal_scores, dtype=float)
    normal_scores = normal_scores[np.isfinite(normal_scores)]
    if len(normal_scores) == 0:
        raise ValueError("No finite out-of-fold calibration-normal scores.")

    candidates = np.unique(normal_scores)
    candidates = np.concatenate(
        [candidates, np.nextafter(candidates, np.inf), [np.inf]]
    )
    candidates = np.unique(candidates)
    feasible = []
    for threshold in candidates:
        specificity = float(np.mean(normal_scores <= threshold))
        if specificity > target_specificity:
            feasible.append((float(threshold), specificity))
    if not feasible:
        return float(np.inf), 1.0
    return min(feasible, key=lambda pair: pair[0])


def make_model():
    """Scaling is fitted within each CV fold to avoid preprocessing leakage."""
    return Pipeline(
        steps=[
            ("scaler", StandardScaler()),
            (
                "logistic_regression",
                LogisticRegression(
                    penalty="l2",
                    C=1.0,
                    solver="liblinear",
                    max_iter=5000,
                    random_state=RANDOM_SEED,
                ),
            ),
        ]
    )


def main():
    if not INPUT_WORKBOOK.exists():
        raise FileNotFoundError(
            f"Existing split workbook not found:\n{INPUT_WORKBOOK}\n"
            "Set INPUT_WORKBOOK to the already-created *_Updated.xlsx workbook. "
            "This script intentionally does not regenerate or reshuffle the split."
        )

    df = pd.read_excel(INPUT_WORKBOOK, sheet_name=SHEET_NAME)
    required = [COHORT_COL, PATIENT_ID_COL, DATA_TYPE_COL, *BIOMARKERS]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(
            f"Missing required columns in sheet {SHEET_NAME!r}: {missing}"
        )

    expected_roles = {
        "reference",
        "regression fit calibration 1",
        "biomarker-panel calibration and selection",
        "test",
    }
    observed_roles = set(
        df[DATA_TYPE_COL].astype(str).str.strip().str.casefold().unique()
    )
    missing_roles = expected_roles - observed_roles
    if missing_roles:
        raise ValueError(
            "The workbook does not contain the expected existing split roles: "
            f"{sorted(missing_roles)}. Observed roles: {sorted(observed_roles)}. "
            "Use the updated workbook created by the biomarker-selection script."
        )

    cohorts = cohort_values(df)
    normal = cohorts.eq(NORMAL_LABEL.casefold())

    # Supervised training/calibration partition: labeled normals and cancers.
    cal2_mask = role_mask(df, "Biomarker-Panel Calibration and Selection")
    cal_df = df.loc[cal2_mask].copy()
    test_df = df.loc[role_mask(df, "Test")].copy()

    if cal_df.empty:
        raise ValueError("The calibration/training partition is empty.")
    if test_df.empty:
        raise ValueError("The existing Test partition is empty.")

    y_cal = (
        ~cohort_values(cal_df).eq(NORMAL_LABEL.casefold())
    ).astype(int).to_numpy()
    y_test = (
        ~cohort_values(test_df).eq(NORMAL_LABEL.casefold())
    ).astype(int).to_numpy()

    if len(np.unique(y_cal)) < 2:
        raise ValueError(
            "The calibration/training partition must contain both normal and cancer patients."
        )
    if len(np.unique(y_test)) < 2:
        raise ValueError("The Test partition must contain both normal and cancer patients.")

    cal_normal_n = int((y_cal == 0).sum())
    cal_cancer_n = int((y_cal == 1).sum())
    min_class_n = int(np.bincount(y_cal).min())
    n_splits = min(MAX_CV_FOLDS, min_class_n)
    if n_splits < 2:
        raise ValueError(
            "Too few patients in one calibration class for out-of-fold threshold calibration."
        )

    x_cal = log_biomarkers(cal_df, BIOMARKERS)
    x_test = log_biomarkers(test_df, BIOMARKERS)

    # Out-of-fold probabilities on Calibration 2. Each patient receives a score
    # from a model that did not train on that patient. The threshold is calibrated
    # only against OOF scores for normal calibration patients.
    cv = StratifiedKFold(
        n_splits=n_splits, shuffle=True, random_state=RANDOM_SEED
    )
    oof_model = make_model()
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", ConvergenceWarning)
        oof_probabilities = cross_val_predict(
            oof_model,
            x_cal,
            y_cal,
            cv=cv,
            method="predict_proba",
            n_jobs=None,
        )[:, 1]

    oof_normal_scores = oof_probabilities[y_cal == 0]
    threshold, calibration_specificity = choose_threshold_for_specificity(
        oof_normal_scores, TARGET_SPECIFICITY
    )

    # Final model refit on the complete calibration/training partition.
    final_model = make_model()
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", ConvergenceWarning)
        final_model.fit(x_cal, y_cal)

    # Larger probability means stronger evidence of cancer.
    test_scores = final_model.predict_proba(x_test)[:, 1]
    test_pred = (test_scores > threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(y_test, test_pred, labels=[0, 1]).ravel()
    test_specificity = float(tn / (tn + fp)) if (tn + fp) else float("nan")
    test_sensitivity = float(tp / (tp + fn)) if (tp + fn) else float("nan")
    test_accuracy = float(accuracy_score(y_test, test_pred))
    test_balanced_accuracy = float(balanced_accuracy_score(y_test, test_pred))
    test_auc = safe_auc(y_test, test_scores)

    # ROC curve is descriptive only; do not use it to alter the locked threshold.
    fpr, tpr, roc_thresholds = roc_curve(y_test, test_scores)
    specificity_curve = 1.0 - fpr
    eligible = np.where(specificity_curve > TARGET_SPECIFICITY)[0]
    if len(eligible):
        best_idx = eligible[np.argmax(tpr[eligible])]
        roc_sens_at_target = float(tpr[best_idx])
        roc_spec_at_target = float(specificity_curve[best_idx])
        roc_threshold_at_target = float(roc_thresholds[best_idx])
    else:
        roc_sens_at_target = float("nan")
        roc_spec_at_target = float("nan")
        roc_threshold_at_target = float("nan")

    predictions = test_df.copy()
    predictions["LogisticRegression_Cancer_Probability"] = test_scores
    predictions[
        "LogisticRegression_Threshold_Calibrated_StrictlyGreaterThanTargetSpecificity"
    ] = threshold
    predictions["LogisticRegression_Prediction"] = np.where(
        test_pred == 1, "Positive", "Negative"
    )
    predictions["True_Label_Binary"] = np.where(y_test == 1, "Cancer", "Normal")
    predictions["Correct_Prediction"] = test_pred == y_test

    cohort_test = cohort_values(test_df)
    per_cohort_rows = []
    for cohort in sorted(cohort_test.unique()):
        mask = cohort_test.eq(cohort).to_numpy()
        n = int(mask.sum())
        if cohort == NORMAL_LABEL.casefold():
            false_positives = int(test_pred[mask].sum())
            per_cohort_rows.append(
                {
                    "Cohort": cohort,
                    "n": n,
                    "True_Positive": np.nan,
                    "False_Negative": np.nan,
                    "False_Positive": false_positives,
                    "True_Negative": int((test_pred[mask] == 0).sum()),
                    "Sensitivity": np.nan,
                    "Specificity": float((test_pred[mask] == 0).mean()),
                }
            )
        else:
            true_positives = int(test_pred[mask].sum())
            false_negatives = int(n - true_positives)
            per_cohort_rows.append(
                {
                    "Cohort": cohort,
                    "n": n,
                    "True_Positive": true_positives,
                    "False_Negative": false_negatives,
                    "False_Positive": np.nan,
                    "True_Negative": np.nan,
                    "Sensitivity": float(true_positives / n) if n else float("nan"),
                    "Specificity": np.nan,
                }
            )

    reference_n = int((role_mask(df, "Reference") & normal).sum())
    cal1_normal_n = int(
        (role_mask(df, "Regression Fit Calibration 1") & normal).sum()
    )
    test_normal_n = int((y_test == 0).sum())
    test_cancer_n = int((y_test == 1).sum())

    summary_rows = [
        ("Method", "Logistic regression; L2 regularization"),
        ("Biomarkers", " | ".join(BIOMARKERS)),
        ("Biomarker count", len(BIOMARKERS)),
        ("Input transformation", "Natural log of positive biomarker concentrations"),
        ("Model training partition", "Biomarker-Panel Calibration and Selection; labeled normals and cancers"),
        ("Model training n", len(cal_df)),
        ("Calibration training normal n", cal_normal_n),
        ("Calibration training cancer n", cal_cancer_n),
        ("Threshold calibration", f"{n_splits}-fold stratified out-of-fold probabilities; calibration normals only"),
        ("Reference normal n (not used by this supervised model)", reference_n),
        ("Calibration 1 normal n (not used by this supervised model)", cal1_normal_n),
        ("Test n", len(test_df)),
        ("Test normal n", test_normal_n),
        ("Test cancer n", test_cancer_n),
        ("Target calibration specificity constraint", f"Strictly greater than {TARGET_SPECIFICITY:.0%}"),
        ("Observed OOF calibration-normal specificity", calibration_specificity),
        ("Calibrated probability threshold", threshold),
        ("Test AUC", test_auc),
        ("Test sensitivity at locked threshold", test_sensitivity),
        ("Test specificity at locked threshold", test_specificity),
        ("Test accuracy at locked threshold", test_accuracy),
        ("Test balanced accuracy", test_balanced_accuracy),
        ("Test true positives", int(tp)),
        ("Test false negatives", int(fn)),
        ("Test true negatives", int(tn)),
        ("Test false positives", int(fp)),
        (f"Descriptive test ROC sensitivity at >{TARGET_SPECIFICITY:.0%} specificity", roc_sens_at_target),
        ("Specificity at selected descriptive test ROC point", roc_spec_at_target),
        ("Test ROC threshold for descriptive point (NOT used for predictions)", roc_threshold_at_target),
        (
            "Classifier",
            "StandardScaler + LogisticRegression(L2, C=1.0, liblinear)",
        ),
        ("Threshold rule", "Cancer positive if predicted cancer probability > calibrated threshold"),
        ("Data split", "Loaded existing Data Type roles; no resplitting"),
        ("Leakage note", "Test patients were not used for model fitting or threshold selection"),
    ]
    summary_df = pd.DataFrame(summary_rows, columns=["Metric", "Value"])
    cohort_df = pd.DataFrame(per_cohort_rows)

    roc_df = pd.DataFrame(
        {
            "False Positive Rate": fpr,
            "True Positive Rate": tpr,
            "Specificity": specificity_curve,
            "Probability Threshold": roc_thresholds,
        }
    )

    # Save a continuous ROC curve using TEST-set probabilities only.
    # The AUC is calculated from the same untouched Test labels and scores.
    plt.figure(figsize=(7.5, 6.0))
    plt.plot(
        fpr,
        tpr,
        linewidth=2.5,
        label=f"Logistic regression (Test AUC = {test_auc:.3f})",
    )
    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        linewidth=1.5,
        label="No-discrimination reference",
    )
    plt.xlim(0.0, 1.0)
    plt.ylim(0.0, 1.05)
    plt.xlabel("False Positive Rate (1 - Specificity)", fontsize=12)
    plt.ylabel("True Positive Rate (Sensitivity)", fontsize=12)
    plt.title("Test-Set ROC Curve — Logistic Regression", fontsize=14)
    plt.legend(loc="lower right", frameon=True)
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.savefig(OUTPUT_ROC_PLOT, dpi=300, bbox_inches="tight")
    plt.close()

    OUTPUT_WORKBOOK.parent.mkdir(parents=True, exist_ok=True)
    with pd.ExcelWriter(OUTPUT_WORKBOOK, engine="openpyxl") as writer:
        summary_df.to_excel(writer, sheet_name="Summary", index=False)
        cohort_df.to_excel(writer, sheet_name="Per_Cohort_Test", index=False)
        predictions.to_excel(writer, sheet_name="Test_Predictions", index=False)
        roc_df.to_excel(writer, sheet_name="Test_ROC_Curve", index=False)

    print("\nLOGISTIC REGRESSION CANCER DETECTION — FINAL TEST RESULTS")
    print(f"Biomarkers: {BIOMARKERS}")
    print(
        f"Calibration training patients: {len(cal_df)} "
        f"(normal={cal_normal_n}, cancer={cal_cancer_n})"
    )
    print(f"Out-of-fold threshold calibration: {n_splits}-fold; normal patients only for specificity")
    print(f"Calibration specificity constraint: strictly greater than {TARGET_SPECIFICITY:.1%}")
    print(f"Observed OOF calibration-normal specificity: {calibration_specificity:.2%}")
    print(f"Calibrated probability threshold: {threshold:.8g}")
    print(
        f"Test patients: {len(test_df)} "
        f"(normal={test_normal_n}, cancer={test_cancer_n})"
    )
    print(f"Test AUC: {test_auc:.4f}")
    print(f"Test sensitivity at locked threshold: {test_sensitivity:.2%}")
    print(f"Test specificity at locked threshold: {test_specificity:.2%}")
    print(f"Test accuracy at locked threshold: {test_accuracy:.2%}")
    print(f"TP={tp}, FN={fn}, TN={tn}, FP={fp}")
    print("\nPer-cancer-type test sensitivity:")
    for row in per_cohort_rows:
        if str(row["Cohort"]).casefold() != NORMAL_LABEL.casefold():
            print(
                f"  {row['Cohort']}: n={row['n']}, "
                f"sensitivity={row['Sensitivity']:.2%}, "
                f"TP={row['True_Positive']}, FN={row['False_Negative']}"
            )
    print(f"\nSaved one results workbook:\n{OUTPUT_WORKBOOK}")
    print(f"Saved test ROC curve image:\n{OUTPUT_ROC_PLOT}")

if __name__ == "__main__":
    main()
```

```python
"""
Support-vector-machine (SVM) baseline for binary cancer detection.

This standalone script evaluates the same existing patient split used by the
other baseline models. It does not create, modify, or reshuffle patient
partitions.

Protocol:
    1. Apply a natural-log transformation to the eight specified biomarkers.
    2. Train a standardized RBF-kernel SVM using labeled normal and cancer
       patients in the existing Biomarker-Panel Calibration and Selection
       partition.
    3. Generate stratified out-of-fold decision scores within that partition
       and select the decision threshold using calibration-normal scores only,
       requiring specificity strictly greater than 95%.
    4. Refit the final SVM pipeline on the complete calibration/training
       partition.
    5. Apply the locked threshold once to the unchanged Test partition.

No Test patients are used for model fitting or threshold selection. The
out-of-fold specificity is an empirical calibration estimate, not a guarantee
of Test-set specificity. Test ROC summaries are descriptive only and do not
alter the locked predictions.

Installation:
    python -m pip install numpy pandas matplotlib scikit-learn openpyxl
"""

from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import (
    roc_auc_score,
    roc_curve,
    confusion_matrix,
    accuracy_score,
    balanced_accuracy_score,
)
from sklearn.model_selection import StratifiedKFold, cross_val_predict


# =============================================================================
# USER SETTINGS — update paths only if your files are stored elsewhere
# =============================================================================
INPUT_WORKBOOK = Path(
    r"D:\Cancer Detection folder\Biomarker Selection\clinical cancer validation data_Updated.xlsx"
)
SHEET_NAME = "Normal and Cancer"
OUTPUT_WORKBOOK = INPUT_WORKBOOK.with_name(
    "SVM_8Biomarker_Cancer_Detection_95pctSpecificity.xlsx"
)

COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
DATA_TYPE_COL = "Data Type"
NORMAL_LABEL = "normal"

BIOMARKERS = [
    "sIL-6Ra",
    "CA-125",
    "CA 19-9",
    "IL-6",
    "IL-8",
    "OPN",
    "TRAIL",
    "Total PSA",
]

TARGET_SPECIFICITY = 0.95
RANDOM_SEED = 20261017
MAX_CV_FOLDS = 5


def role_mask(df, role):
    return (
        df[DATA_TYPE_COL]
        .astype(str)
        .str.strip()
        .str.casefold()
        .eq(role.casefold())
    )


def cohort_values(df):
    return df[COHORT_COL].astype(str).str.strip().str.casefold()


def log_biomarkers(df, columns):
    """Return log-transformed biomarker matrix; require finite positive values."""
    x = df[columns].apply(pd.to_numeric, errors="coerce").to_numpy(dtype=float)
    if not np.isfinite(x).all():
        bad_rows, bad_cols = np.where(~np.isfinite(x))
        examples = [
            f"row index {df.index[r]}, biomarker {columns[c]}"
            for r, c in list(zip(bad_rows, bad_cols))[:10]
        ]
        raise ValueError(
            "Missing/non-finite biomarker values found. Do not silently impute "
            "test data in this benchmark. Resolve using a prespecified "
            "training-only imputation strategy. Examples: " + "; ".join(examples)
        )
    if (x <= 0).any():
        bad_rows, bad_cols = np.where(x <= 0)
        examples = [
            f"row index {df.index[r]}, biomarker {columns[c]}"
            for r, c in list(zip(bad_rows, bad_cols))[:10]
        ]
        raise ValueError(
            "Log transformation requires strictly positive concentrations. "
            "Resolve zero/negative values using a scientifically justified, "
            "predefined procedure. Examples: " + "; ".join(examples)
        )
    return np.log(x)


def safe_auc(y_true, scores):
    if len(np.unique(y_true)) < 2:
        return float("nan")
    return float(roc_auc_score(y_true, scores))


def choose_threshold_for_specificity(normal_scores, target_specificity):
    """
    Choose the smallest threshold for the rule score > threshold that attains
    strictly greater than target_specificity on the supplied calibration-normal scores.
    """
    normal_scores = np.asarray(normal_scores, dtype=float)
    normal_scores = normal_scores[np.isfinite(normal_scores)]
    if len(normal_scores) == 0:
        raise ValueError("No finite out-of-fold calibration-normal scores.")

    candidates = np.unique(normal_scores)
    candidates = np.concatenate(
        [candidates, np.nextafter(candidates, np.inf), [np.inf]]
    )
    candidates = np.unique(candidates)
    feasible = []
    for threshold in candidates:
        specificity = float(np.mean(normal_scores <= threshold))
        if specificity > target_specificity:
            feasible.append((float(threshold), specificity))
    if not feasible:
        return float(np.inf), 1.0
    return min(feasible, key=lambda pair: pair[0])


def make_model():
    """Return a standardized, reproducible RBF-kernel SVM classifier."""
    return Pipeline(
        steps=[
            ("scaler", StandardScaler()),
            (
                "svm",
                SVC(
                    kernel="rbf",
                    C=1.0,
                    gamma="scale",
                    class_weight="balanced",
                    probability=False,
                    random_state=RANDOM_SEED,
                ),
            ),
        ]
    )


def main():
    if not INPUT_WORKBOOK.exists():
        raise FileNotFoundError(
            f"Existing split workbook not found:\n{INPUT_WORKBOOK}\n"
            "Set INPUT_WORKBOOK to the already-created *_Updated.xlsx workbook. "
            "This script intentionally does not regenerate or reshuffle the split."
        )

    df = pd.read_excel(INPUT_WORKBOOK, sheet_name=SHEET_NAME)
    required = [COHORT_COL, PATIENT_ID_COL, DATA_TYPE_COL, *BIOMARKERS]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(
            f"Missing required columns in sheet {SHEET_NAME!r}: {missing}"
        )

    expected_roles = {
        "reference",
        "regression fit calibration 1",
        "biomarker-panel calibration and selection",
        "test",
    }
    observed_roles = set(
        df[DATA_TYPE_COL].astype(str).str.strip().str.casefold().unique()
    )
    missing_roles = expected_roles - observed_roles
    if missing_roles:
        raise ValueError(
            "The workbook does not contain the expected existing split roles: "
            f"{sorted(missing_roles)}. Observed roles: {sorted(observed_roles)}. "
            "Use the updated workbook created by the biomarker-selection script."
        )

    cohorts = cohort_values(df)
    normal = cohorts.eq(NORMAL_LABEL.casefold())

    # Supervised training/calibration partition: labeled normals and cancers.
    cal2_mask = role_mask(df, "Biomarker-Panel Calibration and Selection")
    cal_df = df.loc[cal2_mask].copy()
    test_df = df.loc[role_mask(df, "Test")].copy()

    if cal_df.empty:
        raise ValueError("The calibration/training partition is empty.")
    if test_df.empty:
        raise ValueError("The existing Test partition is empty.")

    y_cal = (
        ~cohort_values(cal_df).eq(NORMAL_LABEL.casefold())
    ).astype(int).to_numpy()
    y_test = (
        ~cohort_values(test_df).eq(NORMAL_LABEL.casefold())
    ).astype(int).to_numpy()

    if len(np.unique(y_cal)) < 2:
        raise ValueError(
            "The calibration/training partition must contain both normal and cancer patients."
        )
    if len(np.unique(y_test)) < 2:
        raise ValueError("The Test partition must contain both normal and cancer patients.")

    cal_normal_n = int((y_cal == 0).sum())
    cal_cancer_n = int((y_cal == 1).sum())
    min_class_n = int(np.bincount(y_cal).min())
    n_splits = min(MAX_CV_FOLDS, min_class_n)
    if n_splits < 2:
        raise ValueError(
            "Too few patients in one calibration class for out-of-fold threshold calibration."
        )

    x_cal = log_biomarkers(cal_df, BIOMARKERS)
    x_test = log_biomarkers(test_df, BIOMARKERS)

    # Out-of-fold SVM decision scores on Calibration 2. Each patient receives a score
    # from a model that did not train on that patient. The threshold is calibrated
    # only against OOF scores for normal calibration patients.
    cv = StratifiedKFold(
        n_splits=n_splits, shuffle=True, random_state=RANDOM_SEED
    )
    oof_model = make_model()
    oof_decision_scores = cross_val_predict(
            oof_model,
            x_cal,
            y_cal,
            cv=cv,
            method="decision_function",
            n_jobs=None,
        )

    oof_normal_scores = oof_decision_scores[y_cal == 0]
    threshold, calibration_specificity = choose_threshold_for_specificity(
        oof_normal_scores, TARGET_SPECIFICITY
    )

    # Final model refit on the complete calibration/training partition.
    final_model = make_model()
    final_model.fit(x_cal, y_cal)

    # Larger SVM decision-function score means stronger evidence of cancer.
    test_scores = final_model.decision_function(x_test)
    test_pred = (test_scores > threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(y_test, test_pred, labels=[0, 1]).ravel()
    test_specificity = float(tn / (tn + fp)) if (tn + fp) else float("nan")
    test_sensitivity = float(tp / (tp + fn)) if (tp + fn) else float("nan")
    test_accuracy = float(accuracy_score(y_test, test_pred))
    test_balanced_accuracy = float(balanced_accuracy_score(y_test, test_pred))
    test_auc = safe_auc(y_test, test_scores)

    # ROC curve is descriptive only; select points with specificity strictly greater than the target. Do not alter the locked threshold.
    fpr, tpr, roc_thresholds = roc_curve(y_test, test_scores)
    specificity_curve = 1.0 - fpr
    eligible = np.where(specificity_curve > TARGET_SPECIFICITY)[0]
    if len(eligible):
        best_idx = eligible[np.argmax(tpr[eligible])]
        roc_sens_at_target = float(tpr[best_idx])
        roc_spec_at_target = float(specificity_curve[best_idx])
        roc_threshold_at_target = float(roc_thresholds[best_idx])
    else:
        roc_sens_at_target = float("nan")
        roc_spec_at_target = float("nan")
        roc_threshold_at_target = float("nan")

    predictions = test_df.copy()
    predictions["SVM_Decision_Score"] = test_scores
    predictions["SVM_Threshold_Calibrated"] = threshold
    predictions["SVM_Prediction"] = np.where(
        test_pred == 1, "Positive", "Negative"
    )
    predictions["True_Label_Binary"] = np.where(y_test == 1, "Cancer", "Normal")
    predictions["Correct_Prediction"] = test_pred == y_test

    cohort_test = cohort_values(test_df)
    per_cohort_rows = []
    for cohort in sorted(cohort_test.unique()):
        mask = cohort_test.eq(cohort).to_numpy()
        n = int(mask.sum())
        if cohort == NORMAL_LABEL.casefold():
            false_positives = int(test_pred[mask].sum())
            per_cohort_rows.append(
                {
                    "Cohort": cohort,
                    "n": n,
                    "True_Positive": np.nan,
                    "False_Negative": np.nan,
                    "False_Positive": false_positives,
                    "True_Negative": int((test_pred[mask] == 0).sum()),
                    "Sensitivity": np.nan,
                    "Specificity": float((test_pred[mask] == 0).mean()),
                }
            )
        else:
            true_positives = int(test_pred[mask].sum())
            false_negatives = int(n - true_positives)
            per_cohort_rows.append(
                {
                    "Cohort": cohort,
                    "n": n,
                    "True_Positive": true_positives,
                    "False_Negative": false_negatives,
                    "False_Positive": np.nan,
                    "True_Negative": np.nan,
                    "Sensitivity": float(true_positives / n) if n else float("nan"),
                    "Specificity": np.nan,
                }
            )

    reference_n = int((role_mask(df, "Reference") & normal).sum())
    cal1_normal_n = int(
        (role_mask(df, "Regression Fit Calibration 1") & normal).sum()
    )
    test_normal_n = int((y_test == 0).sum())
    test_cancer_n = int((y_test == 1).sum())

    summary_rows = [
        ("Method", "Support Vector Machine (RBF kernel), standardized features"),
        ("Biomarkers", " | ".join(BIOMARKERS)),
        ("Biomarker count", len(BIOMARKERS)),
        ("Input transformation", "Natural log of positive biomarker concentrations"),
        ("Model training partition", "Biomarker-Panel Calibration and Selection; labeled normals and cancers"),
        ("Model training n", len(cal_df)),
        ("Calibration training normal n", cal_normal_n),
        ("Calibration training cancer n", cal_cancer_n),
        (
            "Threshold calibration",
            f"{n_splits}-fold stratified out-of-fold SVM decision scores; "
            "calibration normals only",
        ),
        ("Reference normal n (not used by this supervised model)", reference_n),
        ("Calibration 1 normal n (not used by this supervised model)", cal1_normal_n),
        ("Test n", len(test_df)),
        ("Test normal n", test_normal_n),
        ("Test cancer n", test_cancer_n),
        ("Target calibration specificity (strictly greater than)", TARGET_SPECIFICITY),
        ("Observed OOF calibration-normal specificity", calibration_specificity),
        ("Calibrated SVM decision-score threshold", threshold),
        ("Test AUC", test_auc),
        ("Test sensitivity at locked threshold", test_sensitivity),
        ("Test specificity at locked threshold", test_specificity),
        ("Test accuracy at locked threshold", test_accuracy),
        ("Test balanced accuracy", test_balanced_accuracy),
        ("Test true positives", int(tp)),
        ("Test false negatives", int(fn)),
        ("Test true negatives", int(tn)),
        ("Test false positives", int(fp)),
        (
            f"Descriptive Test ROC sensitivity at >{TARGET_SPECIFICITY:.0%} specificity",
            roc_sens_at_target,
        ),
        ("Specificity at selected descriptive test ROC point", roc_spec_at_target),
        ("Test ROC threshold for descriptive point (NOT used for predictions)", roc_threshold_at_target),
        ("Classifier", "StandardScaler + SVC(kernel=rbf, C=1.0, gamma=scale, class_weight=balanced)"),
        ("Threshold rule", "Cancer positive if SVM decision_function score > calibrated threshold"),
        ("Data split", "Loaded existing Data Type roles; no resplitting"),
        ("Leakage note", "Test patients were not used for model fitting or threshold selection"),
    ]
    summary_df = pd.DataFrame(summary_rows, columns=["Metric", "Value"])
    cohort_df = pd.DataFrame(per_cohort_rows)

    roc_df = pd.DataFrame(
        {
            "False Positive Rate": fpr,
            "True Positive Rate": tpr,
            "Specificity": specificity_curve,
            "SVM Decision-Score Threshold": roc_thresholds,
        }
    )

    # Continuous ROC curve using independent Test-set decision scores only.
    # The calibrated operating threshold and all model-fitting steps remain unchanged.
    ROC_PLOT_FILE = INPUT_WORKBOOK.with_name("SVM_Test_ROC_Curve.png")

    plt.figure(figsize=(7, 6))
    plt.plot(
        fpr,
        tpr,
        linewidth=2.2,
        label=f"ROC curve (Test AUC = {test_auc:.3f})",
    )
    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        linewidth=1.2,
        label="No-discrimination reference",
    )
    plt.xlim(0.0, 1.0)
    plt.ylim(0.0, 1.05)
    plt.xlabel("False Positive Rate (1 - Specificity)")
    plt.ylabel("True Positive Rate (Sensitivity)")
    plt.title("SVM Cancer Detection — Test-set ROC Curve")
    plt.legend(loc="lower right")
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.savefig(ROC_PLOT_FILE, dpi=300, bbox_inches="tight")
    plt.close()

    OUTPUT_WORKBOOK.parent.mkdir(parents=True, exist_ok=True)
    with pd.ExcelWriter(OUTPUT_WORKBOOK, engine="openpyxl") as writer:
        summary_df.to_excel(writer, sheet_name="Summary", index=False)
        cohort_df.to_excel(writer, sheet_name="Per_Cohort_Test", index=False)
        predictions.to_excel(writer, sheet_name="Test_Predictions", index=False)
        roc_df.to_excel(writer, sheet_name="Test_ROC_Curve", index=False)

    print("\nSVM CANCER DETECTION — FINAL TEST RESULTS")
    print(f"Biomarkers: {BIOMARKERS}")
    print(
        f"Calibration training patients: {len(cal_df)} "
        f"(normal={cal_normal_n}, cancer={cal_cancer_n})"
    )
    print(f"Out-of-fold threshold calibration: {n_splits}-fold; normal patients only for specificity")
    print(f"Target calibration specificity (strictly greater than): {TARGET_SPECIFICITY:.1%}")
    print(f"Observed OOF calibration-normal specificity: {calibration_specificity:.2%}")
    print(f"Calibrated SVM decision-score threshold: {threshold:.8g}")
    print(
        f"Test patients: {len(test_df)} "
        f"(normal={test_normal_n}, cancer={test_cancer_n})"
    )
    print(f"Test AUC: {test_auc:.4f}")
    print(f"Test sensitivity at locked threshold: {test_sensitivity:.2%}")
    print(f"Test specificity at locked threshold: {test_specificity:.2%}")
    print(f"Test accuracy at locked threshold: {test_accuracy:.2%}")
    print(f"TP={tp}, FN={fn}, TN={tn}, FP={fp}")
    print("\nPer-cancer-type test sensitivity:")
    for row in per_cohort_rows:
        if str(row["Cohort"]).casefold() != NORMAL_LABEL.casefold():
            print(
                f"  {row['Cohort']}: n={row['n']}, "
                f"sensitivity={row['Sensitivity']:.2%}, "
                f"TP={row['True_Positive']}, FN={row['False_Negative']}"
            )
    print(f"\nSaved one results workbook:\n{OUTPUT_WORKBOOK}")
    print(f"Saved test ROC plot (300 DPI):\n{ROC_PLOT_FILE}")

if __name__ == "__main__":
    main()
```

```python
"""
Combined ROC comparison: TDAcancer-detect and baseline models.

The script reads previously saved continuous ROC coordinates and patient-level
Test predictions. It does not retrain models, recalibrate thresholds, or
regenerate patient partitions.

For a valid comparison, it verifies that all four methods contain exactly the
same Test patient IDs and binary labels. Patient-level, class-stratified
bootstrap confidence intervals are calculated from each model's continuous
Test scores. Point AUCs, partial AUCs, and plotted curves are calculated from
the saved ROC coordinates; the TDA continuous-score AUC is checked against its
saved ROC coordinates.

Outputs include a publication-quality ROC figure (PNG, PDF, and 300-dpi TIFF),
an Excel summary, and CSV files containing combined ROC coordinates and
statistics.

Installation:
    python -m pip install numpy pandas matplotlib scikit-learn openpyxl
"""

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import auc, roc_curve


# ============================================================
# 1. USER SETTINGS
# ============================================================

INPUT_DIR = Path(
    r"D:\Cancer Detection folder\Biomarker Selection"
)

# Saved ROC-coordinate files
TDA_ROC_FILE = INPUT_DIR / "TDA_test_ROC_curve.csv"

MAHALANOBIS_FILE = INPUT_DIR / (
    "Mahalanobis_8Biomarker_Cancer_Detection_95pctSpecificity.xlsx"
)

LOGISTIC_FILE = INPUT_DIR / (
    "LogisticRegression_8Biomarker_Cancer_Detection_95pctSpecificity.xlsx"
)

SVM_FILE = INPUT_DIR / (
    "SVM_8Biomarker_Cancer_Detection_95pctSpecificity.xlsx"
)

# Patient-level TDA TEST predictions, used only for bootstrap CI
TDA_PREDICTIONS_FILE = INPUT_DIR / "TDA_test_predictions.csv"

PREDICTION_SHEET = "Test_Predictions"

N_BOOTSTRAP = 2000
RANDOM_SEED = 20261031
PARTIAL_FPR_VALUES = (0.15, 0.35)

# Outputs
OUTPUT_PNG = INPUT_DIR / "Combined_ROC_Curves_AUC_CI_pAUC_sens100.png"
OUTPUT_PDF = INPUT_DIR / "Combined_ROC_Curves_AUC_CI_pAUC_sens100.pdf"
OUTPUT_TIFF = INPUT_DIR / "Combined_ROC_Curves_AUC_CI_pAUC_sens100_300dpi.tiff"
OUTPUT_EXCEL = INPUT_DIR / "Combined_ROC_AUC_CI_pAUC_Summary_sens100.xlsx"
OUTPUT_COORDINATES_CSV = INPUT_DIR / "Combined_ROC_Coordinates_sens100.csv"
OUTPUT_STATISTICS_CSV = INPUT_DIR / "Combined_ROC_Statistics_sens100.csv"


# ============================================================
# 2. COMMON ROC DATA READER
# ============================================================

def validate_roc_data(df, model_name):

    required_columns = [
        "False Positive Rate",
        "True Positive Rate"
    ]

    missing = [
        col for col in required_columns
        if col not in df.columns
    ]

    if missing:
        raise ValueError(
            f"{model_name}: Missing ROC columns: {missing}. "
            f"Available columns: {list(df.columns)}"
        )

    fpr = pd.to_numeric(
        df["False Positive Rate"],
        errors="coerce"
    ).to_numpy(dtype=float)

    tpr = pd.to_numeric(
        df["True Positive Rate"],
        errors="coerce"
    ).to_numpy(dtype=float)

    if len(fpr) < 2 or len(tpr) < 2:
        raise ValueError(
            f"{model_name}: Insufficient ROC coordinates."
        )

    if len(fpr) != len(tpr):
        raise ValueError(
            f"{model_name}: FPR and TPR lengths differ."
        )

    if not (
        np.isfinite(fpr).all()
        and np.isfinite(tpr).all()
    ):
        raise ValueError(
            f"{model_name}: Non-finite ROC coordinates found."
        )

    if (
        (fpr < 0).any()
        or (fpr > 1).any()
        or (tpr < 0).any()
        or (tpr > 1).any()
    ):
        raise ValueError(
            f"{model_name}: ROC coordinates must lie between 0 and 1."
        )

    if np.any(np.diff(fpr) < -1e-12):
        raise ValueError(
            f"{model_name}: FPR coordinates are not in increasing order."
        )

    # Retain the supplied ROC coordinates without reconstructing the curve.
    model_auc = float(auc(fpr, tpr))

    return fpr, tpr, model_auc


# ============================================================
# 3. LOAD SAVED ROC CURVES
# ============================================================

def load_tda_roc():

    if not TDA_ROC_FILE.exists():
        raise FileNotFoundError(
            f"TDA ROC file not found:\n{TDA_ROC_FILE}"
        )

    df = pd.read_csv(TDA_ROC_FILE)

    return validate_roc_data(
        df,
        "TDAcancer-detect"
    )


def load_excel_roc(file_path, model_name):

    if not file_path.exists():
        raise FileNotFoundError(
            f"{model_name} results workbook not found:\n{file_path}"
        )

    df = pd.read_excel(
        file_path,
        sheet_name="Test_ROC_Curve"
    )

    return validate_roc_data(
        df,
        model_name
    )


# ============================================================
# 4. PARTIAL AUC FROM SAVED ROC COORDINATES
# ============================================================

def calculate_raw_partial_auc(fpr, tpr, max_fpr=PARTIAL_FPR_VALUES[0]):
    """
    Calculate raw (unstandardized) partial AUC from FPR=0 to max_fpr.
    The endpoint at max_fpr is linearly interpolated.
    """

    fpr = np.asarray(fpr, dtype=float)
    tpr = np.asarray(tpr, dtype=float)

    order = np.argsort(fpr)
    x = fpr[order]
    y = tpr[order]

    # Keep all points strictly below the requested FPR limit.
    keep = x < max_fpr
    x_part = x[keep]
    y_part = y[keep]

    # Include FPR=0 if it is not already present.
    if len(x_part) == 0 or x_part[0] > 0:
        x_part = np.insert(x_part, 0, 0.0)
        y_part = np.insert(
            y_part, 0, float(np.interp(0.0, x, y))
        )

    # Include an interpolated endpoint exactly at the requested FPR limit.
    x_part = np.append(x_part, max_fpr)
    y_part = np.append(
        y_part, float(np.interp(max_fpr, x, y))
    )

    return float(np.trapz(y_part, x_part))



def calculate_sensitivity_at_100_specificity(fpr, tpr, tolerance=1e-12):
    """
    Sensitivity at 100% specificity, i.e. the maximum TPR attained
    at FPR = 0 on the supplied ROC coordinates.

    Returns sensitivity as a percentage (0-100).
    """
    fpr = np.asarray(fpr, dtype=float)
    tpr = np.asarray(tpr, dtype=float)

    zero_fpr = np.isclose(fpr, 0.0, atol=tolerance, rtol=0.0)

    if not np.any(zero_fpr):
        raise ValueError(
            "The saved ROC curve contains no FPR=0 coordinate; "
            "sensitivity at 100% specificity cannot be determined."
        )

    return float(np.max(tpr[zero_fpr]) * 100.0)



# ============================================================
# 5. PATIENT-LEVEL DATA FOR BOOTSTRAP AUC CONFIDENCE INTERVAL
# ============================================================

def find_column(df, candidates, model_name):

    normalized = {
        str(col).strip().casefold(): col
        for col in df.columns
    }

    for candidate in candidates:
        key = candidate.strip().casefold()
        if key in normalized:
            return normalized[key]

    raise ValueError(
        f"{model_name}: Could not find any of {candidates}. "
        f"Available columns: {list(df.columns)}"
    )


def normalize_patient_ids(series, model_name):
    """Return normalized patient identifiers and reject missing/duplicate IDs."""
    ids = series.astype(str).str.strip()
    invalid = series.isna() | ids.eq("") | ids.str.casefold().isin(
        {"nan", "none", "nat"}
    )
    if invalid.any():
        raise ValueError(
            f"{model_name}: Missing patient IDs found in the Test predictions."
        )
    if ids.duplicated().any():
        duplicates = ids[ids.duplicated(keep=False)].unique()[:10]
        raise ValueError(
            f"{model_name}: Duplicate Test patient IDs found: {duplicates}"
        )
    return ids.to_numpy()


def convert_labels(series, model_name):
    """
    Convert binary labels or cohort names:
        Normal/Healthy -> 0
        Cancer cohort names -> 1
    """

    numeric = pd.to_numeric(series, errors="coerce")

    if numeric.notna().all() and set(
        numeric.unique()
    ).issubset({0, 1}):
        return numeric.astype(int).to_numpy()

    label_map = {
        "normal": 0,
        "healthy": 0,
        "control": 0,
        "negative": 0,
        "0": 0,
        "0.0": 0,

        "cancer": 1,
        "positive": 1,
        "case": 1,
        "1": 1,
        "1.0": 1,

        "bladder": 1,
        "bladder cancer": 1,
        "ovarian": 1,
        "ovary": 1,
        "ovarian cancer": 1,
        "pancreatic": 1,
        "pancreas": 1,
        "pancreatic cancer": 1,
        "breast": 1,
        "breast cancer": 1,
        "colorectal": 1,
        "colorectum": 1,
        "colon": 1,
        "esophageal": 1,
        "esophagus": 1,
        "oesophageal": 1,
        "oesophagus": 1,
        "liver": 1,
        "lung": 1,
        "prostate": 1,
        "stomach": 1,
        "gastric": 1,
    }

    normalized = (
        series.astype(str)
        .str.strip()
        .str.casefold()
    )

    mapped = normalized.map(label_map)

    if mapped.isna().any():
        bad = (
            series[mapped.isna()]
            .dropna()
            .astype(str)
            .unique()[:10]
        )

        raise ValueError(
            f"{model_name}: Unrecognized labels {bad}. "
            "Expected Normal/Healthy, binary 0/1, or cancer cohort names."
        )

    return mapped.astype(int).to_numpy()


def load_patient_predictions(
    file_path,
    model_name,
    score_candidates,
    csv_file=False
):

    if not file_path.exists():
        raise FileNotFoundError(
            f"Patient-level TEST predictions not found for {model_name}:\n"
            f"{file_path}\n\n"
            "A patient-level bootstrap confidence interval cannot be "
            "calculated from ROC coordinates alone. Provide one row per "
            "TEST patient with the true label and continuous prediction score."
        )

    if csv_file:
        df = pd.read_csv(file_path)
    else:
        df = pd.read_excel(
            file_path,
            sheet_name=PREDICTION_SHEET
        )

    label_col = find_column(
        df,
        [
            "True_Label_Binary",
            "True Label Binary",
            "True_Label",
            "True Label",
            "Actual_Label",
            "Actual Label",
            "Cohort",
            "Diagnosis",
            "Label"
        ],
        model_name
    )

    score_col = find_column(
        df,
        score_candidates,
        model_name
    )
    patient_id_col = find_column(
        df,
        [
            "Patient ID #",
            "Patient ID",
            "Patient_ID",
            "PatientID",
            "Patient_ID_Number",
            "ID",
        ],
        model_name,
    )

    patient_ids = normalize_patient_ids(df[patient_id_col], model_name)
    y = convert_labels(
        df[label_col],
        model_name
    )

    scores = pd.to_numeric(
        df[score_col],
        errors="coerce"
    ).to_numpy(dtype=float)

    valid = np.isfinite(scores)

    y = y[valid]
    scores = scores[valid]
    patient_ids = patient_ids[valid]

    if len(np.unique(y)) != 2:
        raise ValueError(
            f"{model_name}: Patient-level TEST predictions must contain "
            "both cancer and normal patients."
        )

    return patient_ids, y, scores


# ============================================================
# 6. STRATIFIED PATIENT-LEVEL BOOTSTRAP AUC 95% CI
# ============================================================

def bootstrap_auc_ci(
    y,
    scores,
    n_bootstrap=2000,
    seed=20261031
):
    """
    Percentile 95% CI using class-stratified patient-level bootstrap.
    Cancer and normal patients are independently resampled with replacement.
    """

    rng = np.random.default_rng(seed)

    y = np.asarray(y, dtype=int)
    scores = np.asarray(scores, dtype=float)

    positive_indices = np.flatnonzero(y == 1)
    negative_indices = np.flatnonzero(y == 0)

    if (
        len(positive_indices) < 2
        or len(negative_indices) < 2
    ):
        raise ValueError(
            "At least two cancer and two normal TEST patients are required "
            "for bootstrap confidence intervals."
        )

    bootstrap_aucs = []

    for _ in range(n_bootstrap):

        sampled_positive = rng.choice(
            positive_indices,
            size=len(positive_indices),
            replace=True
        )

        sampled_negative = rng.choice(
            negative_indices,
            size=len(negative_indices),
            replace=True
        )

        sampled_indices = np.concatenate([
            sampled_positive,
            sampled_negative
        ])

        fpr_boot, tpr_boot, _ = roc_curve(
            y[sampled_indices],
            scores[sampled_indices],
            pos_label=1
        )

        bootstrap_aucs.append(
            auc(fpr_boot, tpr_boot)
        )

    ci_low, ci_high = np.percentile(
        bootstrap_aucs,
        [2.5, 97.5]
    )

    return float(ci_low), float(ci_high)


# ============================================================
# 7. LOAD ALL FOUR SAVED ROC CURVES
# ============================================================

roc_results = {}

roc_results["TDAcancer-detect"] = load_tda_roc()

roc_results["Mahalanobis distance"] = load_excel_roc(
    MAHALANOBIS_FILE,
    "Mahalanobis distance"
)

roc_results["Logistic regression"] = load_excel_roc(
    LOGISTIC_FILE,
    "Logistic regression"
)

roc_results["Support Vector Machine (SVM)"] = load_excel_roc(
    SVM_FILE,
    "Support Vector Machine"
)


# ============================================================
# 8. LOAD PATIENT-LEVEL TEST PREDICTIONS FOR AUC CI
# ============================================================

patient_predictions = {}

patient_predictions["TDAcancer-detect"] = load_patient_predictions(
    TDA_PREDICTIONS_FILE,
    "TDAcancer-detect",
    [
        # This is the exact continuous score used to generate TDA_test_ROC_curve.csv.
        # Do not substitute Positive Conditions Count (a discrete 0/1/2 score).
        "ROC Continuous Score"
    ],
    csv_file=True
)

patient_predictions["Mahalanobis distance"] = load_patient_predictions(
    MAHALANOBIS_FILE,
    "Mahalanobis distance",
    [
        "Mahalanobis_Distance",
        "Mahalanobis Distance",
        "Mahalanobis_Score",
        "Distance",
        "Prediction_Score",
        "Score"
    ]
)

patient_predictions["Logistic regression"] = load_patient_predictions(
    LOGISTIC_FILE,
    "Logistic regression",
    [
        "LogisticRegression_Cancer_Probability",
        "Cancer_Probability",
        "Predicted_Probability",
        "Probability",
        "Prediction_Score",
        "Score"
    ]
)

patient_predictions["Support Vector Machine (SVM)"] = load_patient_predictions(
    SVM_FILE,
    "Support Vector Machine (SVM)",
    [
        "SVM_Decision_Score",
        "Decision_Score",
        "SVM Score",
        "SVM_Score",
        "Prediction_Score",
        "Score"
    ]
)


# Verify exact patient-level Test-set identity across all four methods.
# Matching class counts alone is insufficient: the patient IDs and labels
# must also match. Reorder each method to the TDA patient order.
reference_model = "TDAcancer-detect"
reference_ids, reference_y, _ = patient_predictions[reference_model]
reference_id_set = set(reference_ids)

for model_name, (patient_ids, y_values, score_values) in list(
    patient_predictions.items()
):
    current_id_set = set(patient_ids)
    if current_id_set != reference_id_set:
        missing_ids = sorted(reference_id_set - current_id_set)[:10]
        extra_ids = sorted(current_id_set - reference_id_set)[:10]
        raise ValueError(
            f"{model_name}: Test patient IDs do not match {reference_model}. "
            f"Missing IDs (first 10): {missing_ids}; "
            f"extra IDs (first 10): {extra_ids}. "
            "Use prediction files generated from the identical Test partition."
        )

    index_by_id = {patient_id: index for index, patient_id in enumerate(patient_ids)}
    order = np.asarray([index_by_id[patient_id] for patient_id in reference_ids])
    aligned_y = y_values[order]
    aligned_scores = score_values[order]

    if not np.array_equal(aligned_y, reference_y):
        mismatch = np.flatnonzero(aligned_y != reference_y)[:10]
        mismatch_ids = reference_ids[mismatch]
        raise ValueError(
            f"{model_name}: True labels differ from {reference_model} for "
            f"Test patient IDs: {mismatch_ids}. Check the cohort labels and "
            "prediction files before comparing models."
        )

    patient_predictions[model_name] = (
        reference_ids.copy(),
        aligned_y,
        aligned_scores,
    )

# Explicitly report the common Test-set class counts.
common_cancer_n = int(np.sum(reference_y == 1))
common_normal_n = int(np.sum(reference_y == 0))
print(
    "Verified identical Test patients across all methods: "
    f"n={len(reference_y)} (cancer={common_cancer_n}, normal={common_normal_n})"
)


# ============================================================
# 9. CALCULATE AUC, 95% CI, AND PARTIAL AUC
# ============================================================

statistics_rows = []

for model_index, (model_name, (fpr, tpr, model_auc)) in enumerate(
    roc_results.items()
):

    patient_ids, y_patient, score_patient = patient_predictions[model_name]

    # Recalculate the point AUC directly from the same patient-level scores
    # used for the bootstrap CI. For TDA, score_patient is explicitly
    # "ROC Continuous Score", not the discrete Positive Conditions Count.
    patient_fpr, patient_tpr, _ = roc_curve(
        y_patient,
        score_patient,
        pos_label=1
    )
    patient_level_auc = float(auc(patient_fpr, patient_tpr))

    # The saved TDA ROC coordinates must reproduce the continuous-score AUC.
    # If they do not, stop rather than reporting a mismatched point estimate
    # and confidence interval.
    if model_name == "TDAcancer-detect":
        diagnostic_difference = abs(model_auc - patient_level_auc)
        if diagnostic_difference > 1e-8:
            raise ValueError(
                "TDA AUC mismatch: the saved TDA ROC coordinates give "
                f"{model_auc:.10f}, whereas patient-level ROC Continuous Score "
                f"gives {patient_level_auc:.10f}. Confirm that the ROC CSV and "
                "TDA_test_predictions.csv were generated by the same run."
            )
        model_auc = patient_level_auc

    ci_low, ci_high = bootstrap_auc_ci(
        y_patient,
        score_patient,
        n_bootstrap=N_BOOTSTRAP,
        seed=RANDOM_SEED + model_index
    )

    # Partial AUCs are calculated from the same saved ROC coordinates
    # that are used to draw the curves.
    partial_auc_015 = calculate_raw_partial_auc(
        fpr, tpr, max_fpr=PARTIAL_FPR_VALUES[0]
    )
    partial_auc_035 = calculate_raw_partial_auc(
        fpr, tpr, max_fpr=PARTIAL_FPR_VALUES[1]
    )

    sensitivity_at_100_spec = calculate_sensitivity_at_100_specificity(
        fpr, tpr
    )

    statistics_rows.append({
        "Model": model_name,
        "Overall AUC": model_auc,
        "AUC 95% CI Lower": ci_low,
        "AUC 95% CI Upper": ci_high,
        "Overall AUC (95% CI)": (
            f"{model_auc:.2f} ({ci_low:.2f}-{ci_high:.2f})"
        ),
        "Partial AUC (FPR <= 0.15; raw)": partial_auc_015,
        "Partial AUC (FPR <= 0.35; raw)": partial_auc_035,
        "Sensitivity at 100% Specificity (%)": sensitivity_at_100_spec,
        "Patient-level AUC (diagnostic)": patient_level_auc,
        "Absolute AUC difference (diagnostic)": abs(
            model_auc - patient_level_auc
        ),
        "Number of saved ROC coordinates": len(fpr),
        "Number of TEST patients": len(y_patient),
        "TEST cancer patients": int(np.sum(y_patient == 1)),
        "TEST normal patients": int(np.sum(y_patient == 0)),
        "Bootstrap replicates": N_BOOTSTRAP
    })


statistics_df = pd.DataFrame(statistics_rows)

# Round reported statistics to two decimal places.
# Calculations and bootstrap resampling are performed at full precision.
statistics_df[
    [
        "Overall AUC",
        "AUC 95% CI Lower",
        "AUC 95% CI Upper",
        "Partial AUC (FPR <= 0.15; raw)",
        "Partial AUC (FPR <= 0.35; raw)",
        "Sensitivity at 100% Specificity (%)",
        "Patient-level AUC (diagnostic)",
        "Absolute AUC difference (diagnostic)"
    ]
] = statistics_df[
    [
        "Overall AUC",
        "AUC 95% CI Lower",
        "AUC 95% CI Upper",
        "Partial AUC (FPR <= 0.15; raw)",
        "Partial AUC (FPR <= 0.35; raw)",
        "Sensitivity at 100% Specificity (%)",
        "Patient-level AUC (diagnostic)",
        "Absolute AUC difference (diagnostic)"
    ]
].round(2)



# ============================================================
# 10. PREPARE ALL ORIGINAL ROC COORDINATES
# ============================================================

coordinate_tables = []

for model_name, (fpr, tpr, model_auc) in roc_results.items():

    coordinate_tables.append(
        pd.DataFrame({
            "Model": model_name,
            "False Positive Rate": fpr,
            "True Positive Rate": tpr
        })
    )


combined_coordinates = pd.concat(
    coordinate_tables,
    ignore_index=True
)


# ============================================================
# 11. GENERATE PUBLICATION-QUALITY COMBINED ROC FIGURE
# ============================================================

INPUT_DIR.mkdir(parents=True, exist_ok=True)

# A clean, journal-oriented style. The figure is saved as vector PDF and
# high-resolution raster formats for manuscript submission.
plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman", "Times", "DejaVu Serif"],
    "mathtext.fontset": "stix",
    "font.size": 20,
    "axes.labelsize": 18,
    "axes.titlesize": 22,
    "xtick.labelsize": 20,
    "ytick.labelsize": 20,
    "legend.fontsize": 15,
    "axes.linewidth": 1.0,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "savefig.facecolor": "white",
})

fig, ax = plt.subplots(figsize=(8.2, 7.2))

for model_name, (fpr, tpr, model_auc) in roc_results.items():

    stats_row = statistics_df.loc[
        statistics_df["Model"] == model_name
    ].iloc[0]

    # Show only overall AUC and its 95% CI in the figure legend.
    # Partial AUC values for FPR <= 0.15 and FPR <= 0.35 remain
    # calculated and are retained in the Excel/CSV statistical outputs.
    legend_label = (
        f"{model_name} "
        f"(AUC = {model_auc:.2f}; "
        f"95% CI: {stats_row['AUC 95% CI Lower']:.2f}–"
        f"{stats_row['AUC 95% CI Upper']:.2f})"
    )

    if model_name == "TDAcancer-detect":
        ax.plot(
            fpr, tpr,
            linewidth=3.0,
            linestyle="-",
            label=legend_label,
            zorder=5
        )

    elif model_name == "Mahalanobis distance":
        ax.plot(
            fpr, tpr,
            linewidth=2.0,
            linestyle="--",
            label=legend_label
        )

    elif model_name == "Logistic regression":
        ax.plot(
            fpr, tpr,
            linewidth=2.0,
            linestyle="-.",
            label=legend_label
        )

    elif model_name == "Support Vector Machine (SVM)":
        ax.plot(
            fpr, tpr,
            linewidth=2.0,
            linestyle=":",
            label=legend_label
        )

# No-discrimination reference
ax.plot(
    [0, 1], [0, 1],
    color="black",
    linestyle="--",
    linewidth=1.0,
    label="No-discrimination reference"
)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1.02)
ax.set_xlabel("False Positive Rate (1 − Specificity)", labelpad=7)
ax.set_ylabel("True Positive Rate (Sensitivity)", labelpad=7)

# Avoid a redundant in-figure title; Springer figure captions are normally
# supplied separately in the manuscript.
ax.tick_params(axis="both", which="major", direction="out",
               length=4, width=0.9)
ax.grid(True, linestyle=":", linewidth=0.6, alpha=0.35)
ax.set_aspect("equal", adjustable="box")

# Larger legend text and adequate spacing. A two-column legend would make
# these long statistical labels difficult to read, so retain one column.
legend = ax.legend(
    loc="lower right",
    fontsize=12,
    frameon=True,
    fancybox=False,
    framealpha=1.0,
    edgecolor="0.35",
    borderpad=0.8,
    labelspacing=0.75,
    handlelength=2.8,
    handletextpad=0.8
)
legend.get_frame().set_linewidth(0.8)

fig.tight_layout()

fig.savefig(
    OUTPUT_PNG,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

fig.savefig(
    OUTPUT_PDF,
    bbox_inches="tight",
    facecolor="white"
)

fig.savefig(
    OUTPUT_TIFF,
    dpi=300,
    bbox_inches="tight",
    facecolor="white",
    pil_kwargs={"compression": "tiff_lzw"}
)

plt.show()


# ============================================================
# 12. SAVE STATISTICS AND ROC COORDINATES
# ============================================================

with pd.ExcelWriter(
    OUTPUT_EXCEL,
    engine="openpyxl"
) as writer:

    statistics_df.to_excel(
        writer,
        sheet_name="ROC_Statistics",
        index=False
    )

    combined_coordinates.to_excel(
        writer,
        sheet_name="ROC_Coordinates",
        index=False
    )


combined_coordinates.to_csv(
    OUTPUT_COORDINATES_CSV,
    index=False
)

statistics_df.to_csv(
    OUTPUT_STATISTICS_CSV,
    index=False
)


# ============================================================
# 13. PRINT FINAL RESULTS
# ============================================================

print("\n==============================================")
print("COMBINED ROC CURVE: TEST-SET RESULTS")
print("==============================================")

print(
    statistics_df[
        [
            "Model",
            "Overall AUC (95% CI)",
            "Partial AUC (FPR <= 0.15; raw)",
            "Partial AUC (FPR <= 0.35; raw)",
            "Sensitivity at 100% Specificity (%)",
                "Patient-level AUC (diagnostic)",
            "Absolute AUC difference (diagnostic)"
        ]
    ].to_string(index=False)
)

print("\nSaved files:")
print("Combined ROC PNG:", OUTPUT_PNG)
print("Combined ROC PDF:", OUTPUT_PDF)
print("Combined ROC TIFF (600 dpi):", OUTPUT_TIFF)
print("AUC and pAUC statistics Excel:", OUTPUT_EXCEL)
print("Combined ROC coordinates CSV:", OUTPUT_COORDINATES_CSV)
print("Statistics CSV:", OUTPUT_STATISTICS_CSV)

print("\nAll four saved ROC curves have been combined using the same verified Test patients.")
```

```python
"""
TDA-based biomarker-panel selection using a four-stage constrained search.

Methodological overview
-----------------------
1. Assigns patients to four mutually exclusive roles:
   - Reference: the first configured number of healthy/normal patients.
   - Regression Fit Calibration 1: a configured fraction of the remaining
     healthy/normal patients; used to fit the regression model and calculate
     the reference coordinates for Distance %(1).
   - Biomarker-Panel Calibration and Selection: configured fractions of the
     remaining normal patients and each cancer cohort; used for threshold and
     biomarker-panel selection.
   - Test: the remaining patients, held out until final evaluation.
2. Evaluates exactly TOTAL_PANEL_BUDGET unique biomarker panels, each containing
   PANEL_SIZE biomarkers, across four search stages.
3. Stage 1 promotes balanced biomarker representation and low overlap between panels.
4. Stage 2 uses biomarker recurrence in the strongest specificity-feasible panels
   to guide weighted panel sampling.
5. Stage 3 performs balanced exploration within a pool of frequently recurring
   biomarkers from the strongest feasible panels.
6. Stage 4 explores one-biomarker substitutions around high-performing panels.
7. Selects the final panel from all four stages, rather than from Stage 4 alone.
8. For each panel, fits the linear regression model and calculates the
   Distance %(1) reference coordinates using Calibration 1 healthy/normal
   patients only. Candidate thresholds for Regression Difference Percentage
   and Distance %(1) are evaluated on Calibration 2 using an OR decision rule.
   Threshold pairs must satisfy the pooled Calibration 1 + Calibration 2
   normal-specificity constraint and are ranked primarily by macro-sensitivity
   across the cancer cohorts represented in Calibration 2.
9. Evaluates the selected panel once on the untouched Test partition using the
   frozen thresholds and Calibration 1-fitted reference quantities.

Implementation notes
--------------------
- Set the input workbook path, worksheet, biomarker names, and column names in
  the USER-DEFINED SETTINGS section.
- Cancer cohorts are detected automatically from the Cohort column; the normal
  cohort is identified by NORMAL_COHORT.
- The split is generated in memory; this script does not write an updated
  workbook.
- Selected biomarker values must be finite and strictly positive because the
  feature calculation applies a natural-log transformation.
- The correlation-derived matrix D = 1 - |C| is treated as a dissimilarity
  matrix. It is symmetric and nonnegative with a zero diagonal, but it is not
  assumed to satisfy the triangle inequality.
"""

from pathlib import Path
import math
import random

import gudhi
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression


# =============================================================================
# USER-DEFINED SETTINGS
# Change these values when applying the function to another dataset.
# Algorithmic tuning parameters below this section retain their defaults.
# =============================================================================

# Input workbook
clinical_data_path = Path(
    r"D:\Cancer Detection folder\Biomarker Selection\clinical cancer validation data.xlsx"
)
sheet_name = "Normal and Cancer"

# Dataset-specific biomarker universe (names must match workbook column names).
BIOMARKERS = [
    "Tenascin C", "sAXL", "sE-selectin", "sHGFR/c-Met", "sHer2", "sHer3",
    "sIL-6Ra", "sNeuropilin-1", "sPECAM-1", "sVEGFR1", "sVEGFR3",
    "sc-kit/SCFR", "CA-125", "CA 15-3", "CA 19-9", "CEA", "FGF2", "HE4",
    "HGF", "IL-6", "IL-8", "Leptin", "MIF", "OPN", "Prolactin", "SCF",
    "TNFa", "TRAIL", "Total PSA", "VEGF", "b-HCG", "sFAS", "Cathepsin D",
    "FAP alpha", "Ferritin", "Galectin-3", "IGFBP3", "MIA", "MPO", "SHBG",
    "TIMP1", "TIMP2"
]

# Cohort labels and preferred display/sort order.
# Cancer cohorts are detected from the workbook; this mapping only sets preferred ordering.
NORMAL_COHORT = "normal"
cohort_order = {"normal": 0, "bladder": 1, "ovarian": 2, "pancreatic": 3}
# Replaced with the observed non-normal cohort labels when the workbook is loaded.
CANCER_TYPES = [name.casefold() for name in cohort_order if name.casefold() != NORMAL_COHORT.casefold()]

# Dataset-specific column names.
COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
STAGE_COL = "AJCC Stage"
DATA_TYPE_COL = "Data Type"

# Data split and model constraints.
# Edit the two fractions below whenever needed. They are fractions, not whole percentages:
#   0.20 = 20%, 0.30 = 30%, 0.50 = 50%.
# For NORMAL patients only, the first n rows (from REFERENCE_SAMPLE_SIZES) are Reference;
# the configured Calibration 1 and Calibration 2 fractions are applied to the remaining
# normal patients, with Test receiving the remainder.
# CANCER patients are NEVER assigned to Calibration 1: CALIBRATION2_FRACTION of each
# cancer cohort is assigned to Biomarker-Panel Calibration and Selection, and the rest
# is Test. Thus Calibration 1 contains only healthy/normal patients.
REFERENCE_SAMPLE_SIZES = [19, 29, 39, 49, 59, 69]  # User-defined: one complete search per reference size
PROGRESS_INTERVAL = 300
CALIBRATION1_FRACTION = 0.15
CALIBRATION2_FRACTION = 0.50
FPR = 0.01  # Require pooled Calibration 1 + Calibration 2 normal specificity strictly > 99.00%
SENSITIVITY_TARGET = 0.99  # Diagnostic target only; specificity is the hard constraint
PANEL_SIZE = 8
TOTAL_PANEL_BUDGET = 3000
RANDOM_SEED = 20261017

# Maximum number of candidate thresholds to test for EACH detection score:
# - One set of candidates is generated for Regression Difference Percentage (k1).
# - One set is generated for Distance %(1) (k2).
# - The threshold search evaluates up to MAX_THRESHOLD_CANDIDATES x
#   MAX_THRESHOLD_CANDIDATES candidate pairs for the OR decision rule.
# - Lower values generally run faster but explore fewer threshold combinations;
#   higher values search more combinations and may take longer.
MAX_THRESHOLD_CANDIDATES = 80

# OPTIONAL USER-DEFINED THRESHOLD SEARCH BOUNDS (inclusive)
# Leave ALL FOUR values as None to use the original unbounded candidate search.
# To restrict the search, provide all four numeric values.
# Example bounded search: K1_MIN=-1.0, K1_MAX=5.0, K2_MIN=0.0, K2_MAX=10.0
K1_MIN = None  # Regression Difference Percentage lower bound, or None
K1_MAX = None  # Regression Difference Percentage upper bound, or None
K2_MIN = None  # Distance %(1) lower bound, or None
K2_MAX = None  # Distance %(1) upper bound, or None

# Algorithm defaults: normally leave these unchanged unless tuning the search.
PHASE_WEIGHTS = {
    "balanced_diverse_exploration": 0.40,
    "sensitivity_guided_search": 0.30,
    "recurrence_informed_balanced_search": 0.20,
    "best_panel_neighborhoods": 0.10,
}
CANDIDATE_POOL_SIZE = 300
TOP_PANEL_FRACTION = 0.20
RECURRENCE_POOL_SIZE = 16
PRINT_EACH_PANEL_PERFORMANCE = False


def calculate_phase_budgets(total_budget, phase_weights):
    if not isinstance(total_budget, int) or total_budget < 1:
        raise ValueError("TOTAL_PANEL_BUDGET must be a positive integer.")
    weight_sum = sum(phase_weights.values())
    if not math.isclose(weight_sum, 1.0, rel_tol=0.0, abs_tol=1e-9):
        raise ValueError("PHASE_WEIGHTS must sum to 1.0.")
    exact = {name: total_budget * weight for name, weight in phase_weights.items()}
    budgets = {name: math.floor(value) for name, value in exact.items()}
    remaining = total_budget - sum(budgets.values())
    order = sorted(
        phase_weights,
        key=lambda name: (exact[name] - budgets[name], phase_weights[name]),
        reverse=True,
    )
    for name in order[:remaining]:
        budgets[name] += 1
    return budgets


PHASE_BUDGETS = calculate_phase_budgets(TOTAL_PANEL_BUDGET, PHASE_WEIGHTS)

# TDA features used by the regression model and the Distance %(1) score.
REGRESSION_PREDICTORS = [
    "0-Dim Hole Min Life Range",
    "0-Dim Hole Max Life Range",
    "Median_Death_0D",
    "Std_Death_0D",
]


# -------------------- WORKBOOK / SPLIT --------------------


def validate_split_fractions():
    """Validate configurable calibration fractions; Test is the remainder."""
    for name, value in (("CALIBRATION1_FRACTION", CALIBRATION1_FRACTION),
                        ("CALIBRATION2_FRACTION", CALIBRATION2_FRACTION)):
        if not isinstance(value, (int, float)) or not np.isfinite(value):
            raise ValueError(f"{name} must be a finite number between 0 and 1.")
        if value < 0 or value > 1:
            raise ValueError(f"{name} must be between 0 and 1; received {value}.")
    total_calibration = CALIBRATION1_FRACTION + CALIBRATION2_FRACTION
    if total_calibration >= 1.0:
        raise ValueError(
            "CALIBRATION1_FRACTION + CALIBRATION2_FRACTION must be less than 1.0 "
            "so that Test has a positive fraction. "
            f"Received {CALIBRATION1_FRACTION} + {CALIBRATION2_FRACTION}."
        )


def create_split_dataframe(source_path, sheet, n_reference_normal):
    """Load the worksheet and assign the four patient roles in memory."""
    global CANCER_TYPES
    validate_split_fractions()
    source_path = Path(source_path)
    if not source_path.exists():
        raise FileNotFoundError(f"Source workbook not found: {source_path}\nChange clinical_data_path.")
    df = pd.read_excel(source_path, sheet_name=sheet).copy()
    required = [COHORT_COL, PATIENT_ID_COL, STAGE_COL] + BIOMARKERS
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns in source workbook: {missing}")
    if DATA_TYPE_COL in df.columns:
        df = df.drop(columns=[DATA_TYPE_COL])

    cohort = df[COHORT_COL].astype(str).str.strip().str.casefold()
    if df[COHORT_COL].isna().any():
        raise ValueError(f"Column {COHORT_COL!r} contains missing cohort labels.")
    if cohort.eq("").any():
        raise ValueError(f"Column {COHORT_COL!r} contains blank cohort labels.")
    observed_cohorts = list(dict.fromkeys(cohort.tolist()))
    CANCER_TYPES = [name for name in observed_cohorts if name and name != NORMAL_COHORT.casefold()]
    if not CANCER_TYPES:
        raise ValueError(f"No cancer cohorts were detected in {COHORT_COL!r}.")

    normal_indices = df.index[cohort.eq(NORMAL_COHORT.casefold())].tolist()
    if len(normal_indices) <= n_reference_normal:
        raise ValueError(
            f"Need more than {n_reference_normal} normal patients; found {len(normal_indices)}."
        )

    role = {}
    reference_indices = normal_indices[:n_reference_normal]
    for idx in reference_indices:
        role[idx] = "Reference"

    rng = np.random.RandomState(RANDOM_SEED)
    remaining_normals = list(normal_indices[n_reference_normal:])
    rng.shuffle(remaining_normals)
    n1 = int(np.floor(len(remaining_normals) * CALIBRATION1_FRACTION))
    n2 = int(np.floor(len(remaining_normals) * CALIBRATION2_FRACTION))
    for idx in remaining_normals[:n1]:
        role[idx] = "Regression Fit Calibration 1"
    for idx in remaining_normals[n1:n1+n2]:
        role[idx] = "Biomarker-Panel Calibration and Selection"
    for idx in remaining_normals[n1+n2:]:
        role[idx] = "Test"

    # Keep the original cohort iteration order and RNG consumption for cancer splits.
    for cancer in CANCER_TYPES:
        indices = df.index[cohort.eq(cancer.casefold())].tolist()
        if not indices:
            raise ValueError(f"No rows found for cancer cohort {cancer!r}.")
        shuffled = list(indices)
        rng.shuffle(shuffled)
        n2c = int(np.floor(len(shuffled) * CALIBRATION2_FRACTION))
        for idx in shuffled[:n2c]:
            role[idx] = "Biomarker-Panel Calibration and Selection"
        for idx in shuffled[n2c:]:
            role[idx] = "Test"

    df[DATA_TYPE_COL] = pd.Series(role)
    role_order = {
        "Reference": 0,
        "Regression Fit Calibration 1": 1,
        "Biomarker-Panel Calibration and Selection": 2,
        "Test": 3,
    }
    df["_source_order"] = np.arange(len(df))
    df["_role_order"] = df[DATA_TYPE_COL].map(role_order)
    df["_cohort_order"] = (
        df[COHORT_COL].astype(str).str.strip().str.casefold()
        .map({str(k).casefold(): v for k, v in cohort_order.items()}).fillna(99)
    )
    df = df.sort_values(
        ["_role_order", "_cohort_order", "_source_order"], kind="stable"
    ).drop(columns=["_source_order", "_role_order", "_cohort_order"]).reset_index(drop=True)
    return df


def calculate_tda_features(reference_df, patient_row, panel):
    """Calculate the 0D and 1D persistence summaries used by the model."""
    # Panels are stored as tuples for hashing/caching, but pandas column
    # selection requires a list of column names rather than a tuple key.
    panel = list(panel)

    # Convert the reference matrix and append the patient's values as one row.
    reference_values = reference_df[panel].to_numpy(dtype=float, copy=False)
    patient_values = patient_row[panel].to_numpy(dtype=float, copy=False).reshape(1, -1)
    values = np.vstack((reference_values, patient_values))
    if not np.isfinite(values).all():
        raise ValueError(
            f"Non-finite selected biomarker value for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )
    if (values <= 0).any():
        raise ValueError(
            "Log transformation requires strictly positive values. "
            f"Check selected biomarkers for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )

    logged = np.log(values)
    correlation_matrix = np.corrcoef(logged, rowvar=False)
    dissimilarity_matrix = 1.0 - np.abs(correlation_matrix)
    if not np.isfinite(dissimilarity_matrix).all():
        raise ValueError(
            f"Invalid correlation-derived dissimilarity matrix for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )

    # Enforce numerical symmetry, a zero diagonal, and nonnegative entries.
    dissimilarity_matrix = (
        dissimilarity_matrix + dissimilarity_matrix.T
    ) / 2.0
    np.fill_diagonal(dissimilarity_matrix, 0.0)
    dissimilarity_matrix = np.clip(dissimilarity_matrix, 0.0, 1.0)

    # GUDHI names this input argument distance_matrix; the supplied values
    # represent the correlation-derived dissimilarity matrix.
    rips = gudhi.RipsComplex(distance_matrix=dissimilarity_matrix)
    simplex_tree = rips.create_simplex_tree(max_dimension=2)
    persistence = simplex_tree.persistence()
    finite_pairs = [
        (dimension, (birth, death))
        for dimension, (birth, death) in persistence
        if np.isfinite(death)
    ]
    one_dimensional_pairs = [
        pair for dimension, pair in finite_pairs if dimension == 1
    ]
    zero_dimensional_pairs = [
        pair for dimension, pair in finite_pairs if dimension == 0
    ]

    # Mean 1D birth/death coordinates are used to calculate Distance %(1).
    if one_dimensional_pairs:
        average_birth_1d = float(
            np.mean([birth for birth, _ in one_dimensional_pairs])
        )
        average_death_1d = float(
            np.mean([death for _, death in one_dimensional_pairs])
        )
    else:
        average_birth_1d = average_death_1d = 0.0

    # The 0D lifetime summaries are used as regression predictors/target.
    if zero_dimensional_pairs:
        deaths_0d = [death for _, death in zero_dimensional_pairs]
        lifetimes_0d = [death - birth for birth, death in zero_dimensional_pairs]
        min_lifetime_0d = float(min(lifetimes_0d))
        max_lifetime_0d = float(max(lifetimes_0d))
        average_death_0d = float(np.mean(deaths_0d))
        median_death_0d = float(np.median(deaths_0d))
        std_death_0d = float(np.std(deaths_0d))
    else:
        min_lifetime_0d = 0.0
        max_lifetime_0d = 0.0
        average_death_0d = 0.0
        median_death_0d = 0.0
        std_death_0d = 0.0

    return {
        "0-Dim Hole Min Life Range": min_lifetime_0d,
        "0-Dim Hole Max Life Range": max_lifetime_0d,
        "Median_Death_0D": median_death_0d,
        "Std_Death_0D": std_death_0d,
        "Avg_Death_0D": average_death_0d,
        "Average Birth (1D)": average_birth_1d,
        "Average Death (1D)": average_death_1d,
    }



def get_role_mask(df, role_name):
    return df[DATA_TYPE_COL].astype(str).str.strip().str.casefold().eq(
        role_name.casefold()
    )



def calculate_panel_features(df, reference_df, panel, include_test=False):
    """Extract TDA features for Calibration 1 and Calibration 2, plus Test if requested."""
    allowed = (
        get_role_mask(df, "Regression Fit Calibration 1")
        | get_role_mask(df, "Biomarker-Panel Calibration and Selection")
    )
    if include_test:
        allowed = allowed | get_role_mask(df, "Test")
    rows = []
    for idx, patient in df.loc[allowed].iterrows():
        record = {
            "Source_Row_Index": idx,
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
            COHORT_COL: patient[COHORT_COL],
            STAGE_COL: patient.get(STAGE_COL, np.nan),
            DATA_TYPE_COL: patient[DATA_TYPE_COL],
        }
        record.update(calculate_tda_features(reference_df, patient, panel))
        rows.append(record)
    return pd.DataFrame(rows)



def evaluate_panel(df, reference_df, panel, save_features=False):
    """
    Fit regression/reference quantities on Calibration 1 normals and optimize
    the candidate threshold pair on Calibration 2 under the pooled-normal
    specificity constraint.
    """
    features = calculate_panel_features(df, reference_df, panel, include_test=False)
    if features.empty:
        raise ValueError("No Calibration 1 / Calibration 2 patients were available.")

    role = features[DATA_TYPE_COL].astype(str).str.strip().str.casefold()
    cohort = features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal1_normals = features[
        role.eq("regression fit calibration 1")
        & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()
    cal2 = features[
        role.eq("biomarker-panel calibration and selection")
    ].copy()
    if len(cal1_normals) < 2:
        raise ValueError("At least two Calibration 1 normal patients are required.")
    if cal2.empty:
        raise ValueError("Calibration 2 is empty.")

    model = LinearRegression().fit(
        cal1_normals[REGRESSION_PREDICTORS],
        cal1_normals["Avg_Death_0D"],
    )
    features["Predicted Avg_Death_0D"] = model.predict(features[REGRESSION_PREDICTORS])
    features["Difference"] = features["Avg_Death_0D"] - features["Predicted Avg_Death_0D"]
    denom = features["Avg_Death_0D"].abs().replace(0, np.nan)
    features["Regression Difference Percentage"] = (
        features["Difference"].abs() / denom
    ) * 100.0

    ref_birth = float(cal1_normals["Average Birth (1D)"].mean())
    ref_death = float(cal1_normals["Average Death (1D)"].mean())
    features["Distance %(1)"] = np.sqrt(
        (features["Average Birth (1D)"] - ref_birth) ** 2
        + (features["Average Death (1D)"] - ref_death) ** 2
    ) * 100.0

    # IMPORTANT: cal1_normals was copied before the derived columns were added.
    # Refresh it now so it contains Regression Difference Percentage and Distance %(1).
    cal1_normals = features.loc[
        role.eq("regression fit calibration 1")
        & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()

    # ------------------------------------------------------------------
    # Reduced, vectorized threshold-pair search for the OR decision rule.
    #
    # Candidate thresholds are generated from observed score boundaries and
    # deterministically downsampled to at most MAX_THRESHOLD_CANDIDATES values
    # per score. Feasible pairs must produce pooled Calibration 1 + Calibration 2
    # normal specificity strictly greater than the configured target.
    # ------------------------------------------------------------------
    cal2_mask = role.eq("biomarker-panel calibration and selection")
    cal2_features = features.loc[cal2_mask].copy()
    cal2_cohort = cal2_features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal2_normals = cal2_features[cal2_cohort.eq(NORMAL_COHORT.casefold())]
    cal2_cancers = cal2_features[cal2_cohort.isin(CANCER_TYPES)]

    if cal2_normals.empty:
        raise ValueError("Calibration 2 contains no normal patients for specificity calibration.")
    if cal2_cancers.empty:
        raise ValueError("Calibration 2 contains no cancer patients for threshold selection.")

    cal1_reg_norm = cal1_normals["Regression Difference Percentage"].to_numpy(dtype=float)
    cal1_dist_norm = cal1_normals["Distance %(1)"].to_numpy(dtype=float)
    cal2_reg_norm = cal2_normals["Regression Difference Percentage"].to_numpy(dtype=float)
    cal2_dist_norm = cal2_normals["Distance %(1)"].to_numpy(dtype=float)
    pooled_reg_norm = np.concatenate([cal1_reg_norm, cal2_reg_norm])
    pooled_dist_norm = np.concatenate([cal1_dist_norm, cal2_dist_norm])

    cancer_reg = cal2_cancers["Regression Difference Percentage"].to_numpy(dtype=float)
    cancer_dist = cal2_cancers["Distance %(1)"].to_numpy(dtype=float)
    cancer_cohort_values = (
        cal2_cancers[COHORT_COL].astype(str).str.strip().str.casefold().to_numpy()
    )
    specificity_target = 1.0 - FPR

    def reduced_boundary_candidates(
        normal_scores, cancer_scores, max_candidates, min_value=None, max_value=None, name="threshold"
    ):
        """Build original score-boundary candidates, optionally restricting their range."""
        combined = np.concatenate([
            np.asarray(normal_scores, dtype=float),
            np.asarray(cancer_scores, dtype=float),
        ])
        combined = combined[np.isfinite(combined)]
        if combined.size == 0:
            raise ValueError("No finite scores available for threshold search.")
        if not isinstance(max_candidates, int) or max_candidates < 2:
            raise ValueError("MAX_THRESHOLD_CANDIDATES must be an integer >= 2.")

        # This is the original candidate-generation procedure.
        unique_values = np.unique(combined)
        boundaries = np.unique(np.concatenate([
            unique_values, np.nextafter(unique_values, np.inf)
        ]))

        # Apply bounds only when a complete min/max pair is provided.
        bounded = min_value is not None and max_value is not None
        if bounded:
            if not np.isfinite(min_value) or not np.isfinite(max_value):
                raise ValueError(f"{name} bounds must be finite numbers.")
            if min_value > max_value:
                raise ValueError(
                    f"{name}_MIN ({min_value}) must be <= {name}_MAX ({max_value})."
                )
            boundaries = boundaries[
                (boundaries >= float(min_value)) & (boundaries <= float(max_value))
            ]
            # Include exact bounds, even when no observed score lies there.
            boundaries = np.unique(np.concatenate([
                boundaries, np.array([float(min_value), float(max_value)])
            ]))

        if boundaries.size > max_candidates:
            # Keep the original deterministic downsampling behavior.
            indices = np.linspace(0, boundaries.size - 1, max_candidates, dtype=int)
            boundaries = boundaries[np.unique(indices)]
        return boundaries

    # Bounds are active only when all four are supplied. If all are None, this
    # exactly follows the original code's unbounded threshold candidate search.
    bound_values = (K1_MIN, K1_MAX, K2_MIN, K2_MAX)
    bounds_provided = [value is not None for value in bound_values]
    if any(bounds_provided) and not all(bounds_provided):
        raise ValueError(
            "Provide all four bounds (K1_MIN, K1_MAX, K2_MIN, K2_MAX), or leave all four as None."
        )
    USE_THRESHOLD_BOUNDS = all(bounds_provided)

    k1_candidates = reduced_boundary_candidates(
        pooled_reg_norm, cancer_reg, MAX_THRESHOLD_CANDIDATES,
        K1_MIN if USE_THRESHOLD_BOUNDS else None,
        K1_MAX if USE_THRESHOLD_BOUNDS else None, "K1"
    )
    k2_candidates = reduced_boundary_candidates(
        pooled_dist_norm, cancer_dist, MAX_THRESHOLD_CANDIDATES,
        K2_MIN if USE_THRESHOLD_BOUNDS else None,
        K2_MAX if USE_THRESHOLD_BOUNDS else None, "K2"
    )
    if USE_THRESHOLD_BOUNDS:
        print(
            f"Bounded threshold search: k1 in [{K1_MIN}, {K1_MAX}] "
            f"({len(k1_candidates)} candidates); "
            f"k2 in [{K2_MIN}, {K2_MAX}] ({len(k2_candidates)} candidates); "
            f"up to {len(k1_candidates) * len(k2_candidates):,} pairs."
        )
    # Evaluate each k1 against all k2 values in vectorized NumPy operations.
    n_normals = len(pooled_reg_norm)
    best_pair = None
    best_key = None
    best_macro_sensitivity = -1.0
    best_overall_sensitivity = -1.0
    best_specificity = -1.0
    best_fp = None
    best_tn = None

    cancer_reg_pos = cancer_reg[:, None] > k1_candidates[None, :]
    cancer_dist_pos = cancer_dist[:, None] > k2_candidates[None, :]
    active_cal2_cancers = [
        cancer for cancer in CANCER_TYPES
        if np.any(cancer_cohort_values == cancer.casefold())
    ]
    per_type_masks = {
        cancer: cancer_cohort_values == cancer.casefold()
        for cancer in active_cal2_cancers
    }

    for i, k1 in enumerate(k1_candidates):
        normal_positive = (
            (pooled_reg_norm > k1)[:, None]
            | (pooled_dist_norm[:, None] > k2_candidates[None, :])
        )
        fp_counts = normal_positive.sum(axis=0)
        specificities = 1.0 - fp_counts / n_normals
        feasible = np.flatnonzero(specificities > specificity_target + 1e-12)
        if feasible.size == 0:
            continue

        cancer_positive_by_k2 = cancer_reg_pos[:, i, None] | cancer_dist_pos[:, feasible]
        overall_sens = cancer_positive_by_k2.mean(axis=0)
        type_sens = []
        for cancer in active_cal2_cancers:
            mask = per_type_masks[cancer]
            type_sens.append(cancer_positive_by_k2[mask].mean(axis=0))
        if not type_sens:
            continue
        macro_sens = np.mean(np.vstack(type_sens), axis=0)

        # Lexicographic objective matches original intent: macro sensitivity,
        # overall sensitivity, specificity, then larger thresholds.
        for j_local, j in enumerate(feasible):
            candidate_key = (
                float(macro_sens[j_local]),
                float(overall_sens[j_local]),
                float(specificities[j]),
                float(k1),
                float(k2_candidates[j]),
            )
            if best_key is None or candidate_key > best_key:
                best_key = candidate_key
                best_pair = (float(k1), float(k2_candidates[j]))
                best_macro_sensitivity = candidate_key[0]
                best_overall_sensitivity = candidate_key[1]
                best_specificity = candidate_key[2]
                best_fp = int(fp_counts[j])
                best_tn = int(n_normals - best_fp)

    if best_pair is None:
        if USE_THRESHOLD_BOUNDS:
            raise RuntimeError(
                "No k1/k2 pair within the user-defined bounds satisfied the pooled "
                f"Calibration 1 + Calibration 2 specificity requirement (> {specificity_target:.2%}). "
                f"Current bounds: k1=[{K1_MIN}, {K1_MAX}], k2=[{K2_MIN}, {K2_MAX}]. "
                "Adjust the bounds or review the specificity target."
            )
        # Preserve the original unbounded fallback: thresholds just above the
        # maximum pooled normal scores guarantee zero false positives.
        finite_reg = pooled_reg_norm[np.isfinite(pooled_reg_norm)]
        finite_dist = pooled_dist_norm[np.isfinite(pooled_dist_norm)]
        if finite_reg.size == 0 or finite_dist.size == 0:
            raise ValueError("Pooled calibration normal scores contain no finite values.")
        k1_fallback = float(np.nextafter(np.max(finite_reg), np.inf))
        k2_fallback = float(np.nextafter(np.max(finite_dist), np.inf))
        best_pair = (k1_fallback, k2_fallback)
        fallback_positive = (cancer_reg > k1_fallback) | (cancer_dist > k2_fallback)
        sens_by_type = [
            float(fallback_positive[per_type_masks[cancer]].mean())
            for cancer in CANCER_TYPES if per_type_masks[cancer].any()
        ]
        best_macro_sensitivity = (
            float(np.mean(sens_by_type)) if len(sens_by_type) == len(CANCER_TYPES) else float("nan")
        )
        best_overall_sensitivity = float(fallback_positive.mean()) if fallback_positive.size else float("nan")
        best_fp, best_tn = 0, n_normals
        best_specificity = 1.0
    best_threshold, distance_threshold = best_pair

    # Report each partition's specificity descriptively, but only the pooled
    # specificity is a threshold-selection constraint.
    cal1_positive_final = (
        (cal1_normals["Regression Difference Percentage"] > best_threshold)
        | (cal1_normals["Distance %(1)"] > distance_threshold)
    )
    best_cal1_specificity = float((~cal1_positive_final).mean())

    features["Result(Cancer)"] = np.where(
        (features["Regression Difference Percentage"] > best_threshold)
        | (features["Distance %(1)"] > distance_threshold),
        "Positive", "Negative"
    )
    cal1_normal_positive_final = (
        (cal1_normals["Regression Difference Percentage"] > best_threshold)
        | (cal1_normals["Distance %(1)"] > distance_threshold)
    )
    cal2_normal_positive = (
        (cal2_normals["Regression Difference Percentage"] > best_threshold)
        | (cal2_normals["Distance %(1)"] > distance_threshold)
    )
    combined_normal_positive_final = np.concatenate([
        cal1_normal_positive_final.to_numpy(dtype=bool),
        cal2_normal_positive.to_numpy(dtype=bool),
    ])
    tn = int((~combined_normal_positive_final).sum())
    fp = int(combined_normal_positive_final.sum())
    specificity_combined = (
        tn / (tn + fp) if (tn + fp) else float("nan")
    )
    specificity_cal2 = (
        float((~cal2_normal_positive).mean()) if len(cal2_normal_positive) else float("nan")
    )
    if not specificity_combined > specificity_target + 1e-12:
        raise RuntimeError(
            "Selected thresholds failed the pooled Calibration 1 + Calibration 2 "
            "specificity constraint."
        )

    cancer_results = {}
    sensitivities = []
    all_predictions = (
        (cal2_cancers["Regression Difference Percentage"] > best_threshold)
        | (cal2_cancers["Distance %(1)"] > distance_threshold)
    )
    for cancer in CANCER_TYPES:
        subset = cal2_cancers[
            cal2_cancers[COHORT_COL].astype(str).str.strip().str.casefold().eq(cancer)
        ]
        pred = (
            (subset["Regression Difference Percentage"] > best_threshold)
            | (subset["Distance %(1)"] > distance_threshold)
        )
        sens = float(pred.mean()) if len(subset) else float("nan")
        if len(subset):
            sensitivities.append(sens)
        cancer_results[cancer] = {
            "n": int(len(subset)),
            "sensitivity": sens,
            "true_positives": int(pred.sum()),
        }
    macro_sensitivity = float(np.mean(sensitivities)) if sensitivities else float("nan")
    overall_sensitivity = float(all_predictions.mean()) if len(cal2_cancers) else float("nan")



    return {
        "n_biomarkers": len(panel),
        "panel": list(panel),
        "regression_threshold": float(best_threshold),
        "distance_threshold": float(distance_threshold),
        "threshold_bounds_enabled": bool(USE_THRESHOLD_BOUNDS),
        "k1_min_search_bound": float(K1_MIN) if USE_THRESHOLD_BOUNDS else None,
        "k1_max_search_bound": float(K1_MAX) if USE_THRESHOLD_BOUNDS else None,
        "k2_min_search_bound": float(K2_MIN) if USE_THRESHOLD_BOUNDS else None,
        "k2_max_search_bound": float(K2_MAX) if USE_THRESHOLD_BOUNDS else None,
        "k1_candidate_count": int(len(k1_candidates)),
        "k2_candidate_count": int(len(k2_candidates)),
        "threshold_pairs_considered_max": int(len(k1_candidates) * len(k2_candidates)),
        "calibration1_specificity": float(best_cal1_specificity),
        "calibration2_specificity": float(specificity_cal2),
        "specificity": float(specificity_combined),
        "combined_calibration_specificity": float(specificity_combined),
        "false_positives": fp,
        "true_negatives": tn,
        "macro_sensitivity": macro_sensitivity,
        "overall_cancer_sensitivity": overall_sensitivity,
        "cancer_results": cancer_results,
        "regression_fit_normal_n": int(len(cal1_normals)),
        "calibration2_normal_n": int(len(cal2_normals)),
        "calibration_cancer_n": int(len(cal2_cancers)),
        "meets_sensitivity_target": bool(
            np.isfinite(macro_sensitivity) and macro_sensitivity > SENSITIVITY_TARGET
        ),
        "threshold_rule": (
            "Positive iff Regression Difference Percentage > regression_threshold "
            "OR Distance %(1) > distance_threshold"
        ),
    }, features


def canonical_panel(panel):
    return tuple(sorted(panel, key=BIOMARKERS.index))


def panel_key(panel):
    return tuple(panel)


def random_panel(rng):
    return canonical_panel(rng.sample(BIOMARKERS, PANEL_SIZE))


def balanced_diverse_panels(rng, count, used, marker_pool=None, prior_counts=None):
    """Generate unique panels while balancing marker appearances and reducing overlap.

    In Stage 1, marker_pool is all 42 biomarkers and prior_counts starts at zero,
    so all markers have approximately equal expected representation. In Stage 3,
    the pool is restricted to recurring markers found among the strongest feasible
    panels, and representation is balanced within that pool.
    """
    pool = list(marker_pool or BIOMARKERS)
    if len(pool) < PANEL_SIZE:
        pool = list(BIOMARKERS)
    counts = {b: int((prior_counts or {}).get(b, 0)) for b in pool}
    selected = []
    selected_set = set(used)
    max_attempts = max(100000, count * 1000)

    while len(selected) < count:
        candidates = set()
        attempts = 0
        while len(candidates) < CANDIDATE_POOL_SIZE and attempts < max_attempts:
            attempts += 1
            candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            if candidate not in selected_set:
                candidates.add(candidate)
        if not candidates:
            # This is only expected if the candidate space is nearly exhausted.
            candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            if candidate in selected_set:
                # Fall back to the full marker universe to maintain the exact budget.
                pool = list(BIOMARKERS)
                candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
                while candidate in selected_set:
                    candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            candidates.add(candidate)

        def balance_diversity_key(candidate):
            projected = dict(counts)
            for b in candidate:
                if b in projected:
                    projected[b] += 1
            values = list(projected.values())
            # First balance total appearances; then minimize overlap with earlier
            # panels from this stage; randomized final tie-break avoids ordering bias.
            imbalance = float(np.std(values)) if values else 0.0
            max_count = max(values) if values else 0
            if selected:
                overlaps = [len(set(candidate) & set(old)) for old in selected]
                mean_overlap = float(np.mean(overlaps))
                max_overlap = max(overlaps)
            else:
                mean_overlap = max_overlap = 0.0
            return (imbalance, max_count, mean_overlap, max_overlap, rng.random())

        # IMPORTANT FOR REPRODUCIBILITY:
        # `candidates` is a set, whose iteration order can vary between Python
        # processes. The key function consumes seeded RNG values for tie-breaking;
        # therefore, iterate in a stable sorted order before calling it.
        chosen = min(sorted(candidates), key=balance_diversity_key)
        selected.append(chosen)
        selected_set.add(chosen)
        # Keep the shared cross-stage registry synchronized. Without this line,
        # panels generated in this stage are not counted in `used`, even though
        # they are appended to all_rows later.
        used.add(chosen)
        for b in chosen:
            if b in counts:
                counts[b] += 1

    return selected


def top_feasible_rows(panel_rows):
    feasible = [item for item in panel_rows if item[1]["specificity"] > (1.0 - FPR) + 1e-12]
    feasible.sort(key=lambda item: item[1]["macro_sensitivity"], reverse=True)
    if not feasible:
        return []
    n_top = max(1, int(math.ceil(len(feasible) * TOP_PANEL_FRACTION)))
    return feasible[:n_top]


def biomarker_weights(panel_rows):
    """Estimate recurring-marker weights from the strongest feasible panels."""
    top = top_feasible_rows(panel_rows)
    counts = {b: 1.0 for b in BIOMARKERS}  # small baseline keeps exploration possible
    for panel, _summary in top:
        for biomarker in panel:
            counts[biomarker] += 1.0
    return counts


def weighted_panel(rng, weights):
    """Sample a unique-size panel without replacement, weighted by marker frequency."""
    available = list(BIOMARKERS)
    chosen = []
    for _ in range(PANEL_SIZE):
        total = sum(max(0.001, weights[b]) for b in available)
        pick = rng.random() * total
        cumulative = 0.0
        selected = available[-1]
        for biomarker in available:
            cumulative += max(0.001, weights[biomarker])
            if cumulative >= pick:
                selected = biomarker
                break
        chosen.append(selected)
        available.remove(selected)
    return canonical_panel(chosen)


def make_unique_weighted_panels(rng, count, used, weights):
    panels = []
    attempts = 0
    max_attempts = max(10000, count * 200)
    while len(panels) < count and attempts < max_attempts:
        attempts += 1
        candidate = weighted_panel(rng, weights)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    while len(panels) < count:
        candidate = random_panel(rng)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    return panels


def make_neighborhood_panels(rng, count, used, panel_rows):
    """Generate swaps around the best feasible panels, with random unique fill."""
    feasible = [item for item in panel_rows
                if item[1]["specificity"] > (1.0 - FPR) + 1e-12
                and np.isfinite(item[1]["macro_sensitivity"])]
    ranked_source = feasible if feasible else panel_rows
    ranked = sorted(ranked_source,
                    key=lambda item: item[1]["macro_sensitivity"]
                    if np.isfinite(item[1]["macro_sensitivity"]) else -1.0,
                    reverse=True)
    panels = []
    for panel, _summary in ranked:
        candidates = local_neighbors(panel)
        rng.shuffle(candidates)
        for candidate in candidates:
            if candidate not in used:
                used.add(candidate)
                panels.append(candidate)
                if len(panels) >= count:
                    return panels
    while len(panels) < count:
        candidate = random_panel(rng)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    return panels

def score_panel(df, reference_df, panel, cache):
    key = panel_key(panel)
    if key not in cache:
        summary, _ = evaluate_panel(df, reference_df, panel, save_features=False)
        cache[key] = summary
    return cache[key]


def rank_key(summary):
    # Specificity the configured specificity target is enforced as a hard feasibility constraint in
    # evaluate_panel. Among feasible panels, rank ONLY by macro-sensitivity.
    return summary["macro_sensitivity"]


def local_neighbors(panel):
    panel_set = set(panel)
    not_in_panel = [b for b in BIOMARKERS if b not in panel_set]
    neighbors = set()
    for removed in panel:
        for added in not_in_panel:
            candidate = canonical_panel((panel_set - {removed}) | {added})
            if len(candidate) == PANEL_SIZE:
                neighbors.add(candidate)
    return sorted(
        neighbors,
        key=lambda p: tuple(BIOMARKERS.index(x) for x in p),
    )



def evaluate_selected_panel_on_test(df, reference_df, panel, thresholds):
    """Evaluate the selected panel once on the untouched Test partition."""
    test_df = df[get_role_mask(df, "Test")].copy()
    if test_df.empty:
        print("No Test patients were available.")
        return None

    # Fit the regression model and reference coordinates from Calibration 1 normals,
    # then apply the already frozen thresholds to Test patients.
    cal1_features = calculate_panel_features(df, reference_df, panel, include_test=False)
    role = cal1_features[DATA_TYPE_COL].astype(str).str.strip().str.casefold()
    cohort = cal1_features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal1_normals = cal1_features[
        role.eq("regression fit calibration 1") & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()
    model = LinearRegression().fit(
        cal1_normals[REGRESSION_PREDICTORS], cal1_normals["Avg_Death_0D"]
    )
    test_features = []
    for idx, patient in test_df.iterrows():
        rec = {
            "Source_Row_Index": idx,
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
            COHORT_COL: patient[COHORT_COL],
            STAGE_COL: patient.get(STAGE_COL, np.nan),
            DATA_TYPE_COL: patient[DATA_TYPE_COL],
        }
        rec.update(calculate_tda_features(reference_df, patient, panel))
        test_features.append(rec)
    test_features = pd.DataFrame(test_features)
    test_features["Predicted Avg_Death_0D"] = model.predict(test_features[REGRESSION_PREDICTORS])
    test_features["Difference"] = (
        test_features["Avg_Death_0D"] - test_features["Predicted Avg_Death_0D"]
    )
    denom = test_features["Avg_Death_0D"].abs().replace(0, np.nan)
    test_features["Regression Difference Percentage"] = (
        test_features["Difference"].abs() / denom
    ) * 100.0
    ref_birth = float(cal1_normals["Average Birth (1D)"].mean())
    ref_death = float(cal1_normals["Average Death (1D)"].mean())
    test_features["Distance %(1)"] = np.sqrt(
        (test_features["Average Birth (1D)"] - ref_birth) ** 2
        + (test_features["Average Death (1D)"] - ref_death) ** 2
    ) * 100.0
    reg_t = thresholds["regression_threshold"]
    dist_t = thresholds["distance_threshold"]
    positive = (
        (test_features["Regression Difference Percentage"] > reg_t)
        | (test_features["Distance %(1)"] > dist_t)
    )
    test_features["Result(Cancer)"] = np.where(positive, "Positive", "Negative")
    cohort_values = test_features[COHORT_COL].astype(str).str.strip().str.casefold()
    normal_mask = cohort_values.eq(NORMAL_COHORT.casefold())
    normal_pred = positive[normal_mask]
    specificity = float((~normal_pred).mean()) if normal_mask.any() else float("nan")
    sens_values = [
        float(positive[cohort_values.eq(cancer)].mean())
        for cancer in CANCER_TYPES if cohort_values.eq(cancer).any()
    ]
    macro_sens = float(np.mean(sens_values)) if sens_values else float("nan")
    print(f"Test patients: {len(test_features)} | Normal specificity: {specificity:.4%}")
    print(f"Test macro-sensitivity: {macro_sens:.4%}")
    return {"specificity": specificity, "macro_sensitivity": macro_sens,
            "n_test": int(len(test_features))}


def Optimized_Biomarker_Selection(n_reference_normal):
    """Run the staged, constrained TDA biomarker-selection procedure.

    Dataset-specific values are configured in the USER-DEFINED SETTINGS section.
    Returns the selected biomarker panel and its calibration summary.
    """
    validate_split_fractions()
    bound_values = (K1_MIN, K1_MAX, K2_MIN, K2_MAX)
    bounds_provided = [value is not None for value in bound_values]
    if any(bounds_provided) and not all(bounds_provided):
        raise ValueError(
            "Provide all four bounds (K1_MIN, K1_MAX, K2_MIN, K2_MAX), or leave all four as None."
        )
    if all(bounds_provided):
        for name, value in zip(("K1_MIN", "K1_MAX", "K2_MIN", "K2_MAX"), bound_values):
            if not isinstance(value, (int, float)) or not np.isfinite(value):
                raise ValueError(f"{name} must be a finite number or None when bounds are disabled.")
        if K1_MIN > K1_MAX:
            raise ValueError(f"K1_MIN ({K1_MIN}) must be <= K1_MAX ({K1_MAX}).")
        if K2_MIN > K2_MAX:
            raise ValueError(f"K2_MIN ({K2_MIN}) must be <= K2_MAX ({K2_MAX}).")
    if not isinstance(MAX_THRESHOLD_CANDIDATES, int) or MAX_THRESHOLD_CANDIDATES < 2:
        raise ValueError("MAX_THRESHOLD_CANDIDATES must be an integer >= 2.")
    # Do not validate CANCER_TYPES here: it is populated automatically from the
    # workbook Cohort column inside create_split_dataframe(), which runs below.
    # cohort_order is only a preferred sorting map and must not control which
    # cancer cohorts are included in analysis.
    if len(BIOMARKERS) != len(set(BIOMARKERS)):
        raise ValueError("BIOMARKERS must contain unique names.")
    if PANEL_SIZE < 1 or PANEL_SIZE > len(BIOMARKERS):
        raise ValueError("PANEL_SIZE must be between 1 and the number of BIOMARKERS.")
    if sum(PHASE_BUDGETS.values()) != TOTAL_PANEL_BUDGET:
        raise ValueError("The phase budgets must sum exactly to TOTAL_PANEL_BUDGET.")

    df = create_split_dataframe(clinical_data_path, sheet_name, n_reference_normal)
    reference_df = df[
        get_role_mask(df, "Reference")
        & df[COHORT_COL].astype(str).str.strip().str.casefold().eq(NORMAL_COHORT.casefold())
    ].copy()
    if len(reference_df) != n_reference_normal:
        raise ValueError(
            f"Expected {n_reference_normal} reference normals; found {len(reference_df)}."
        )

    rng = random.Random(RANDOM_SEED)
    cache = {}
    used = set()
    all_rows = []
    stage_by_panel = {}

    def evaluate_phase(label, panels, budget):
        if len(panels) != budget:
            raise RuntimeError(f"{label}: expected {budget} panels, got {len(panels)}")
        phase_rows = []
        for i, panel in enumerate(panels, start=1):
            summary = score_panel(df, reference_df, panel, cache)
            phase_rows.append((panel, summary))
            all_rows.append((panel, summary))
            stage_by_panel[panel] = label

            # Compact progress only at the user-configured interval.
            completed = len(all_rows)
            if completed % PROGRESS_INTERVAL == 0 or completed == TOTAL_PANEL_BUDGET:
                feasible_so_far = [
                    item for item in all_rows
                    if item[1]["specificity"] > (1.0 - FPR) + 1e-12
                    and np.isfinite(item[1]["macro_sensitivity"])
                ]
                if feasible_so_far:
                    progress_best_panel, progress_best = max(
                        feasible_so_far,
                        key=lambda item: (
                            item[1]["macro_sensitivity"],
                            item[1]["overall_cancer_sensitivity"],
                            item[1]["specificity"],
                        ),
                    )
                    print(
                        f"Search {completed}/{TOTAL_PANEL_BUDGET} | Stage {label} | "
                        f"best panel so far: {list(progress_best_panel)} | "
                        f"calibration specificity: {progress_best['specificity']:.4%} | "
                        f"calibration macro-sensitivity: {progress_best['macro_sensitivity']:.4%}"
                    )
                else:
                    print(
                        f"Search {completed}/{TOTAL_PANEL_BUDGET} | Stage {label} | "
                        "no panel has met the calibration specificity constraint yet"
                    )
        return phase_rows

    print(f"\nReference normals: {n_reference_normal} | Searching {TOTAL_PANEL_BUDGET} unique panels")

    # STAGE 1: all 42 biomarkers are treated equally at the outset. A greedy
    # balancing criterion keeps their appearance counts as even as possible,
    # while minimizing overlap between panels. No outcome-guided marker weights.
    stage1_panels = balanced_diverse_panels(
        rng, PHASE_BUDGETS["balanced_diverse_exploration"], used,
        marker_pool=BIOMARKERS, prior_counts={b: 0 for b in BIOMARKERS}
    )
    used.update(stage1_panels)
    evaluate_phase("1/4 balanced diverse exploration", stage1_panels,
                   PHASE_BUDGETS["balanced_diverse_exploration"])

    # STAGE 2: rank only feasible panels by macro-sensitivity, then estimate marker
    # recurrence from the top feasible fraction and use it to guide new combinations.
    weights = biomarker_weights(all_rows)
    stage2_panels = make_unique_weighted_panels(
        rng, PHASE_BUDGETS["sensitivity_guided_search"], used, weights
    )
    evaluate_phase("2/4 sensitivity-guided search", stage2_panels,
                   PHASE_BUDGETS["sensitivity_guided_search"])

    # STAGE 3: identify recurring biomarkers in the top feasible panels from Stages
    # 1–2, then repeat balanced/low-overlap exploration within the recurrence-informed
    # pool. This keeps the marker frequencies balanced within that selected pool.
    top_before_stage3 = top_feasible_rows(all_rows)
    recurrence_counts = {b: 0 for b in BIOMARKERS}
    for panel, _summary in top_before_stage3:
        for b in panel:
            recurrence_counts[b] += 1
    recurring_pool = [b for b, _count in sorted(
        recurrence_counts.items(), key=lambda item: (-item[1], BIOMARKERS.index(item[0]))
    )[:RECURRENCE_POOL_SIZE]]
    if len(recurring_pool) < PANEL_SIZE:
        recurring_pool = list(BIOMARKERS)
    stage3_panels = balanced_diverse_panels(
        rng, PHASE_BUDGETS["recurrence_informed_balanced_search"], used,
        marker_pool=recurring_pool, prior_counts={b: 0 for b in recurring_pool}
    )
    evaluate_phase("3/4 recurrence-informed balanced search", stage3_panels,
                   PHASE_BUDGETS["recurrence_informed_balanced_search"])

    # STAGE 4: one-biomarker swaps around the strongest panels from all previous
    # stages, with random unique panels as a fallback. The final selection below
    # still compares every feasible panel from all four stages.
    stage4_panels = make_neighborhood_panels(
        rng, PHASE_BUDGETS["best_panel_neighborhoods"], used, all_rows
    )
    evaluate_phase("4/4 best-panel neighborhood search", stage4_panels,
                   PHASE_BUDGETS["best_panel_neighborhoods"])

    if len(used) != TOTAL_PANEL_BUDGET or len(all_rows) != TOTAL_PANEL_BUDGET:
        raise RuntimeError(
            f"Search budget error: evaluated {len(all_rows)} panels, expected {TOTAL_PANEL_BUDGET}."
        )
    if len(set(panel for panel, _ in all_rows)) != TOTAL_PANEL_BUDGET:
        raise RuntimeError("Duplicate panels were evaluated; every panel must be unique.")

    # IMPORTANT: the winning panel is selected across all four stages, not only Stage 4.
    specificity_feasible_rows = [
        item for item in all_rows
        if item[1]["specificity"] > (1.0 - FPR) + 1e-12
        and np.isfinite(item[1]["macro_sensitivity"])
    ]
    finite_rows = [
        item for item in all_rows
        if np.isfinite(item[1]["macro_sensitivity"])
    ]
    if not finite_rows:
        raise RuntimeError("All evaluated panels had undefined macro-sensitivity.")
    # Detection-based panel selection:
    # 1. Keep only panels whose pooled Calibration 1 + Calibration 2 normal specificity is strictly above 1 - FPR.
    # 2. Among those feasible panels, select the panel with the highest
    #    macro-sensitivity across the cancer cohorts.
    # 3. Do not require macro-sensitivity itself to exceed 99%; it is the
    #    optimization objective, while specificity is the hard constraint.
    if specificity_feasible_rows:
        ranked_rows = sorted(
            specificity_feasible_rows,
            key=lambda item: (
                item[1]["macro_sensitivity"],
                item[1]["overall_cancer_sensitivity"],
                item[1]["specificity"],
            ),
            reverse=True,
        )
    else:
        # This is a diagnostic fallback only. The unbounded candidate search
        # includes score boundaries above the pooled-normal maxima, which should
        # normally permit a threshold pair with 100% pooled-normal specificity.
        ranked_rows = sorted(
            finite_rows,
            key=lambda item: (
                item[1]["macro_sensitivity"],
                item[1]["overall_cancer_sensitivity"],
                item[1]["specificity"],
            ),
            reverse=True,
        )
        print(
            "\nWARNING: No panel met the strict pooled Calibration 1 + Calibration 2 "
            "specificity constraint. The selected panel is diagnostic only."
        )
    best_panel, _ = ranked_rows[0]

    # Reuse the already-computed summary for the winning panel; no duplicate calibration evaluation.
    best_summary = ranked_rows[0][1]
    print("\nBEST BIOMARKER PANEL — SELECTED ACROSS ALL FOUR STAGES")
    print(f"Reference normals: {n_reference_normal}")
    print(f"Winning stage: {stage_by_panel.get(best_panel, 'unknown')}")
    print(f"Panel: {list(best_panel)!r}")
    print(f"Calibration specificity (pooled Calibration 1 + Calibration 2 normals): {best_summary['specificity']:.4%}")
    print(f"Calibration 1 normal specificity: {best_summary['calibration1_specificity']:.4%}")
    print(f"Calibration 2 normal specificity: {best_summary['calibration2_specificity']:.4%}")
    print(f"Calibration macro-sensitivity: {best_summary['macro_sensitivity']:.4%}")
    print(f"Regression Difference Percentage threshold (k1): {best_summary['regression_threshold']:.8g}")
    print(f"Distance %(1) threshold (k2): {best_summary['distance_threshold']:.8g}")
    print(f"Meets pooled calibration specificity target: {best_summary['specificity'] > (1.0 - FPR) + 1e-12}")
    print("FINAL UNTOUCHED TEST-SET PERFORMANCE (only after panel selection)")
    test_summary = evaluate_selected_panel_on_test(df, reference_df, best_panel, best_summary)
    return {
        "reference_normals": n_reference_normal,
        "panel": list(best_panel),
        "winning_stage": stage_by_panel.get(best_panel, 'unknown'),
        "calibration_specificity": best_summary["specificity"],
        "calibration1_specificity": best_summary["calibration1_specificity"],
        "calibration2_specificity": best_summary["calibration2_specificity"],
        "calibration_macro_sensitivity": best_summary["macro_sensitivity"],
        "test_specificity": test_summary["specificity"] if test_summary else float("nan"),
        "test_macro_sensitivity": test_summary["macro_sensitivity"] if test_summary else float("nan"),
    }




if __name__ == "__main__":
    if not REFERENCE_SAMPLE_SIZES or len(set(REFERENCE_SAMPLE_SIZES)) != len(REFERENCE_SAMPLE_SIZES):
        raise ValueError("REFERENCE_SAMPLE_SIZES must be a non-empty list of unique positive integers.")
    if any(not isinstance(n, int) or n < 2 for n in REFERENCE_SAMPLE_SIZES):
        raise ValueError("Every reference sample size must be an integer >= 2.")

    all_run_results = []
    for run_number, reference_size in enumerate(REFERENCE_SAMPLE_SIZES, start=1):
        print("\n" + "#" * 78)
        print(f"RUN {run_number}/{len(REFERENCE_SAMPLE_SIZES)} — reference sample size = {reference_size}")
        print("#" * 78)
        result = Optimized_Biomarker_Selection(reference_size)
        all_run_results.append(result)

    print("\n" + "=" * 78)
    print("SUMMARY ACROSS REFERENCE SAMPLE SIZES")
    print("=" * 78)
    for result in all_run_results:
        print(
            f"Reference n={result['reference_normals']} | "
            f"Calibration specificity={result['calibration_specificity']:.4%} | "
            f"Calibration macro-sensitivity={result['calibration_macro_sensitivity']:.4%} | "
            f"Test specificity={result['test_specificity']:.4%} | "
            f"Test macro-sensitivity={result['test_macro_sensitivity']:.4%} | "
            f"Panel={result['panel']}"
        )
```

```python
"""
Four-stage constrained random search for TDA-based biomarker-panel selection.

Workflow
--------
1. Loads the specified clinical workbook and validates the required cohorts and
   biomarker columns.
2. Creates five reproducible split workbooks. In each repetition, 59 normal
   patients are randomly selected as the Reference group. The remaining normal
   patients are divided among Regression Fit Calibration 1, Biomarker-Panel
   Calibration and Selection, and Test using the configured fractions.
3. Cancer-patient assignments to Calibration 2 and Test are generated once and
   held fixed across all five repetitions. Cancer patients are not assigned to
   Regression Fit Calibration 1.
4. In each repetition, evaluates exactly TOTAL_PANEL_BUDGET distinct panels of
   PANEL_SIZE biomarkers using four search stages:
   - balanced, low-overlap exploration across the full biomarker set;
   - sensitivity-guided, recurrence-weighted sampling;
   - balanced exploration within a recurrence-informed biomarker pool; and
   - one-biomarker neighborhood swaps around high-performing panels.
5. For each panel, fits the regression model and calculates the Distance %(1)
   reference coordinates using Regression Fit Calibration 1 normal patients.
   Candidate threshold pairs are evaluated on Calibration 2 with an OR rule.
   Feasibility requires pooled Regression Fit Calibration 1 and Calibration 2
   normal specificity to be strictly greater than 1 - FPR. Feasible panels are
   ranked primarily by Calibration 2 macro-sensitivity across cancer cohorts.
6. Selects the winning panel from all four search stages and evaluates it once
   on the untouched Test partition using the selected panel and frozen thresholds.
7. Saves the five split workbooks and a combined summary of split-level results.

Configuration
-------------
Update the input/output paths, biomarker names, cohort labels, and column names
to match the workbook. Cancer cohorts are validated against COHORT_ORDER.
The log transformation requires selected biomarker values to be finite and
strictly positive in the Reference and evaluated patient rows.
"""

from pathlib import Path
import math
import random

import gudhi
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression


# =============================================================================
# USER-DEFINED SETTINGS
# Change these values when applying the function to another dataset.
# Algorithmic tuning parameters below this section retain their defaults.
# =============================================================================

# Input/output files
clinical_data_path = Path(
    r"D:\Cancer Detection folder\Biomarker Selection\clinical cancer validation data.xlsx"
)
sheet_name = "Normal and Cancer"
output_folder_path = Path(r"D:\Cancer Detection folder\Biomarker Selection")

# Dataset-specific biomarker universe (names must match workbook column names).
BIOMARKERS = [
    "Tenascin C", "sAXL", "sE-selectin", "sHGFR/c-Met", "sHer2", "sHer3",
    "sIL-6Ra", "sNeuropilin-1", "sPECAM-1", "sVEGFR1", "sVEGFR3",
    "sc-kit/SCFR", "CA-125", "CA 15-3", "CA 19-9", "CEA", "FGF2", "HE4",
    "HGF", "IL-6", "IL-8", "Leptin", "MIF", "OPN", "Prolactin", "SCF",
    "TNFa", "TRAIL", "Total PSA", "VEGF", "b-HCG", "sFAS", "Cathepsin D",
    "FAP alpha", "Ferritin", "Galectin-3", "IGFBP3", "MIA", "MPO", "SHBG",
    "TIMP1", "TIMP2"
]

# Cohort names and their display/sort order.
# Keep the normal cohort in this mapping; all other keys are treated as cancer cohorts.
NORMAL_COHORT = "normal"
cohort_order = {"normal": 0, "bladder": 1, "ovarian": 2, "pancreatic": 3}

# Five reproducible random splits; split generation uses a separate RNG seed.
N_REPETITIONS = 5
SPLIT_RANDOM_SEED = 20261022
COHORT_ORDER = {
    "normal": 0,
    "bladder": 1,
    "ovarian": 2,
    "pancreatic": 3,
}
SOURCE_ROW_COL = "Original_Source_Row_Zero_Based"
ROLE_REFERENCE = "Reference"
ROLE_CAL1 = "Regression Fit Calibration 1"
ROLE_CAL2 = "Biomarker-Panel Calibration and Selection"
ROLE_TEST = "Test"
ROLE_ORDER = {
    ROLE_REFERENCE: 0,
    ROLE_CAL1: 1,
    ROLE_CAL2: 2,
    ROLE_TEST: 3,
}
SPLIT_OUTPUT_DIR = output_folder_path / "59 Random Splits"
COMBINED_RESULTS_PATH = output_folder_path / "RandomSplit_BiomarkerSelection_Results.xlsx"
CANCER_TYPES = [name.casefold() for name in cohort_order if name.casefold() != NORMAL_COHORT.casefold()]

# Dataset-specific column names.
COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
STAGE_COL = "AJCC Stage"
DATA_TYPE_COL = "Data Type"

# Data split and model constraints.
# Edit the two fractions below whenever needed. They are fractions, not whole percentages:
#   0.20 = 20%, 0.30 = 30%, 0.50 = 50%.
# For NORMAL patients only, N_REFERENCE_NORMAL patients are randomly sampled as Reference in each repetition;
# the configured Calibration 1 and Calibration 2 fractions are applied to the remaining
# normal patients, with Test receiving the remainder.
# CANCER patients are NEVER assigned to Calibration 1: CALIBRATION2_FRACTION of each
# cancer cohort is assigned to Biomarker-Panel Calibration and Selection, and the rest
# is Test. Thus Calibration 1 contains only healthy/normal patients.
N_REFERENCE_NORMAL = 59
CALIBRATION1_FRACTION = 0.15
CALIBRATION2_FRACTION = 0.50
FPR = 0.01  # Require pooled Calibration 1 + 2 normal specificity strictly > 99.00%
SENSITIVITY_TARGET = 0.99  # Diagnostic target only; specificity is the hard constraint
PANEL_SIZE = 8
TOTAL_PANEL_BUDGET = 3000
RANDOM_SEED = 20261017

# Maximum number of candidate thresholds to test for EACH detection score:
# - One set of candidates is generated for Regression Difference Percentage (k1).
# - One set is generated for Distance %(1) (k2).
# - The threshold search evaluates up to MAX_THRESHOLD_CANDIDATES x
#   MAX_THRESHOLD_CANDIDATES candidate pairs for the OR decision rule.
# - Lower values generally run faster but explore fewer threshold combinations;
#   higher values search more combinations and may take longer.
MAX_THRESHOLD_CANDIDATES = 80

# OPTIONAL USER-DEFINED THRESHOLD SEARCH BOUNDS (inclusive)
# Leave ALL FOUR values as None to use the original unbounded candidate search.
# To restrict the search, provide all four numeric values.
# Example bounded search: K1_MIN=-1.0, K1_MAX=5.0, K2_MIN=0.0, K2_MAX=10.0
K1_MIN = None  # Regression Difference Percentage lower bound, or None
K1_MAX = None  # Regression Difference Percentage upper bound, or None
K2_MIN = None  # Distance %(1) lower bound, or None
K2_MAX = None  # Distance %(1) upper bound, or None

# Output names (saved inside output_folder_path).
UPDATED_WORKBOOK_PATH = clinical_data_path.with_name(
    "clinical cancer validation data_Updated.xlsx"
)
FEATURE_VECTOR_PATH = output_folder_path / "FeatureVectorCancerDet.csv"
SEARCH_RESULTS_PATH = output_folder_path / "ConstrainedRandomSearch_Results.csv"
SELECTED_PANEL_SUMMARY_CSV_PATH = (
    output_folder_path / "SelectedPanel_Calibration2_Summary.csv"
)
RECURRENCE_TABLE_PATH = output_folder_path / "Biomarker_Frequency_Top_Feasible_Panels.csv"


# Algorithm defaults: normally leave these unchanged unless tuning the search.
PHASE_WEIGHTS = {
    "balanced_diverse_exploration": 0.40,
    "sensitivity_guided_search": 0.30,
    "recurrence_informed_balanced_search": 0.20,
    "best_panel_neighborhoods": 0.10,
}
CANDIDATE_POOL_SIZE = 300
TOP_PANEL_FRACTION = 0.20
RECURRENCE_POOL_SIZE = 16
PRINT_EACH_PANEL_PERFORMANCE = False


def calculate_phase_budgets(total_budget, phase_weights):
    if not isinstance(total_budget, int) or total_budget < 1:
        raise ValueError("TOTAL_PANEL_BUDGET must be a positive integer.")
    weight_sum = sum(phase_weights.values())
    if not math.isclose(weight_sum, 1.0, rel_tol=0.0, abs_tol=1e-9):
        raise ValueError("PHASE_WEIGHTS must sum to 1.0.")
    exact = {name: total_budget * weight for name, weight in phase_weights.items()}
    budgets = {name: math.floor(value) for name, value in exact.items()}
    remaining = total_budget - sum(budgets.values())
    order = sorted(
        phase_weights,
        key=lambda name: (exact[name] - budgets[name], phase_weights[name]),
        reverse=True,
    )
    for name in order[:remaining]:
        budgets[name] += 1
    return budgets


PHASE_BUDGETS = calculate_phase_budgets(TOTAL_PANEL_BUDGET, PHASE_WEIGHTS)

# Regression predictors used to estimate Avg_Death_0D.
REGRESSION_PREDICTORS = [
    "0-Dim Hole Min Life Range",
    "0-Dim Hole Max Life Range",
    "Median_Death_0D",
    "Std_Death_0D",
]


# -------------------- WORKBOOK / SPLIT --------------------


def validate_split_fractions():
    """Validate configurable calibration fractions; Test is the remainder."""
    for name, value in (("CALIBRATION1_FRACTION", CALIBRATION1_FRACTION),
                        ("CALIBRATION2_FRACTION", CALIBRATION2_FRACTION)):
        if not isinstance(value, (int, float)) or not np.isfinite(value):
            raise ValueError(f"{name} must be a finite number between 0 and 1.")
        if value < 0 or value > 1:
            raise ValueError(f"{name} must be between 0 and 1; received {value}.")
    total_calibration = CALIBRATION1_FRACTION + CALIBRATION2_FRACTION
    if total_calibration >= 1.0:
        raise ValueError(
            "CALIBRATION1_FRACTION + CALIBRATION2_FRACTION must be less than 1.0 "
            "so that Test has a positive fraction. "
            f"Received {CALIBRATION1_FRACTION} + {CALIBRATION2_FRACTION}."
        )


def create_updated_workbook(source_path, updated_path, sheet):
    """Create and save one split workbook using the configured cohort-wise fractions.

    This helper is retained for standalone use; the main repeated-split workflow
    creates its five split workbooks through create_one_split().
    """
    global CANCER_TYPES
    validate_split_fractions()
    source_path, updated_path = Path(source_path), Path(updated_path)
    if not source_path.exists():
        raise FileNotFoundError(f"Source workbook not found: {source_path}\nChange clinical_data_path.")
    sheets = pd.read_excel(source_path, sheet_name=None)
    if sheet not in sheets:
        raise ValueError(f"Sheet {sheet!r} not found. Available: {list(sheets)}")
    df = sheets[sheet].copy()
    required = [COHORT_COL, PATIENT_ID_COL, STAGE_COL] + BIOMARKERS
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns in source workbook: {missing}")
    if DATA_TYPE_COL in df.columns:
        df = df.drop(columns=[DATA_TYPE_COL])

    cohort = df[COHORT_COL].astype(str).str.strip().str.casefold()
    if df[COHORT_COL].isna().any():
        raise ValueError(f"Column {COHORT_COL!r} contains missing cohort labels.")
    if cohort.eq("").any():
        raise ValueError(f"Column {COHORT_COL!r} contains blank cohort labels.")
    observed_cohorts = list(dict.fromkeys(cohort.tolist()))
    CANCER_TYPES = [name for name in observed_cohorts if name and name != NORMAL_COHORT.casefold()]
    if not CANCER_TYPES:
        raise ValueError(
            f"No cancer cohorts were detected in {COHORT_COL!r}; expected at least one cohort besides {NORMAL_COHORT!r}."
        )
    print("Automatically detected cancer types:", ", ".join(CANCER_TYPES))
    normal_indices = df.index[cohort.eq(NORMAL_COHORT.casefold())].tolist()
    if len(normal_indices) <= N_REFERENCE_NORMAL:
        raise ValueError(f"Need more than {N_REFERENCE_NORMAL} normal patients; found {len(normal_indices)}.")

    role = {}
    reference_indices = normal_indices[:N_REFERENCE_NORMAL]
    for idx in reference_indices:
        role[idx] = "Reference"

    rng = np.random.RandomState(RANDOM_SEED)
    counts = {"reference_normal": len(reference_indices)}

    # Remaining normal patients: configurable Calibration 1 and Calibration 2
    # percentages; Test receives all remaining patients. Reference normals are
    # excluded from the percentage calculations.
    remaining_normals = list(normal_indices[N_REFERENCE_NORMAL:])
    rng.shuffle(remaining_normals)
    n1 = int(np.floor(len(remaining_normals) * CALIBRATION1_FRACTION))
    n2 = int(np.floor(len(remaining_normals) * CALIBRATION2_FRACTION))
    for idx in remaining_normals[:n1]:
        role[idx] = "Regression Fit Calibration 1"
    for idx in remaining_normals[n1:n1+n2]:
        role[idx] = "Biomarker-Panel Calibration and Selection"
    for idx in remaining_normals[n1+n2:]:
        role[idx] = "Test"
    counts.update({
        "normal_regression_fit_calibration1": n1,
        "normal_calibration2": n2,
        "normal_test": len(remaining_normals) - n1 - n2,
    })

    # Cancer cohorts are split ONLY between Calibration 2 and Test.
    # Cancer patients must never be labelled Regression Fit Calibration 1.
    for cancer in CANCER_TYPES:
        indices = df.index[cohort.eq(cancer.casefold())].tolist()
        if not indices:
            raise ValueError(f"No rows found for cancer cohort {cancer!r}.")
        shuffled = list(indices)
        rng.shuffle(shuffled)
        n2c = int(np.floor(len(shuffled) * CALIBRATION2_FRACTION))
        for idx in shuffled[:n2c]:
            role[idx] = "Biomarker-Panel Calibration and Selection"
        for idx in shuffled[n2c:]:
            role[idx] = "Test"
        counts[f"{cancer}_calibration2"] = n2c
        counts[f"{cancer}_test"] = len(shuffled) - n2c

    df[DATA_TYPE_COL] = pd.Series(role)
    role_order = {
        "Reference": 0,
        "Regression Fit Calibration 1": 1,
        "Biomarker-Panel Calibration and Selection": 2,
        "Test": 3,
    }
    df["_source_order"] = np.arange(len(df))
    df["_role_order"] = df[DATA_TYPE_COL].map(role_order)
    df["_cohort_order"] = (
        df[COHORT_COL].astype(str).str.strip().str.casefold()
        .map({str(k).casefold(): v for k, v in cohort_order.items()}).fillna(99)
    )
    df = df.sort_values(
        ["_role_order", "_cohort_order", "_source_order"], kind="stable"
    ).drop(columns=["_source_order", "_role_order", "_cohort_order"]).reset_index(drop=True)
    sheets[sheet] = df
    updated_path.parent.mkdir(parents=True, exist_ok=True)
    with pd.ExcelWriter(updated_path, engine="openpyxl") as writer:
        for sheet_name_out, sheet_df in sheets.items():
            sheet_df.to_excel(writer, sheet_name=sheet_name_out, index=False)

    test_fraction = 1.0 - CALIBRATION1_FRACTION - CALIBRATION2_FRACTION
    print(f"Updated workbook created: {updated_path}")
    print(
        "Normal split after the separately reserved Reference group: "
        f"{CALIBRATION1_FRACTION:.1%} Calibration 1 (normal patients only) / "
        f"{CALIBRATION2_FRACTION:.1%} Calibration 2 / {test_fraction:.1%} Test."
    )
    print(
        "Cancer split: "
        f"{CALIBRATION2_FRACTION:.1%} Calibration 2 / "
        f"{1.0 - CALIBRATION2_FRACTION:.1%} Test; cancer patients are never assigned to Calibration 1."
    )
    print(
        f"The first {N_REFERENCE_NORMAL} normal patients are separately labelled Reference; "
        "the percentages for normal patients apply only after those reference rows are removed."
    )
    print("\nData Type counts:")
    print(df[DATA_TYPE_COL].value_counts().to_string())
    print("\nCohort x Data Type counts:")
    print(pd.crosstab(df[COHORT_COL], df[DATA_TYPE_COL]).to_string())
    return updated_path, counts


def calculate_tda_features(reference_df, patient_row, panel):
    """Compute the TDA features used by the regression and detection scores."""
    # Convert tuple panels to a list for pandas column selection.
    panel = list(panel)

    # Append the patient to the Reference biomarker matrix.
    reference_values = reference_df[panel].to_numpy(dtype=float, copy=False)
    patient_values = patient_row[panel].to_numpy(dtype=float, copy=False).reshape(1, -1)
    values = np.vstack((reference_values, patient_values))
    if not np.isfinite(values).all():
        raise ValueError(
            f"Non-finite selected biomarker value for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )
    if (values <= 0).any():
        raise ValueError(
            "Log transformation requires strictly positive values. "
            f"Check selected biomarkers for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )

    logged = np.log(values)
    corr = np.corrcoef(logged, rowvar=False)
    dissimilarity_matrix = 1.0 - np.abs(corr)
    if not np.isfinite(dissimilarity_matrix).all():
        raise ValueError(
            f"Invalid correlation/dissimilarity matrix for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )

    # Enforce numerical symmetry, a zero diagonal, and nonnegative entries.
    # The correlation-derived matrix is a dissimilarity matrix; it is not
    # assumed to satisfy the triangle inequality.
    dissimilarity_matrix = (
        dissimilarity_matrix + dissimilarity_matrix.T
    ) / 2.0
    np.fill_diagonal(dissimilarity_matrix, 0.0)
    dissimilarity_matrix = np.clip(dissimilarity_matrix, 0.0, 1.0)

    rips = gudhi.RipsComplex(distance_matrix=dissimilarity_matrix)
    simplex_tree = rips.create_simplex_tree(max_dimension=2)
    persistence = simplex_tree.persistence()
    finite = [
        (dim, (birth, death))
        for dim, (birth, death) in persistence
        if np.isfinite(death)
    ]
    holes1 = [pair for dim, pair in finite if dim == 1]
    holes0 = [pair for dim, pair in finite if dim == 0]

    if holes1:
        avg_birth1 = float(np.mean([birth for birth, _ in holes1]))
        avg_death1 = float(np.mean([death for _, death in holes1]))
    else:
        avg_birth1 = avg_death1 = 0.0

    if holes0:
        deaths0 = [death for _, death in holes0]
        lifetimes0 = [death - birth for birth, death in holes0]
        min_life0 = min(lifetimes0)
        max_life0 = max(lifetimes0)
        avg_death0 = float(np.mean(deaths0))
        median_death0 = float(np.median(deaths0))
        std_death0 = float(np.std(deaths0))
    else:
        min_life0 = max_life0 = avg_death0 = 0.0
        median_death0 = std_death0 = 0.0

    return {
        "0-Dim Hole Min Life Range": min_life0,
        "0-Dim Hole Max Life Range": max_life0,
        "Median_Death_0D": median_death0,
        "Std_Death_0D": std_death0,
        "Avg_Death_0D": avg_death0,
        "Average Birth (1D)": avg_birth1,
        "Average Death (1D)": avg_death1,
    }


def get_role_mask(df, role_name):
    return df[DATA_TYPE_COL].astype(str).str.strip().str.casefold().eq(
        role_name.casefold()
    )



def calculate_panel_features(df, reference_df, panel, include_test=False):
    """Extract features only for the requested development roles."""
    allowed = (
        get_role_mask(df, "Regression Fit Calibration 1")
        | get_role_mask(df, "Biomarker-Panel Calibration and Selection")
    )
    if include_test:
        allowed = allowed | get_role_mask(df, "Test")
    rows = []
    for idx, patient in df.loc[allowed].iterrows():
        record = {
            "Source_Row_Index": idx,
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
            COHORT_COL: patient[COHORT_COL],
            STAGE_COL: patient.get(STAGE_COL, np.nan),
            DATA_TYPE_COL: patient[DATA_TYPE_COL],
        }
        record.update(calculate_tda_features(reference_df, patient, panel))
        rows.append(record)
    return pd.DataFrame(rows)



def evaluate_panel(df, reference_df, panel, save_features=False):
    """
    Fit regression and thresholds on Calibration 1 NORMALS only.
    Evaluate the frozen rule on the independent Calibration 2 partition.
    """
    features = calculate_panel_features(df, reference_df, panel, include_test=False)
    if features.empty:
        raise ValueError("No Calibration 1 / Calibration 2 patients were available.")

    role = features[DATA_TYPE_COL].astype(str).str.strip().str.casefold()
    cohort = features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal1_normals = features[
        role.eq("regression fit calibration 1")
        & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()
    cal2 = features[
        role.eq("biomarker-panel calibration and selection")
    ].copy()
    if len(cal1_normals) < 2:
        raise ValueError("At least two Calibration 1 normal patients are required.")
    if cal2.empty:
        raise ValueError("Calibration 2 is empty.")

    model = LinearRegression().fit(
        cal1_normals[REGRESSION_PREDICTORS],
        cal1_normals["Avg_Death_0D"],
    )
    features["Predicted Avg_Death_0D"] = model.predict(features[REGRESSION_PREDICTORS])
    features["Difference"] = features["Avg_Death_0D"] - features["Predicted Avg_Death_0D"]
    denom = features["Avg_Death_0D"].abs().replace(0, np.nan)
    features["Regression Difference Percentage"] = (
        features["Difference"].abs() / denom
    ) * 100.0

    ref_birth = float(cal1_normals["Average Birth (1D)"].mean())
    ref_death = float(cal1_normals["Average Death (1D)"].mean())
    features["Distance %(1)"] = np.sqrt(
        (features["Average Birth (1D)"] - ref_birth) ** 2
        + (features["Average Death (1D)"] - ref_death) ** 2
    ) * 100.0

    # IMPORTANT: cal1_normals was copied before the derived columns were added.
    # Refresh it now so it contains Regression Difference Percentage and Distance %(1).
    cal1_normals = features.loc[
        role.eq("regression fit calibration 1")
        & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()

    # ------------------------------------------------------------------
    # REDUCED, VECTORIZED candidate threshold-pair search for the OR rule.
    #
    # Thresholds are optimized over candidate score boundaries. To reduce
    # runtime, the candidate grid is deterministically downsampled to at most
    # MAX_THRESHOLD_CANDIDATES values per score. Specificity is constrained
    # on the pooled Calibration 1 + Calibration 2 normal patients, strictly
    # above the configured target (1 - FPR).
    # ------------------------------------------------------------------
    cal2_mask = role.eq("biomarker-panel calibration and selection")
    cal2_features = features.loc[cal2_mask].copy()
    cal2_cohort = cal2_features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal2_normals = cal2_features[cal2_cohort.eq(NORMAL_COHORT.casefold())]
    cal2_cancers = cal2_features[cal2_cohort.isin(CANCER_TYPES)]

    if cal2_normals.empty:
        raise ValueError("Calibration 2 contains no normal patients for specificity calibration.")
    if cal2_cancers.empty:
        raise ValueError("Calibration 2 contains no cancer patients for threshold selection.")

    cal1_reg_norm = cal1_normals["Regression Difference Percentage"].to_numpy(dtype=float)
    cal1_dist_norm = cal1_normals["Distance %(1)"].to_numpy(dtype=float)
    cal2_reg_norm = cal2_normals["Regression Difference Percentage"].to_numpy(dtype=float)
    cal2_dist_norm = cal2_normals["Distance %(1)"].to_numpy(dtype=float)
    pooled_reg_norm = np.concatenate([cal1_reg_norm, cal2_reg_norm])
    pooled_dist_norm = np.concatenate([cal1_dist_norm, cal2_dist_norm])

    cancer_reg = cal2_cancers["Regression Difference Percentage"].to_numpy(dtype=float)
    cancer_dist = cal2_cancers["Distance %(1)"].to_numpy(dtype=float)
    cancer_cohort_values = (
        cal2_cancers[COHORT_COL].astype(str).str.strip().str.casefold().to_numpy()
    )
    specificity_target = 1.0 - FPR

    def reduced_boundary_candidates(
        normal_scores, cancer_scores, max_candidates, min_value=None, max_value=None, name="threshold"
    ):
        """Build original score-boundary candidates, optionally restricting their range."""
        combined = np.concatenate([
            np.asarray(normal_scores, dtype=float),
            np.asarray(cancer_scores, dtype=float),
        ])
        combined = combined[np.isfinite(combined)]
        if combined.size == 0:
            raise ValueError("No finite scores available for threshold search.")
        if not isinstance(max_candidates, int) or max_candidates < 2:
            raise ValueError("MAX_THRESHOLD_CANDIDATES must be an integer >= 2.")

        # This is the original candidate-generation procedure.
        unique_values = np.unique(combined)
        boundaries = np.unique(np.concatenate([
            unique_values, np.nextafter(unique_values, np.inf)
        ]))

        # Apply bounds only when a complete min/max pair is provided.
        bounded = min_value is not None and max_value is not None
        if bounded:
            if not np.isfinite(min_value) or not np.isfinite(max_value):
                raise ValueError(f"{name} bounds must be finite numbers.")
            if min_value > max_value:
                raise ValueError(
                    f"{name}_MIN ({min_value}) must be <= {name}_MAX ({max_value})."
                )
            boundaries = boundaries[
                (boundaries >= float(min_value)) & (boundaries <= float(max_value))
            ]
            # Include exact bounds, even when no observed score lies there.
            boundaries = np.unique(np.concatenate([
                boundaries, np.array([float(min_value), float(max_value)])
            ]))

        if boundaries.size > max_candidates:
            # Keep the original deterministic downsampling behavior.
            indices = np.linspace(0, boundaries.size - 1, max_candidates, dtype=int)
            boundaries = boundaries[np.unique(indices)]
        return boundaries

    # Bounds are active only when all four are supplied. If all are None, this
    # exactly follows the original code's unbounded threshold candidate search.
    bound_values = (K1_MIN, K1_MAX, K2_MIN, K2_MAX)
    bounds_provided = [value is not None for value in bound_values]
    if any(bounds_provided) and not all(bounds_provided):
        raise ValueError(
            "Provide all four bounds (K1_MIN, K1_MAX, K2_MIN, K2_MAX), or leave all four as None."
        )
    USE_THRESHOLD_BOUNDS = all(bounds_provided)

    k1_candidates = reduced_boundary_candidates(
        pooled_reg_norm, cancer_reg, MAX_THRESHOLD_CANDIDATES,
        K1_MIN if USE_THRESHOLD_BOUNDS else None,
        K1_MAX if USE_THRESHOLD_BOUNDS else None, "K1"
    )
    k2_candidates = reduced_boundary_candidates(
        pooled_dist_norm, cancer_dist, MAX_THRESHOLD_CANDIDATES,
        K2_MIN if USE_THRESHOLD_BOUNDS else None,
        K2_MAX if USE_THRESHOLD_BOUNDS else None, "K2"
    )
    if USE_THRESHOLD_BOUNDS:
        print(
            f"Bounded threshold search: k1 in [{K1_MIN}, {K1_MAX}] "
            f"({len(k1_candidates)} candidates); "
            f"k2 in [{K2_MIN}, {K2_MAX}] ({len(k2_candidates)} candidates); "
            f"up to {len(k1_candidates) * len(k2_candidates):,} pairs."
        )
    # Evaluate each k1 against all k2 values in vectorized NumPy operations.
    n_normals = len(pooled_reg_norm)
    best_pair = None
    best_key = None
    best_macro_sensitivity = -1.0
    best_overall_sensitivity = -1.0
    best_specificity = -1.0
    best_fp = None
    best_tn = None

    cancer_reg_pos = cancer_reg[:, None] > k1_candidates[None, :]
    cancer_dist_pos = cancer_dist[:, None] > k2_candidates[None, :]
    active_cal2_cancers = [
        cancer for cancer in CANCER_TYPES
        if np.any(cancer_cohort_values == cancer.casefold())
    ]
    per_type_masks = {
        cancer: cancer_cohort_values == cancer.casefold()
        for cancer in active_cal2_cancers
    }

    for i, k1 in enumerate(k1_candidates):
        normal_positive = (
            (pooled_reg_norm > k1)[:, None]
            | (pooled_dist_norm[:, None] > k2_candidates[None, :])
        )
        fp_counts = normal_positive.sum(axis=0)
        specificities = 1.0 - fp_counts / n_normals
        feasible = np.flatnonzero(specificities > specificity_target + 1e-12)
        if feasible.size == 0:
            continue

        cancer_positive_by_k2 = cancer_reg_pos[:, i, None] | cancer_dist_pos[:, feasible]
        overall_sens = cancer_positive_by_k2.mean(axis=0)
        type_sens = []
        for cancer in active_cal2_cancers:
            mask = per_type_masks[cancer]
            type_sens.append(cancer_positive_by_k2[mask].mean(axis=0))
        if not type_sens:
            continue
        macro_sens = np.mean(np.vstack(type_sens), axis=0)

        # Lexicographic objective matches original intent: macro sensitivity,
        # overall sensitivity, specificity, then larger thresholds.
        for j_local, j in enumerate(feasible):
            candidate_key = (
                float(macro_sens[j_local]),
                float(overall_sens[j_local]),
                float(specificities[j]),
                float(k1),
                float(k2_candidates[j]),
            )
            if best_key is None or candidate_key > best_key:
                best_key = candidate_key
                best_pair = (float(k1), float(k2_candidates[j]))
                best_macro_sensitivity = candidate_key[0]
                best_overall_sensitivity = candidate_key[1]
                best_specificity = candidate_key[2]
                best_fp = int(fp_counts[j])
                best_tn = int(n_normals - best_fp)

    if best_pair is None:
        if USE_THRESHOLD_BOUNDS:
            raise RuntimeError(
                "No k1/k2 pair within the user-defined bounds satisfied the pooled "
                f"Calibration 1 + Calibration 2 specificity requirement (> {specificity_target:.2%}). "
                f"Current bounds: k1=[{K1_MIN}, {K1_MAX}], k2=[{K2_MIN}, {K2_MAX}]. "
                "Adjust the bounds or review the specificity target."
            )
        # Preserve the original unbounded fallback: thresholds just above the
        # maximum pooled normal scores guarantee zero false positives.
        finite_reg = pooled_reg_norm[np.isfinite(pooled_reg_norm)]
        finite_dist = pooled_dist_norm[np.isfinite(pooled_dist_norm)]
        if finite_reg.size == 0 or finite_dist.size == 0:
            raise ValueError("Pooled calibration normal scores contain no finite values.")
        k1_fallback = float(np.nextafter(np.max(finite_reg), np.inf))
        k2_fallback = float(np.nextafter(np.max(finite_dist), np.inf))
        best_pair = (k1_fallback, k2_fallback)
        fallback_positive = (cancer_reg > k1_fallback) | (cancer_dist > k2_fallback)
        sens_by_type = [
            float(fallback_positive[per_type_masks[cancer]].mean())
            for cancer in CANCER_TYPES if per_type_masks[cancer].any()
        ]
        best_macro_sensitivity = (
            float(np.mean(sens_by_type)) if len(sens_by_type) == len(CANCER_TYPES) else float("nan")
        )
        best_overall_sensitivity = float(fallback_positive.mean()) if fallback_positive.size else float("nan")
        best_fp, best_tn = 0, n_normals
        best_specificity = 1.0
    best_threshold, distance_threshold = best_pair

    # Report each partition's specificity descriptively, but only the pooled
    # specificity is a threshold-selection constraint.
    cal1_positive_final = (
        (cal1_normals["Regression Difference Percentage"] > best_threshold)
        | (cal1_normals["Distance %(1)"] > distance_threshold)
    )
    best_cal1_specificity = float((~cal1_positive_final).mean())

    features["Result(Cancer)"] = np.where(
        (features["Regression Difference Percentage"] > best_threshold)
        | (features["Distance %(1)"] > distance_threshold),
        "Positive", "Negative"
    )
    cal1_normal_positive_final = (
        (cal1_normals["Regression Difference Percentage"] > best_threshold)
        | (cal1_normals["Distance %(1)"] > distance_threshold)
    )
    cal2_normal_positive = (
        (cal2_normals["Regression Difference Percentage"] > best_threshold)
        | (cal2_normals["Distance %(1)"] > distance_threshold)
    )
    combined_normal_positive_final = np.concatenate([
        cal1_normal_positive_final.to_numpy(dtype=bool),
        cal2_normal_positive.to_numpy(dtype=bool),
    ])
    tn = int((~combined_normal_positive_final).sum())
    fp = int(combined_normal_positive_final.sum())
    specificity_combined = (
        tn / (tn + fp) if (tn + fp) else float("nan")
    )
    specificity_cal2 = (
        float((~cal2_normal_positive).mean()) if len(cal2_normal_positive) else float("nan")
    )
    if not specificity_combined > specificity_target + 1e-12:
        raise RuntimeError(
            "Selected thresholds failed the pooled Calibration 1 + Calibration 2 "
            "specificity constraint."
        )

    cancer_results = {}
    sensitivities = []
    all_predictions = (
        (cal2_cancers["Regression Difference Percentage"] > best_threshold)
        | (cal2_cancers["Distance %(1)"] > distance_threshold)
    )
    for cancer in CANCER_TYPES:
        subset = cal2_cancers[
            cal2_cancers[COHORT_COL].astype(str).str.strip().str.casefold().eq(cancer)
        ]
        pred = (
            (subset["Regression Difference Percentage"] > best_threshold)
            | (subset["Distance %(1)"] > distance_threshold)
        )
        sens = float(pred.mean()) if len(subset) else float("nan")
        if len(subset):
            sensitivities.append(sens)
        cancer_results[cancer] = {
            "n": int(len(subset)),
            "sensitivity": sens,
            "true_positives": int(pred.sum()),
        }
    macro_sensitivity = float(np.mean(sensitivities)) if sensitivities else float("nan")
    overall_sensitivity = float(all_predictions.mean()) if len(cal2_cancers) else float("nan")

    if save_features:
        export_features = features.copy()
        export_features.insert(
            0, "Unnamed: 0",
            export_features["Source_Row_Index"].map(lambda value: f"P_{int(value)}")
        )
        requested_columns = [
            "Unnamed: 0", PATIENT_ID_COL, COHORT_COL, STAGE_COL, DATA_TYPE_COL,
            "0-Dim Hole Min Life Range", "0-Dim Hole Max Life Range",
            "Median_Death_0D", "Std_Death_0D", "Avg_Death_0D",
            "Average Birth (1D)", "Average Death (1D)",
            "Predicted Avg_Death_0D", "Difference",
            "Regression Difference Percentage", "Distance %(1)", "Result(Cancer)",
        ]
        export_features[[c for c in requested_columns if c in export_features.columns]].to_csv(
            FEATURE_VECTOR_PATH, index=False
        )

    return {
        "n_biomarkers": len(panel),
        "panel": list(panel),
        "regression_threshold": float(best_threshold),
        "distance_threshold": float(distance_threshold),
        "threshold_bounds_enabled": bool(USE_THRESHOLD_BOUNDS),
        "k1_min_search_bound": float(K1_MIN) if USE_THRESHOLD_BOUNDS else None,
        "k1_max_search_bound": float(K1_MAX) if USE_THRESHOLD_BOUNDS else None,
        "k2_min_search_bound": float(K2_MIN) if USE_THRESHOLD_BOUNDS else None,
        "k2_max_search_bound": float(K2_MAX) if USE_THRESHOLD_BOUNDS else None,
        "k1_candidate_count": int(len(k1_candidates)),
        "k2_candidate_count": int(len(k2_candidates)),
        "threshold_pairs_considered_max": int(len(k1_candidates) * len(k2_candidates)),
        "calibration1_specificity": float(best_cal1_specificity),
        "calibration2_specificity": float(specificity_cal2),
        "specificity": float(specificity_combined),
        "combined_calibration_specificity": float(specificity_combined),
        "false_positives": fp,
        "true_negatives": tn,
        "macro_sensitivity": macro_sensitivity,
        "overall_cancer_sensitivity": overall_sensitivity,
        "cancer_results": cancer_results,
        "regression_fit_normal_n": int(len(cal1_normals)),
        "calibration2_normal_n": int(len(cal2_normals)),
        "calibration_cancer_n": int(len(cal2_cancers)),
        "meets_sensitivity_target": bool(
            np.isfinite(macro_sensitivity) and macro_sensitivity > SENSITIVITY_TARGET
        ),
        "threshold_rule": (
            "Positive iff Regression Difference Percentage > regression_threshold "
            "OR Distance %(1) > distance_threshold"
        ),
    }, features


def canonical_panel(panel):
    return tuple(sorted(panel, key=BIOMARKERS.index))


def panel_key(panel):
    return tuple(panel)


def random_panel(rng):
    return canonical_panel(rng.sample(BIOMARKERS, PANEL_SIZE))


def balanced_diverse_panels(rng, count, used, marker_pool=None, prior_counts=None):
    """Generate unique panels while balancing marker appearances and reducing overlap.

    In Stage 1, marker_pool is all 42 biomarkers and prior_counts starts at zero,
    so all markers have approximately equal expected representation. In Stage 3,
    the pool is restricted to recurring markers found among the strongest feasible
    panels, and representation is balanced within that pool.
    """
    pool = list(marker_pool or BIOMARKERS)
    if len(pool) < PANEL_SIZE:
        pool = list(BIOMARKERS)
    counts = {b: int((prior_counts or {}).get(b, 0)) for b in pool}
    selected = []
    selected_set = set(used)
    max_attempts = max(100000, count * 1000)

    while len(selected) < count:
        candidates = set()
        attempts = 0
        while len(candidates) < CANDIDATE_POOL_SIZE and attempts < max_attempts:
            attempts += 1
            candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            if candidate not in selected_set:
                candidates.add(candidate)
        if not candidates:
            # This is only expected if the candidate space is nearly exhausted.
            candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            if candidate in selected_set:
                # Fall back to the full marker universe to maintain the exact budget.
                pool = list(BIOMARKERS)
                candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
                while candidate in selected_set:
                    candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            candidates.add(candidate)

        def balance_diversity_key(candidate):
            projected = dict(counts)
            for b in candidate:
                if b in projected:
                    projected[b] += 1
            values = list(projected.values())
            # First balance total appearances; then minimize overlap with earlier
            # panels from this stage; randomized final tie-break avoids ordering bias.
            imbalance = float(np.std(values)) if values else 0.0
            max_count = max(values) if values else 0
            if selected:
                overlaps = [len(set(candidate) & set(old)) for old in selected]
                mean_overlap = float(np.mean(overlaps))
                max_overlap = max(overlaps)
            else:
                mean_overlap = max_overlap = 0.0
            return (imbalance, max_count, mean_overlap, max_overlap, rng.random())

        # IMPORTANT FOR REPRODUCIBILITY:
        # `candidates` is a set, whose iteration order can vary between Python
        # processes. The key function consumes seeded RNG values for tie-breaking;
        # therefore, iterate in a stable sorted order before calling it.
        chosen = min(sorted(candidates), key=balance_diversity_key)
        selected.append(chosen)
        selected_set.add(chosen)
        # Keep the shared cross-stage registry synchronized. Without this line,
        # panels generated in this stage are not counted in `used`, even though
        # they are appended to all_rows later.
        used.add(chosen)
        for b in chosen:
            if b in counts:
                counts[b] += 1

    return selected


def top_feasible_rows(panel_rows):
    feasible = [item for item in panel_rows if item[1]["specificity"] > (1.0 - FPR) + 1e-12]
    feasible.sort(key=lambda item: item[1]["macro_sensitivity"], reverse=True)
    if not feasible:
        return []
    n_top = max(1, int(math.ceil(len(feasible) * TOP_PANEL_FRACTION)))
    return feasible[:n_top]


def biomarker_weights(panel_rows):
    """Estimate recurring-marker weights from the strongest feasible panels."""
    top = top_feasible_rows(panel_rows)
    counts = {b: 1.0 for b in BIOMARKERS}  # small baseline keeps exploration possible
    for panel, _summary in top:
        for biomarker in panel:
            counts[biomarker] += 1.0
    return counts


def weighted_panel(rng, weights):
    """Sample a unique-size panel without replacement, weighted by marker frequency."""
    available = list(BIOMARKERS)
    chosen = []
    for _ in range(PANEL_SIZE):
        total = sum(max(0.001, weights[b]) for b in available)
        pick = rng.random() * total
        cumulative = 0.0
        selected = available[-1]
        for biomarker in available:
            cumulative += max(0.001, weights[biomarker])
            if cumulative >= pick:
                selected = biomarker
                break
        chosen.append(selected)
        available.remove(selected)
    return canonical_panel(chosen)


def make_unique_weighted_panels(rng, count, used, weights):
    panels = []
    attempts = 0
    max_attempts = max(10000, count * 200)
    while len(panels) < count and attempts < max_attempts:
        attempts += 1
        candidate = weighted_panel(rng, weights)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    while len(panels) < count:
        candidate = random_panel(rng)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    return panels


def make_neighborhood_panels(rng, count, used, panel_rows):
    """Generate swaps around the best feasible panels, with random unique fill."""
    feasible = [item for item in panel_rows
                if item[1]["specificity"] > (1.0 - FPR) + 1e-12
                and np.isfinite(item[1]["macro_sensitivity"])]
    ranked_source = feasible if feasible else panel_rows
    ranked = sorted(ranked_source,
                    key=lambda item: item[1]["macro_sensitivity"]
                    if np.isfinite(item[1]["macro_sensitivity"]) else -1.0,
                    reverse=True)
    panels = []
    for panel, _summary in ranked:
        candidates = local_neighbors(panel)
        rng.shuffle(candidates)
        for candidate in candidates:
            if candidate not in used:
                used.add(candidate)
                panels.append(candidate)
                if len(panels) >= count:
                    return panels
    while len(panels) < count:
        candidate = random_panel(rng)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    return panels

def score_panel(df, reference_df, panel, cache):
    key = panel_key(panel)
    if key not in cache:
        summary, _ = evaluate_panel(df, reference_df, panel, save_features=False)
        cache[key] = summary
    return cache[key]


def rank_key(summary):
    # Specificity the configured specificity target is enforced as a hard feasibility constraint in
    # evaluate_panel. Among feasible panels, rank ONLY by macro-sensitivity.
    return summary["macro_sensitivity"]


def local_neighbors(panel):
    panel_set = set(panel)
    not_in_panel = [b for b in BIOMARKERS if b not in panel_set]
    neighbors = set()
    for removed in panel:
        for added in not_in_panel:
            candidate = canonical_panel((panel_set - {removed}) | {added})
            if len(candidate) == PANEL_SIZE:
                neighbors.add(candidate)
    return sorted(
        neighbors,
        key=lambda p: tuple(BIOMARKERS.index(x) for x in p),
    )



def evaluate_selected_panel_on_test(df, reference_df, panel, thresholds):
    """Evaluate the selected panel once on the untouched Test partition."""
    test_df = df[get_role_mask(df, "Test")].copy()
    if test_df.empty:
        print("No Test patients were available.")
        return None

    # Fit the regression model and reference coordinates from Calibration 1 normals,
    # then apply the already frozen thresholds to Test patients.
    cal1_features = calculate_panel_features(df, reference_df, panel, include_test=False)
    role = cal1_features[DATA_TYPE_COL].astype(str).str.strip().str.casefold()
    cohort = cal1_features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal1_normals = cal1_features[
        role.eq("regression fit calibration 1") & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()
    model = LinearRegression().fit(
        cal1_normals[REGRESSION_PREDICTORS], cal1_normals["Avg_Death_0D"]
    )
    test_features = []
    for idx, patient in test_df.iterrows():
        rec = {
            "Source_Row_Index": idx,
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
            COHORT_COL: patient[COHORT_COL],
            STAGE_COL: patient.get(STAGE_COL, np.nan),
            DATA_TYPE_COL: patient[DATA_TYPE_COL],
        }
        rec.update(calculate_tda_features(reference_df, patient, panel))
        test_features.append(rec)
    test_features = pd.DataFrame(test_features)
    test_features["Predicted Avg_Death_0D"] = model.predict(test_features[REGRESSION_PREDICTORS])
    test_features["Difference"] = (
        test_features["Avg_Death_0D"] - test_features["Predicted Avg_Death_0D"]
    )
    denom = test_features["Avg_Death_0D"].abs().replace(0, np.nan)
    test_features["Regression Difference Percentage"] = (
        test_features["Difference"].abs() / denom
    ) * 100.0
    ref_birth = float(cal1_normals["Average Birth (1D)"].mean())
    ref_death = float(cal1_normals["Average Death (1D)"].mean())
    test_features["Distance %(1)"] = np.sqrt(
        (test_features["Average Birth (1D)"] - ref_birth) ** 2
        + (test_features["Average Death (1D)"] - ref_death) ** 2
    ) * 100.0
    reg_t = thresholds["regression_threshold"]
    dist_t = thresholds["distance_threshold"]
    positive = (
        (test_features["Regression Difference Percentage"] > reg_t)
        | (test_features["Distance %(1)"] > dist_t)
    )
    test_features["Result(Cancer)"] = np.where(positive, "Positive", "Negative")
    cohort_values = test_features[COHORT_COL].astype(str).str.strip().str.casefold()
    normal_mask = cohort_values.eq(NORMAL_COHORT.casefold())
    normal_pred = positive[normal_mask]
    specificity = float((~normal_pred).mean()) if normal_mask.any() else float("nan")
    sens_values = [
        float(positive[cohort_values.eq(cancer)].mean())
        for cancer in CANCER_TYPES
        if cohort_values.eq(cancer).any()
    ]
    macro_sens = float(np.mean(sens_values)) if sens_values else float("nan")
    print("\nFINAL UNTOUCHED TEST-SET PERFORMANCE (reported only after panel selection)")
    print(f"Normal specificity: {specificity:.4%}")
    print(f"Macro-sensitivity: {macro_sens:.4%}")
    return {"specificity": specificity, "macro_sensitivity": macro_sens,
            "n_test": int(len(test_features))}


def Optimized_Biomarker_Selection(df, split_number):
    """Run the original staged selection algorithm on one already-created random split."""
    global CANCER_TYPES
    output_folder_path.mkdir(parents=True, exist_ok=True)
    validate_split_fractions()

    bound_values = (K1_MIN, K1_MAX, K2_MIN, K2_MAX)
    bounds_provided = [value is not None for value in bound_values]
    if any(bounds_provided) and not all(bounds_provided):
        raise ValueError(
            "Provide all four bounds (K1_MIN, K1_MAX, K2_MIN, K2_MAX), or leave all four as None."
        )
    if all(bounds_provided):
        for name, value in zip(("K1_MIN", "K1_MAX", "K2_MIN", "K2_MAX"), bound_values):
            if not isinstance(value, (int, float)) or not np.isfinite(value):
                raise ValueError(f"{name} must be a finite number or None when bounds are disabled.")
        if K1_MIN > K1_MAX:
            raise ValueError(f"K1_MIN ({K1_MIN}) must be <= K1_MAX ({K1_MAX}).")
        if K2_MIN > K2_MAX:
            raise ValueError(f"K2_MIN ({K2_MIN}) must be <= K2_MAX ({K2_MAX}).")
    if not isinstance(MAX_THRESHOLD_CANDIDATES, int) or MAX_THRESHOLD_CANDIDATES < 2:
        raise ValueError("MAX_THRESHOLD_CANDIDATES must be an integer >= 2.")
    if len(BIOMARKERS) != len(set(BIOMARKERS)):
        raise ValueError("BIOMARKERS must contain unique names.")
    if PANEL_SIZE < 1 or PANEL_SIZE > len(BIOMARKERS):
        raise ValueError("PANEL_SIZE must be between 1 and the number of BIOMARKERS.")
    if sum(PHASE_BUDGETS.values()) != TOTAL_PANEL_BUDGET:
        raise ValueError("The phase budgets must sum exactly to TOTAL_PANEL_BUDGET.")

    required = [COHORT_COL, PATIENT_ID_COL, STAGE_COL, DATA_TYPE_COL] + BIOMARKERS
    missing = [column for column in required if column not in df.columns]
    if missing:
        raise ValueError(f"Split {split_number}: missing required columns: {missing}")
    df = df.copy()
    observed_cohorts = list(dict.fromkeys(
        df[COHORT_COL].astype(str).str.strip().str.casefold().tolist()
    ))
    CANCER_TYPES = [
        name for name in observed_cohorts
        if name and name != NORMAL_COHORT.casefold()
    ]
    if not CANCER_TYPES:
        raise ValueError(f"Split {split_number}: no cancer cohorts found.")
    reference_df = df[
        get_role_mask(df, "Reference")
        & df[COHORT_COL].astype(str).str.strip().str.casefold().eq(NORMAL_COHORT.casefold())
    ].copy()
    if len(reference_df) != N_REFERENCE_NORMAL:
        raise ValueError(
            f"Split {split_number}: expected {N_REFERENCE_NORMAL} reference normals; found {len(reference_df)}."
        )

    print(f"\n{'=' * 78}")
    print(f"RANDOM SPLIT {split_number}/{N_REPETITIONS}")
    print("=" * 78)

    rng = random.Random(RANDOM_SEED)
    cache = {}
    used = set()
    all_rows = []
    stage_by_panel = {}

    def evaluate_phase(label, panels, budget):
        if len(panels) != budget:
            raise RuntimeError(f"{label}: expected {budget} panels, got {len(panels)}")
        phase_rows = []
        for i, panel in enumerate(panels, start=1):
            summary = score_panel(df, reference_df, panel, cache)
            phase_rows.append((panel, summary))
            all_rows.append((panel, summary))
            stage_by_panel[panel] = label

            # Print compact progress only every 300 panels, plus the final panel.
            completed = len(all_rows)
            if completed % 300 == 0 or completed == TOTAL_PANEL_BUDGET:
                feasible = [
                    row_summary["macro_sensitivity"]
                    for _, row_summary in all_rows
                    if row_summary["specificity"] > (1.0 - FPR) + 1e-12
                    and np.isfinite(row_summary["macro_sensitivity"])
                ]
                best_so_far = max(feasible) if feasible else float("nan")
                print(
                    f"Random search: {completed}/{TOTAL_PANEL_BUDGET} | "
                    f"Stage: {label} | "
                    f"best pooled-specificity-feasible macro-sensitivity so far: {best_so_far:.4f}"
                )
        return phase_rows

    # STAGE 1: all 42 biomarkers are treated equally at the outset. A greedy
    # balancing criterion keeps their appearance counts as even as possible,
    # while minimizing overlap between panels. No outcome-guided marker weights.
    stage1_panels = balanced_diverse_panels(
        rng, PHASE_BUDGETS["balanced_diverse_exploration"], used,
        marker_pool=BIOMARKERS, prior_counts={b: 0 for b in BIOMARKERS}
    )
    used.update(stage1_panels)
    evaluate_phase("1/4 balanced diverse exploration", stage1_panels,
                   PHASE_BUDGETS["balanced_diverse_exploration"])

    # STAGE 2: rank only feasible panels by macro-sensitivity, then estimate marker
    # recurrence from the top feasible fraction and use it to guide new combinations.
    weights = biomarker_weights(all_rows)
    stage2_panels = make_unique_weighted_panels(
        rng, PHASE_BUDGETS["sensitivity_guided_search"], used, weights
    )
    evaluate_phase("2/4 sensitivity-guided search", stage2_panels,
                   PHASE_BUDGETS["sensitivity_guided_search"])

    # STAGE 3: identify recurring biomarkers in the top feasible panels from Stages
    # 1–2, then repeat balanced/low-overlap exploration within the recurrence-informed
    # pool. This keeps the marker frequencies balanced within that selected pool.
    top_before_stage3 = top_feasible_rows(all_rows)
    recurrence_counts = {b: 0 for b in BIOMARKERS}
    for panel, _summary in top_before_stage3:
        for b in panel:
            recurrence_counts[b] += 1
    recurring_pool = [b for b, _count in sorted(
        recurrence_counts.items(), key=lambda item: (-item[1], BIOMARKERS.index(item[0]))
    )[:RECURRENCE_POOL_SIZE]]
    if len(recurring_pool) < PANEL_SIZE:
        recurring_pool = list(BIOMARKERS)
    stage3_panels = balanced_diverse_panels(
        rng, PHASE_BUDGETS["recurrence_informed_balanced_search"], used,
        marker_pool=recurring_pool, prior_counts={b: 0 for b in recurring_pool}
    )
    evaluate_phase("3/4 recurrence-informed balanced search", stage3_panels,
                   PHASE_BUDGETS["recurrence_informed_balanced_search"])

    # STAGE 4: one-biomarker swaps around the strongest panels from all previous
    # stages, with random unique panels as a fallback. The final selection below
    # still compares every feasible panel from all four stages.
    stage4_panels = make_neighborhood_panels(
        rng, PHASE_BUDGETS["best_panel_neighborhoods"], used, all_rows
    )
    evaluate_phase("4/4 best-panel neighborhood search", stage4_panels,
                   PHASE_BUDGETS["best_panel_neighborhoods"])

    if len(used) != TOTAL_PANEL_BUDGET or len(all_rows) != TOTAL_PANEL_BUDGET:
        raise RuntimeError(
            f"Search budget error: evaluated {len(all_rows)} panels, expected {TOTAL_PANEL_BUDGET}."
        )
    if len(set(panel for panel, _ in all_rows)) != TOTAL_PANEL_BUDGET:
        raise RuntimeError("Duplicate panels were evaluated; every panel must be unique.")

    # IMPORTANT: the winning panel is selected across all four stages, not only Stage 4.
    specificity_feasible_rows = [
        item for item in all_rows
        if item[1]["specificity"] > (1.0 - FPR) + 1e-12
        and np.isfinite(item[1]["macro_sensitivity"])
    ]
    finite_rows = [
        item for item in all_rows
        if np.isfinite(item[1]["macro_sensitivity"])
    ]
    if not finite_rows:
        raise RuntimeError("All evaluated panels had undefined macro-sensitivity.")
    # Detection-based panel selection:
    # 1. Keep only panels whose pooled Calibration 1 + Calibration 2 normal specificity is strictly above 1 - FPR.
    # 2. Among those feasible panels, select the panel with the highest
    #    macro-sensitivity across the cancer cohorts.
    # 3. Do not require macro-sensitivity itself to exceed 99%; it is the
    #    optimization objective, while specificity is the hard constraint.
    if specificity_feasible_rows:
        ranked_rows = sorted(
            specificity_feasible_rows,
            key=lambda item: (
                item[1]["macro_sensitivity"],
                item[1]["overall_cancer_sensitivity"],
                item[1]["specificity"],
            ),
            reverse=True,
        )
    else:
        # This is a diagnostic fallback for the unbounded search. Thresholds
        # immediately above the maximum pooled-normal scores yield 100% specificity.
        ranked_rows = sorted(
            finite_rows,
            key=lambda item: (
                item[1]["macro_sensitivity"],
                item[1]["overall_cancer_sensitivity"],
                item[1]["specificity"],
            ),
            reverse=True,
        )
        print(
            "\nWARNING: No panel met the strict pooled Calibration 1 + Calibration 2 "
            "specificity constraint. The selected panel is diagnostic only."
        )
    best_panel, _ = ranked_rows[0]

    best_summary = ranked_rows[0][1].copy()
    best_macro_sensitivity = best_summary["macro_sensitivity"]
    print("\nBEST BIOMARKER PANEL — SELECTED ACROSS ALL FOUR STAGES")
    print(f"Winning stage: {stage_by_panel.get(best_panel, 'unknown')}")
    print(f"Final biomarker set: {list(best_panel)!r}")
    print(f"Calibration specificity: {best_summary['specificity']:.4%}")
    print(f"Calibration macro-sensitivity: {best_macro_sensitivity:.4%}")
    print(f"Meets pooled specificity target: {best_summary['specificity'] > (1.0 - FPR) + 1e-12}")

    test_summary = evaluate_selected_panel_on_test(df, reference_df, best_panel, best_summary)
    return {
        "split": split_number,
        "winning_stage": stage_by_panel.get(best_panel, "unknown"),
        "biomarker_panel": " | ".join(best_panel),
        "calibration_specificity": float(best_summary["specificity"]),
        "calibration_macro_sensitivity": float(best_macro_sensitivity),
        "regression_threshold": float(best_summary["regression_threshold"]),
        "distance_threshold": float(best_summary["distance_threshold"]),
        "test_normal_specificity": float(test_summary["specificity"]),
        "test_macro_sensitivity": float(test_summary["macro_sensitivity"]),
        "n_test": int(test_summary["n_test"]),
        "total_unique_panels_evaluated": int(len(all_rows)),
    }


def cohort_key(series):
    return series.astype(str).str.strip().str.casefold()


def create_one_split(source_df, rng, repetition, fixed_cancer_roles):
    """
    Create one split with a newly sampled Reference group of 59 NORMAL patients.

    Only the identity of the 59 reference-normal patients changes between
    repetitions. Every non-normal cohort keeps the same source rows, biomarker
    values, and Calibration 2/Test assignment in all five repetitions.
    The remaining NORMAL patients are divided into Calibration 1, Calibration 2,
    and Test using the configured fractions.
    """
    df = source_df.copy()
    cohort = cohort_key(df[COHORT_COL])
    if df[COHORT_COL].isna().any() or cohort.eq("").any():
        raise ValueError("Cohort labels must not be missing or blank.")

    normal_indices = df.index[cohort.eq(NORMAL_COHORT.casefold())].tolist()
    if len(normal_indices) <= N_REFERENCE_NORMAL:
        raise ValueError(
            f"Need more than {N_REFERENCE_NORMAL} Normal patients; found {len(normal_indices)}."
        )

    # Randomly select 59 reference patients from the entire normal cohort.
    # This is the only group whose membership is deliberately resampled
    # between repetitions.
    reference_indices = rng.choice(
        np.asarray(normal_indices, dtype=int),
        size=N_REFERENCE_NORMAL,
        replace=False,
    ).tolist()
    reference_set = set(reference_indices)
    role = {idx: ROLE_REFERENCE for idx in reference_indices}

    # Split all remaining normal patients. No normal patient is discarded.
    remaining_normals = [idx for idx in normal_indices if idx not in reference_set]
    rng.shuffle(remaining_normals)
    n_cal1 = int(np.floor(len(remaining_normals) * CALIBRATION1_FRACTION))
    n_cal2_normal = int(np.floor(len(remaining_normals) * CALIBRATION2_FRACTION))

    for idx in remaining_normals[:n_cal1]:
        role[idx] = ROLE_CAL1
    for idx in remaining_normals[n_cal1:n_cal1 + n_cal2_normal]:
        role[idx] = ROLE_CAL2
    for idx in remaining_normals[n_cal1 + n_cal2_normal:]:
        role[idx] = ROLE_TEST

    # Non-normal cohorts use one fixed role assignment shared by all five
    # repetitions. Their source rows and biomarker values are never modified.
    for idx, assigned_role in fixed_cancer_roles.items():
        role[idx] = assigned_role

    if len(role) != len(df):
        unassigned = [idx for idx in df.index if idx not in role]
        raise ValueError(
            f"Split {repetition}: some rows were not assigned a Data Type; "
            f"check cohort labels: {sorted(set(cohort.loc[unassigned]))}"
        )

    df[DATA_TYPE_COL] = pd.Series(role)
    df[SOURCE_ROW_COL] = np.arange(len(df), dtype=int)
    df["_role_order"] = df[DATA_TYPE_COL].map(ROLE_ORDER)
    df["_cohort_order"] = cohort.map(COHORT_ORDER).fillna(999).astype(int)
    df = df.sort_values(
        ["_role_order", "_cohort_order", SOURCE_ROW_COL], kind="stable"
    ).drop(columns=["_role_order", "_cohort_order"]).reset_index(drop=True)

    if len(df) != len(source_df) or df[SOURCE_ROW_COL].duplicated().any():
        raise RuntimeError(f"Split {repetition}: source rows were lost or duplicated.")
    if set(df[SOURCE_ROW_COL]) != set(range(len(source_df))):
        raise RuntimeError(f"Split {repetition}: source-row audit failed.")

    split_cohort = cohort_key(df[COHORT_COL])
    reference_mask = (
        df[DATA_TYPE_COL].astype(str).str.casefold().eq(ROLE_REFERENCE.casefold())
        & split_cohort.eq(NORMAL_COHORT.casefold())
    )
    if int(reference_mask.sum()) != N_REFERENCE_NORMAL:
        raise RuntimeError(
            f"Split {repetition}: expected exactly {N_REFERENCE_NORMAL} reference normals; "
            f"found {int(reference_mask.sum())}."
        )

    return df


def main():
    validate_split_fractions()
    output_folder_path.mkdir(parents=True, exist_ok=True)
    SPLIT_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    if not clinical_data_path.exists():
        raise FileNotFoundError(
            f"Input workbook not found: {clinical_data_path}\\n"
            "Update clinical_data_path to the correct workbook."
        )

    all_sheets = pd.read_excel(clinical_data_path, sheet_name=None)
    if sheet_name not in all_sheets:
        raise ValueError(
            f"Sheet {sheet_name!r} not found. Available sheets: {list(all_sheets)}"
        )
    source_df = all_sheets[sheet_name].copy()
    required = [COHORT_COL, PATIENT_ID_COL, STAGE_COL] + BIOMARKERS
    missing = [column for column in required if column not in source_df.columns]
    if missing:
        raise ValueError(f"Missing required columns in source workbook: {missing}")
    if DATA_TYPE_COL in source_df.columns:
        source_df = source_df.drop(columns=[DATA_TYPE_COL])
    if SOURCE_ROW_COL in source_df.columns:
        source_df = source_df.drop(columns=[SOURCE_ROW_COL])
    source_df = source_df.reset_index(drop=True)

    cohort = cohort_key(source_df[COHORT_COL])
    if source_df[COHORT_COL].isna().any() or cohort.eq("").any():
        raise ValueError("Cohort labels must not be missing or blank.")
    observed = set(cohort.unique())
    expected = set(COHORT_ORDER)
    unexpected = observed - expected
    if unexpected:
        raise ValueError(
            f"Unexpected cohort name(s): {sorted(unexpected)}. "
            "Update COHORT_ORDER to include the intended cohorts."
        )
    missing_cohorts = expected - observed
    if missing_cohorts:
        raise ValueError(
            f"Expected cohort(s) missing from source workbook: {sorted(missing_cohorts)}."
        )
    if int(cohort.eq(NORMAL_COHORT.casefold()).sum()) <= N_REFERENCE_NORMAL:
        raise ValueError(
            f"Need more than {N_REFERENCE_NORMAL} Normal patients in the source workbook."
        )

    # Generate non-normal cohort assignments ONCE. The same original cancer
    # patient rows, biomarker values, and Calibration 2/Test labels are retained
    # in every repetition.
    cancer_rng = np.random.RandomState(SPLIT_RANDOM_SEED + 1)
    fixed_cancer_roles = {}
    cancer_cohorts = [
        name for name in COHORT_ORDER
        if name.casefold() != NORMAL_COHORT.casefold()
    ]
    for cancer in cancer_cohorts:
        indices = source_df.index[cohort.eq(cancer.casefold())].tolist()
        if not indices:
            raise ValueError(f"No rows found for cancer cohort {cancer!r}.")
        shuffled = list(indices)
        cancer_rng.shuffle(shuffled)
        n_cal2_cancer = int(np.floor(len(shuffled) * CALIBRATION2_FRACTION))
        for idx in shuffled[:n_cal2_cancer]:
            fixed_cancer_roles[idx] = ROLE_CAL2
        for idx in shuffled[n_cal2_cancer:]:
            fixed_cancer_roles[idx] = ROLE_TEST

    # Generate five reproducible samples of 59 reference patients from the
    # complete normal cohort. Cancer-patient assignments remain fixed across
    # repetitions; the remaining normal-patient assignments are regenerated.
    split_rng = np.random.RandomState(SPLIT_RANDOM_SEED)
    split_runs = []
    reference_patient_sets = []

    for repetition in range(1, N_REPETITIONS + 1):
        split_df = create_one_split(
            source_df, split_rng, repetition, fixed_cancer_roles
        )

        reference_rows = split_df[
            split_df[DATA_TYPE_COL].astype(str).str.casefold().eq(ROLE_REFERENCE.casefold())
            & cohort_key(split_df[COHORT_COL]).eq(NORMAL_COHORT.casefold())
        ]
        reference_patient_sets.append(
            set(reference_rows[PATIENT_ID_COL].astype(str).tolist())
        )

        split_path = SPLIT_OUTPUT_DIR / (
            f"clinical cancer validation data_RandomSplit_{repetition:02d}.xlsx"
        )
        output_sheets = {name: sheet.copy() for name, sheet in all_sheets.items()}
        output_sheets[sheet_name] = split_df
        with pd.ExcelWriter(split_path, engine="openpyxl") as writer:
            for sheet_name_out, sheet_df in output_sheets.items():
                sheet_df.to_excel(writer, sheet_name=sheet_name_out, index=False)

        split_runs.append((repetition, split_df, split_path))
        print(f"Created random split {repetition}/{N_REPETITIONS}: {split_path.name}")
        print(f"  Randomly selected reference-normal patients: {len(reference_rows)}")

    print("\\nREFERENCE-GROUP OVERLAP ACROSS THE FIVE SPLITS")
    for i in range(len(reference_patient_sets)):
        for j in range(i + 1, len(reference_patient_sets)):
            overlap = len(reference_patient_sets[i] & reference_patient_sets[j])
            print(f"Split {i + 1} vs Split {j + 1}: {overlap}/59 shared patients")

    # Run the configured four-stage panel search independently on each split.
    # Panel generation, threshold calibration, selection criteria, and reporting
    # retain the same settings for every repetition.
    results = []
    for repetition, split_df, _split_path in split_runs:
        result = Optimized_Biomarker_Selection(split_df, repetition)
        results.append(result)

    results_df = pd.DataFrame(results)
    results_df.to_excel(COMBINED_RESULTS_PATH, index=False)
    print("\\n" + "=" * 78)
    print("ALL FIVE RANDOM SPLITS COMPLETED")
    print(f"Combined results: {COMBINED_RESULTS_PATH}")
    print(f"Split workbooks: {SPLIT_OUTPUT_DIR}")
    print("=" * 78)


if __name__ == "__main__":
    main()
```

```python
"""
Four-stage constrained random search for TDA-based biomarker-panel selection.

Workflow
--------
1. Creates five reproducible, stratified split workbooks.
   - The first N_REFERENCE_NORMAL normal patients in source-row order form the
     fixed Reference cohort in every split.
   - Remaining normal patients are randomly assigned to Regression Fit
     Calibration 1, Biomarker-Panel Calibration and Selection, and Test.
   - Cancer patients are assigned only to Calibration 2 and Test; they are
     never included in Regression Fit Calibration 1.
2. Evaluates TOTAL_PANEL_BUDGET unique biomarker panels of size PANEL_SIZE
   separately within each split.
3. Divides the panel-search budget across four stages:
   - Stage 1: balanced biomarker representation and reduced panel overlap.
   - Stage 2: weighted sampling informed by markers in the top feasible panels.
   - Stage 3: balanced exploration within a recurrence-informed marker pool.
   - Stage 4: one-biomarker neighbourhood swaps around high-performing panels.
4. For each panel, fits the linear regression model and calculates the
   Distance %(1) reference coordinates using Calibration 1 normal patients only.
5. Selects candidate thresholds for Regression Difference Percentage and
   Distance %(1) on Calibration 2, using an OR decision rule. Specificity is
   constrained on the pooled Calibration 1 and Calibration 2 normal patients
   to be strictly greater than 1 - FPR. The objective is to maximize
   Calibration 2 macro-sensitivity across the cancer cohorts.
6. Selects the final panel from all four search stages, ranking feasible panels
   by macro-sensitivity, then overall cancer sensitivity, then pooled
   calibration specificity. The selected panel is evaluated once on the
   untouched Test partition using the frozen thresholds.

Configuration and implementation notes
--------------------------------------
- Update the input/output paths, biomarker names, cohort mapping, and column
  names to match the workbook.
- COHORT_ORDER defines the expected cohorts and their display order; the
  script validates the workbook against this mapping.
- SPLIT_RANDOM_SEED controls the five split assignments. RANDOM_SEED controls
  the panel-search sampling sequence used within each split.
- Each threshold score contributes at most MAX_THRESHOLD_CANDIDATES values,
  so the vectorized search evaluates at most the square of that number of
  threshold pairs.
- The log transformation requires finite, strictly positive selected
  biomarker values in the reference and evaluated patient rows.
"""

from pathlib import Path
import math
import random

import gudhi
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression


# =============================================================================
# USER-DEFINED SETTINGS
# Change these values when applying the function to another dataset.
# Algorithmic tuning parameters below this section retain their defaults.
# =============================================================================

# Input and output files
clinical_data_path = Path(
    r"D:\Cancer Detection folder\Biomarker Selection\clinical cancer validation data.xlsx"
)
sheet_name = "Normal and Cancer"
output_folder_path = Path(r"D:\Cancer Detection folder\Biomarker Selection")

# Dataset-specific biomarker universe (names must match workbook column names).
BIOMARKERS = [
    "Tenascin C", "sAXL", "sE-selectin", "sHGFR/c-Met", "sHer2", "sHer3",
    "sIL-6Ra", "sNeuropilin-1", "sPECAM-1", "sVEGFR1", "sVEGFR3",
    "sc-kit/SCFR", "CA-125", "CA 15-3", "CA 19-9", "CEA", "FGF2", "HE4",
    "HGF", "IL-6", "IL-8", "Leptin", "MIF", "OPN", "Prolactin", "SCF",
    "TNFa", "TRAIL", "Total PSA", "VEGF", "b-HCG", "sFAS", "Cathepsin D",
    "FAP alpha", "Ferritin", "Galectin-3", "IGFBP3", "MIA", "MPO", "SHBG",
    "TIMP1", "TIMP2"
]

# Cohort names and their display/sort order.
# Keep the normal cohort in this mapping; all other keys are treated as cancer cohorts.
NORMAL_COHORT = "normal"
cohort_order = {"normal": 0, "bladder": 1, "ovarian": 2, "pancreatic": 3}

# Number of reproducible split repetitions; split assignments use a separate RNG seed.
N_REPETITIONS = 5
SPLIT_RANDOM_SEED = 20261022
COHORT_ORDER = {
    "normal": 0,
    "bladder": 1,
    "ovarian": 2,
    "pancreatic": 3,
}
SOURCE_ROW_COL = "Original_Source_Row_Zero_Based"
ROLE_REFERENCE = "Reference"
ROLE_CAL1 = "Regression Fit Calibration 1"
ROLE_CAL2 = "Biomarker-Panel Calibration and Selection"
ROLE_TEST = "Test"
ROLE_ORDER = {
    ROLE_REFERENCE: 0,
    ROLE_CAL1: 1,
    ROLE_CAL2: 2,
    ROLE_TEST: 3,
}
SPLIT_OUTPUT_DIR = output_folder_path / "Random Splits"
COMBINED_RESULTS_PATH = output_folder_path / "RandomSplit_BiomarkerSelection_Results.xlsx"
CANCER_TYPES = [name.casefold() for name in cohort_order if name.casefold() != NORMAL_COHORT.casefold()]

# Dataset-specific column names.
COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
STAGE_COL = "AJCC Stage"
DATA_TYPE_COL = "Data Type"

# Data split and model constraints.
# Edit the two fractions below whenever needed. They are fractions, not whole percentages:
#   0.20 = 20%, 0.30 = 30%, 0.50 = 50%.
# For NORMAL patients only, the first N_REFERENCE_NORMAL rows are reserved as Reference;
# the configured Calibration 1 and Calibration 2 fractions are applied to the remaining
# normal patients, with Test receiving the remainder.
# CANCER patients are NEVER assigned to Calibration 1: CALIBRATION2_FRACTION of each
# cancer cohort is assigned to Biomarker-Panel Calibration and Selection, and the rest
# is Test. Thus Calibration 1 contains only healthy/normal patients.
N_REFERENCE_NORMAL = 59
CALIBRATION1_FRACTION = 0.15
CALIBRATION2_FRACTION = 0.50
FPR = 0.01  # Require pooled Calibration 1 + Calibration 2 normal specificity strictly > 99%.
SENSITIVITY_TARGET = 0.99  # Diagnostic target only; specificity is the hard constraint
PANEL_SIZE = 8
TOTAL_PANEL_BUDGET = 3000
RANDOM_SEED = 20261017

# Maximum number of candidate thresholds to test for EACH detection score:
# - One set of candidates is generated for Regression Difference Percentage (k1).
# - One set is generated for Distance %(1) (k2).
# - The threshold search evaluates up to MAX_THRESHOLD_CANDIDATES x
#   MAX_THRESHOLD_CANDIDATES candidate pairs for the OR decision rule.
# - Lower values generally run faster but explore fewer threshold combinations;
#   higher values search more combinations and may take longer.
MAX_THRESHOLD_CANDIDATES = 80

# OPTIONAL USER-DEFINED THRESHOLD SEARCH BOUNDS (inclusive)
# Leave ALL FOUR values as None to use the original unbounded candidate search.
# To restrict the search, provide all four numeric values.
# Example bounded search: K1_MIN=-1.0, K1_MAX=5.0, K2_MIN=0.0, K2_MAX=10.0
K1_MIN = None  # Regression Difference Percentage lower bound, or None
K1_MAX = None  # Regression Difference Percentage upper bound, or None
K2_MIN = None  # Distance %(1) lower bound, or None
K2_MAX = None  # Distance %(1) upper bound, or None

# Output names (saved inside output_folder_path).
UPDATED_WORKBOOK_PATH = clinical_data_path.with_name(
    "clinical cancer validation data_Updated.xlsx"
)
FEATURE_VECTOR_PATH = output_folder_path / "FeatureVectorCancerDet.csv"
SEARCH_RESULTS_PATH = output_folder_path / "ConstrainedRandomSearch_Results.csv"
SELECTED_PANEL_SUMMARY_CSV_PATH = (
    output_folder_path / "SelectedPanel_Calibration2_Summary.csv"
)
RECURRENCE_TABLE_PATH = output_folder_path / "Biomarker_Frequency_Top_Feasible_Panels.csv"


# Algorithm defaults: normally leave these unchanged unless tuning the search.
PHASE_WEIGHTS = {
    "balanced_diverse_exploration": 0.40,
    "sensitivity_guided_search": 0.30,
    "recurrence_informed_balanced_search": 0.20,
    "best_panel_neighborhoods": 0.10,
}
CANDIDATE_POOL_SIZE = 300
TOP_PANEL_FRACTION = 0.20
RECURRENCE_POOL_SIZE = 16
PRINT_EACH_PANEL_PERFORMANCE = False


def calculate_phase_budgets(total_budget, phase_weights):
    if not isinstance(total_budget, int) or total_budget < 1:
        raise ValueError("TOTAL_PANEL_BUDGET must be a positive integer.")
    weight_sum = sum(phase_weights.values())
    if not math.isclose(weight_sum, 1.0, rel_tol=0.0, abs_tol=1e-9):
        raise ValueError("PHASE_WEIGHTS must sum to 1.0.")
    exact = {name: total_budget * weight for name, weight in phase_weights.items()}
    budgets = {name: math.floor(value) for name, value in exact.items()}
    remaining = total_budget - sum(budgets.values())
    order = sorted(
        phase_weights,
        key=lambda name: (exact[name] - budgets[name], phase_weights[name]),
        reverse=True,
    )
    for name in order[:remaining]:
        budgets[name] += 1
    return budgets


PHASE_BUDGETS = calculate_phase_budgets(TOTAL_PANEL_BUDGET, PHASE_WEIGHTS)

# TDA features required by the regression model and detection scores.
REGRESSION_PREDICTORS = [
    "0-Dim Hole Min Life Range",
    "0-Dim Hole Max Life Range",
    "Median_Death_0D",
    "Std_Death_0D",
]
FEATURE_COLUMNS = [
    "0-Dim Hole Min Life Range",
    "0-Dim Hole Max Life Range",
    "Median_Death_0D",
    "Std_Death_0D",
    "Avg_Death_0D",
    "Average Birth (1D)",
    "Average Death (1D)",
]


# -------------------- WORKBOOK / SPLIT --------------------


def validate_split_fractions():
    """Validate configurable calibration fractions; Test is the remainder."""
    for name, value in (("CALIBRATION1_FRACTION", CALIBRATION1_FRACTION),
                        ("CALIBRATION2_FRACTION", CALIBRATION2_FRACTION)):
        if not isinstance(value, (int, float)) or not np.isfinite(value):
            raise ValueError(f"{name} must be a finite number between 0 and 1.")
        if value < 0 or value > 1:
            raise ValueError(f"{name} must be between 0 and 1; received {value}.")
    total_calibration = CALIBRATION1_FRACTION + CALIBRATION2_FRACTION
    if total_calibration >= 1.0:
        raise ValueError(
            "CALIBRATION1_FRACTION + CALIBRATION2_FRACTION must be less than 1.0 "
            "so that Test has a positive fraction. "
            f"Received {CALIBRATION1_FRACTION} + {CALIBRATION2_FRACTION}."
        )


def create_updated_workbook(source_path, updated_path, sheet):
    """Create one split workbook and detect cancer cohorts from the source dataset."""
    global CANCER_TYPES
    validate_split_fractions()
    source_path, updated_path = Path(source_path), Path(updated_path)
    if not source_path.exists():
        raise FileNotFoundError(f"Source workbook not found: {source_path}\nChange clinical_data_path.")
    sheets = pd.read_excel(source_path, sheet_name=None)
    if sheet not in sheets:
        raise ValueError(f"Sheet {sheet!r} not found. Available: {list(sheets)}")
    df = sheets[sheet].copy()
    required = [COHORT_COL, PATIENT_ID_COL, STAGE_COL] + BIOMARKERS
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns in source workbook: {missing}")
    if DATA_TYPE_COL in df.columns:
        df = df.drop(columns=[DATA_TYPE_COL])

    cohort = df[COHORT_COL].astype(str).str.strip().str.casefold()
    if df[COHORT_COL].isna().any():
        raise ValueError(f"Column {COHORT_COL!r} contains missing cohort labels.")
    if cohort.eq("").any():
        raise ValueError(f"Column {COHORT_COL!r} contains blank cohort labels.")
    observed_cohorts = list(dict.fromkeys(cohort.tolist()))
    CANCER_TYPES = [name for name in observed_cohorts if name and name != NORMAL_COHORT.casefold()]
    if not CANCER_TYPES:
        raise ValueError(
            f"No cancer cohorts were detected in {COHORT_COL!r}; expected at least one cohort besides {NORMAL_COHORT!r}."
        )
    print("Automatically detected cancer types:", ", ".join(CANCER_TYPES))
    normal_indices = df.index[cohort.eq(NORMAL_COHORT.casefold())].tolist()
    if len(normal_indices) <= N_REFERENCE_NORMAL:
        raise ValueError(f"Need more than {N_REFERENCE_NORMAL} normal patients; found {len(normal_indices)}.")

    role = {}
    reference_indices = normal_indices[:N_REFERENCE_NORMAL]
    for idx in reference_indices:
        role[idx] = "Reference"

    rng = np.random.RandomState(RANDOM_SEED)
    counts = {"reference_normal": len(reference_indices)}

    # Remaining normal patients: configurable Calibration 1 and Calibration 2
    # percentages; Test receives all remaining patients. Reference normals are
    # excluded from the percentage calculations.
    remaining_normals = list(normal_indices[N_REFERENCE_NORMAL:])
    rng.shuffle(remaining_normals)
    n1 = int(np.floor(len(remaining_normals) * CALIBRATION1_FRACTION))
    n2 = int(np.floor(len(remaining_normals) * CALIBRATION2_FRACTION))
    for idx in remaining_normals[:n1]:
        role[idx] = "Regression Fit Calibration 1"
    for idx in remaining_normals[n1:n1+n2]:
        role[idx] = "Biomarker-Panel Calibration and Selection"
    for idx in remaining_normals[n1+n2:]:
        role[idx] = "Test"
    counts.update({
        "normal_regression_fit_calibration1": n1,
        "normal_calibration2": n2,
        "normal_test": len(remaining_normals) - n1 - n2,
    })

    # Cancer cohorts are split ONLY between Calibration 2 and Test.
    # Cancer patients must never be labelled Regression Fit Calibration 1.
    for cancer in CANCER_TYPES:
        indices = df.index[cohort.eq(cancer.casefold())].tolist()
        if not indices:
            raise ValueError(f"No rows found for cancer cohort {cancer!r}.")
        shuffled = list(indices)
        rng.shuffle(shuffled)
        n2c = int(np.floor(len(shuffled) * CALIBRATION2_FRACTION))
        for idx in shuffled[:n2c]:
            role[idx] = "Biomarker-Panel Calibration and Selection"
        for idx in shuffled[n2c:]:
            role[idx] = "Test"
        counts[f"{cancer}_calibration2"] = n2c
        counts[f"{cancer}_test"] = len(shuffled) - n2c

    df[DATA_TYPE_COL] = pd.Series(role)
    role_order = {
        "Reference": 0,
        "Regression Fit Calibration 1": 1,
        "Biomarker-Panel Calibration and Selection": 2,
        "Test": 3,
    }
    df["_source_order"] = np.arange(len(df))
    df["_role_order"] = df[DATA_TYPE_COL].map(role_order)
    df["_cohort_order"] = (
        df[COHORT_COL].astype(str).str.strip().str.casefold()
        .map({str(k).casefold(): v for k, v in cohort_order.items()}).fillna(99)
    )
    df = df.sort_values(
        ["_role_order", "_cohort_order", "_source_order"], kind="stable"
    ).drop(columns=["_source_order", "_role_order", "_cohort_order"]).reset_index(drop=True)
    sheets[sheet] = df
    updated_path.parent.mkdir(parents=True, exist_ok=True)
    with pd.ExcelWriter(updated_path, engine="openpyxl") as writer:
        for sheet_name_out, sheet_df in sheets.items():
            sheet_df.to_excel(writer, sheet_name=sheet_name_out, index=False)

    test_fraction = 1.0 - CALIBRATION1_FRACTION - CALIBRATION2_FRACTION
    print(f"Updated workbook created: {updated_path}")
    print(
        "Normal split after the separately reserved Reference group: "
        f"{CALIBRATION1_FRACTION:.1%} Calibration 1 (normal patients only) / "
        f"{CALIBRATION2_FRACTION:.1%} Calibration 2 / {test_fraction:.1%} Test."
    )
    print(
        "Cancer split: "
        f"{CALIBRATION2_FRACTION:.1%} Calibration 2 / "
        f"{1.0 - CALIBRATION2_FRACTION:.1%} Test; cancer patients are never assigned to Calibration 1."
    )
    print(
        f"The first {N_REFERENCE_NORMAL} normal patients are separately labelled Reference; "
        "the percentages for normal patients apply only after those reference rows are removed."
    )
    print("\nData Type counts:")
    print(df[DATA_TYPE_COL].value_counts().to_string())
    print("\nCohort x Data Type counts:")
    print(pd.crosstab(df[COHORT_COL], df[DATA_TYPE_COL]).to_string())
    return updated_path, counts


def calculate_tda_features(reference_df, patient_row, panel):
    """Calculate the seven TDA features used by the detection model."""
    # Panels are stored as tuples for hashing/caching, but pandas column
    # selection requires a list of column names rather than a tuple key.
    panel = list(panel)

    # Convert the reference matrix once per call and append the patient's values
    # directly as a NumPy row, avoiding a temporary pandas DataFrame.
    reference_values = reference_df[panel].to_numpy(dtype=float, copy=False)
    patient_values = patient_row[panel].to_numpy(dtype=float, copy=False).reshape(1, -1)
    values = np.vstack((reference_values, patient_values))

    if not np.isfinite(values).all():
        raise ValueError(
            f"Non-finite selected biomarker value for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )
    if (values <= 0).any():
        raise ValueError(
            "Log transformation requires strictly positive values. "
            f"Check selected biomarkers for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )

    logged = np.log(values)
    correlation_matrix = np.corrcoef(logged, rowvar=False)
    dissimilarity_matrix = 1.0 - np.abs(correlation_matrix)

    if not np.isfinite(dissimilarity_matrix).all():
        raise ValueError(
            f"Invalid correlation-derived dissimilarity matrix for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )

    # Enforce numerical symmetry, a zero diagonal, and values in [0, 1].
    # This is a dissimilarity matrix and is not assumed to satisfy the
    # triangle inequality.
    dissimilarity_matrix = (
        dissimilarity_matrix + dissimilarity_matrix.T
    ) / 2.0
    np.fill_diagonal(dissimilarity_matrix, 0.0)
    dissimilarity_matrix = np.clip(dissimilarity_matrix, 0.0, 1.0)

    rips = gudhi.RipsComplex(distance_matrix=dissimilarity_matrix)
    simplex_tree = rips.create_simplex_tree(max_dimension=2)
    persistence = simplex_tree.persistence()
    finite = [
        (dim, (birth, death))
        for dim, (birth, death) in persistence
        if np.isfinite(death)
    ]
    holes1 = [pair for dim, pair in finite if dim == 1]
    holes0 = [pair for dim, pair in finite if dim == 0]

    if holes1:
        avg_birth1 = float(np.mean([birth for birth, _ in holes1]))
        avg_death1 = float(np.mean([death for _, death in holes1]))
    else:
        avg_birth1 = avg_death1 = 0.0

    if holes0:
        deaths0 = [death for _, death in holes0]
        lifetimes0 = [death - birth for birth, death in holes0]
        min_life0 = min(lifetimes0)
        max_life0 = max(lifetimes0)
        avg_death0 = float(np.mean(deaths0))
        median_death0 = float(np.median(deaths0))
        std_death0 = float(np.std(deaths0))
    else:
        min_life0 = max_life0 = avg_death0 = 0.0
        median_death0 = std_death0 = 0.0

    return {
        "0-Dim Hole Min Life Range": min_life0,
        "0-Dim Hole Max Life Range": max_life0,
        "Median_Death_0D": median_death0,
        "Std_Death_0D": std_death0,
        "Avg_Death_0D": avg_death0,
        "Average Birth (1D)": avg_birth1,
        "Average Death (1D)": avg_death1,
    }


def get_role_mask(df, role_name):
    return df[DATA_TYPE_COL].astype(str).str.strip().str.casefold().eq(
        role_name.casefold()
    )



def calculate_panel_features(df, reference_df, panel, include_test=False):
    """Extract features only for the requested development roles."""
    allowed = (
        get_role_mask(df, "Regression Fit Calibration 1")
        | get_role_mask(df, "Biomarker-Panel Calibration and Selection")
    )
    if include_test:
        allowed = allowed | get_role_mask(df, "Test")
    rows = []
    for idx, patient in df.loc[allowed].iterrows():
        record = {
            "Source_Row_Index": idx,
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
            COHORT_COL: patient[COHORT_COL],
            STAGE_COL: patient.get(STAGE_COL, np.nan),
            DATA_TYPE_COL: patient[DATA_TYPE_COL],
        }
        record.update(calculate_tda_features(reference_df, patient, panel))
        rows.append(record)
    return pd.DataFrame(rows)



def evaluate_panel(df, reference_df, panel, save_features=False):
    """
    Fit the regression model on Calibration 1 normals and select thresholds
    using Calibration 2 cancer cases subject to pooled calibration specificity.
    """
    features = calculate_panel_features(df, reference_df, panel, include_test=False)
    if features.empty:
        raise ValueError("No Calibration 1 / Calibration 2 patients were available.")

    role = features[DATA_TYPE_COL].astype(str).str.strip().str.casefold()
    cohort = features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal1_normals = features[
        role.eq("regression fit calibration 1")
        & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()
    cal2 = features[
        role.eq("biomarker-panel calibration and selection")
    ].copy()
    if len(cal1_normals) < 2:
        raise ValueError("At least two Calibration 1 normal patients are required.")
    if cal2.empty:
        raise ValueError("Calibration 2 is empty.")

    model = LinearRegression().fit(
        cal1_normals[REGRESSION_PREDICTORS],
        cal1_normals["Avg_Death_0D"],
    )
    features["Predicted Avg_Death_0D"] = model.predict(features[REGRESSION_PREDICTORS])
    features["Difference"] = features["Avg_Death_0D"] - features["Predicted Avg_Death_0D"]
    denom = features["Avg_Death_0D"].abs().replace(0, np.nan)
    features["Regression Difference Percentage"] = (
        features["Difference"].abs() / denom
    ) * 100.0

    ref_birth = float(cal1_normals["Average Birth (1D)"].mean())
    ref_death = float(cal1_normals["Average Death (1D)"].mean())
    features["Distance %(1)"] = np.sqrt(
        (features["Average Birth (1D)"] - ref_birth) ** 2
        + (features["Average Death (1D)"] - ref_death) ** 2
    ) * 100.0

    # IMPORTANT: cal1_normals was copied before the derived columns were added.
    # Refresh it now so it contains Regression Difference Percentage and Distance %(1).
    cal1_normals = features.loc[
        role.eq("regression fit calibration 1")
        & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()

    # ------------------------------------------------------------------
    # Reduced, vectorized candidate-threshold search for the OR decision rule.
    # Candidate boundaries are generated from observed scores and their next
    # representable floating-point values, then deterministically downsampled
    # to at most MAX_THRESHOLD_CANDIDATES values per score. The specificity
    # constraint is applied to pooled Calibration 1 and Calibration 2 normals.
    # ------------------------------------------------------------------
    cal2_mask = role.eq("biomarker-panel calibration and selection")
    cal2_features = features.loc[cal2_mask].copy()
    cal2_cohort = cal2_features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal2_normals = cal2_features[cal2_cohort.eq(NORMAL_COHORT.casefold())]
    cal2_cancers = cal2_features[cal2_cohort.isin(CANCER_TYPES)]

    if cal2_normals.empty:
        raise ValueError("Calibration 2 contains no normal patients for specificity calibration.")
    if cal2_cancers.empty:
        raise ValueError("Calibration 2 contains no cancer patients for threshold selection.")

    cal1_reg_norm = cal1_normals["Regression Difference Percentage"].to_numpy(dtype=float)
    cal1_dist_norm = cal1_normals["Distance %(1)"].to_numpy(dtype=float)
    cal2_reg_norm = cal2_normals["Regression Difference Percentage"].to_numpy(dtype=float)
    cal2_dist_norm = cal2_normals["Distance %(1)"].to_numpy(dtype=float)
    pooled_reg_norm = np.concatenate([cal1_reg_norm, cal2_reg_norm])
    pooled_dist_norm = np.concatenate([cal1_dist_norm, cal2_dist_norm])

    cancer_reg = cal2_cancers["Regression Difference Percentage"].to_numpy(dtype=float)
    cancer_dist = cal2_cancers["Distance %(1)"].to_numpy(dtype=float)
    cancer_cohort_values = (
        cal2_cancers[COHORT_COL].astype(str).str.strip().str.casefold().to_numpy()
    )
    specificity_target = 1.0 - FPR

    def reduced_boundary_candidates(
        normal_scores, cancer_scores, max_candidates, min_value=None, max_value=None, name="threshold"
    ):
        """Build original score-boundary candidates, optionally restricting their range."""
        combined = np.concatenate([
            np.asarray(normal_scores, dtype=float),
            np.asarray(cancer_scores, dtype=float),
        ])
        combined = combined[np.isfinite(combined)]
        if combined.size == 0:
            raise ValueError("No finite scores available for threshold search.")
        if not isinstance(max_candidates, int) or max_candidates < 2:
            raise ValueError("MAX_THRESHOLD_CANDIDATES must be an integer >= 2.")

        # This is the original candidate-generation procedure.
        unique_values = np.unique(combined)
        boundaries = np.unique(np.concatenate([
            unique_values, np.nextafter(unique_values, np.inf)
        ]))

        # Apply bounds only when a complete min/max pair is provided.
        bounded = min_value is not None and max_value is not None
        if bounded:
            if not np.isfinite(min_value) or not np.isfinite(max_value):
                raise ValueError(f"{name} bounds must be finite numbers.")
            if min_value > max_value:
                raise ValueError(
                    f"{name}_MIN ({min_value}) must be <= {name}_MAX ({max_value})."
                )
            boundaries = boundaries[
                (boundaries >= float(min_value)) & (boundaries <= float(max_value))
            ]
            # Include exact bounds, even when no observed score lies there.
            boundaries = np.unique(np.concatenate([
                boundaries, np.array([float(min_value), float(max_value)])
            ]))

        if boundaries.size > max_candidates:
            # Keep the original deterministic downsampling behavior.
            indices = np.linspace(0, boundaries.size - 1, max_candidates, dtype=int)
            boundaries = boundaries[np.unique(indices)]
        return boundaries

    # Bounds are active only when all four are supplied. If all are None, this
    # exactly follows the original code's unbounded threshold candidate search.
    bound_values = (K1_MIN, K1_MAX, K2_MIN, K2_MAX)
    bounds_provided = [value is not None for value in bound_values]
    if any(bounds_provided) and not all(bounds_provided):
        raise ValueError(
            "Provide all four bounds (K1_MIN, K1_MAX, K2_MIN, K2_MAX), or leave all four as None."
        )
    USE_THRESHOLD_BOUNDS = all(bounds_provided)

    k1_candidates = reduced_boundary_candidates(
        pooled_reg_norm, cancer_reg, MAX_THRESHOLD_CANDIDATES,
        K1_MIN if USE_THRESHOLD_BOUNDS else None,
        K1_MAX if USE_THRESHOLD_BOUNDS else None, "K1"
    )
    k2_candidates = reduced_boundary_candidates(
        pooled_dist_norm, cancer_dist, MAX_THRESHOLD_CANDIDATES,
        K2_MIN if USE_THRESHOLD_BOUNDS else None,
        K2_MAX if USE_THRESHOLD_BOUNDS else None, "K2"
    )
    if USE_THRESHOLD_BOUNDS:
        print(
            f"Bounded threshold search: k1 in [{K1_MIN}, {K1_MAX}] "
            f"({len(k1_candidates)} candidates); "
            f"k2 in [{K2_MIN}, {K2_MAX}] ({len(k2_candidates)} candidates); "
            f"up to {len(k1_candidates) * len(k2_candidates):,} pairs."
        )
    # Evaluate each k1 against all k2 values in vectorized NumPy operations.
    n_normals = len(pooled_reg_norm)
    best_pair = None
    best_key = None
    best_macro_sensitivity = -1.0
    best_overall_sensitivity = -1.0
    best_specificity = -1.0
    best_fp = None
    best_tn = None

    cancer_reg_pos = cancer_reg[:, None] > k1_candidates[None, :]
    cancer_dist_pos = cancer_dist[:, None] > k2_candidates[None, :]
    active_cal2_cancers = [
        cancer for cancer in CANCER_TYPES
        if np.any(cancer_cohort_values == cancer.casefold())
    ]
    per_type_masks = {
        cancer: cancer_cohort_values == cancer.casefold()
        for cancer in active_cal2_cancers
    }

    for i, k1 in enumerate(k1_candidates):
        normal_positive = (
            (pooled_reg_norm > k1)[:, None]
            | (pooled_dist_norm[:, None] > k2_candidates[None, :])
        )
        fp_counts = normal_positive.sum(axis=0)
        specificities = 1.0 - fp_counts / n_normals
        feasible = np.flatnonzero(specificities > specificity_target + 1e-12)
        if feasible.size == 0:
            continue

        cancer_positive_by_k2 = cancer_reg_pos[:, i, None] | cancer_dist_pos[:, feasible]
        overall_sens = cancer_positive_by_k2.mean(axis=0)
        type_sens = []
        for cancer in active_cal2_cancers:
            mask = per_type_masks[cancer]
            type_sens.append(cancer_positive_by_k2[mask].mean(axis=0))
        if not type_sens:
            continue
        macro_sens = np.mean(np.vstack(type_sens), axis=0)

        # Lexicographic objective matches original intent: macro sensitivity,
        # overall sensitivity, specificity, then larger thresholds.
        for j_local, j in enumerate(feasible):
            candidate_key = (
                float(macro_sens[j_local]),
                float(overall_sens[j_local]),
                float(specificities[j]),
                float(k1),
                float(k2_candidates[j]),
            )
            if best_key is None or candidate_key > best_key:
                best_key = candidate_key
                best_pair = (float(k1), float(k2_candidates[j]))
                best_macro_sensitivity = candidate_key[0]
                best_overall_sensitivity = candidate_key[1]
                best_specificity = candidate_key[2]
                best_fp = int(fp_counts[j])
                best_tn = int(n_normals - best_fp)

    if best_pair is None:
        if USE_THRESHOLD_BOUNDS:
            raise RuntimeError(
                "No k1/k2 pair within the user-defined bounds satisfied the pooled "
                f"Calibration 1 + Calibration 2 specificity requirement (> {specificity_target:.2%}). "
                f"Current bounds: k1=[{K1_MIN}, {K1_MAX}], k2=[{K2_MIN}, {K2_MAX}]. "
                "Adjust the bounds or review the specificity target."
            )
        # Preserve the original unbounded fallback: thresholds just above the
        # maximum pooled normal scores guarantee zero false positives.
        finite_reg = pooled_reg_norm[np.isfinite(pooled_reg_norm)]
        finite_dist = pooled_dist_norm[np.isfinite(pooled_dist_norm)]
        if finite_reg.size == 0 or finite_dist.size == 0:
            raise ValueError("Pooled calibration normal scores contain no finite values.")
        k1_fallback = float(np.nextafter(np.max(finite_reg), np.inf))
        k2_fallback = float(np.nextafter(np.max(finite_dist), np.inf))
        best_pair = (k1_fallback, k2_fallback)
        fallback_positive = (cancer_reg > k1_fallback) | (cancer_dist > k2_fallback)
        sens_by_type = [
            float(fallback_positive[per_type_masks[cancer]].mean())
            for cancer in CANCER_TYPES if per_type_masks[cancer].any()
        ]
        best_macro_sensitivity = (
            float(np.mean(sens_by_type)) if len(sens_by_type) == len(CANCER_TYPES) else float("nan")
        )
        best_overall_sensitivity = float(fallback_positive.mean()) if fallback_positive.size else float("nan")
        best_fp, best_tn = 0, n_normals
        best_specificity = 1.0
    best_threshold, distance_threshold = best_pair

    # Report each partition's specificity descriptively, but only the pooled
    # specificity is a threshold-selection constraint.
    cal1_positive_final = (
        (cal1_normals["Regression Difference Percentage"] > best_threshold)
        | (cal1_normals["Distance %(1)"] > distance_threshold)
    )
    best_cal1_specificity = float((~cal1_positive_final).mean())

    features["Result(Cancer)"] = np.where(
        (features["Regression Difference Percentage"] > best_threshold)
        | (features["Distance %(1)"] > distance_threshold),
        "Positive", "Negative"
    )
    cal1_normal_positive_final = (
        (cal1_normals["Regression Difference Percentage"] > best_threshold)
        | (cal1_normals["Distance %(1)"] > distance_threshold)
    )
    cal2_normal_positive = (
        (cal2_normals["Regression Difference Percentage"] > best_threshold)
        | (cal2_normals["Distance %(1)"] > distance_threshold)
    )
    combined_normal_positive_final = np.concatenate([
        cal1_normal_positive_final.to_numpy(dtype=bool),
        cal2_normal_positive.to_numpy(dtype=bool),
    ])
    tn = int((~combined_normal_positive_final).sum())
    fp = int(combined_normal_positive_final.sum())
    specificity_combined = (
        tn / (tn + fp) if (tn + fp) else float("nan")
    )
    specificity_cal2 = (
        float((~cal2_normal_positive).mean()) if len(cal2_normal_positive) else float("nan")
    )
    if not specificity_combined > specificity_target + 1e-12:
        raise RuntimeError(
            "Selected thresholds failed the pooled Calibration 1 + Calibration 2 "
            "specificity constraint."
        )

    cancer_results = {}
    sensitivities = []
    all_predictions = (
        (cal2_cancers["Regression Difference Percentage"] > best_threshold)
        | (cal2_cancers["Distance %(1)"] > distance_threshold)
    )
    for cancer in CANCER_TYPES:
        subset = cal2_cancers[
            cal2_cancers[COHORT_COL].astype(str).str.strip().str.casefold().eq(cancer)
        ]
        pred = (
            (subset["Regression Difference Percentage"] > best_threshold)
            | (subset["Distance %(1)"] > distance_threshold)
        )
        sens = float(pred.mean()) if len(subset) else float("nan")
        if len(subset):
            sensitivities.append(sens)
        cancer_results[cancer] = {
            "n": int(len(subset)),
            "sensitivity": sens,
            "true_positives": int(pred.sum()),
        }
    macro_sensitivity = float(np.mean(sensitivities)) if sensitivities else float("nan")
    overall_sensitivity = float(all_predictions.mean()) if len(cal2_cancers) else float("nan")

    if save_features:
        export_features = features.copy()
        export_features.insert(
            0, "Unnamed: 0",
            export_features["Source_Row_Index"].map(lambda value: f"P_{int(value)}")
        )
        requested_columns = [
            "Unnamed: 0", PATIENT_ID_COL, COHORT_COL, STAGE_COL, DATA_TYPE_COL,
            *FEATURE_COLUMNS, "Predicted Avg_Death_0D", "Difference",
            "Regression Difference Percentage", "Distance %(1)", "Result(Cancer)",
        ]
        export_features[[c for c in requested_columns if c in export_features.columns]].to_csv(
            FEATURE_VECTOR_PATH, index=False
        )

    return {
        "n_biomarkers": len(panel),
        "panel": list(panel),
        "regression_threshold": float(best_threshold),
        "distance_threshold": float(distance_threshold),
        "threshold_bounds_enabled": bool(USE_THRESHOLD_BOUNDS),
        "k1_min_search_bound": float(K1_MIN) if USE_THRESHOLD_BOUNDS else None,
        "k1_max_search_bound": float(K1_MAX) if USE_THRESHOLD_BOUNDS else None,
        "k2_min_search_bound": float(K2_MIN) if USE_THRESHOLD_BOUNDS else None,
        "k2_max_search_bound": float(K2_MAX) if USE_THRESHOLD_BOUNDS else None,
        "k1_candidate_count": int(len(k1_candidates)),
        "k2_candidate_count": int(len(k2_candidates)),
        "threshold_pairs_considered_max": int(len(k1_candidates) * len(k2_candidates)),
        "calibration1_specificity": float(best_cal1_specificity),
        "calibration2_specificity": float(specificity_cal2),
        "specificity": float(specificity_combined),
        "combined_calibration_specificity": float(specificity_combined),
        "false_positives": fp,
        "true_negatives": tn,
        "macro_sensitivity": macro_sensitivity,
        "overall_cancer_sensitivity": overall_sensitivity,
        "cancer_results": cancer_results,
        "regression_fit_normal_n": int(len(cal1_normals)),
        "calibration2_normal_n": int(len(cal2_normals)),
        "calibration_cancer_n": int(len(cal2_cancers)),
        "meets_sensitivity_target": bool(
            np.isfinite(macro_sensitivity) and macro_sensitivity > SENSITIVITY_TARGET
        ),
        "threshold_rule": (
            "Positive iff Regression Difference Percentage > regression_threshold "
            "OR Distance %(1) > distance_threshold"
        ),
    }, features


def canonical_panel(panel):
    return tuple(sorted(panel, key=BIOMARKERS.index))


def panel_key(panel):
    return tuple(panel)


def random_panel(rng):
    return canonical_panel(rng.sample(BIOMARKERS, PANEL_SIZE))


def balanced_diverse_panels(rng, count, used, marker_pool=None, prior_counts=None):
    """Generate unique panels while balancing marker appearances and reducing overlap.

    In Stage 1, marker_pool is all 42 biomarkers and prior_counts starts at zero,
    so all markers have approximately equal expected representation. In Stage 3,
    the pool is restricted to recurring markers found among the strongest feasible
    panels, and representation is balanced within that pool.
    """
    pool = list(marker_pool or BIOMARKERS)
    if len(pool) < PANEL_SIZE:
        pool = list(BIOMARKERS)
    counts = {b: int((prior_counts or {}).get(b, 0)) for b in pool}
    selected = []
    selected_set = set(used)
    max_attempts = max(100000, count * 1000)

    while len(selected) < count:
        candidates = set()
        attempts = 0
        while len(candidates) < CANDIDATE_POOL_SIZE and attempts < max_attempts:
            attempts += 1
            candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            if candidate not in selected_set:
                candidates.add(candidate)
        if not candidates:
            # This is only expected if the candidate space is nearly exhausted.
            candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            if candidate in selected_set:
                # Fall back to the full marker universe to maintain the exact budget.
                pool = list(BIOMARKERS)
                candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
                while candidate in selected_set:
                    candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            candidates.add(candidate)

        def balance_diversity_key(candidate):
            projected = dict(counts)
            for b in candidate:
                if b in projected:
                    projected[b] += 1
            values = list(projected.values())
            # First balance total appearances; then minimize overlap with earlier
            # panels from this stage; randomized final tie-break avoids ordering bias.
            imbalance = float(np.std(values)) if values else 0.0
            max_count = max(values) if values else 0
            if selected:
                overlaps = [len(set(candidate) & set(old)) for old in selected]
                mean_overlap = float(np.mean(overlaps))
                max_overlap = max(overlaps)
            else:
                mean_overlap = max_overlap = 0.0
            return (imbalance, max_count, mean_overlap, max_overlap, rng.random())

        # IMPORTANT FOR REPRODUCIBILITY:
        # `candidates` is a set, whose iteration order can vary between Python
        # processes. The key function consumes seeded RNG values for tie-breaking;
        # therefore, iterate in a stable sorted order before calling it.
        chosen = min(sorted(candidates), key=balance_diversity_key)
        selected.append(chosen)
        selected_set.add(chosen)
        # Keep the shared cross-stage registry synchronized. Without this line,
        # panels generated in this stage are not counted in `used`, even though
        # they are appended to all_rows later.
        used.add(chosen)
        for b in chosen:
            if b in counts:
                counts[b] += 1

    return selected


def top_feasible_rows(panel_rows):
    feasible = [item for item in panel_rows if item[1]["specificity"] > (1.0 - FPR) + 1e-12]
    feasible.sort(key=lambda item: item[1]["macro_sensitivity"], reverse=True)
    if not feasible:
        return []
    n_top = max(1, int(math.ceil(len(feasible) * TOP_PANEL_FRACTION)))
    return feasible[:n_top]


def biomarker_weights(panel_rows):
    """Estimate recurring-marker weights from the strongest feasible panels."""
    top = top_feasible_rows(panel_rows)
    counts = {b: 1.0 for b in BIOMARKERS}  # small baseline keeps exploration possible
    for panel, _summary in top:
        for biomarker in panel:
            counts[biomarker] += 1.0
    return counts


def weighted_panel(rng, weights):
    """Sample a unique-size panel without replacement, weighted by marker frequency."""
    available = list(BIOMARKERS)
    chosen = []
    for _ in range(PANEL_SIZE):
        total = sum(max(0.001, weights[b]) for b in available)
        pick = rng.random() * total
        cumulative = 0.0
        selected = available[-1]
        for biomarker in available:
            cumulative += max(0.001, weights[biomarker])
            if cumulative >= pick:
                selected = biomarker
                break
        chosen.append(selected)
        available.remove(selected)
    return canonical_panel(chosen)


def make_unique_weighted_panels(rng, count, used, weights):
    panels = []
    attempts = 0
    max_attempts = max(10000, count * 200)
    while len(panels) < count and attempts < max_attempts:
        attempts += 1
        candidate = weighted_panel(rng, weights)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    while len(panels) < count:
        candidate = random_panel(rng)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    return panels


def make_neighborhood_panels(rng, count, used, panel_rows):
    """Generate swaps around the best feasible panels, with random unique fill."""
    feasible = [item for item in panel_rows
                if item[1]["specificity"] > (1.0 - FPR) + 1e-12
                and np.isfinite(item[1]["macro_sensitivity"])]
    ranked_source = feasible if feasible else panel_rows
    ranked = sorted(ranked_source,
                    key=lambda item: item[1]["macro_sensitivity"]
                    if np.isfinite(item[1]["macro_sensitivity"]) else -1.0,
                    reverse=True)
    panels = []
    for panel, _summary in ranked:
        candidates = local_neighbors(panel)
        rng.shuffle(candidates)
        for candidate in candidates:
            if candidate not in used:
                used.add(candidate)
                panels.append(candidate)
                if len(panels) >= count:
                    return panels
    while len(panels) < count:
        candidate = random_panel(rng)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    return panels

def score_panel(df, reference_df, panel, cache):
    key = panel_key(panel)
    if key not in cache:
        summary, _ = evaluate_panel(df, reference_df, panel, save_features=False)
        cache[key] = summary
    return cache[key]


def rank_key(summary):
    # The configured specificity target is enforced as a hard feasibility constraint
    # in evaluate_panel. Among feasible panels, rank by macro-sensitivity.
    return summary["macro_sensitivity"]


def local_neighbors(panel):
    panel_set = set(panel)
    not_in_panel = [b for b in BIOMARKERS if b not in panel_set]
    neighbors = set()
    for removed in panel:
        for added in not_in_panel:
            candidate = canonical_panel((panel_set - {removed}) | {added})
            if len(candidate) == PANEL_SIZE:
                neighbors.add(candidate)
    return sorted(
        neighbors,
        key=lambda p: tuple(BIOMARKERS.index(x) for x in p),
    )



def evaluate_selected_panel_on_test(df, reference_df, panel, thresholds):
    """Evaluate the selected panel once on the untouched Test partition."""
    test_df = df[get_role_mask(df, "Test")].copy()
    if test_df.empty:
        print("No Test patients were available.")
        return None

    # Fit the regression model and reference coordinates from Calibration 1 normals,
    # then apply the already frozen thresholds to Test patients.
    cal1_features = calculate_panel_features(df, reference_df, panel, include_test=False)
    role = cal1_features[DATA_TYPE_COL].astype(str).str.strip().str.casefold()
    cohort = cal1_features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal1_normals = cal1_features[
        role.eq("regression fit calibration 1") & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()
    model = LinearRegression().fit(
        cal1_normals[REGRESSION_PREDICTORS], cal1_normals["Avg_Death_0D"]
    )
    test_features = []
    for idx, patient in test_df.iterrows():
        rec = {
            "Source_Row_Index": idx,
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
            COHORT_COL: patient[COHORT_COL],
            STAGE_COL: patient.get(STAGE_COL, np.nan),
            DATA_TYPE_COL: patient[DATA_TYPE_COL],
        }
        rec.update(calculate_tda_features(reference_df, patient, panel))
        test_features.append(rec)
    test_features = pd.DataFrame(test_features)
    test_features["Predicted Avg_Death_0D"] = model.predict(test_features[REGRESSION_PREDICTORS])
    test_features["Difference"] = (
        test_features["Avg_Death_0D"] - test_features["Predicted Avg_Death_0D"]
    )
    denom = test_features["Avg_Death_0D"].abs().replace(0, np.nan)
    test_features["Regression Difference Percentage"] = (
        test_features["Difference"].abs() / denom
    ) * 100.0
    ref_birth = float(cal1_normals["Average Birth (1D)"].mean())
    ref_death = float(cal1_normals["Average Death (1D)"].mean())
    test_features["Distance %(1)"] = np.sqrt(
        (test_features["Average Birth (1D)"] - ref_birth) ** 2
        + (test_features["Average Death (1D)"] - ref_death) ** 2
    ) * 100.0
    reg_t = thresholds["regression_threshold"]
    dist_t = thresholds["distance_threshold"]
    positive = (
        (test_features["Regression Difference Percentage"] > reg_t)
        | (test_features["Distance %(1)"] > dist_t)
    )
    test_features["Result(Cancer)"] = np.where(positive, "Positive", "Negative")
    cohort_values = test_features[COHORT_COL].astype(str).str.strip().str.casefold()
    normal_mask = cohort_values.eq(NORMAL_COHORT.casefold())
    normal_pred = positive[normal_mask]
    specificity = float((~normal_pred).mean()) if normal_mask.any() else float("nan")
    sens_values = [
        float(positive[cohort_values.eq(cancer)].mean())
        for cancer in CANCER_TYPES
        if cohort_values.eq(cancer).any()
    ]
    macro_sens = float(np.mean(sens_values)) if sens_values else float("nan")
    print("\nFINAL UNTOUCHED TEST-SET PERFORMANCE (reported only after panel selection)")
    print(f"Normal specificity: {specificity:.4%}")
    print(f"Macro-sensitivity: {macro_sens:.4%}")
    return {"specificity": specificity, "macro_sensitivity": macro_sens,
            "n_test": int(len(test_features))}


def Optimized_Biomarker_Selection(df, split_number):
    """Run the original staged selection algorithm on one already-created random split."""
    global CANCER_TYPES
    output_folder_path.mkdir(parents=True, exist_ok=True)
    validate_split_fractions()

    bound_values = (K1_MIN, K1_MAX, K2_MIN, K2_MAX)
    bounds_provided = [value is not None for value in bound_values]
    if any(bounds_provided) and not all(bounds_provided):
        raise ValueError(
            "Provide all four bounds (K1_MIN, K1_MAX, K2_MIN, K2_MAX), or leave all four as None."
        )
    if all(bounds_provided):
        for name, value in zip(("K1_MIN", "K1_MAX", "K2_MIN", "K2_MAX"), bound_values):
            if not isinstance(value, (int, float)) or not np.isfinite(value):
                raise ValueError(f"{name} must be a finite number or None when bounds are disabled.")
        if K1_MIN > K1_MAX:
            raise ValueError(f"K1_MIN ({K1_MIN}) must be <= K1_MAX ({K1_MAX}).")
        if K2_MIN > K2_MAX:
            raise ValueError(f"K2_MIN ({K2_MIN}) must be <= K2_MAX ({K2_MAX}).")
    if not isinstance(MAX_THRESHOLD_CANDIDATES, int) or MAX_THRESHOLD_CANDIDATES < 2:
        raise ValueError("MAX_THRESHOLD_CANDIDATES must be an integer >= 2.")
    if len(BIOMARKERS) != len(set(BIOMARKERS)):
        raise ValueError("BIOMARKERS must contain unique names.")
    if PANEL_SIZE < 1 or PANEL_SIZE > len(BIOMARKERS):
        raise ValueError("PANEL_SIZE must be between 1 and the number of BIOMARKERS.")
    if sum(PHASE_BUDGETS.values()) != TOTAL_PANEL_BUDGET:
        raise ValueError("The phase budgets must sum exactly to TOTAL_PANEL_BUDGET.")

    required = [COHORT_COL, PATIENT_ID_COL, STAGE_COL, DATA_TYPE_COL] + BIOMARKERS
    missing = [column for column in required if column not in df.columns]
    if missing:
        raise ValueError(f"Split {split_number}: missing required columns: {missing}")
    df = df.copy()
    observed_cohorts = list(dict.fromkeys(
        df[COHORT_COL].astype(str).str.strip().str.casefold().tolist()
    ))
    CANCER_TYPES = [
        name for name in observed_cohorts
        if name and name != NORMAL_COHORT.casefold()
    ]
    if not CANCER_TYPES:
        raise ValueError(f"Split {split_number}: no cancer cohorts found.")
    reference_df = df[
        get_role_mask(df, "Reference")
        & df[COHORT_COL].astype(str).str.strip().str.casefold().eq(NORMAL_COHORT.casefold())
    ].copy()
    if len(reference_df) != N_REFERENCE_NORMAL:
        raise ValueError(
            f"Split {split_number}: expected {N_REFERENCE_NORMAL} reference normals; found {len(reference_df)}."
        )

    print(f"\n{'=' * 78}")
    print(f"RANDOM SPLIT {split_number}/{N_REPETITIONS}")
    print("=" * 78)

    rng = random.Random(RANDOM_SEED)
    cache = {}
    used = set()
    all_rows = []
    stage_by_panel = {}

    def evaluate_phase(label, panels, budget):
        if len(panels) != budget:
            raise RuntimeError(f"{label}: expected {budget} panels, got {len(panels)}")
        phase_rows = []
        for i, panel in enumerate(panels, start=1):
            summary = score_panel(df, reference_df, panel, cache)
            phase_rows.append((panel, summary))
            all_rows.append((panel, summary))
            stage_by_panel[panel] = label

            # Print compact progress only every 300 panels, plus the final panel.
            completed = len(all_rows)
            if completed % 300 == 0 or completed == TOTAL_PANEL_BUDGET:
                feasible = [
                    row_summary["macro_sensitivity"]
                    for _, row_summary in all_rows
                    if row_summary["specificity"] > (1.0 - FPR) + 1e-12
                    and np.isfinite(row_summary["macro_sensitivity"])
                ]
                best_so_far = max(feasible) if feasible else float("nan")
                print(
                    f"Random search: {completed}/{TOTAL_PANEL_BUDGET} | "
                    f"Stage: {label} | "
                    f"best pooled-specificity-feasible macro-sensitivity so far: {best_so_far:.4f}"
                )
        return phase_rows

    # STAGE 1: all candidate biomarkers are treated equally at the outset. A greedy
    # balancing criterion keeps their appearance counts as even as possible,
    # while minimizing overlap between panels. No outcome-guided marker weights.
    stage1_panels = balanced_diverse_panels(
        rng, PHASE_BUDGETS["balanced_diverse_exploration"], used,
        marker_pool=BIOMARKERS, prior_counts={b: 0 for b in BIOMARKERS}
    )
    used.update(stage1_panels)
    evaluate_phase("1/4 balanced diverse exploration", stage1_panels,
                   PHASE_BUDGETS["balanced_diverse_exploration"])

    # STAGE 2: rank only feasible panels by macro-sensitivity, then estimate marker
    # recurrence from the top feasible fraction and use it to guide new combinations.
    weights = biomarker_weights(all_rows)
    stage2_panels = make_unique_weighted_panels(
        rng, PHASE_BUDGETS["sensitivity_guided_search"], used, weights
    )
    evaluate_phase("2/4 sensitivity-guided search", stage2_panels,
                   PHASE_BUDGETS["sensitivity_guided_search"])

    # STAGE 3: identify recurring biomarkers in the top feasible panels from Stages
    # 1–2, then repeat balanced/low-overlap exploration within the recurrence-informed
    # pool. This keeps the marker frequencies balanced within that selected pool.
    top_before_stage3 = top_feasible_rows(all_rows)
    recurrence_counts = {b: 0 for b in BIOMARKERS}
    for panel, _summary in top_before_stage3:
        for b in panel:
            recurrence_counts[b] += 1
    recurring_pool = [b for b, _count in sorted(
        recurrence_counts.items(), key=lambda item: (-item[1], BIOMARKERS.index(item[0]))
    )[:RECURRENCE_POOL_SIZE]]
    if len(recurring_pool) < PANEL_SIZE:
        recurring_pool = list(BIOMARKERS)
    stage3_panels = balanced_diverse_panels(
        rng, PHASE_BUDGETS["recurrence_informed_balanced_search"], used,
        marker_pool=recurring_pool, prior_counts={b: 0 for b in recurring_pool}
    )
    evaluate_phase("3/4 recurrence-informed balanced search", stage3_panels,
                   PHASE_BUDGETS["recurrence_informed_balanced_search"])

    # STAGE 4: one-biomarker swaps around the strongest panels from all previous
    # stages, with random unique panels as a fallback. The final selection below
    # still compares every feasible panel from all four stages.
    stage4_panels = make_neighborhood_panels(
        rng, PHASE_BUDGETS["best_panel_neighborhoods"], used, all_rows
    )
    evaluate_phase("4/4 best-panel neighborhood search", stage4_panels,
                   PHASE_BUDGETS["best_panel_neighborhoods"])

    if len(used) != TOTAL_PANEL_BUDGET or len(all_rows) != TOTAL_PANEL_BUDGET:
        raise RuntimeError(
            f"Search budget error: evaluated {len(all_rows)} panels, expected {TOTAL_PANEL_BUDGET}."
        )
    if len(set(panel for panel, _ in all_rows)) != TOTAL_PANEL_BUDGET:
        raise RuntimeError("Duplicate panels were evaluated; every panel must be unique.")

    # IMPORTANT: the winning panel is selected across all four stages, not only Stage 4.
    specificity_feasible_rows = [
        item for item in all_rows
        if item[1]["specificity"] > (1.0 - FPR) + 1e-12
        and np.isfinite(item[1]["macro_sensitivity"])
    ]
    finite_rows = [
        item for item in all_rows
        if np.isfinite(item[1]["macro_sensitivity"])
    ]
    if not finite_rows:
        raise RuntimeError("All evaluated panels had undefined macro-sensitivity.")
    # Detection-based panel selection:
    # 1. Keep only panels whose pooled Calibration 1 + Calibration 2 normal specificity is strictly above 1 - FPR.
    # 2. Among those feasible panels, select the panel with the highest
    #    macro-sensitivity across the cancer cohorts.
    # 3. Do not require macro-sensitivity itself to exceed 99%; it is the
    #    optimization objective, while specificity is the hard constraint.
    if specificity_feasible_rows:
        ranked_rows = sorted(
            specificity_feasible_rows,
            key=lambda item: (
                item[1]["macro_sensitivity"],
                item[1]["overall_cancer_sensitivity"],
                item[1]["specificity"],
            ),
            reverse=True,
        )
    else:
        # This is a diagnostic fallback only. In the unbounded search, thresholds
        # above all pooled-normal scores yield 100% specificity.
        ranked_rows = sorted(
            finite_rows,
            key=lambda item: (
                item[1]["macro_sensitivity"],
                item[1]["overall_cancer_sensitivity"],
                item[1]["specificity"],
            ),
            reverse=True,
        )
        print(
            "\nWARNING: No panel met the strict pooled Calibration 1 + Calibration 2 "
            "specificity constraint. The selected panel is diagnostic only."
        )
    best_panel, _ = ranked_rows[0]

    best_summary = ranked_rows[0][1].copy()
    best_macro_sensitivity = best_summary["macro_sensitivity"]
    print("\nBEST BIOMARKER PANEL — SELECTED ACROSS ALL FOUR STAGES")
    print(f"Winning stage: {stage_by_panel.get(best_panel, 'unknown')}")
    print(f"Final biomarker set: {list(best_panel)!r}")
    print(f"Calibration specificity: {best_summary['specificity']:.4%}")
    print(f"Calibration macro-sensitivity: {best_macro_sensitivity:.4%}")
    print(f"Meets pooled specificity target: {best_summary['specificity'] > (1.0 - FPR) + 1e-12}")

    test_summary = evaluate_selected_panel_on_test(df, reference_df, best_panel, best_summary)
    return {
        "split": split_number,
        "winning_stage": stage_by_panel.get(best_panel, "unknown"),
        "biomarker_panel": " | ".join(best_panel),
        "calibration_specificity": float(best_summary["specificity"]),
        "calibration_macro_sensitivity": float(best_macro_sensitivity),
        "regression_threshold": float(best_summary["regression_threshold"]),
        "distance_threshold": float(best_summary["distance_threshold"]),
        "test_normal_specificity": float(test_summary["specificity"]),
        "test_macro_sensitivity": float(test_summary["macro_sensitivity"]),
        "n_test": int(test_summary["n_test"]),
        "total_unique_panels_evaluated": int(len(all_rows)),
    }


def cohort_key(series):
    return series.astype(str).str.strip().str.casefold()


def create_one_split(source_df, rng, repetition):
    """Create one split, retaining the first N_REFERENCE_NORMAL normal rows as Reference."""
    df = source_df.copy()
    cohort = cohort_key(df[COHORT_COL])
    if df[COHORT_COL].isna().any() or cohort.eq("").any():
        raise ValueError("Cohort labels must not be missing or blank.")

    normal_indices = df.index[cohort.eq(NORMAL_COHORT.casefold())].tolist()
    if len(normal_indices) <= N_REFERENCE_NORMAL:
        raise ValueError(
            f"Need more than {N_REFERENCE_NORMAL} Normal patients; found {len(normal_indices)}."
        )

    role = {}
    for idx in normal_indices[:N_REFERENCE_NORMAL]:
        role[idx] = ROLE_REFERENCE

    remaining_normals = list(normal_indices[N_REFERENCE_NORMAL:])
    rng.shuffle(remaining_normals)
    n_cal1 = int(np.floor(len(remaining_normals) * CALIBRATION1_FRACTION))
    n_cal2_normal = int(np.floor(len(remaining_normals) * CALIBRATION2_FRACTION))
    for idx in remaining_normals[:n_cal1]:
        role[idx] = ROLE_CAL1
    for idx in remaining_normals[n_cal1:n_cal1 + n_cal2_normal]:
        role[idx] = ROLE_CAL2
    for idx in remaining_normals[n_cal1 + n_cal2_normal:]:
        role[idx] = ROLE_TEST

    cancer_cohorts = [
        name for name in COHORT_ORDER
        if name.casefold() != NORMAL_COHORT.casefold()
    ]
    for cancer in cancer_cohorts:
        indices = df.index[cohort.eq(cancer)].tolist()
        if not indices:
            continue
        shuffled = list(indices)
        rng.shuffle(shuffled)
        n_cal2_cancer = int(np.floor(len(shuffled) * CALIBRATION2_FRACTION))
        for idx in shuffled[:n_cal2_cancer]:
            role[idx] = ROLE_CAL2
        for idx in shuffled[n_cal2_cancer:]:
            role[idx] = ROLE_TEST

    if len(role) != len(df):
        unassigned = [idx for idx in df.index if idx not in role]
        raise ValueError(
            f"Split {repetition}: some rows were not assigned a Data Type; "
            f"check cohort labels: {sorted(set(cohort.loc[unassigned]))}"
        )

    df[DATA_TYPE_COL] = pd.Series(role)
    df[SOURCE_ROW_COL] = np.arange(len(df), dtype=int)
    df["_role_order"] = df[DATA_TYPE_COL].map(ROLE_ORDER)
    df["_cohort_order"] = cohort.map(COHORT_ORDER).fillna(999).astype(int)
    df = df.sort_values(
        ["_role_order", "_cohort_order", SOURCE_ROW_COL], kind="stable"
    ).drop(columns=["_role_order", "_cohort_order"]).reset_index(drop=True)

    if len(df) != len(source_df) or df[SOURCE_ROW_COL].duplicated().any():
        raise RuntimeError(f"Split {repetition}: source rows were lost or duplicated.")
    if set(df[SOURCE_ROW_COL]) != set(range(len(source_df))):
        raise RuntimeError(f"Split {repetition}: source-row audit failed.")
    return df


def main():
    validate_split_fractions()
    output_folder_path.mkdir(parents=True, exist_ok=True)
    SPLIT_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    if not clinical_data_path.exists():
        raise FileNotFoundError(
            f"Input workbook not found: {clinical_data_path}\\n"
            "Update clinical_data_path to the correct workbook."
        )

    all_sheets = pd.read_excel(clinical_data_path, sheet_name=None)
    if sheet_name not in all_sheets:
        raise ValueError(
            f"Sheet {sheet_name!r} not found. Available sheets: {list(all_sheets)}"
        )
    source_df = all_sheets[sheet_name].copy()
    required = [COHORT_COL, PATIENT_ID_COL, STAGE_COL] + BIOMARKERS
    missing = [column for column in required if column not in source_df.columns]
    if missing:
        raise ValueError(f"Missing required columns in source workbook: {missing}")
    if DATA_TYPE_COL in source_df.columns:
        source_df = source_df.drop(columns=[DATA_TYPE_COL])
    if SOURCE_ROW_COL in source_df.columns:
        source_df = source_df.drop(columns=[SOURCE_ROW_COL])
    source_df = source_df.reset_index(drop=True)

    cohort = cohort_key(source_df[COHORT_COL])
    if source_df[COHORT_COL].isna().any() or cohort.eq("").any():
        raise ValueError("Cohort labels must not be missing or blank.")
    observed = set(cohort.unique())
    expected = set(COHORT_ORDER)
    unexpected = observed - expected
    if unexpected:
        raise ValueError(
            f"Unexpected cohort name(s): {sorted(unexpected)}. "
            "Update COHORT_ORDER to include the intended cohorts."
        )
    missing_cohorts = expected - observed
    if missing_cohorts:
        raise ValueError(
            f"Expected cohort(s) missing from source workbook: {sorted(missing_cohorts)}."
        )
    if int(cohort.eq(NORMAL_COHORT.casefold()).sum()) <= N_REFERENCE_NORMAL:
        raise ValueError(
            f"Need more than {N_REFERENCE_NORMAL} Normal patients in the source workbook."
        )

    # Create and save all five split workbooks using one seeded RNG sequence.
    # The Reference cohort remains fixed; the other assignments vary by split.
    split_rng = np.random.RandomState(SPLIT_RANDOM_SEED)
    split_runs = []
    for repetition in range(1, N_REPETITIONS + 1):
        split_df = create_one_split(source_df, split_rng, repetition)
        split_path = SPLIT_OUTPUT_DIR / (
            f"clinical cancer validation data_RandomSplit_{repetition:02d}.xlsx"
        )
        output_sheets = {name: sheet.copy() for name, sheet in all_sheets.items()}
        output_sheets[sheet_name] = split_df
        with pd.ExcelWriter(split_path, engine="openpyxl") as writer:
            for sheet_name_out, sheet_df in output_sheets.items():
                sheet_df.to_excel(writer, sheet_name=sheet_name_out, index=False)
        split_runs.append((repetition, split_df, split_path))
        print(f"Created random split {repetition}/{N_REPETITIONS}: {split_path.name}")

    # Run the configured panel budget and four-stage selection algorithm on each split.
    # Existing Data Type assignments are used directly; patients are not re-split.
    results = []
    for repetition, split_df, _split_path in split_runs:
        result = Optimized_Biomarker_Selection(split_df, repetition)
        results.append(result)

    results_df = pd.DataFrame(results)
    results_df.to_excel(COMBINED_RESULTS_PATH, index=False)
    print("\n" + "=" * 78)
    print("ALL FIVE RANDOM SPLITS COMPLETED")
    print(f"Combined results: {COMBINED_RESULTS_PATH}")
    print(f"Split workbooks: {SPLIT_OUTPUT_DIR}")
    print("=" * 78)


if __name__ == "__main__":
    main()
```

```python
# Stability of Model Performance Across 100 Random Data Partitions
```

```python
"""
Repeated random-split evaluation of the TDAcancer_detect detection rule.

Purpose
-------
Generate reproducible patient-level random splits in memory, evaluate the
predefined TDA-based detection rule on each Test partition, and summarize
performance across successful repetitions. No separate split workbooks or
CSV files are created. The script writes a main Excel workbook containing
split-level results and summary statistics, plus a separate workbook with
the compact mean-performance table.

Patient partitioning
--------------------
- The first N_REFERENCE_NORMAL patients in the Normal cohort, in source-row
  order, form the fixed Reference group in every repetition.
- The remaining Normal patients are shuffled and allocated to Regression Fit
  Calibration 1, Biomarker-Panel Calibration and Selection, and Test according
  to CALIBRATION1_FRACTION and CALIBRATION2_FRACTION; Test receives the remainder.
- Patients in each cancer cohort listed in COHORT_ORDER are shuffled and
  allocated to Calibration 2 and Test according to CALIBRATION2_FRACTION.
- Cancer patients are not assigned to Calibration 1.

Detection procedure
-------------------
For each repetition, the fixed Reference group is used to construct TDA
features for Calibration 1 Normal patients and Test patients. A linear
regression model is fitted using Calibration 1 Normal patients only. The
predefined thresholds K1 and K2 are applied to Test without recalibration,
using the OR decision rule:

    Positive if Regression Difference Percentage > K1
    OR Distance %(1) > K2.

Performance is reported for overall cancer sensitivity, normal specificity,
and cancer sensitivity stratified by AJCC stage.

Across-repetition summaries
---------------------------
The summary reports descriptive means, standard deviations, medians, minima,
maxima, and 95% t-based confidence intervals for split-level performance.
It also reports mean split-level numerators and denominators and the mean
within-split Wilson confidence-interval limits. These are summaries across
randomized repetitions, not pooled patient-level estimates.

Configuration
-------------
Update the input/output paths, biomarker names, cohort labels, split settings,
and fixed thresholds as required for the dataset and analysis.
"""
from pathlib import Path
import numpy as np
import pandas as pd
import gudhi
from scipy.stats import t
from sklearn.linear_model import LinearRegression


# =============================================================================
# USER-DEFINED SETTINGS
# =============================================================================
INPUT_FILE = Path(r"D:\Cancer Detection folder\Biomarker Selection\clinical cancer validation data.xlsx")
SHEET_NAME = "Normal and Cancer"
OUTPUT_FILE = Path(r"D:\Cancer Detection folder\Biomarker Selection\TDAcancer_detect_Random_Split_Performance.xlsx")
FINAL_MEAN_OUTPUT_FILE = OUTPUT_FILE.parent / "TDAcancer_detect_Final_Mean_Performance.xlsx"

N_REPETITIONS = 100
RANDOM_SEED = 20261022
N_REFERENCE_NORMAL = 59
CALIBRATION1_FRACTION = 0.15
CALIBRATION2_FRACTION = 0.50

BIOMARKERS = ['sIL-6Ra', 'CA-125', 'CA 19-9', 'IL-6', 'IL-8', 'OPN', 'TRAIL', 'Total PSA']
K1, K2 = 2.04457, 5.25692

COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
STAGE_COL = "AJCC Stage"
DATA_TYPE_COL = "Data Type"
SOURCE_ROW_COL = "Original_Source_Row_Zero_Based"
NORMAL_COHORT = "normal"
COHORT_ORDER = {"normal": 0, "bladder": 1, "ovarian": 2, "pancreatic": 3}
ROLE_REFERENCE = "Reference"
ROLE_CAL1 = "Regression Fit Calibration 1"
ROLE_CAL2 = "Biomarker-Panel Calibration and Selection"
ROLE_TEST = "Test"
ROLE_ORDER = {ROLE_REFERENCE: 0, ROLE_CAL1: 1, ROLE_CAL2: 2, ROLE_TEST: 3}


# =============================================================================
# VALIDATION AND RANDOM SPLITS
# =============================================================================
def cohort_key(series):
    return series.astype(str).str.strip().str.casefold()


def validate_settings():
    if not isinstance(N_REPETITIONS, int) or N_REPETITIONS < 1:
        raise ValueError("N_REPETITIONS must be a positive integer.")
    for name, value in (("CALIBRATION1_FRACTION", CALIBRATION1_FRACTION),
                        ("CALIBRATION2_FRACTION", CALIBRATION2_FRACTION)):
        if not isinstance(value, (int, float)) or not np.isfinite(value) or not 0 <= value <= 1:
            raise ValueError(f"{name} must be a finite number between 0 and 1.")
    if CALIBRATION1_FRACTION + CALIBRATION2_FRACTION >= 1:
        raise ValueError("Calibration 1 + Calibration 2 fractions must be less than 1.")


def create_one_split(source_df, rng, repetition):
    """Create one reproducible split while retaining every source patient exactly once."""
    df = source_df.copy().reset_index(drop=True)
    df[SOURCE_ROW_COL] = np.arange(len(df), dtype=int)
    cohort = cohort_key(df[COHORT_COL])
    normal_indices = df.index[cohort.eq(NORMAL_COHORT)].tolist()
    if len(normal_indices) <= N_REFERENCE_NORMAL:
        raise ValueError(f"Need more than {N_REFERENCE_NORMAL} Normal patients; found {len(normal_indices)}.")

    roles = {}
    for idx in normal_indices[:N_REFERENCE_NORMAL]:
        roles[idx] = ROLE_REFERENCE

    remaining_normals = normal_indices[N_REFERENCE_NORMAL:].copy()
    rng.shuffle(remaining_normals)
    n_cal1 = int(np.floor(len(remaining_normals) * CALIBRATION1_FRACTION))
    n_cal2_normal = int(np.floor(len(remaining_normals) * CALIBRATION2_FRACTION))
    for idx in remaining_normals[:n_cal1]:
        roles[idx] = ROLE_CAL1
    for idx in remaining_normals[n_cal1:n_cal1 + n_cal2_normal]:
        roles[idx] = ROLE_CAL2
    for idx in remaining_normals[n_cal1 + n_cal2_normal:]:
        roles[idx] = ROLE_TEST

    for cancer in [c for c in COHORT_ORDER if c != NORMAL_COHORT]:
        indices = df.index[cohort.eq(cancer)].tolist()
        if not indices:
            raise ValueError(f"No rows found for cancer cohort {cancer!r}; check COHORT_ORDER.")
        shuffled = indices.copy()
        rng.shuffle(shuffled)
        n_cal2 = int(np.floor(len(shuffled) * CALIBRATION2_FRACTION))
        for idx in shuffled[:n_cal2]:
            roles[idx] = ROLE_CAL2
        for idx in shuffled[n_cal2:]:
            roles[idx] = ROLE_TEST

    missing = [idx for idx in df.index if idx not in roles]
    if missing:
        unknown = sorted(set(cohort.loc[missing].tolist()))
        raise ValueError(f"Some rows were not assigned a role. Unrecognized cohorts: {unknown}")

    df[DATA_TYPE_COL] = pd.Series(roles)
    df["_role_order"] = df[DATA_TYPE_COL].map(ROLE_ORDER)
    df["_cohort_order"] = cohort.map(COHORT_ORDER).fillna(999).astype(int)
    df = df.sort_values(["_role_order", "_cohort_order", SOURCE_ROW_COL], kind="stable")
    df = df.drop(columns=["_role_order", "_cohort_order"]).reset_index(drop=True)

    if len(df) != len(source_df) or df[SOURCE_ROW_COL].duplicated().any() or set(df[SOURCE_ROW_COL]) != set(range(len(source_df))):
        raise RuntimeError(f"Split {repetition}: source rows are missing or duplicated.")
    return df


# =============================================================================
# TDA FEATURE EXTRACTION AND TEST-SET PERFORMANCE
# =============================================================================
def wilson_ci(successes, total, z=1.959963984540054):
    if total <= 0:
        return np.nan, np.nan
    p = successes / total
    den = 1 + z * z / total
    centre = (p + z * z / (2 * total)) / den
    half = z * np.sqrt(p * (1 - p) / total + z * z / (4 * total * total)) / den
    return max(0.0, centre - half), min(1.0, centre + half)


def numeric_values(frame, cols, label):
    x = frame[cols].apply(pd.to_numeric, errors="coerce")
    if not np.isfinite(x.to_numpy(dtype=float)).all():
        raise ValueError(f"Missing/non-finite biomarker values in {label}.")
    return x.to_numpy(dtype=float)


def tda_features(reference_values, patient_values):
    combined = np.vstack([reference_values, patient_values.reshape(1, -1)])
    if (combined <= 0).any():
        raise ValueError("Log transform requires strictly positive biomarker values.")
    corr = np.corrcoef(np.log(combined), rowvar=False)
    if not np.isfinite(corr).all():
        raise ValueError("Non-finite correlation matrix; check constant biomarkers.")
    dissimilarity_matrix = np.clip(1 - np.abs(corr), 0.0, 1.0)
    dissimilarity_matrix = (dissimilarity_matrix + dissimilarity_matrix.T) / 2.0
    np.fill_diagonal(dissimilarity_matrix, 0.0)
    simplex_tree = gudhi.RipsComplex(distance_matrix=dissimilarity_matrix).create_simplex_tree(max_dimension=2)
    persistence = simplex_tree.persistence()
    finite = [(dim, pair) for dim, pair in persistence if np.isfinite(pair[1])]
    h0 = [pair for dim, pair in finite if dim == 0]
    h1 = [pair for dim, pair in finite if dim == 1]
    deaths0 = [death for _, death in h0]
    lives0 = [death - birth for birth, death in h0]
    births1 = [birth for birth, _ in h1]
    deaths1 = [death for _, death in h1]
    return {
        "0-Dim Hole Min Life Range": min(lives0, default=0.0),
        "0-Dim Hole Max Life Range": max(lives0, default=0.0),
        "Median_Death_0D": float(np.median(deaths0)) if deaths0 else 0.0,
        "Std_Death_0D": float(np.std(deaths0)) if deaths0 else 0.0,
        "Avg_Death_0D": float(np.mean(deaths0)) if deaths0 else 0.0,
        "Average Birth (1D)": float(np.mean(births1)) if births1 else 0.0,
        "Average Death (1D)": float(np.mean(deaths1)) if deaths1 else 0.0,
    }


def add_metric_record(rows, split_id, scope, group, metric, successes, total, **extras):
    lo, hi = wilson_ci(successes, total)
    rows.append({
        "Split": split_id, "Scope": scope, "Group": group, "Metric": metric,
        "Numerator": int(successes), "Denominator": int(total),
        "Value (%)": 100.0 * successes / total if total else np.nan,
        "Within-split Wilson 95% CI Lower (%)": 100 * lo,
        "Within-split Wilson 95% CI Upper (%)": 100 * hi,
        **extras,
    })


def evaluate_split(df, split_id):
    required = [DATA_TYPE_COL, COHORT_COL, STAGE_COL, PATIENT_ID_COL, *BIOMARKERS]
    missing = [col for col in required if col not in df.columns]
    if missing:
        raise ValueError(f"Missing columns: {missing}")
    df = df.copy()
    df[DATA_TYPE_COL] = df[DATA_TYPE_COL].astype(str).str.strip()
    df[COHORT_COL] = df[COHORT_COL].astype(str).str.strip()
    ckey = cohort_key(df[COHORT_COL])
    normal_indices = df.index[ckey.eq(NORMAL_COHORT)].tolist()
    if SOURCE_ROW_COL in df.columns:
        source_rows = pd.to_numeric(df[SOURCE_ROW_COL], errors="coerce")
        normal_indices.sort(key=lambda idx: source_rows.loc[idx])
    if len(normal_indices) < N_REFERENCE_NORMAL:
        raise ValueError(f"Only {len(normal_indices)} Normal patients; need {N_REFERENCE_NORMAL}.")
    reference_indices = normal_indices[:N_REFERENCE_NORMAL]
    reference_values = numeric_values(df.loc[reference_indices], BIOMARKERS, "fixed Reference group")

    cal1 = df[df[DATA_TYPE_COL].eq(ROLE_CAL1) & ckey.eq(NORMAL_COHORT) & ~df.index.isin(reference_indices)]
    test = df[df[DATA_TYPE_COL].eq(ROLE_TEST)].copy()
    if len(cal1) < 2:
        raise ValueError(f"Need at least 2 Calibration 1 Normal patients; found {len(cal1)}.")
    if test.empty:
        raise ValueError("No rows labelled Test.")

    # Calculate TDA features only for Calibration 1 normal patients and Test patients.
    feature_map = {}
    for idx in list(cal1.index) + list(test.index):
        values = numeric_values(df.loc[[idx]], BIOMARKERS, f"split {split_id}, row {idx}")[0]
        feature_map[idx] = tda_features(reference_values, values)
    cal1_features = pd.DataFrame.from_dict(feature_map, orient="index").loc[cal1.index]
    test_features = pd.DataFrame.from_dict(feature_map, orient="index").loc[test.index]

    predictors = ["0-Dim Hole Min Life Range", "0-Dim Hole Max Life Range", "Median_Death_0D", "Std_Death_0D"]
    regression = LinearRegression().fit(cal1_features[predictors], cal1_features["Avg_Death_0D"])
    ev = pd.concat([test[[COHORT_COL, STAGE_COL, PATIENT_ID_COL]], test_features], axis=1)
    ev["Predicted Avg_Death_0D"] = regression.predict(ev[predictors])
    ev["Difference"] = ev["Avg_Death_0D"] - ev["Predicted Avg_Death_0D"]
    denominator = ev["Avg_Death_0D"].abs().replace(0, np.nan)
    ev["Regression Difference Percentage"] = ev["Difference"].abs() / denominator * 100
    reference_birth = cal1_features["Average Birth (1D)"].mean()
    reference_death = cal1_features["Average Death (1D)"].mean()
    ev["Distance %(1)"] = np.sqrt(
        (ev["Average Birth (1D)"] - reference_birth) ** 2
        + (ev["Average Death (1D)"] - reference_death) ** 2
    ) * 100

    predicted_cancer = (
        (ev["Regression Difference Percentage"] > K1)
        | (ev["Distance %(1)"] > K2)
    )
    actual_cancer = ~ev[COHORT_COL].astype(str).str.casefold().eq(NORMAL_COHORT)
    actual_normal = ~actual_cancer
    cancer_tp = int((predicted_cancer & actual_cancer).sum())
    cancer_n = int(actual_cancer.sum())
    normal_tn = int((~predicted_cancer & actual_normal).sum())
    normal_n = int(actual_normal.sum())
    test_n = len(ev)

    rows = []
    common = {"Test N": test_n, "Cancer N": cancer_n, "Normal N": normal_n,
              "Cancer TP": cancer_tp, "Normal TN": normal_tn}
    add_metric_record(rows, split_id, "Overall", "All cancer types", "Sensitivity", cancer_tp, cancer_n, **common)
    add_metric_record(rows, split_id, "Overall", "Normal patients", "Specificity", normal_tn, normal_n, **common)

    stages = ev[STAGE_COL].fillna("Unknown").astype(str).str.strip()
    for stage in sorted(stages[actual_cancer].unique()):
        mask = stages.eq(stage) & actual_cancer
        stage_n = int(mask.sum())
        stage_tp = int(predicted_cancer[mask].sum())
        add_metric_record(rows, split_id, "Stage-wise", f"All cancer types | Stage {stage}",
                          "Sensitivity", stage_tp, stage_n, **common)
    return rows


# =============================================================================
# SUMMARY ACROSS RANDOMIZED SPLITS
# =============================================================================
def summarize_splits(each):
    rows = []
    for (scope, group, metric), frame in each.groupby(["Scope", "Group", "Metric"], dropna=False, sort=True):
        vals = pd.to_numeric(frame["Value (%)"], errors="coerce").dropna()
        # Summarize split-level numerators and denominators descriptively. Each
        # repetition is a separate randomized split; these are not pooled counts.
        nums = pd.to_numeric(frame["Numerator"], errors="coerce").dropna()
        dens = pd.to_numeric(frame["Denominator"], errors="coerce").dropna()
        n = len(vals)
        mean = vals.mean() if n else np.nan
        sd = vals.std(ddof=1) if n > 1 else np.nan
        half = t.ppf(0.975, n - 1) * sd / np.sqrt(n) if n > 1 else np.nan
        rows.append({
            "Scope": scope, "Group": group, "Metric": metric,
            "Number of successful splits": int(frame["Split"].nunique()),
            "Mean numerator": nums.mean() if len(nums) else np.nan,
            "SD numerator": nums.std(ddof=1) if len(nums) > 1 else np.nan,
            "Mean denominator": dens.mean() if len(dens) else np.nan,
            "SD denominator": dens.std(ddof=1) if len(dens) > 1 else np.nan,
            "Mean performance (%)": mean, "SD performance (%)": sd,
            "Median performance (%)": vals.median() if n else np.nan,
            "Minimum performance (%)": vals.min() if n else np.nan,
            "Maximum performance (%)": vals.max() if n else np.nan,
            "Across-split 95% t-CI Lower (%)": mean - half if n > 1 else np.nan,
            "Across-split 95% t-CI Upper (%)": mean + half if n > 1 else np.nan,
            "Mean within-split Wilson CI Lower (%)": pd.to_numeric(frame["Within-split Wilson 95% CI Lower (%)"], errors="coerce").mean(),
            "Mean within-split Wilson CI Upper (%)": pd.to_numeric(frame["Within-split Wilson 95% CI Upper (%)"], errors="coerce").mean(),
        })
    return pd.DataFrame(rows)


def main():
    validate_settings()
    if not INPUT_FILE.exists():
        raise FileNotFoundError(f"Input workbook not found: {INPUT_FILE}\nUpdate INPUT_FILE in the settings.")
    sheets = pd.read_excel(INPUT_FILE, sheet_name=None)
    if SHEET_NAME not in sheets:
        raise ValueError(f"Sheet {SHEET_NAME!r} not found. Available sheets: {list(sheets)}")
    source_df = sheets[SHEET_NAME].copy()
    required = [COHORT_COL, PATIENT_ID_COL, STAGE_COL, *BIOMARKERS]
    missing = [col for col in required if col not in source_df.columns]
    if missing:
        raise ValueError(f"Missing required columns in source sheet: {missing}")
    if DATA_TYPE_COL in source_df.columns:
        source_df = source_df.drop(columns=[DATA_TYPE_COL])
    if SOURCE_ROW_COL in source_df.columns:
        source_df = source_df.drop(columns=[SOURCE_ROW_COL])
    source_df = source_df.reset_index(drop=True)

    observed = set(cohort_key(source_df[COHORT_COL]).unique())
    expected = set(COHORT_ORDER)
    unexpected = observed - expected
    if unexpected:
        raise ValueError(f"Unexpected cohort names: {sorted(unexpected)}. Update COHORT_ORDER if intended.")
    n_normal = int(cohort_key(source_df[COHORT_COL]).eq(NORMAL_COHORT).sum())
    if n_normal <= N_REFERENCE_NORMAL:
        raise ValueError(f"Need more than {N_REFERENCE_NORMAL} Normal patients; found {n_normal}.")

    rng = np.random.RandomState(RANDOM_SEED)
    all_records, failures = [], []
    for split_id in range(1, N_REPETITIONS + 1):
        try:
            split_df = create_one_split(source_df, rng, split_id)
            records = evaluate_split(split_df, split_id)
            all_records.extend(records)
            print(f"\nSplit {split_id}/{N_REPETITIONS}")
            for row in records:
                print(f"  {row['Scope']} | {row['Group']} | {row['Metric']}: "
                      f"{row['Numerator']}/{row['Denominator']} = {row['Value (%)']:.2f}%")
        except Exception as exc:
            failures.append({"Split": split_id, "Error": f"{type(exc).__name__}: {exc}"})
            print(f"\nSplit {split_id}/{N_REPETITIONS} FAILED: {type(exc).__name__}: {exc}")

    if not all_records:
        raise RuntimeError("No splits were evaluated successfully. Check the input data and error messages.")
    each = pd.DataFrame(all_records)
    summary = summarize_splits(each)
    split_status = pd.DataFrame({
        "Requested splits": [N_REPETITIONS],
        "Successful splits": [int(each["Split"].nunique())],
        "Failed splits": [len(failures)],
        "Random seed": [RANDOM_SEED],
        "Fixed reference normals": [N_REFERENCE_NORMAL],
        "Normal Calibration 1 fraction (after Reference)": [CALIBRATION1_FRACTION],
        "Normal Calibration 2 fraction (after Reference)": [CALIBRATION2_FRACTION],
        "Cancer Calibration 2 fraction (within each cancer cohort)": [CALIBRATION2_FRACTION],
        "Threshold k1": [K1], "Threshold k2": [K2],
    })
    failures_df = pd.DataFrame(failures, columns=["Split", "Error"])

    OUTPUT_FILE.parent.mkdir(parents=True, exist_ok=True)

    # Main workbook: split-level results, across-split summaries, run settings,
# and failure details when one or more repetitions could not be evaluated.
    with pd.ExcelWriter(OUTPUT_FILE, engine="openpyxl") as writer:
        each.to_excel(writer, sheet_name="Each Split Performance", index=False)
        summary.to_excel(writer, sheet_name="Mean Performance Summary", index=False)
        split_status.to_excel(writer, sheet_name="Run Settings", index=False)
        if not failures_df.empty:
            failures_df.to_excel(writer, sheet_name="Failed Splits", index=False)

    # Separate results workbook containing the compact mean-performance table.
    display_cols = ["Scope", "Group", "Metric", "Number of successful splits",
                    "Mean numerator", "Mean denominator", "Mean performance (%)",
                    "SD performance (%)", "Across-split 95% t-CI Lower (%)",
                    "Across-split 95% t-CI Upper (%)"]
    final_mean_results = summary[display_cols].copy()
    with pd.ExcelWriter(FINAL_MEAN_OUTPUT_FILE, engine="openpyxl") as writer:
        final_mean_results.to_excel(writer, sheet_name="Final Mean Performance", index=False)
        split_status.to_excel(writer, sheet_name="Run Settings", index=False)
        if not failures_df.empty:
            failures_df.to_excel(writer, sheet_name="Failed Splits", index=False)

    print("\nFINAL MEAN PERFORMANCE ACROSS SUCCESSFUL SPLITS")
    print(final_mean_results.to_string(index=False, float_format=lambda x: f"{x:.2f}"))
    print(f"\nSaved main performance workbook: {OUTPUT_FILE}")
    print(f"Saved separate final mean-performance workbook: {FINAL_MEAN_OUTPUT_FILE}")
    print(f"Successful splits: {each['Split'].nunique()}/{N_REPETITIONS}; failed splits: {len(failures)}")


if __name__ == "__main__":
    main()
```

```python
"""
Constrained random-search framework for TDA-based biomarker-panel selection.

Overview
--------
For each panel size listed in PANEL_SIZES, the script evaluates exactly
TOTAL_PANEL_BUDGET distinct biomarker panels using a four-stage search strategy.
The source workbook is loaded and partitioned in memory; no revised workbook
is written to disk.

Data partitioning
-----------------
1. The first N_REFERENCE_NORMAL normal-cohort rows, in source-row order, form
   the fixed Reference cohort.
2. The remaining normal patients are shuffled and allocated to:
   - Regression Fit Calibration 1: CALIBRATION1_FRACTION;
   - Biomarker-Panel Calibration and Selection: CALIBRATION2_FRACTION;
   - Test: all remaining normal patients.
3. Patients in each detected cancer cohort are shuffled and allocated to
   Biomarker-Panel Calibration and Selection and Test using
   CALIBRATION2_FRACTION. Cancer patients are not assigned to Calibration 1.

Panel search and calibration
----------------------------
1. Stage 1 performs balanced biomarker exploration while discouraging overlap
   between panels.
2. Stage 2 uses biomarker frequencies in the strongest feasible panels to
   guide weighted panel sampling.
3. Stage 3 performs balanced exploration within a pool of biomarkers recurring
   in the strongest feasible panels from the preceding stages.
4. Stage 4 explores one-biomarker substitutions around high-performing panels.
5. The final panel is selected from the results of all four stages using
   Calibration 1 and Calibration 2 results only.

For each candidate panel, TDA features are computed from a Vietoris-Rips
filtration constructed using the correlation-derived dissimilarity matrix
D = 1 - |C|. A linear regression model and the reference coordinates used by
Distance %(1) are fitted from Calibration 1 normal patients only. Candidate
thresholds for Regression Difference Percentage and Distance %(1) are
evaluated using the OR decision rule. The threshold search uses observed score
boundaries and their next representable floating-point values, with the
candidate grid capped by MAX_THRESHOLD_CANDIDATES for each score. Optional
user-defined bounds can further restrict the candidate ranges.

Threshold selection requires pooled Calibration 1 and Calibration 2 normal
specificity to be strictly greater than 1 - FPR. Among panels satisfying this
constraint, the primary selection objective is macro-sensitivity across the
cancer cohorts represented in Calibration 2; overall sensitivity and
specificity are secondary tie-breakers. No TDA features, predictions, or
performance metrics are calculated for the Test partition.

Configuration notes
-------------------
- Update the input path, sheet name, biomarker names, and column names for the
  dataset being analysed.
- Cancer cohorts are detected from the Cohort column; NORMAL_COHORT identifies
  the healthy cohort.
- PANEL_SIZES controls the separate panel-size runs. The same seeded data
  partition is recreated for each run.
- The logarithmic transformation requires finite, strictly positive values
  for every selected biomarker in the reference and evaluated patients.
"""

from pathlib import Path
import math
import random

import gudhi
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression


# =============================================================================
# USER-DEFINED SETTINGS
# Change these values when applying the function to another dataset.
# Algorithmic tuning parameters below this section retain their defaults.
# =============================================================================

# Input workbook
clinical_data_path = Path(
    r"D:\Cancer Detection folder\Biomarker Selection\clinical cancer validation data.xlsx"
)
sheet_name = "Normal and Cancer"

# Dataset-specific biomarker universe (names must match workbook column names).
BIOMARKERS = [
    "Tenascin C", "sAXL", "sE-selectin", "sHGFR/c-Met", "sHer2", "sHer3",
    "sIL-6Ra", "sNeuropilin-1", "sPECAM-1", "sVEGFR1", "sVEGFR3",
    "sc-kit/SCFR", "CA-125", "CA 15-3", "CA 19-9", "CEA", "FGF2", "HE4",
    "HGF", "IL-6", "IL-8", "Leptin", "MIF", "OPN", "Prolactin", "SCF",
    "TNFa", "TRAIL", "Total PSA", "VEGF", "b-HCG", "sFAS", "Cathepsin D",
    "FAP alpha", "Ferritin", "Galectin-3", "IGFBP3", "MIA", "MPO", "SHBG",
    "TIMP1", "TIMP2"
]

# Cohort names and their display/sort order.
# Keep the normal cohort in this mapping; all other keys are treated as cancer cohorts.
NORMAL_COHORT = "normal"
cohort_order = {"normal": 0, "bladder": 1, "ovarian": 2, "pancreatic": 3}
CANCER_TYPES = [name.casefold() for name in cohort_order if name.casefold() != NORMAL_COHORT.casefold()]

# Dataset-specific column names.
COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
STAGE_COL = "AJCC Stage"
DATA_TYPE_COL = "Data Type"

# Data split and model constraints.
# Edit the two fractions below whenever needed. They are fractions, not whole percentages:
#   0.20 = 20%, 0.30 = 30%, 0.50 = 50%.
# For NORMAL patients only, the first n rows (from REFERENCE_SAMPLE_SIZES) are Reference;
# the configured Calibration 1 and Calibration 2 fractions are applied to the remaining
# normal patients, with Test receiving the remainder.
# CANCER patients are NEVER assigned to Calibration 1: CALIBRATION2_FRACTION of each
# cancer cohort is assigned to Biomarker-Panel Calibration and Selection, and the rest
# is Test. Thus Calibration 1 contains only healthy/normal patients.
N_REFERENCE_NORMAL = 59  # User-defined fixed reference sample size
PANEL_SIZES = [5, 8, 12]  # Run one complete search for each panel size
PROGRESS_INTERVAL = 300
CALIBRATION1_FRACTION = 0.15
CALIBRATION2_FRACTION = 0.50
FPR = 0.01  # Require pooled Calibration 1 + 2 normal specificity strictly > 99.00%
SENSITIVITY_TARGET = 0.99  # Diagnostic target only; specificity is the hard constraint
PANEL_SIZE = 8
TOTAL_PANEL_BUDGET = 3000
RANDOM_SEED = 20261017

# Maximum number of candidate thresholds to test for EACH detection score:
# - One set of candidates is generated for Regression Difference Percentage (k1).
# - One set is generated for Distance %(1) (k2).
# - The threshold search evaluates up to MAX_THRESHOLD_CANDIDATES x
#   MAX_THRESHOLD_CANDIDATES candidate pairs for the OR decision rule.
# - Lower values generally run faster but explore fewer threshold combinations;
#   higher values search more combinations and may take longer.
MAX_THRESHOLD_CANDIDATES = 80

# OPTIONAL USER-DEFINED THRESHOLD SEARCH BOUNDS (inclusive)
# Leave ALL FOUR values as None to use the original unbounded candidate search.
# To restrict the search, provide all four numeric values.
# Example bounded search: K1_MIN=-1.0, K1_MAX=5.0, K2_MIN=0.0, K2_MAX=10.0
K1_MIN = None  # Regression Difference Percentage lower bound, or None
K1_MAX = None  # Regression Difference Percentage upper bound, or None
K2_MIN = None  # Distance %(1) lower bound, or None
K2_MAX = None  # Distance %(1) upper bound, or None

# Algorithm defaults: normally leave these unchanged unless tuning the search.
PHASE_WEIGHTS = {
    "balanced_diverse_exploration": 0.40,
    "sensitivity_guided_search": 0.30,
    "recurrence_informed_balanced_search": 0.20,
    "best_panel_neighborhoods": 0.10,
}
CANDIDATE_POOL_SIZE = 300
TOP_PANEL_FRACTION = 0.20
RECURRENCE_POOL_SIZE = 16
PRINT_EACH_PANEL_PERFORMANCE = False


def calculate_phase_budgets(total_budget, phase_weights):
    if not isinstance(total_budget, int) or total_budget < 1:
        raise ValueError("TOTAL_PANEL_BUDGET must be a positive integer.")
    weight_sum = sum(phase_weights.values())
    if not math.isclose(weight_sum, 1.0, rel_tol=0.0, abs_tol=1e-9):
        raise ValueError("PHASE_WEIGHTS must sum to 1.0.")
    exact = {name: total_budget * weight for name, weight in phase_weights.items()}
    budgets = {name: math.floor(value) for name, value in exact.items()}
    remaining = total_budget - sum(budgets.values())
    order = sorted(
        phase_weights,
        key=lambda name: (exact[name] - budgets[name], phase_weights[name]),
        reverse=True,
    )
    for name in order[:remaining]:
        budgets[name] += 1
    return budgets


PHASE_BUDGETS = calculate_phase_budgets(TOTAL_PANEL_BUDGET, PHASE_WEIGHTS)

# Features used by the regression model and the two detection scores.
# The TDA feature dictionary is defined in calculate_tda_features().


# -------------------- DATA LOADING AND PARTITIONING --------------------


def validate_split_fractions():
    """Validate configurable calibration fractions; Test is the remainder."""
    for name, value in (("CALIBRATION1_FRACTION", CALIBRATION1_FRACTION),
                        ("CALIBRATION2_FRACTION", CALIBRATION2_FRACTION)):
        if not isinstance(value, (int, float)) or not np.isfinite(value):
            raise ValueError(f"{name} must be a finite number between 0 and 1.")
        if value < 0 or value > 1:
            raise ValueError(f"{name} must be between 0 and 1; received {value}.")
    total_calibration = CALIBRATION1_FRACTION + CALIBRATION2_FRACTION
    if total_calibration >= 1.0:
        raise ValueError(
            "CALIBRATION1_FRACTION + CALIBRATION2_FRACTION must be less than 1.0 "
            "so that Test has a positive fraction. "
            f"Received {CALIBRATION1_FRACTION} + {CALIBRATION2_FRACTION}."
        )


def create_split_dataframe(source_path, sheet, n_reference_normal):
    """Load the source sheet and assign mutually exclusive roles in memory."""
    global CANCER_TYPES
    validate_split_fractions()
    source_path = Path(source_path)
    if not source_path.exists():
        raise FileNotFoundError(f"Source workbook not found: {source_path}\nChange clinical_data_path.")
    df = pd.read_excel(source_path, sheet_name=sheet).copy()
    required = [COHORT_COL, PATIENT_ID_COL, STAGE_COL] + BIOMARKERS
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns in source workbook: {missing}")
    if DATA_TYPE_COL in df.columns:
        df = df.drop(columns=[DATA_TYPE_COL])

    cohort = df[COHORT_COL].astype(str).str.strip().str.casefold()
    if df[COHORT_COL].isna().any():
        raise ValueError(f"Column {COHORT_COL!r} contains missing cohort labels.")
    if cohort.eq("").any():
        raise ValueError(f"Column {COHORT_COL!r} contains blank cohort labels.")
    observed_cohorts = list(dict.fromkeys(cohort.tolist()))
    CANCER_TYPES = [name for name in observed_cohorts if name and name != NORMAL_COHORT.casefold()]
    if not CANCER_TYPES:
        raise ValueError(f"No cancer cohorts were detected in {COHORT_COL!r}.")

    normal_indices = df.index[cohort.eq(NORMAL_COHORT.casefold())].tolist()
    if len(normal_indices) <= n_reference_normal:
        raise ValueError(
            f"Need more than {n_reference_normal} normal patients; found {len(normal_indices)}."
        )

    role = {}
    reference_indices = normal_indices[:n_reference_normal]
    for idx in reference_indices:
        role[idx] = "Reference"

    rng = np.random.RandomState(RANDOM_SEED)
    remaining_normals = list(normal_indices[n_reference_normal:])
    rng.shuffle(remaining_normals)
    n1 = int(np.floor(len(remaining_normals) * CALIBRATION1_FRACTION))
    n2 = int(np.floor(len(remaining_normals) * CALIBRATION2_FRACTION))
    for idx in remaining_normals[:n1]:
        role[idx] = "Regression Fit Calibration 1"
    for idx in remaining_normals[n1:n1+n2]:
        role[idx] = "Biomarker-Panel Calibration and Selection"
    for idx in remaining_normals[n1+n2:]:
        role[idx] = "Test"

    # Keep the original cohort iteration order and RNG consumption for cancer splits.
    for cancer in CANCER_TYPES:
        indices = df.index[cohort.eq(cancer.casefold())].tolist()
        if not indices:
            raise ValueError(f"No rows found for cancer cohort {cancer!r}.")
        shuffled = list(indices)
        rng.shuffle(shuffled)
        n2c = int(np.floor(len(shuffled) * CALIBRATION2_FRACTION))
        for idx in shuffled[:n2c]:
            role[idx] = "Biomarker-Panel Calibration and Selection"
        for idx in shuffled[n2c:]:
            role[idx] = "Test"

    df[DATA_TYPE_COL] = pd.Series(role)
    role_order = {
        "Reference": 0,
        "Regression Fit Calibration 1": 1,
        "Biomarker-Panel Calibration and Selection": 2,
        "Test": 3,
    }
    df["_source_order"] = np.arange(len(df))
    df["_role_order"] = df[DATA_TYPE_COL].map(role_order)
    df["_cohort_order"] = (
        df[COHORT_COL].astype(str).str.strip().str.casefold()
        .map({str(k).casefold(): v for k, v in cohort_order.items()}).fillna(99)
    )
    df = df.sort_values(
        ["_role_order", "_cohort_order", "_source_order"], kind="stable"
    ).drop(columns=["_source_order", "_role_order", "_cohort_order"]).reset_index(drop=True)
    return df


def calculate_tda_features(reference_df, patient_row, panel):
    """Calculate only the TDA features required by regression and detection."""
    # Panels are stored as tuples for hashing/caching, but pandas column
    # selection requires a list of column names rather than a tuple key.
    panel = list(panel)

    # Convert the reference matrix once per call and append the patient's values
    # directly as a NumPy row.
    reference_values = reference_df[panel].to_numpy(dtype=float, copy=False)
    patient_values = patient_row[panel].to_numpy(dtype=float, copy=False).reshape(1, -1)
    values = np.vstack((reference_values, patient_values))
    if not np.isfinite(values).all():
        raise ValueError(
            f"Non-finite selected biomarker value for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )
    if (values <= 0).any():
        raise ValueError(
            "Log transformation requires strictly positive values. "
            f"Check selected biomarkers for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )

    logged = np.log(values)
    corr = np.corrcoef(logged, rowvar=False)
    dissimilarity_matrix = 1.0 - np.abs(corr)
    if not np.isfinite(dissimilarity_matrix).all():
        raise ValueError(
            f"Invalid correlation/dissimilarity matrix for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )

    # Enforce numerical symmetry, a zero diagonal, and nonnegative entries.
    dissimilarity_matrix = (dissimilarity_matrix + dissimilarity_matrix.T) / 2.0
    np.fill_diagonal(dissimilarity_matrix, 0.0)
    dissimilarity_matrix = np.clip(dissimilarity_matrix, 0.0, 1.0)

    # GUDHI names this input argument distance_matrix; the supplied matrix is
    # a dissimilarity matrix and is not assumed to satisfy the triangle inequality.
    rips = gudhi.RipsComplex(distance_matrix=dissimilarity_matrix)
    simplex_tree = rips.create_simplex_tree(max_dimension=2)
    persistence = simplex_tree.persistence()
    finite = [
        (dim, (birth, death))
        for dim, (birth, death) in persistence
        if np.isfinite(death)
    ]
    holes1 = [pair for dim, pair in finite if dim == 1]
    holes0 = [pair for dim, pair in finite if dim == 0]

    if holes1:
        avg_birth1 = float(np.mean([birth for birth, _ in holes1]))
        avg_death1 = float(np.mean([death for _, death in holes1]))
    else:
        avg_birth1 = avg_death1 = 0.0

    if holes0:
        deaths0 = [death for _, death in holes0]
        lifetimes0 = [death - birth for birth, death in holes0]
        min_life0 = min(lifetimes0)
        max_life0 = max(lifetimes0)
        avg_death0 = float(np.mean(deaths0))
        median_death0 = float(np.median(deaths0))
        std_death0 = float(np.std(deaths0))
    else:
        min_life0 = max_life0 = avg_death0 = 0.0
        median_death0 = std_death0 = 0.0

    return {
        "0-Dim Hole Min Life Range": min_life0,
        "0-Dim Hole Max Life Range": max_life0,
        "Median_Death_0D": median_death0,
        "Std_Death_0D": std_death0,
        "Avg_Death_0D": avg_death0,
        "Average Birth (1D)": avg_birth1,
        "Average Death (1D)": avg_death1,
    }


def get_role_mask(df, role_name):
    return df[DATA_TYPE_COL].astype(str).str.strip().str.casefold().eq(
        role_name.casefold()
    )



def calculate_panel_features(df, reference_df, panel):
    """Extract TDA features for Calibration 1 and Calibration 2 patients only."""
    allowed = (
        get_role_mask(df, "Regression Fit Calibration 1")
        | get_role_mask(df, "Biomarker-Panel Calibration and Selection")
    )
    rows = []
    for idx, patient in df.loc[allowed].iterrows():
        record = {
            "Source_Row_Index": idx,
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
            COHORT_COL: patient[COHORT_COL],
            STAGE_COL: patient.get(STAGE_COL, np.nan),
            DATA_TYPE_COL: patient[DATA_TYPE_COL],
        }
        record.update(calculate_tda_features(reference_df, patient, panel))
        rows.append(record)
    return pd.DataFrame(rows)



def evaluate_panel(df, reference_df, panel, save_features=False):
    """
    Fit regression and thresholds on Calibration 1 NORMALS only.
    Evaluate the frozen rule on the independent Calibration 2 partition.
    """
    features = calculate_panel_features(df, reference_df, panel)
    if features.empty:
        raise ValueError("No Calibration 1 / Calibration 2 patients were available.")

    role = features[DATA_TYPE_COL].astype(str).str.strip().str.casefold()
    cohort = features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal1_normals = features[
        role.eq("regression fit calibration 1")
        & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()
    cal2 = features[
        role.eq("biomarker-panel calibration and selection")
    ].copy()
    if len(cal1_normals) < 2:
        raise ValueError("At least two Calibration 1 normal patients are required.")
    if cal2.empty:
        raise ValueError("Calibration 2 is empty.")

    model = LinearRegression().fit(
        cal1_normals[REGRESSION_PREDICTORS],
        cal1_normals["Avg_Death_0D"],
    )
    features["Predicted Avg_Death_0D"] = model.predict(features[REGRESSION_PREDICTORS])
    features["Difference"] = features["Avg_Death_0D"] - features["Predicted Avg_Death_0D"]
    denom = features["Avg_Death_0D"].abs().replace(0, np.nan)
    features["Regression Difference Percentage"] = (
        features["Difference"].abs() / denom
    ) * 100.0

    ref_birth = float(cal1_normals["Average Birth (1D)"].mean())
    ref_death = float(cal1_normals["Average Death (1D)"].mean())
    features["Distance %(1)"] = np.sqrt(
        (features["Average Birth (1D)"] - ref_birth) ** 2
        + (features["Average Death (1D)"] - ref_death) ** 2
    ) * 100.0

    # IMPORTANT: cal1_normals was copied before the derived columns were added.
    # Refresh it now so it contains Regression Difference Percentage and Distance %(1).
    cal1_normals = features.loc[
        role.eq("regression fit calibration 1")
        & cohort.eq(NORMAL_COHORT.casefold())
    ].copy()

    # ------------------------------------------------------------------
    # REDUCED, VECTORIZED threshold-pair search for the OR decision rule.
    #
    # Thresholds are still optimized over candidate values, rather than set
    # directly to the pooled-normal maxima. To reduce runtime, the candidate
    # grid is deterministically downsampled to at most MAX_THRESHOLD_CANDIDATES
    # values per score. Specificity is constrained on the pooled Calibration 1
    # + Calibration 2 normal patients, strictly above the configured target.
    # ------------------------------------------------------------------
    cal2_mask = role.eq("biomarker-panel calibration and selection")
    cal2_features = features.loc[cal2_mask].copy()
    cal2_cohort = cal2_features[COHORT_COL].astype(str).str.strip().str.casefold()
    cal2_normals = cal2_features[cal2_cohort.eq(NORMAL_COHORT.casefold())]
    cal2_cancers = cal2_features[cal2_cohort.isin(CANCER_TYPES)]

    if cal2_normals.empty:
        raise ValueError("Calibration 2 contains no normal patients for specificity calibration.")
    if cal2_cancers.empty:
        raise ValueError("Calibration 2 contains no cancer patients for threshold selection.")

    cal1_reg_norm = cal1_normals["Regression Difference Percentage"].to_numpy(dtype=float)
    cal1_dist_norm = cal1_normals["Distance %(1)"].to_numpy(dtype=float)
    cal2_reg_norm = cal2_normals["Regression Difference Percentage"].to_numpy(dtype=float)
    cal2_dist_norm = cal2_normals["Distance %(1)"].to_numpy(dtype=float)
    pooled_reg_norm = np.concatenate([cal1_reg_norm, cal2_reg_norm])
    pooled_dist_norm = np.concatenate([cal1_dist_norm, cal2_dist_norm])

    cancer_reg = cal2_cancers["Regression Difference Percentage"].to_numpy(dtype=float)
    cancer_dist = cal2_cancers["Distance %(1)"].to_numpy(dtype=float)
    cancer_cohort_values = (
        cal2_cancers[COHORT_COL].astype(str).str.strip().str.casefold().to_numpy()
    )
    specificity_target = 1.0 - FPR

    def reduced_boundary_candidates(
        normal_scores, cancer_scores, max_candidates, min_value=None, max_value=None, name="threshold"
    ):
        """Build score-boundary candidates, optionally restricting their range."""
        combined = np.concatenate([
            np.asarray(normal_scores, dtype=float),
            np.asarray(cancer_scores, dtype=float),
        ])
        combined = combined[np.isfinite(combined)]
        if combined.size == 0:
            raise ValueError("No finite scores available for threshold search.")
        if not isinstance(max_candidates, int) or max_candidates < 2:
            raise ValueError("MAX_THRESHOLD_CANDIDATES must be an integer >= 2.")

        # This is the original candidate-generation procedure.
        unique_values = np.unique(combined)
        boundaries = np.unique(np.concatenate([
            unique_values, np.nextafter(unique_values, np.inf)
        ]))

        # Apply bounds only when a complete min/max pair is provided.
        bounded = min_value is not None and max_value is not None
        if bounded:
            if not np.isfinite(min_value) or not np.isfinite(max_value):
                raise ValueError(f"{name} bounds must be finite numbers.")
            if min_value > max_value:
                raise ValueError(
                    f"{name}_MIN ({min_value}) must be <= {name}_MAX ({max_value})."
                )
            boundaries = boundaries[
                (boundaries >= float(min_value)) & (boundaries <= float(max_value))
            ]
            # Include exact bounds, even when no observed score lies there.
            boundaries = np.unique(np.concatenate([
                boundaries, np.array([float(min_value), float(max_value)])
            ]))

        if boundaries.size > max_candidates:
            # Keep the original deterministic downsampling behavior.
            indices = np.linspace(0, boundaries.size - 1, max_candidates, dtype=int)
            boundaries = boundaries[np.unique(indices)]
        return boundaries

    # Bounds are active only when all four are supplied. If all are None, this
    # exactly follows the original code's unbounded threshold candidate search.
    bound_values = (K1_MIN, K1_MAX, K2_MIN, K2_MAX)
    bounds_provided = [value is not None for value in bound_values]
    if any(bounds_provided) and not all(bounds_provided):
        raise ValueError(
            "Provide all four bounds (K1_MIN, K1_MAX, K2_MIN, K2_MAX), or leave all four as None."
        )
    USE_THRESHOLD_BOUNDS = all(bounds_provided)

    k1_candidates = reduced_boundary_candidates(
        pooled_reg_norm, cancer_reg, MAX_THRESHOLD_CANDIDATES,
        K1_MIN if USE_THRESHOLD_BOUNDS else None,
        K1_MAX if USE_THRESHOLD_BOUNDS else None, "K1"
    )
    k2_candidates = reduced_boundary_candidates(
        pooled_dist_norm, cancer_dist, MAX_THRESHOLD_CANDIDATES,
        K2_MIN if USE_THRESHOLD_BOUNDS else None,
        K2_MAX if USE_THRESHOLD_BOUNDS else None, "K2"
    )
    if USE_THRESHOLD_BOUNDS:
        print(
            f"Bounded threshold search: k1 in [{K1_MIN}, {K1_MAX}] "
            f"({len(k1_candidates)} candidates); "
            f"k2 in [{K2_MIN}, {K2_MAX}] ({len(k2_candidates)} candidates); "
            f"up to {len(k1_candidates) * len(k2_candidates):,} pairs."
        )
    # Evaluate each k1 against all k2 values in vectorized NumPy operations.
    n_normals = len(pooled_reg_norm)
    best_pair = None
    best_key = None
    best_macro_sensitivity = -1.0
    best_overall_sensitivity = -1.0
    best_specificity = -1.0
    best_fp = None
    best_tn = None

    cancer_reg_pos = cancer_reg[:, None] > k1_candidates[None, :]
    cancer_dist_pos = cancer_dist[:, None] > k2_candidates[None, :]
    active_cal2_cancers = [
        cancer for cancer in CANCER_TYPES
        if np.any(cancer_cohort_values == cancer.casefold())
    ]
    per_type_masks = {
        cancer: cancer_cohort_values == cancer.casefold()
        for cancer in active_cal2_cancers
    }

    for i, k1 in enumerate(k1_candidates):
        normal_positive = (
            (pooled_reg_norm > k1)[:, None]
            | (pooled_dist_norm[:, None] > k2_candidates[None, :])
        )
        fp_counts = normal_positive.sum(axis=0)
        specificities = 1.0 - fp_counts / n_normals
        feasible = np.flatnonzero(specificities > specificity_target + 1e-12)
        if feasible.size == 0:
            continue

        cancer_positive_by_k2 = cancer_reg_pos[:, i, None] | cancer_dist_pos[:, feasible]
        overall_sens = cancer_positive_by_k2.mean(axis=0)
        type_sens = []
        for cancer in active_cal2_cancers:
            mask = per_type_masks[cancer]
            type_sens.append(cancer_positive_by_k2[mask].mean(axis=0))
        if not type_sens:
            continue
        macro_sens = np.mean(np.vstack(type_sens), axis=0)

        # Lexicographic objective matches original intent: macro sensitivity,
        # overall sensitivity, specificity, then larger thresholds.
        for j_local, j in enumerate(feasible):
            candidate_key = (
                float(macro_sens[j_local]),
                float(overall_sens[j_local]),
                float(specificities[j]),
                float(k1),
                float(k2_candidates[j]),
            )
            if best_key is None or candidate_key > best_key:
                best_key = candidate_key
                best_pair = (float(k1), float(k2_candidates[j]))
                best_macro_sensitivity = candidate_key[0]
                best_overall_sensitivity = candidate_key[1]
                best_specificity = candidate_key[2]
                best_fp = int(fp_counts[j])
                best_tn = int(n_normals - best_fp)

    if best_pair is None:
        if USE_THRESHOLD_BOUNDS:
            raise RuntimeError(
                "No k1/k2 pair within the user-defined bounds satisfied the pooled "
                f"Calibration 1 + Calibration 2 specificity requirement (> {specificity_target:.2%}). "
                f"Current bounds: k1=[{K1_MIN}, {K1_MAX}], k2=[{K2_MIN}, {K2_MAX}]. "
                "Adjust the bounds or review the specificity target."
            )
        # Preserve the original unbounded fallback: thresholds just above the
        # maximum pooled normal scores guarantee zero false positives.
        finite_reg = pooled_reg_norm[np.isfinite(pooled_reg_norm)]
        finite_dist = pooled_dist_norm[np.isfinite(pooled_dist_norm)]
        if finite_reg.size == 0 or finite_dist.size == 0:
            raise ValueError("Pooled calibration normal scores contain no finite values.")
        k1_fallback = float(np.nextafter(np.max(finite_reg), np.inf))
        k2_fallback = float(np.nextafter(np.max(finite_dist), np.inf))
        best_pair = (k1_fallback, k2_fallback)
        fallback_positive = (cancer_reg > k1_fallback) | (cancer_dist > k2_fallback)
        sens_by_type = [
            float(fallback_positive[per_type_masks[cancer]].mean())
            for cancer in CANCER_TYPES if per_type_masks[cancer].any()
        ]
        best_macro_sensitivity = (
            float(np.mean(sens_by_type)) if len(sens_by_type) == len(CANCER_TYPES) else float("nan")
        )
        best_overall_sensitivity = float(fallback_positive.mean()) if fallback_positive.size else float("nan")
        best_fp, best_tn = 0, n_normals
        best_specificity = 1.0
    best_threshold, distance_threshold = best_pair

    # Report each partition's specificity descriptively, but only the pooled
    # specificity is a threshold-selection constraint.
    cal1_positive_final = (
        (cal1_normals["Regression Difference Percentage"] > best_threshold)
        | (cal1_normals["Distance %(1)"] > distance_threshold)
    )
    best_cal1_specificity = float((~cal1_positive_final).mean())

    features["Result(Cancer)"] = np.where(
        (features["Regression Difference Percentage"] > best_threshold)
        | (features["Distance %(1)"] > distance_threshold),
        "Positive", "Negative"
    )
    cal1_normal_positive_final = (
        (cal1_normals["Regression Difference Percentage"] > best_threshold)
        | (cal1_normals["Distance %(1)"] > distance_threshold)
    )
    cal2_normal_positive = (
        (cal2_normals["Regression Difference Percentage"] > best_threshold)
        | (cal2_normals["Distance %(1)"] > distance_threshold)
    )
    combined_normal_positive_final = np.concatenate([
        cal1_normal_positive_final.to_numpy(dtype=bool),
        cal2_normal_positive.to_numpy(dtype=bool),
    ])
    tn = int((~combined_normal_positive_final).sum())
    fp = int(combined_normal_positive_final.sum())
    specificity_combined = (
        tn / (tn + fp) if (tn + fp) else float("nan")
    )
    specificity_cal2 = (
        float((~cal2_normal_positive).mean()) if len(cal2_normal_positive) else float("nan")
    )
    if not specificity_combined > specificity_target + 1e-12:
        raise RuntimeError(
            "Selected thresholds failed the pooled Calibration 1 + Calibration 2 "
            "specificity constraint."
        )

    cancer_results = {}
    sensitivities = []
    all_predictions = (
        (cal2_cancers["Regression Difference Percentage"] > best_threshold)
        | (cal2_cancers["Distance %(1)"] > distance_threshold)
    )
    for cancer in CANCER_TYPES:
        subset = cal2_cancers[
            cal2_cancers[COHORT_COL].astype(str).str.strip().str.casefold().eq(cancer)
        ]
        pred = (
            (subset["Regression Difference Percentage"] > best_threshold)
            | (subset["Distance %(1)"] > distance_threshold)
        )
        sens = float(pred.mean()) if len(subset) else float("nan")
        if len(subset):
            sensitivities.append(sens)
        cancer_results[cancer] = {
            "n": int(len(subset)),
            "sensitivity": sens,
            "true_positives": int(pred.sum()),
        }
    macro_sensitivity = float(np.mean(sensitivities)) if sensitivities else float("nan")
    overall_sensitivity = float(all_predictions.mean()) if len(cal2_cancers) else float("nan")



    return {
        "n_biomarkers": len(panel),
        "panel": list(panel),
        "regression_threshold": float(best_threshold),
        "distance_threshold": float(distance_threshold),
        "threshold_bounds_enabled": bool(USE_THRESHOLD_BOUNDS),
        "k1_min_search_bound": float(K1_MIN) if USE_THRESHOLD_BOUNDS else None,
        "k1_max_search_bound": float(K1_MAX) if USE_THRESHOLD_BOUNDS else None,
        "k2_min_search_bound": float(K2_MIN) if USE_THRESHOLD_BOUNDS else None,
        "k2_max_search_bound": float(K2_MAX) if USE_THRESHOLD_BOUNDS else None,
        "k1_candidate_count": int(len(k1_candidates)),
        "k2_candidate_count": int(len(k2_candidates)),
        "threshold_pairs_considered_max": int(len(k1_candidates) * len(k2_candidates)),
        "calibration1_specificity": float(best_cal1_specificity),
        "calibration2_specificity": float(specificity_cal2),
        "specificity": float(specificity_combined),
        "combined_calibration_specificity": float(specificity_combined),
        "false_positives": fp,
        "true_negatives": tn,
        "macro_sensitivity": macro_sensitivity,
        "overall_cancer_sensitivity": overall_sensitivity,
        "cancer_results": cancer_results,
        "regression_fit_normal_n": int(len(cal1_normals)),
        "calibration2_normal_n": int(len(cal2_normals)),
        "calibration_cancer_n": int(len(cal2_cancers)),
        "meets_sensitivity_target": bool(
            np.isfinite(macro_sensitivity) and macro_sensitivity > SENSITIVITY_TARGET
        ),
        "threshold_rule": (
            "Positive iff Regression Difference Percentage > regression_threshold "
            "OR Distance %(1) > distance_threshold"
        ),
    }, features


def canonical_panel(panel):
    return tuple(sorted(panel, key=BIOMARKERS.index))


def panel_key(panel):
    return tuple(panel)


def random_panel(rng):
    return canonical_panel(rng.sample(BIOMARKERS, PANEL_SIZE))


def balanced_diverse_panels(rng, count, used, marker_pool=None, prior_counts=None):
    """Generate unique panels while balancing marker appearances and reducing overlap.

    In Stage 1, marker_pool is all 42 biomarkers and prior_counts starts at zero,
    so all markers have approximately equal expected representation. In Stage 3,
    the pool is restricted to recurring markers found among the strongest feasible
    panels, and representation is balanced within that pool.
    """
    pool = list(marker_pool or BIOMARKERS)
    if len(pool) < PANEL_SIZE:
        pool = list(BIOMARKERS)
    counts = {b: int((prior_counts or {}).get(b, 0)) for b in pool}
    selected = []
    selected_set = set(used)
    max_attempts = max(100000, count * 1000)

    while len(selected) < count:
        candidates = set()
        attempts = 0
        while len(candidates) < CANDIDATE_POOL_SIZE and attempts < max_attempts:
            attempts += 1
            candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            if candidate not in selected_set:
                candidates.add(candidate)
        if not candidates:
            # This is only expected if the candidate space is nearly exhausted.
            candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            if candidate in selected_set:
                # Fall back to the full marker universe to maintain the exact budget.
                pool = list(BIOMARKERS)
                candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
                while candidate in selected_set:
                    candidate = canonical_panel(rng.sample(pool, PANEL_SIZE))
            candidates.add(candidate)

        def balance_diversity_key(candidate):
            projected = dict(counts)
            for b in candidate:
                if b in projected:
                    projected[b] += 1
            values = list(projected.values())
            # First balance total appearances; then minimize overlap with earlier
            # panels from this stage; randomized final tie-break avoids ordering bias.
            imbalance = float(np.std(values)) if values else 0.0
            max_count = max(values) if values else 0
            if selected:
                overlaps = [len(set(candidate) & set(old)) for old in selected]
                mean_overlap = float(np.mean(overlaps))
                max_overlap = max(overlaps)
            else:
                mean_overlap = max_overlap = 0.0
            return (imbalance, max_count, mean_overlap, max_overlap, rng.random())

        # IMPORTANT FOR REPRODUCIBILITY:
        # `candidates` is a set, whose iteration order can vary between Python
        # processes. The key function consumes seeded RNG values for tie-breaking;
        # therefore, iterate in a stable sorted order before calling it.
        chosen = min(sorted(candidates), key=balance_diversity_key)
        selected.append(chosen)
        selected_set.add(chosen)
        # Keep the shared cross-stage registry synchronized. Without this line,
        # panels generated in this stage are not counted in `used`, even though
        # they are appended to all_rows later.
        used.add(chosen)
        for b in chosen:
            if b in counts:
                counts[b] += 1

    return selected


def top_feasible_rows(panel_rows):
    feasible = [item for item in panel_rows if item[1]["specificity"] > (1.0 - FPR) + 1e-12]
    feasible.sort(key=lambda item: item[1]["macro_sensitivity"], reverse=True)
    if not feasible:
        return []
    n_top = max(1, int(math.ceil(len(feasible) * TOP_PANEL_FRACTION)))
    return feasible[:n_top]


def biomarker_weights(panel_rows):
    """Estimate recurring-marker weights from the strongest feasible panels."""
    top = top_feasible_rows(panel_rows)
    counts = {b: 1.0 for b in BIOMARKERS}  # small baseline keeps exploration possible
    for panel, _summary in top:
        for biomarker in panel:
            counts[biomarker] += 1.0
    return counts


def weighted_panel(rng, weights):
    """Sample a unique-size panel without replacement, weighted by marker frequency."""
    available = list(BIOMARKERS)
    chosen = []
    for _ in range(PANEL_SIZE):
        total = sum(max(0.001, weights[b]) for b in available)
        pick = rng.random() * total
        cumulative = 0.0
        selected = available[-1]
        for biomarker in available:
            cumulative += max(0.001, weights[biomarker])
            if cumulative >= pick:
                selected = biomarker
                break
        chosen.append(selected)
        available.remove(selected)
    return canonical_panel(chosen)


def make_unique_weighted_panels(rng, count, used, weights):
    panels = []
    attempts = 0
    max_attempts = max(10000, count * 200)
    while len(panels) < count and attempts < max_attempts:
        attempts += 1
        candidate = weighted_panel(rng, weights)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    while len(panels) < count:
        candidate = random_panel(rng)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    return panels


def make_neighborhood_panels(rng, count, used, panel_rows):
    """Generate swaps around the best feasible panels, with random unique fill."""
    feasible = [item for item in panel_rows
                if item[1]["specificity"] > (1.0 - FPR) + 1e-12
                and np.isfinite(item[1]["macro_sensitivity"])]
    ranked_source = feasible if feasible else panel_rows
    ranked = sorted(ranked_source,
                    key=lambda item: item[1]["macro_sensitivity"]
                    if np.isfinite(item[1]["macro_sensitivity"]) else -1.0,
                    reverse=True)
    panels = []
    for panel, _summary in ranked:
        candidates = local_neighbors(panel)
        rng.shuffle(candidates)
        for candidate in candidates:
            if candidate not in used:
                used.add(candidate)
                panels.append(candidate)
                if len(panels) >= count:
                    return panels
    while len(panels) < count:
        candidate = random_panel(rng)
        if candidate not in used:
            used.add(candidate)
            panels.append(candidate)
    return panels

def score_panel(df, reference_df, panel, cache):
    key = panel_key(panel)
    if key not in cache:
        summary, _ = evaluate_panel(df, reference_df, panel, save_features=False)
        cache[key] = summary
    return cache[key]


def rank_key(summary):
    # The configured specificity target is enforced as a hard feasibility
    # constraint in evaluate_panel. Among feasible panels, rank by macro-sensitivity.
    return summary["macro_sensitivity"]


def local_neighbors(panel):
    panel_set = set(panel)
    not_in_panel = [b for b in BIOMARKERS if b not in panel_set]
    neighbors = set()
    for removed in panel:
        for added in not_in_panel:
            candidate = canonical_panel((panel_set - {removed}) | {added})
            if len(candidate) == PANEL_SIZE:
                neighbors.add(candidate)
    return sorted(
        neighbors,
        key=lambda p: tuple(BIOMARKERS.index(x) for x in p),
    )



def Optimized_Biomarker_Selection(n_reference_normal):
    """Run the staged, specificity-constrained TDA biomarker-selection procedure.

    Dataset-specific values are configured in the USER-DEFINED SETTINGS section.
    Returns the selected biomarker panel and its calibration summary.
    """
    validate_split_fractions()
    bound_values = (K1_MIN, K1_MAX, K2_MIN, K2_MAX)
    bounds_provided = [value is not None for value in bound_values]
    if any(bounds_provided) and not all(bounds_provided):
        raise ValueError(
            "Provide all four bounds (K1_MIN, K1_MAX, K2_MIN, K2_MAX), or leave all four as None."
        )
    if all(bounds_provided):
        for name, value in zip(("K1_MIN", "K1_MAX", "K2_MIN", "K2_MAX"), bound_values):
            if not isinstance(value, (int, float)) or not np.isfinite(value):
                raise ValueError(f"{name} must be a finite number or None when bounds are disabled.")
        if K1_MIN > K1_MAX:
            raise ValueError(f"K1_MIN ({K1_MIN}) must be <= K1_MAX ({K1_MAX}).")
        if K2_MIN > K2_MAX:
            raise ValueError(f"K2_MIN ({K2_MIN}) must be <= K2_MAX ({K2_MAX}).")
    if not isinstance(MAX_THRESHOLD_CANDIDATES, int) or MAX_THRESHOLD_CANDIDATES < 2:
        raise ValueError("MAX_THRESHOLD_CANDIDATES must be an integer >= 2.")
    # Do not validate CANCER_TYPES here: it is populated automatically from the
    # workbook Cohort column inside create_split_dataframe(), which runs below.
    # cohort_order is only a preferred sorting map and must not control which
    # cancer cohorts are included in analysis.
    if len(BIOMARKERS) != len(set(BIOMARKERS)):
        raise ValueError("BIOMARKERS must contain unique names.")
    if PANEL_SIZE < 1 or PANEL_SIZE > len(BIOMARKERS):
        raise ValueError("PANEL_SIZE must be between 1 and the number of BIOMARKERS.")
    if sum(PHASE_BUDGETS.values()) != TOTAL_PANEL_BUDGET:
        raise ValueError("The phase budgets must sum exactly to TOTAL_PANEL_BUDGET.")

    df = create_split_dataframe(clinical_data_path, sheet_name, n_reference_normal)
    reference_df = df[
        get_role_mask(df, "Reference")
        & df[COHORT_COL].astype(str).str.strip().str.casefold().eq(NORMAL_COHORT.casefold())
    ].copy()
    if len(reference_df) != n_reference_normal:
        raise ValueError(
            f"Expected {n_reference_normal} reference normals; found {len(reference_df)}."
        )

    rng = random.Random(RANDOM_SEED)
    cache = {}
    used = set()
    all_rows = []
    stage_by_panel = {}

    def evaluate_phase(label, panels, budget):
        if len(panels) != budget:
            raise RuntimeError(f"{label}: expected {budget} panels, got {len(panels)}")
        phase_rows = []
        for i, panel in enumerate(panels, start=1):
            summary = score_panel(df, reference_df, panel, cache)
            phase_rows.append((panel, summary))
            all_rows.append((panel, summary))
            stage_by_panel[panel] = label

            # Compact progress only at the user-configured interval.
            completed = len(all_rows)
            if completed % PROGRESS_INTERVAL == 0 or completed == TOTAL_PANEL_BUDGET:
                feasible_so_far = [
                    item for item in all_rows
                    if item[1]["specificity"] > (1.0 - FPR) + 1e-12
                    and np.isfinite(item[1]["macro_sensitivity"])
                ]
                if feasible_so_far:
                    progress_best_panel, progress_best = max(
                        feasible_so_far,
                        key=lambda item: (
                            item[1]["macro_sensitivity"],
                            item[1]["overall_cancer_sensitivity"],
                            item[1]["specificity"],
                        ),
                    )
                    print(
                        f"Search {completed}/{TOTAL_PANEL_BUDGET} | Stage {label} | "
                        f"best panel so far: {list(progress_best_panel)} | "
                        f"calibration specificity: {progress_best['specificity']:.4%} | "
                        f"calibration macro-sensitivity: {progress_best['macro_sensitivity']:.4%}"
                    )
                else:
                    print(
                        f"Search {completed}/{TOTAL_PANEL_BUDGET} | Stage {label} | "
                        "no panel has met the calibration specificity constraint yet"
                    )
        return phase_rows

    print(f"\nReference normals: {n_reference_normal} | Searching {TOTAL_PANEL_BUDGET} unique panels")

    # STAGE 1: all 42 biomarkers are treated equally at the outset. A greedy
    # balancing criterion keeps their appearance counts as even as possible,
    # while minimizing overlap between panels. No outcome-guided marker weights.
    stage1_panels = balanced_diverse_panels(
        rng, PHASE_BUDGETS["balanced_diverse_exploration"], used,
        marker_pool=BIOMARKERS, prior_counts={b: 0 for b in BIOMARKERS}
    )
    used.update(stage1_panels)
    evaluate_phase("1/4 balanced diverse exploration", stage1_panels,
                   PHASE_BUDGETS["balanced_diverse_exploration"])

    # STAGE 2: rank only feasible panels by macro-sensitivity, then estimate marker
    # recurrence from the top feasible fraction and use it to guide new combinations.
    weights = biomarker_weights(all_rows)
    stage2_panels = make_unique_weighted_panels(
        rng, PHASE_BUDGETS["sensitivity_guided_search"], used, weights
    )
    evaluate_phase("2/4 sensitivity-guided search", stage2_panels,
                   PHASE_BUDGETS["sensitivity_guided_search"])

    # STAGE 3: identify recurring biomarkers in the top feasible panels from Stages
    # 1–2, then repeat balanced/low-overlap exploration within the recurrence-informed
    # pool. This keeps the marker frequencies balanced within that selected pool.
    top_before_stage3 = top_feasible_rows(all_rows)
    recurrence_counts = {b: 0 for b in BIOMARKERS}
    for panel, _summary in top_before_stage3:
        for b in panel:
            recurrence_counts[b] += 1
    recurring_pool = [b for b, _count in sorted(
        recurrence_counts.items(), key=lambda item: (-item[1], BIOMARKERS.index(item[0]))
    )[:RECURRENCE_POOL_SIZE]]
    if len(recurring_pool) < PANEL_SIZE:
        recurring_pool = list(BIOMARKERS)
    stage3_panels = balanced_diverse_panels(
        rng, PHASE_BUDGETS["recurrence_informed_balanced_search"], used,
        marker_pool=recurring_pool, prior_counts={b: 0 for b in recurring_pool}
    )
    evaluate_phase("3/4 recurrence-informed balanced search", stage3_panels,
                   PHASE_BUDGETS["recurrence_informed_balanced_search"])

    # STAGE 4: one-biomarker swaps around the strongest panels from all previous
    # stages, with random unique panels as a fallback. The final selection below
    # still compares every feasible panel from all four stages.
    stage4_panels = make_neighborhood_panels(
        rng, PHASE_BUDGETS["best_panel_neighborhoods"], used, all_rows
    )
    evaluate_phase("4/4 best-panel neighborhood search", stage4_panels,
                   PHASE_BUDGETS["best_panel_neighborhoods"])

    if len(used) != TOTAL_PANEL_BUDGET or len(all_rows) != TOTAL_PANEL_BUDGET:
        raise RuntimeError(
            f"Search budget error: evaluated {len(all_rows)} panels, expected {TOTAL_PANEL_BUDGET}."
        )
    if len(set(panel for panel, _ in all_rows)) != TOTAL_PANEL_BUDGET:
        raise RuntimeError("Duplicate panels were evaluated; every panel must be unique.")

    # IMPORTANT: the winning panel is selected across all four stages, not only Stage 4.
    specificity_feasible_rows = [
        item for item in all_rows
        if item[1]["specificity"] > (1.0 - FPR) + 1e-12
        and np.isfinite(item[1]["macro_sensitivity"])
    ]
    finite_rows = [
        item for item in all_rows
        if np.isfinite(item[1]["macro_sensitivity"])
    ]
    if not finite_rows:
        raise RuntimeError("All evaluated panels had undefined macro-sensitivity.")
    # Detection-based panel selection:
    # 1. Keep only panels whose pooled Calibration 1 + Calibration 2 normal specificity is strictly above 1 - FPR.
    # 2. Among those feasible panels, select the panel with the highest
    #    macro-sensitivity across the cancer cohorts.
    # 3. Do not require macro-sensitivity itself to exceed 99%; it is the
    #    optimization objective, while specificity is the hard constraint.
    if specificity_feasible_rows:
        ranked_rows = sorted(
            specificity_feasible_rows,
            key=lambda item: (
                item[1]["macro_sensitivity"],
                item[1]["overall_cancer_sensitivity"],
                item[1]["specificity"],
            ),
            reverse=True,
        )
    else:
        # This is a diagnostic fallback only. Thresholds just above the maximum
        # pooled-normal scores yield 100% pooled-normal specificity.
        ranked_rows = sorted(
            finite_rows,
            key=lambda item: (
                item[1]["macro_sensitivity"],
                item[1]["overall_cancer_sensitivity"],
                item[1]["specificity"],
            ),
            reverse=True,
        )
        print(
            "\nWARNING: No panel met the strict pooled Calibration 1 + Calibration 2 "
            "specificity constraint. The selected panel is diagnostic only."
        )
    best_panel, _ = ranked_rows[0]

    # Reuse the already-computed summary for the winning panel; no duplicate calibration evaluation.
    best_summary = ranked_rows[0][1]
    print("\nBEST BIOMARKER PANEL — SELECTED ACROSS ALL FOUR STAGES")
    print(f"Reference normals: {n_reference_normal}")
    print(f"Winning stage: {stage_by_panel.get(best_panel, 'unknown')}")
    print(f"Panel: {list(best_panel)!r}")
    print(f"Calibration specificity (pooled Calibration 1 + Calibration 2 normals): {best_summary['specificity']:.4%}")
    print(f"Calibration 1 normal specificity: {best_summary['calibration1_specificity']:.4%}")
    print(f"Calibration 2 normal specificity: {best_summary['calibration2_specificity']:.4%}")
    print(f"Calibration macro-sensitivity: {best_summary['macro_sensitivity']:.4%}")
    print(f"Regression Difference Percentage threshold (k1): {best_summary['regression_threshold']:.8g}")
    print(f"Distance %(1) threshold (k2): {best_summary['distance_threshold']:.8g}")
    print(f"Meets pooled calibration specificity target: {best_summary['specificity'] > (1.0 - FPR) + 1e-12}")
    return {
        "reference_normals": n_reference_normal,
        "panel": list(best_panel),
        "winning_stage": stage_by_panel.get(best_panel, 'unknown'),
        "calibration_specificity": best_summary["specificity"],
        "calibration1_specificity": best_summary["calibration1_specificity"],
        "calibration2_specificity": best_summary["calibration2_specificity"],
        "calibration_macro_sensitivity": best_summary["macro_sensitivity"],
    }





if __name__ == "__main__":
    if not PANEL_SIZES or len(set(PANEL_SIZES)) != len(PANEL_SIZES):
        raise ValueError("PANEL_SIZES must be a non-empty list of unique positive integers.")
    if any(not isinstance(size, int) or size < 1 or size > len(BIOMARKERS) for size in PANEL_SIZES):
        raise ValueError("Every panel size must be an integer between 1 and the number of biomarkers.")
    if not isinstance(N_REFERENCE_NORMAL, int) or N_REFERENCE_NORMAL < 2:
        raise ValueError("N_REFERENCE_NORMAL must be an integer >= 2.")

    all_run_results = []
    for run_number, panel_size in enumerate(PANEL_SIZES, start=1):
        # PANEL_SIZE is intentionally changed for each independent run; all
        # search stages, thresholds, specificity constraints, and ranking rules
        # remain unchanged.
        PANEL_SIZE = panel_size
        print("\n" + "#" * 78)
        print(
            f"RUN {run_number}/{len(PANEL_SIZES)} — panel size = {panel_size} | "
            f"reference normals = {N_REFERENCE_NORMAL}"
        )
        print("#" * 78)
        result = Optimized_Biomarker_Selection(N_REFERENCE_NORMAL)
        result["panel_size"] = panel_size
        all_run_results.append(result)

    print("\n" + "=" * 78)
    print("SUMMARY ACROSS PANEL SIZES")
    print("=" * 78)
    for result in all_run_results:
        print(
            f"Panel size={result['panel_size']} | "
            f"Reference n={result['reference_normals']} | "
            f"Calibration specificity={result['calibration_specificity']:.4%} | "
            f"Calibration macro-sensitivity={result['calibration_macro_sensitivity']:.4%} | "
            f"Panel={result['panel']}"
        )
```

```python
"""
Variance Inflation Factor (VIF) Assessment of the TDAcancer-detect Regression Predictors.

Purpose
-------
Calculate Variance Inflation Factors (VIFs) for the four predictors used in
the regression model, using TDA features from healthy patients in the
"Regression Fit Calibration 1" group.

The assessment is restricted to VIF calculation; no regression fitting,
additional diagnostic procedures, threshold selection, cancer detection,
test-set evaluation, or plotting is carried out.

The first 59 Normal patients in the original workbook are retained as the
fixed TDA reference cohort. The workbook is then partitioned using the same
preprocessing settings as the supplied analysis script.
"""

from pathlib import Path

import numpy as np
import pandas as pd
import gudhi
from statsmodels.stats.outliers_influence import variance_inflation_factor


# ---------------------------------------------------------------------
# USER SETTINGS — retained from the supplied analysis script
# ---------------------------------------------------------------------

ORIGINAL_INPUT_FILE = Path(
    r"D:\Cancer Detection folder\Biomarker Selection"
    r"\clinical cancer validation data.xlsx"
)

UPDATED_INPUT_FILE = ORIGINAL_INPUT_FILE.with_name(
    "clinical cancer validation data_Updated.xlsx"
)

SHEET_NAME = "Normal and Cancer"

# Normal-patient partition settings retained from the supplied code.
CALIBRATION1_FRACTION = 0.15
CALIBRATION2_FRACTION = 0.50
SPLIT_RANDOM_SEED = 20261017

COHORT_COL = "Cohort"
PATIENT_ID_COL = "Patient ID #"
DATA_TYPE_COL = "Data Type"
NORMAL_LABEL = "normal"

REFERENCE_ROLE = "Reference"
CAL1_ROLE = "Regression Fit Calibration 1"
CAL2_ROLE = "Biomarker-Panel Calibration and Selection"
TEST_ROLE = "Test"

N_REFERENCE_NORMAL = 59

# The exact biomarker panel used in the supplied code.
PANEL = [
    "sIL-6Ra",
    "CA-125",
    "CA 19-9",
    "IL-6",
    "IL-8",
    "OPN",
    "TRAIL",
    "Total PSA",
]

# Exact predictors used in the regression model under verification.
REGRESSION_PREDICTORS = [
    "0-Dim Hole Min Life Range",
    "0-Dim Hole Max Life Range",
    "Median_Death_0D",
    "Std_Death_0D",
]

# Save the VIF results in a dedicated output folder.
OUTPUT_DIR = ORIGINAL_INPUT_FILE.parent / "Regression_VIF_Analysis"
OUTPUT_EXCEL = OUTPUT_DIR / "Regression_Predictor_VIF_Results.xlsx"
OUTPUT_CSV = OUTPUT_DIR / "Regression_Predictor_VIF_Results.csv"


# ---------------------------------------------------------------------
# WORKBOOK PREPROCESSING — same split rules as the supplied code
# ---------------------------------------------------------------------

def normalized_text(series):
    return series.astype(str).str.strip().str.casefold()


def create_updated_workbook():
    """
    Recreate the updated workbook using the same reference and partition
    rules as the supplied analysis script.
    """
    if not ORIGINAL_INPUT_FILE.exists():
        raise FileNotFoundError(
            f"Original workbook not found: {ORIGINAL_INPUT_FILE}\n"
            "Please update ORIGINAL_INPUT_FILE to the correct workbook path."
        )

    if not (0 <= CALIBRATION1_FRACTION <= 1):
        raise ValueError("CALIBRATION1_FRACTION must be between 0 and 1.")
    if not (0 <= CALIBRATION2_FRACTION <= 1):
        raise ValueError("CALIBRATION2_FRACTION must be between 0 and 1.")
    if CALIBRATION1_FRACTION + CALIBRATION2_FRACTION >= 1:
        raise ValueError(
            "The Calibration 1 and Calibration 2 fractions must sum to less than 1."
        )

    all_sheets = pd.read_excel(ORIGINAL_INPUT_FILE, sheet_name=None)
    if SHEET_NAME not in all_sheets:
        raise ValueError(
            f"Sheet {SHEET_NAME!r} not found. Available sheets: {list(all_sheets)}"
        )

    df = all_sheets[SHEET_NAME].copy()

    required_columns = [COHORT_COL, *PANEL]
    missing = [column for column in required_columns if column not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns in workbook: {missing}")

    if DATA_TYPE_COL in df.columns:
        df = df.drop(columns=[DATA_TYPE_COL])

    cohort = normalized_text(df[COHORT_COL])
    if df[COHORT_COL].isna().any() or cohort.eq("").any():
        raise ValueError(f"Missing or blank values found in {COHORT_COL!r}.")

    normal_mask = cohort.eq(NORMAL_LABEL.casefold())
    normal_indices = df.index[normal_mask].tolist()

    if len(normal_indices) <= N_REFERENCE_NORMAL:
        raise ValueError(
            f"More than {N_REFERENCE_NORMAL} normal patients are required; "
            f"found {len(normal_indices)}."
        )

    role = {}

    # Keep the first 59 Normal patients in original workbook order as Reference.
    reference_indices = normal_indices[:N_REFERENCE_NORMAL]
    for index in reference_indices:
        role[index] = REFERENCE_ROLE

    rng = np.random.RandomState(SPLIT_RANDOM_SEED)

    # Split remaining healthy patients using the original fractions.
    remaining_normals = list(normal_indices[N_REFERENCE_NORMAL:])
    rng.shuffle(remaining_normals)

    n_cal1 = int(np.floor(len(remaining_normals) * CALIBRATION1_FRACTION))
    n_cal2 = int(np.floor(len(remaining_normals) * CALIBRATION2_FRACTION))

    for index in remaining_normals[:n_cal1]:
        role[index] = CAL1_ROLE
    for index in remaining_normals[n_cal1:n_cal1 + n_cal2]:
        role[index] = CAL2_ROLE
    for index in remaining_normals[n_cal1 + n_cal2:]:
        role[index] = TEST_ROLE

    # Cancer patients are assigned to Calibration 2 or Test, as in source code.
    cancer_cohorts = list(dict.fromkeys(cohort.loc[~normal_mask].tolist()))
    for cancer in cancer_cohorts:
        indices = df.index[cohort.eq(cancer)].tolist()
        shuffled_indices = list(indices)
        rng.shuffle(shuffled_indices)

        n_cal2_cancer = int(np.floor(
            len(shuffled_indices) * CALIBRATION2_FRACTION
        ))

        for index in shuffled_indices[:n_cal2_cancer]:
            role[index] = CAL2_ROLE
        for index in shuffled_indices[n_cal2_cancer:]:
            role[index] = TEST_ROLE

    df[DATA_TYPE_COL] = pd.Series(role)

    role_order = {
        REFERENCE_ROLE: 0,
        CAL1_ROLE: 1,
        CAL2_ROLE: 2,
        TEST_ROLE: 3,
    }

    df["_original_order"] = np.arange(len(df))
    df["_role_order"] = df[DATA_TYPE_COL].map(role_order)
    df = (
        df.sort_values(["_role_order", "_original_order"], kind="stable")
        .drop(columns=["_original_order", "_role_order"])
        .reset_index(drop=True)
    )

    all_sheets[SHEET_NAME] = df
    UPDATED_INPUT_FILE.parent.mkdir(parents=True, exist_ok=True)

    with pd.ExcelWriter(UPDATED_INPUT_FILE, engine="openpyxl") as writer:
        for sheet_name, sheet_df in all_sheets.items():
            sheet_df.to_excel(
                writer,
                sheet_name=str(sheet_name)[:31],
                index=False,
            )

    print("Updated workbook created:", UPDATED_INPUT_FILE)
    print(f"Fixed Reference normal patients: {len(reference_indices)}")
    print(f"Regression Fit Calibration 1 healthy patients: {n_cal1}")
    print("\nCohort x Data Type counts:")
    print(pd.crosstab(df[COHORT_COL], df[DATA_TYPE_COL]).to_string())


# ---------------------------------------------------------------------
# TDA FEATURE CALCULATION — predictors used in the regression model
# ---------------------------------------------------------------------

def validate_biomarkers(frame, description):
    values = frame[PANEL].apply(pd.to_numeric, errors="coerce")

    if values.isna().any().any():
        raise ValueError(
            f"{description} contains missing or non-numeric biomarker values."
        )

    array = values.to_numpy(dtype=float)

    if not np.isfinite(array).all():
        raise ValueError(f"{description} contains NaN or infinite values.")

    if (array <= 0).any():
        raise ValueError(
            f"{description} contains zero or negative biomarker values. "
            "Log transformation requires strictly positive values."
        )

    return values


def calculate_regression_predictors(reference_df, patient_row):
    """
    Calculate only the four 0D TDA predictors used by the regression model.
    """
    combined = pd.concat(
        [
            reference_df[PANEL].astype(float),
            patient_row[PANEL].astype(float).to_frame().T,
        ],
        ignore_index=True,
    )

    values = combined.to_numpy(dtype=float)

    if not np.isfinite(values).all() or (values <= 0).any():
        raise ValueError(
            f"Invalid biomarker values for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}."
        )

    logged = np.log(values)
    correlation = np.corrcoef(logged, rowvar=False)

    if not np.isfinite(correlation).all():
        raise ValueError(
            f"Invalid correlation matrix for patient "
            f"{patient_row.get(PATIENT_ID_COL, 'unknown')}. "
            "Check for constant biomarker columns."
        )

    dissimilarity = 1.0 - np.abs(correlation)
    dissimilarity = (dissimilarity + dissimilarity.T) / 2.0
    np.fill_diagonal(dissimilarity, 0.0)
    dissimilarity = np.clip(dissimilarity, 0.0, 1.0)

    rips = gudhi.RipsComplex(distance_matrix=dissimilarity)
    simplex_tree = rips.create_simplex_tree(max_dimension=2)
    persistence = simplex_tree.persistence()

    finite_0d = [
        (birth, death)
        for dimension, (birth, death) in persistence
        if dimension == 0 and np.isfinite(death)
    ]

    if not finite_0d:
        return {
            "0-Dim Hole Min Life Range": 0.0,
            "0-Dim Hole Max Life Range": 0.0,
            "Median_Death_0D": 0.0,
            "Std_Death_0D": 0.0,
        }

    deaths_0d = np.asarray([death for _, death in finite_0d], dtype=float)
    lifetimes_0d = np.asarray(
        [death - birth for birth, death in finite_0d],
        dtype=float,
    )

    return {
        "0-Dim Hole Min Life Range": float(np.min(lifetimes_0d)),
        "0-Dim Hole Max Life Range": float(np.max(lifetimes_0d)),
        "Median_Death_0D": float(np.median(deaths_0d)),
        "Std_Death_0D": float(np.std(deaths_0d)),
    }


def build_calibration1_predictors(reference_df, calibration1_df):
    records = []

    for _, patient in calibration1_df.iterrows():
        row = {
            PATIENT_ID_COL: patient.get(PATIENT_ID_COL, np.nan),
        }
        row.update(calculate_regression_predictors(reference_df, patient))
        records.append(row)

    result = pd.DataFrame(records)

    if result.empty:
        raise ValueError("No healthy Calibration 1 patients were available for VIF calculation.")

    return result


# ---------------------------------------------------------------------
# VARIANCE INFLATION FACTOR (VIF) CALCULATION
# ---------------------------------------------------------------------

def calculate_vif(predictor_data):
    """
    Calculate VIF for the four regression predictors.
    An intercept is included in each auxiliary regression, as is standard
    for VIF calculation in a regression model with an intercept.
    """
    X = predictor_data[REGRESSION_PREDICTORS].astype(float)

    if not np.isfinite(X.to_numpy()).all():
        raise ValueError("Non-finite values found in the regression predictors.")

    if len(X) <= len(REGRESSION_PREDICTORS) + 1:
        raise ValueError(
            "There are too few Calibration 1 observations to calculate VIF "
            "for the four predictors with an intercept."
        )

    # Add intercept for the auxiliary regressions used by VIF.
    X_with_intercept = np.column_stack([
        np.ones(len(X), dtype=float),
        X.to_numpy(dtype=float),
    ])

    vif_results = []

    for column_index, predictor in enumerate(REGRESSION_PREDICTORS, start=1):
        try:
            vif_value = float(
                variance_inflation_factor(X_with_intercept, column_index)
            )
        except Exception:
            vif_value = np.inf

        vif_results.append({
            "Regression Predictor": predictor,
            "VIF": vif_value,
        })

    return pd.DataFrame(vif_results)


# ---------------------------------------------------------------------
# MAIN — VIF assessment
# ---------------------------------------------------------------------

def main():
    create_updated_workbook()

    all_sheets = pd.read_excel(UPDATED_INPUT_FILE, sheet_name=None)
    if SHEET_NAME not in all_sheets:
        raise ValueError(f"Sheet {SHEET_NAME!r} not found in updated workbook.")

    df = all_sheets[SHEET_NAME].copy()

    required_columns = [COHORT_COL, DATA_TYPE_COL, *PANEL]
    missing = [column for column in required_columns if column not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns: {missing}")

    cohort = normalized_text(df[COHORT_COL])
    role = normalized_text(df[DATA_TYPE_COL])

    # Select first 59 Normal patients in workbook order as fixed reference.
    normal_positions = np.flatnonzero(cohort.eq(NORMAL_LABEL.casefold()).to_numpy())
    if len(normal_positions) < N_REFERENCE_NORMAL:
        raise ValueError(
            f"At least {N_REFERENCE_NORMAL} Normal patients are required."
        )

    reference_indices = df.index[normal_positions[:N_REFERENCE_NORMAL]]
    reference_df = df.loc[reference_indices].copy()

    # Calibration 1 consists only of healthy patients, excluding Reference.
    reference_index_set = set(reference_indices)
    is_reference = df.index.to_series().isin(reference_index_set).to_numpy()

    calibration1_mask = (
        role.eq(CAL1_ROLE.casefold()).to_numpy()
        & cohort.eq(NORMAL_LABEL.casefold()).to_numpy()
        & ~is_reference
    )
    calibration1_df = df.loc[calibration1_mask].copy()

    if calibration1_df.empty:
        raise ValueError(
            f"No healthy patients labelled {CAL1_ROLE!r} were found."
        )

    reference_values = validate_biomarkers(
        reference_df,
        "Fixed Reference cohort",
    )
    calibration1_values = validate_biomarkers(
        calibration1_df,
        "Regression Fit Calibration 1 healthy cohort",
    )

    reference_df.loc[:, PANEL] = reference_values
    calibration1_df.loc[:, PANEL] = calibration1_values

    print("\nCalculating the four regression predictors for healthy "
          "patients in Regression Fit Calibration 1...")
    predictor_data = build_calibration1_predictors(
        reference_df,
        calibration1_df,
    )

    # Calculate VIF for the four regression predictors.
    vif_results = calculate_vif(predictor_data)

    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    with pd.ExcelWriter(OUTPUT_EXCEL, engine="openpyxl") as writer:
        vif_results.to_excel(writer, sheet_name="VIF Results", index=False)

    vif_results.to_csv(OUTPUT_CSV, index=False)

    print("\n" + "=" * 65)
    print("VARIANCE INFLATION FACTOR (VIF) RESULTS")
    print("=" * 65)
    print(f"Reference cohort size: {len(reference_df)}")
    print(f"Calibration 1 healthy sample size: {len(predictor_data)}")
    print(vif_results.to_string(index=False))
    print("\nVIF results saved to:")
    print(OUTPUT_EXCEL)
    print(OUTPUT_CSV)
    print("\nVIF assessment completed.")


if __name__ == "__main__":
    main()
```
